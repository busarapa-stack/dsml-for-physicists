# บทที่ 8 การลดมิติและตัวแปรรวมทางฟิสิกส์: โน้ตบุ๊กสำหรับชั้นเรียน
โน้ตบุ๊กประกอบตำรา *วิทยาศาสตร์ข้อมูลและการเรียนรู้ของเครื่องสำหรับนักฟิสิกส์*

โน้ตบุ๊กนี้ใช้ในชั้นเรียนได้ด้วยไฟล์นี้ไฟล์เดียว ไม่ต้องดึงโค้ดจาก GitHub และไม่ต้องดาวน์โหลดข้อมูล เนื้อหาเรียงตามบทที่ 8 ในตำรา แบ่งเป็นสามคาบ คาบละประมาณ 60 นาที รวม 180 นาทีตามแผนเวลาในคู่มือผู้สอน หัวข้อที่มีป้าย **โค้ดที่ 8.N** ตรงกับตัวอย่างโค้ดเลขเดียวกันในตำรา

## วัตถุประสงค์
เมื่อทำโน้ตบุ๊กนี้จนจบ ผู้เรียนควรสามารถ

1. อธิบายว่าทำไมข้อมูลฟิสิกส์ที่มีมิติสูงจึงมักมีโครงสร้างมิติต่ำซ่อนอยู่ และการลดมิติ (dimensionality reduction) วัดจำนวนองศาเสรีที่มีความหมายของระบบได้อย่างไร
2. ใช้การวิเคราะห์องค์ประกอบหลัก (principal component analysis, PCA) อ่านสัดส่วนความแปรปรวน และเลือกจำนวนองค์ประกอบด้วยหลายเกณฑ์
3. อธิบายและตรวจด้วยข้อมูลว่าองค์ประกอบหลักของการแกว่งเชิงความร้อนในระบบฮาร์มอนิกคือโหมดปกติ และใช้ผลนี้ตีความสายพอลิเมอร์ตามแบบจำลองของเราส์
4. เลือกระหว่าง PCA การวิเคราะห์องค์ประกอบอิสระ (ICA) และการแยกตัวประกอบเมทริกซ์ไม่เป็นลบ (NMF) จากสมมติฐานทางฟิสิกส์ของข้อมูล
5. ใช้ t-SNE และ UMAP วาดภาพข้อมูล และบอกได้ว่าอะไรในภาพตีความได้และอะไรตีความไม่ได้
6. ประเมินคุณภาพของการลดมิติ และประมาณช่วงความเชื่อมั่นของค่าเจาะจงอย่างถูกต้องเมื่อข้อมูลมีสหสัมพันธ์ในตัว

## ข้อมูลในโน้ตบุ๊กนี้
ข้อมูลทุกชุดเป็น **ข้อมูลจำลอง (SYNTHETIC)** ที่สร้างด้วยโค้ดในตำรา และตัวเลขตรงกับตำรา (ยกเว้นผลของ t-SNE และ UMAP ซึ่งอาจต่างเล็กน้อยตามเครื่องและรุ่นของไลบรารี)

- **AFM:** ลักษณะเด่นเฉพาะที่ 21 ตัวจากแผนที่แรง AFM จำลองของบทที่ 7 (ฟังก์ชัน `make_scaffold_map` จากโค้ดที่ 7.7) ไม่ใช่ข้อมูลหรือผลของงานวิจัย
- **สัญญาณผสม:** สัญญาณเวลาสามแหล่งที่ผสมกันเป็นสามช่อง (โค้ดที่ 8.3) และสเปกตรัมผสมในแบบฝึกหัด E8.3
- **โซ่ฮาร์มอนิก:** โครงรูปที่สุ่มจากการแจกแจงโบลต์ซมันน์ (แบบฝึก E8.2)
- **สายพอลิเมอร์:** เส้นทางการเคลื่อนที่ตามแบบจำลองของเราส์ ซึ่งสุ่มแม่นตรงในฐานของโหมด (โค้ดที่ 8.5)

หัวกราฟที่มาจากข้อมูลจำลองมีคำว่า SYNTHETIC กำกับ ตัวเลขใช้เพื่อการเรียนวิธีการ ไม่ใช่ผลการวิจัย

เซลล์ถัดไปนิยามฟังก์ชัน `make_scaffold_map` ซึ่งสร้างแผนที่แรง AFM จำลองขนาด $128 \times 128$ พิกเซล (ช่อง H, S และ A) พร้อมแผนที่อ้างอิงของโครงร่างและแก้ว โค้ดเป็นชุดเดียวกับในโค้ดที่ 7.7 ของบทที่ 7 ถูกซ่อนไว้ ดับเบิลคลิกเพื่อดูได้ โค้ดที่ 8.1 และ 8.4 ใช้ฟังก์ชันนี้

```python
#@title ฟังก์ชัน make_scaffold_map จากโค้ดที่ 7.7 (ดับเบิลคลิกเพื่อดูโค้ด)
from scipy.ndimage import gaussian_filter

def make_scaffold_map(n=128, seed=2026):
    """Toy AFM force map: scaffold (label 1) vs substrate (paraffin or glass).
    Channels: height H (nm), contact stiffness S (N/m), adhesion A (nN)."""
    rng = np.random.default_rng(seed)
    phi = gaussian_filter(rng.normal(size=(n, n)), sigma=6); phi /= phi.std()
    scaffold = phi > -0.55                                    # about 70 % scaffold
    g = gaussian_filter(rng.normal(size=(n, n)), sigma=4); g /= g.std()
    glass = (~scaffold) & (g > 0.8)                           # exposed glass
    het = gaussian_filter(rng.normal(size=(n, n)), sigma=3); het /= het.std()
    H = np.where(scaffold, 250 + 72 * het, np.where(glass, 20.0, 130.0))
    S = np.where(scaffold, 0.8 + 0.18 * het, np.where(glass, 1.5, 0.5))
    A = np.where(scaffold, 4.5 - 0.72 * het, np.where(glass, 2.5, 6.5))
    H, S, A = (gaussian_filter(m, sigma=1.0) for m in (H, S, A))   # tip blur
    heavy = lambda s: s * rng.standard_t(df=3, size=(n, n))        # heavy tails
    H += heavy(30.0); S += heavy(0.09); A += heavy(0.65)
    X = np.column_stack([H.ravel(), S.ravel(), A.ravel()])
    return X, scaffold.ravel().astype(int), glass.ravel()

print('make_scaffold_map() พร้อมใช้ (แผนที่ AFM จำลอง, SYNTHETIC)')
```

# ส่วนที่ 1 แรงจูงใจ คำสาปของมิติ และองศาเสรีทางฟิสิกส์ (หัวข้อ 8.1)

### แนวคิดสำคัญ
บทที่ 7 จบลงด้วยคำเตือนสองข้อ ข้อแรก เมื่อข้อมูลมีมิติสูง ระยะทางระหว่างจุดจะเสียความหมาย ข้อที่สอง วิธีจัดกลุ่มแบ่งข้อมูลตามทิศทางที่ข้อมูลกระจายมากที่สุด ซึ่งอาจไม่ใช่ทิศทางที่เราสนใจ ทั้งสองข้อชี้ไปที่คำถามเดียวกัน คือ **ข้อมูลของเรามีโครงสร้างจริงกี่มิติ และมิติเหล่านั้นมีความหมายทางฟิสิกส์อะไร**

**การลดมิติ (dimensionality reduction)** คือการแทนข้อมูลที่มี $D$ ตัวแปรด้วยตัวแปรใหม่เพียง $d \ll D$ ตัว โดยยังรักษาโครงสร้างที่สำคัญไว้ได้ นักฟิสิกส์คุ้นกับความคิดนี้มานาน ผลึกที่มีอะตอมนับล้านตัวสั่นเป็นโหมดปกติไม่กี่แบบ สายพอลิเมอร์เคลื่อนไหวช้า ๆ ด้วยโหมดรวมไม่กี่โหมด และการเปลี่ยนสถานะของแม่เหล็กอธิบายได้ด้วยพารามิเตอร์อันดับตัวเดียว **การลดมิติที่ดีไม่ได้เพียงบีบข้อมูล แต่วัดจำนวนองศาเสรีที่มีความหมายของระบบ**

### คำสาปของมิติกับแมนิโฟลด์
บทที่ 7 แสดงว่าเมื่อจำนวนมิติเพิ่มขึ้น จุดที่ใกล้ที่สุดกับจุดที่ไกลที่สุดอยู่ห่างเกือบเท่ากัน ผลนี้จริงสำหรับข้อมูลที่ลักษณะเด่นเป็นอิสระต่อกันและกระจายเต็มปริภูมิ แต่ข้อมูลฟิสิกส์แทบไม่เคยเป็นเช่นนั้น ภาพใบหน้าขนาด $100 \times 100$ พิกเซลเป็นจุดในปริภูมิหนึ่งหมื่นมิติ แต่ภาพใบหน้าจริงทั้งหมดอยู่บนผิวบาง ๆ ที่กำหนดด้วยปัจจัยไม่กี่อย่าง ผิวมิติต่ำที่ฝังอยู่ในปริภูมิมิติสูงนี้เรียกว่า **แมนิโฟลด์ (manifold)** ตำแหน่งของมอนอเมอร์ 50 ตัวในสายพอลิเมอร์ให้ 150 พิกัด แต่มอนอเมอร์ที่ติดกันต้องอยู่ห่างกันราวหนึ่งความยาวพันธะเสมอ พิกัดจึงไม่อิสระต่อกัน การลดมิติหาฟังก์ชัน $\Phi: \mathbb{R}^D \to \mathbb{R}^d$ ที่ $d \ll D$ และรักษาโครงสร้างสำคัญไว้

### ประเภทของการลดมิติ
| กลุ่ม | ตัวแทน | สิ่งที่รักษาไว้ | เหมาะกับ |
|---|---|---|---|
| เชิงเส้น | PCA, ICA, NMF | ความแปรปรวนโดยรวม ความเป็นอิสระ หรือความไม่เป็นลบ | ตัวแปรสัมพันธ์กันแบบเชิงเส้น และงานที่ต้องตีความองค์ประกอบ |
| ไม่เชิงเส้น | t-SNE, UMAP | ความสัมพันธ์ของเพื่อนบ้านใกล้เคียง | วาดภาพข้อมูลในสองหรือสามมิติเพื่อสำรวจ |

# ส่วนที่ 2 การวิเคราะห์องค์ประกอบหลัก (หัวข้อ 8.2.1)

### แนวคิดสำคัญ
ลองนึกถึงการถ่ายเงาของวัตถุสามมิติบนผนัง ถ้าต้องการให้เงาบอกรูปร่างได้มากที่สุด เราจะหมุนด้านที่ยาวที่สุดให้ขนานกับผนัง **การวิเคราะห์องค์ประกอบหลัก (principal component analysis, PCA)** ทำแบบเดียวกันกับข้อมูล มันหาทิศทางที่ข้อมูลกระจายมากที่สุด แล้วฉายข้อมูลลงบนทิศทางนั้น

> **นิยาม (PCA)** ให้ $\mathbf X$ เป็นเมทริกซ์ข้อมูล $N \times D$ ที่ลบค่าเฉลี่ยของแต่ละหลักแล้ว และ $\mathbf C = \mathbf X^\top \mathbf X/(N-1)$ เป็นเมทริกซ์ความแปรปรวนร่วม องค์ประกอบหลักที่หนึ่งคือเวกเตอร์หนึ่งหน่วยที่ทำให้ความแปรปรวนของข้อมูลที่ฉายลงบนมันมากที่สุด
> $$\mathbf v_1 = \arg\max_{\lVert\mathbf v\rVert = 1} \mathbf v^\top \mathbf C \mathbf v$$
> องค์ประกอบถัดไปหาแบบเดียวกันโดยบังคับให้ตั้งฉากกับองค์ประกอบก่อนหน้า ผลเฉลยคือ **เวกเตอร์เจาะจง (eigenvector)** ของ $\mathbf C$ เรียงตาม **ค่าเจาะจง (eigenvalue)** จากมากไปน้อย และ $\lambda_k$ คือความแปรปรวนตามองค์ประกอบที่ $k$

ผลเฉลยได้จากตัวคูณลากรานจ์ ซึ่งให้ $\mathbf C\mathbf v = \lambda\mathbf v$ ทันที ทางเรขาคณิต PCA หมุนแกนไปตามแกนหลักของทรงรีที่ข้อมูลกระจาย ในแกนใหม่ ตัวแปรทุกตัวไม่สัมพันธ์กัน และ $\mathbf C$ กลายเป็นเมทริกซ์ทแยง ซึ่งเป็นปัญหาเดียวกับการหาแกนหลักของเทนเซอร์ความเฉื่อย

### ข้อมูลของโค้ดที่ 8.1
ในตำรา โค้ดที่ 8.1 ใช้เมทริกซ์ `X` แบบทั่วไป โน้ตบุ๊กนี้ใส่ลักษณะเด่นเฉพาะที่ 21 ตัวของแผนที่ AFM จำลองเป็นตัวอย่าง คือ ค่าของแต่ละช่อง (H, S, A) กับค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐานเฉพาะที่ในหน้าต่าง $3\times3$, $7\times7$ และ $15\times15$ พิกเซล ได้ $3 \times (1 + 2 \times 3) = 21$ ตัวต่อพิกเซล ลักษณะเด่นเหล่านี้มีหน่วยต่างกัน (nm, N/m, nN) จึงต้องปรับเป็นมาตรฐาน (ลักษณะเด่นชุดเดียวกันนี้กลับมาในโค้ดที่ 8.4)

```python
from scipy.ndimage import uniform_filter

_X_afm, _, _ = make_scaffold_map()
_maps = []
for _ch in range(3):
    _m = _X_afm[:, _ch].reshape(128, 128); _maps.append(_m)
    for _w in (3, 7, 15):
        _mu = uniform_filter(_m, _w)
        _maps += [_mu, np.sqrt(np.maximum(uniform_filter(_m * _m, _w) - _mu * _mu, 0))]
X = np.column_stack([_f.ravel() for _f in _maps])        # (16384, 21) in physical units
feat_names = [f"{c}{s}" for c in "HSA" for s in ["", "_mean3", "_sd3", "_mean7", "_sd7", "_mean15", "_sd15"]]
print("X:", X.shape)
```

### โค้ดที่ 8.1 — การวิเคราะห์องค์ประกอบหลักด้วย scikit-learn

```python
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

X_scaled = StandardScaler().fit_transform(X)     # X: (n_samples, n_features)
pca = PCA(n_components=10)
X_pca = pca.fit_transform(X_scaled)              # scores on the new axes
print("explained variance ratio:", pca.explained_variance_ratio_)
print("cumulative:", np.cumsum(pca.explained_variance_ratio_))
print("first component:", pca.components_[0])    # loadings on the original features
```

น้ำหนัก (loadings) ของตัวแปรเดิมในองค์ประกอบแรกอ่านง่ายขึ้นเมื่อเรียงตามชื่อ เซลล์ถัดไปแสดงเป็นแผนภูมิแท่ง

```python
fig, ax = plt.subplots(figsize=(10, 3))
ax.bar(feat_names, pca.components_[0]); ax.axhline(0, c='k', lw=0.8)
ax.set_ylabel('loading on PC1'); ax.set_title('PC1 of the 21 AFM features (SYNTHETIC)')
ax.tick_params(axis='x', rotation=90); plt.show()
```

องค์ประกอบแรกอธิบายความแปรปรวนราวร้อยละ 37 ค่าและค่าเฉลี่ยเฉพาะที่ของ H และ S มีเครื่องหมายเดียวกัน ของ A มีเครื่องหมายตรงข้าม ส่วนส่วนเบี่ยงเบนมาตรฐานเฉพาะที่แทบไม่มีน้ำหนักในองค์ประกอบนี้ นั่นคือองค์ประกอบนี้แยก "บริเวณสูงและแข็ง แรงยึดติดต่ำ" ออกจาก "บริเวณต่ำ นิ่ม แรงยึดติดสูง" ซึ่งสอดคล้องกับความต่างระหว่างเนื้อโครงร่างกับพาราฟินในบทที่ 7 การอ่านแบบนี้ใช้ **รูปแบบสัมพัทธ์** ของเครื่องหมาย ไม่ใช่เครื่องหมายสัมบูรณ์

### การปรับเป็นมาตรฐานเป็นการตัดสินใจ ไม่ใช่ขั้นตอนที่ทำเสมอ
ถ้าตัวแปรมีหน่วยต่างกัน ต้องปรับ มิฉะนั้นองค์ประกอบแรกจะชี้ไปตามตัวแปรที่ตัวเลขใหญ่ที่สุด แต่ถ้าตัวแปรทุกตัวเป็นปริมาณชนิดเดียวกันในหน่วยเดียวกัน เช่น พิกัดของมอนอเมอร์ การปรับจะบิดเบือนฟิสิกส์ เพราะมอนอเมอร์ที่เคลื่อนไหวมากควรมีน้ำหนักมากกว่า (แบบฝึก 8.D ในส่วนที่ 7)

> **ความเข้าใจคลาดเคลื่อนที่พบบ่อย:** "องค์ประกอบหลักมีเครื่องหมายและลำดับที่แน่นอน" ถ้า $\mathbf v$ เป็นเวกเตอร์เจาะจง $-\mathbf v$ ก็เป็นด้วย เครื่องหมายจึงไม่มีความหมาย และอาจกลับไปมาเมื่อเปลี่ยนข้อมูลเล็กน้อยหรือเปลี่ยนรุ่นของไลบรารี ปัญหาที่ร้ายแรงกว่าเกิดเมื่อค่าเจาะจงสองค่าขึ้นไปใกล้กัน เวกเตอร์เจาะจงแต่ละตัวไม่มีเอกลักษณ์ ปริภูมิที่มันแผ่ออกมีความหมาย แต่ทิศทางของแต่ละตัวภายในปริภูมินั้นเปลี่ยนตามสัญญาณรบกวน (ส่วนที่ 7 จะเห็นกรณีนี้กับสายพอลิเมอร์)

# ส่วนที่ 3 องค์ประกอบหลักคือโหมดปกติ (หัวข้อ 8.2.2)

หัวข้อนี้เป็นแกนของบท มันแสดงว่าในระบบที่อยู่ในสมดุลความร้อน PCA คือการวัดโหมดปกติของระบบโดยตรง

พิจารณาระบบที่มีพิกัด $\mathbf q$ จำนวน $D$ ตัว วัดจากตำแหน่งสมดุล และมีพลังงานศักย์ฮาร์มอนิก $U = \tfrac{1}{2}\mathbf q^\top \mathbf K \mathbf q$ เมื่อ $\mathbf K$ คือเมทริกซ์ความแข็งที่สมมาตรและเป็นบวกแน่นอน โหมดปกติคือเวกเตอร์เจาะจงของ $\mathbf K$ และถ้ามวลเท่ากันหมด ความถี่คือ $\omega_k = \sqrt{\kappa_k/m}$ เมื่อ $\kappa_k$ คือค่าเจาะจงของ $\mathbf K$

ในสมดุลความร้อนที่อุณหภูมิ $T$ ความน่าจะเป็นของโครงรูปเป็นไปตามการแจกแจงโบลต์ซมันน์ $P(\mathbf q) \propto \exp(-\mathbf q^\top\mathbf K\mathbf q/2k_BT)$ ซึ่งเป็นการแจกแจงเกาส์หลายตัวแปรที่มีเมทริกซ์ความแปรปรวนร่วม
$$\mathbf C = \langle \mathbf q\,\mathbf q^\top \rangle = k_BT\,\mathbf K^{-1}$$
$\mathbf K$ กับ $\mathbf K^{-1}$ มีเวกเตอร์เจาะจงชุดเดียวกัน และค่าเจาะจงเป็นส่วนกลับของกัน จึงได้

> **ทฤษฎีบท (องค์ประกอบหลักกับโหมดปกติ)** สำหรับระบบฮาร์มอนิกที่มีมวลเท่ากันและอยู่ในสมดุลความร้อน องค์ประกอบหลักของการแกว่งเชิงความร้อนเป็นเวกเตอร์ชุดเดียวกับโหมดปกติ และ
> $$\lambda_k = \frac{k_BT}{\kappa_k} = \frac{k_BT}{m\,\omega_k^2}$$
> องค์ประกอบหลักที่หนึ่งจึงเป็นโหมดที่มีความถี่ต่ำที่สุด ถ้ามวลไม่เท่ากัน ผลเดียวกันใช้ได้กับพิกัดที่ถ่วงด้วยรากที่สองของมวล

ผลนี้เป็นกรณีหนึ่งของ **หลักการแบ่งเท่ากันของพลังงาน (equipartition theorem)** ทุกโหมดได้พลังงานศักย์เฉลี่ย $\tfrac{1}{2}k_BT$ เท่ากัน โหมดที่นิ่มจึงต้องแกว่งด้วยแอมพลิจูดใหญ่ ลองนึกถึงเชือกที่ขึงระหว่างเสาสองต้นแล้วถูกลมพัดแบบสุ่ม เชือกส่วนใหญ่แกว่งเป็นโค้งยาวโค้งเดียว ถ้าบันทึกรูปร่างของเชือกหลายพันครั้งแล้วทำ PCA องค์ประกอบแรกจะเป็นโค้งยาวนั้นพอดี ในชีวฟิสิกส์ การวิเคราะห์แบบนี้กับโปรตีนเรียกว่า **พลวัตที่จำเป็น (essential dynamics)**

> **ความเข้าใจคลาดเคลื่อนที่พบบ่อย:** "PCA ของข้อมูลจากการจำลองให้โหมดปกติเสมอ" ทฤษฎีบทมีเงื่อนไขสามข้อ (1) ศักย์ต้องเป็นฮาร์มอนิก ถ้าไม่ใช่ ผลของ PCA เรียกว่า **โหมดกึ่งฮาร์มอนิก (quasi-harmonic modes)** (2) ระบบต้องอยู่ในสมดุล และข้อมูลต้องยาวกว่าเวลาผ่อนคลายของโหมดที่ช้าที่สุดมาก (3) ต้องลบการเคลื่อนที่และการหมุนของทั้งระบบออกก่อน มิฉะนั้นองค์ประกอบแรกจะเป็นการเคลื่อนที่ทั้งก้อน

### ✏️ แบบฝึก E8.2 ในชั้นเรียน (5 นาที) PCA ของโซ่ฮาร์มอนิก
โซ่ของมวล $N = 20$ ตัวที่ปลายทั้งสองยึดกับผนัง ใช้หน่วยที่ $m = k = k_BT = 1$ เมทริกซ์ความแข็ง $\mathbf K$ เป็นไตรทแยงที่มี 2 บนเส้นทแยงและ $-1$ ข้างเคียง เซลล์ถัดไปสุ่มโครงรูป 5,000 ครั้งจากการแจกแจงโบลต์ซมันน์ (ความแปรปรวนร่วม $\mathbf K^{-1}$) ทำ PCA และเตรียมโหมดปกติเชิงวิเคราะห์ $u_j(n) = \sqrt{2/(N+1)}\,\sin[j\pi n/(N+1)]$ ไว้ในแถวของ `u` ให้แล้ว

1. คำนวณขนาดของผลคูณภายในระหว่างองค์ประกอบหลักที่ $j$ กับโหมดที่ $j$ สำหรับ $j = 1, \ldots, 5$ เก็บเป็นรายการใน `my_overlap` (คำใบ้: `np.abs((pca_chain.components_[:5] * u).sum(axis=1))`)
2. เก็บค่าเจาะจงห้าค่าแรกใน `my_ev` แล้วเทียบกับค่าทฤษฎี `theory` $= 1/\omega_j^2$

```python
N_chain = 20
K = 2 * np.eye(N_chain) - np.eye(N_chain, k=1) - np.eye(N_chain, k=-1)
n_ = np.arange(1, N_chain + 1); j_ = np.arange(1, 6)
u = np.array([np.sqrt(2 / (N_chain + 1)) * np.sin(jj * np.pi * n_ / (N_chain + 1)) for jj in j_])   # analytic modes
theory = 1 / (4 * np.sin(j_ * np.pi / (2 * (N_chain + 1)))**2)                                  # 1 / omega_j^2
Q_chain = np.random.default_rng(2026).multivariate_normal(np.zeros(N_chain), np.linalg.inv(K), size=5000)
pca_chain = PCA().fit(Q_chain)
print("theory 1/omega^2:", theory.round(1))

my_overlap = []   # ✏️ |<PC_j, u_j>| for j = 1..5
my_ev = []        # ✏️ first five eigenvalues of pca_chain
```

```python
# ✅ ตรวจคำตอบ
_ov = np.abs((pca_chain.components_[:5] * u).sum(axis=1)); _ev = pca_chain.explained_variance_[:5]
print("ผลคูณภายใน:", "(ยังไม่ได้ตอบ)" if len(my_overlap) == 0 else ("✔" if np.allclose(my_overlap, _ov, atol=1e-3) else "✘"), "—", _ov.round(3))
print("ค่าเจาะจง:  ", "(ยังไม่ได้ตอบ)" if len(my_ev) == 0 else ("✔" if np.allclose(my_ev, _ev, rtol=1e-3) else "✘"), "—", _ev.round(1),
      "| ทฤษฎี", theory.round(1))

fig, ax = plt.subplots(1, 2, figsize=(10, 3.4))
s1 = np.sign(pca_chain.components_[0] @ u[0])                  # the sign of a PC is arbitrary
ax[0].plot(n_, s1 * pca_chain.components_[0], 'o', label='PC1'); ax[0].plot(n_, u[0], '-', label='analytic mode 1')
ax[0].set_xlabel('mass index n'); ax[0].legend(); ax[0].set_title('harmonic chain (SYNTHETIC)')
ax[1].semilogy(j_, _ev, 'o', label='PCA'); ax[1].semilogy(j_, theory, 'k_', ms=14, label='1/omega^2')
ax[1].set_xlabel('mode j'); ax[1].legend()
fig.tight_layout(); plt.show()
```

$$\mathbf x = \mathbf W\mathbf z + \boldsymbol\mu + \boldsymbol\epsilon, \qquad \mathbf z \sim \mathcal N(\mathbf 0, \mathbf I_d), \quad \boldsymbol\epsilon \sim \mathcal N(\mathbf 0, \sigma^2\mathbf I_D)$$
คือโครงสร้างมิติต่ำบวกสัญญาณรบกวนที่เท่ากันทุกทิศทาง ผลเฉลยภาวะน่าจะเป็นสูงสุดของ $\mathbf W$ แผ่ปริภูมิเดียวกับองค์ประกอบหลัก $d$ ตัวแรก และ $\sigma^2$ คือค่าเฉลี่ยของค่าเจาะจงที่ทิ้งไป scikit-learn เก็บค่านี้ไว้ใน `noise_variance_` เซลล์ถัดไปตรวจข้อนี้

```python
for d in (3, 5):
    p_d = PCA(n_components=d).fit(X_scaled)
    print(f"d = {d}: noise_variance_ = {p_d.noise_variance_:.4f}, mean of discarded eigenvalues = "
          f"{pca_full.explained_variance_[d:].mean():.4f}")
```

### ความเป็นอิสระและความไม่เป็นเกาส์
งานเลี้ยงที่มีคนสามคนพูดพร้อมกัน และมีไมโครโฟนสามตัววางคนละมุม ไมโครโฟนแต่ละตัวบันทึกเสียงผสมในสัดส่วนต่างกัน โจทย์คือแยกเสียงของแต่ละคนกลับออกมาโดยไม่รู้ตำแหน่งของใครเลย ปัญหานี้เรียกว่า **ปัญหางานเลี้ยงค็อกเทล (cocktail party problem)** และเป็นตัวอย่างของ **การแยกแหล่งกำเนิดแบบไม่รู้ล่วงหน้า (blind source separation)**

PCA แก้ปัญหานี้ไม่ได้ เพราะหาเพียงทิศทางที่ *ไม่สัมพันธ์กัน* ซึ่งมีได้ไม่จำกัดแบบ **การวิเคราะห์องค์ประกอบอิสระ (independent component analysis, ICA)** ใช้เงื่อนไขที่เข้มกว่า คือให้องค์ประกอบ *เป็นอิสระต่อกันทางสถิติ* แบบจำลองคือ $\mathbf x = \mathbf A\mathbf s$ เมื่อ $\mathbf s$ คือแหล่งกำเนิดอิสระและ $\mathbf A$ คือเมทริกซ์ผสมที่ไม่รู้ค่า ICA แยกได้ภายใต้เงื่อนไขว่ามีแหล่งกำเนิดแบบเกาส์ได้ไม่เกินหนึ่งตัว เหตุผลคือทฤษฎีบทขีดจำกัดกลาง ผลรวมของตัวแปรอิสระ "เป็นเกาส์มากกว่า" ตัวแปรเดิม ICA จึงหาทิศทางที่ผลการฉาย "ไม่เป็นเกาส์มากที่สุด" ซึ่งวัดได้ด้วยความโด่งหรือเนเกนโทรปี

### โค้ดที่ 8.3 — การแยกแหล่งกำเนิดด้วย FastICA

```python
from scipy import signal
from sklearn.decomposition import FastICA

rng = np.random.default_rng(seed=2026)
t = np.linspace(0, 8, 2000)
S_true = np.c_[np.sin(2 * np.pi * 1.3 * t),              # three independent sources
               signal.square(2 * np.pi * 0.7 * t),
               signal.sawtooth(2 * np.pi * 0.45 * t)]
S_true += 0.05 * rng.normal(size=S_true.shape)
A = np.array([[1.0, 0.6, 0.4], [0.5, 1.0, 0.7], [0.3, 0.8, 1.0]])
X_mix = S_true @ A.T                                       # three observed mixtures

ica = FastICA(n_components=3, whiten='unit-variance', random_state=42)
S_ica = ica.fit_transform(X_mix)                           # recovered sources
S_pca = PCA(n_components=3).fit_transform(X_mix)           # for comparison
```

ลำดับ เครื่องหมาย และขนาดขององค์ประกอบที่แยกได้เป็นเรื่องสุ่ม เซลล์ถัดไปจึงจับคู่องค์ประกอบกับแหล่งกำเนิดจริงด้วยวิธีฮังการี (`linear_sum_assignment`) บนขนาดของสหสัมพันธ์ แล้ววาดเทียบกัน

```python
from scipy.optimize import linear_sum_assignment

def matched_corr(S, E):
    C = np.abs(np.corrcoef(S.T, E.T)[:3, 3:])
    r, c = linear_sum_assignment(-C)
    return C[r, c]

print("|correlation| with the true sources (matched):")
print("  ICA:", matched_corr(S_true, S_ica).round(3))
print("  PCA:", matched_corr(S_true, S_pca).round(3))

fig, axes = plt.subplots(3, 4, figsize=(13, 5.5), sharex=True)
for j in range(3):
    axes[j, 0].plot(t, S_true[:, j]); axes[j, 1].plot(t, X_mix[:, j]); axes[j, 2].plot(t, S_ica[:, j]); axes[j, 3].plot(t, S_pca[:, j])
for ax, title in zip(axes[0], ['true sources', 'mixtures', 'FastICA', 'PCA']):
    ax.set_title(title + ' (SYNTHETIC)')
fig.tight_layout(); plt.show()
```

ICA แยกได้ด้วยสหสัมพันธ์ 0.996 ถึง 1.000 ส่วน PCA ได้เพียง 0.81 ถึง 0.90 เพราะองค์ประกอบของ PCA ยังเป็นเสียงผสม ICA ไม่รู้ขนาด เครื่องหมาย และลำดับของแหล่งกำเนิด การตีความจึงต้องใช้รูปร่างของสัญญาณ

### การตีความทางฟิสิกส์: การแยกสเปกตรัม
งานที่มักเสนอให้ใช้ ICA คือ **การแยกสเปกตรัม (spectral unmixing)** สเปกตรัมรามาน อินฟราเรด หรือการเรืองรังสีเอกซ์ของสารผสมเป็นผลรวมเชิงเส้นของสเปกตรัมบริสุทธิ์ถ่วงด้วยความเข้มข้น แต่สเปกตรัมบริสุทธิ์ของสารสามชนิด **ไม่มีเหตุผลใดที่จะต้องเป็นอิสระต่อกัน** ในทางกลับกัน สเปกตรัมและความเข้มข้นมีสมบัติทางฟิสิกส์ที่แน่นอนคือ *ไม่เป็นลบ* **การแยกตัวประกอบเมทริกซ์ไม่เป็นลบ (non-negative matrix factorization, NMF)** ใช้เงื่อนไขนี้แทน โดยเขียน $\mathbf X \approx \mathbf W\mathbf H$ ที่ทุกสมาชิกไม่เป็นลบ

กับสเปกตรัมจำลองสามเส้นที่ผสมเป็น 40 เส้น (แบบฝึกหัด E8.3 นอกชั้น) NMF แยกคืนได้ด้วยสหสัมพันธ์ 0.99 ถึง 1.000 ทุกองค์ประกอบ ขณะที่ ICA ได้ราว 0.66 ถึง 0.97 ถ้าความเข้มข้นรวมกันเป็นหนึ่งเสมอ ซึ่งเรียกว่า **ข้อมูลเชิงองค์ประกอบ (compositional data)** ข้อมูลหลังลบค่าเฉลี่ยจะมีเพียงสองมิติ องค์ประกอบที่สามของ PCA จึงเป็นเพียงสัญญาณรบกวน

> **ความเข้าใจคลาดเคลื่อนที่พบบ่อย:** "ICA ดีกว่า PCA เสมอเพราะองค์ประกอบตีความได้ทางฟิสิกส์" ICA ให้องค์ประกอบที่มีความหมายก็ต่อเมื่อแหล่งกำเนิดเป็นอิสระต่อกันจริงและไม่เป็นเกาส์ การเลือกระหว่าง PCA, ICA และ NMF จึงเป็นการเลือกสมมติฐานเกี่ยวกับฟิสิกส์ของระบบ PCA สมมติว่าโครงสร้างสำคัญอยู่ในทิศทางที่ความแปรปรวนสูง ICA สมมติว่าแหล่งกำเนิดเป็นอิสระต่อกัน และ NMF สมมติว่าทุกอย่างเป็นบวก ซึ่งเป็นจริงสำหรับความเข้ม ความเข้มข้น และจำนวนนับ

### ข้อจำกัดของ PCA และแมนิโฟลด์ไม่เชิงเส้น
PCA ฉายข้อมูลลงบนระนาบ ถ้าข้อมูลอยู่บนผิวที่โค้งหรือม้วน การฉายจะทับบริเวณที่อยู่ห่างกันบนผิวลงมาที่ตำแหน่งเดียวกัน ตัวอย่างมาตรฐานคือ **ม้วนสวิส (Swiss roll)** ซึ่งเป็นแผ่นสองมิติที่ม้วนเป็นก้นหอยในสามมิติ เซลล์ถัดไปแสดงม้วนสวิสระบายสีตามตำแหน่งบนแผ่น และภาพฉายของ PCA

```python
from sklearn.datasets import make_swiss_roll

X_roll, pos_roll = make_swiss_roll(n_samples=1500, noise=0.05, random_state=0)
fig = plt.figure(figsize=(10, 3.8))
ax3 = fig.add_subplot(1, 2, 1, projection='3d')
ax3.scatter(*X_roll.T, c=pos_roll, s=4, cmap='viridis'); ax3.set_title('Swiss roll (SYNTHETIC)')
ax2 = fig.add_subplot(1, 2, 2)
ax2.scatter(*PCA(2).fit_transform(X_roll).T, c=pos_roll, s=4, cmap='viridis'); ax2.set_title('PCA, 2 components')
fig.tight_layout(); plt.show()
```

PCA เพียงกดม้วนให้แบนเป็นก้นหอย ไม่ได้คลี่แผ่นออก จุดบนชั้นที่ติดกันของม้วน ซึ่งมีสีต่างกันมากเพราะอยู่ห่างกันเมื่อเดินไปตามแผ่น ยังอยู่ใกล้กันในภาพ เพราะ PCA วัดระยะเป็นเส้นตรงในปริภูมิเดิม ถ้าลดลงอีกเหลือมิติเดียว ชั้นเหล่านั้นจะทับกันจริง วิธีที่ "คลี่ม้วน" ได้ต้องสนใจเฉพาะเพื่อนบ้านที่ใกล้จริงบนแผ่น วิธีกลุ่มนี้เรียกว่า **การเรียนรู้แมนิโฟลด์ (manifold learning)**

### t-SNE
**t-SNE** (t-distributed stochastic neighbor embedding) พยายามรักษา *ว่าใครเป็นเพื่อนบ้านของใคร* มันแปลงระยะในปริภูมิเดิมเป็นความน่าจะเป็น $p_{ij}$ ด้วยฟังก์ชันเกาส์ ซึ่งความกว้างปรับให้แต่ละจุดมีเพื่อนบ้านประสิทธิผลเท่ากันตามพารามิเตอร์ `perplexity` (มักอยู่ระหว่าง 5 ถึง 50) แล้ววางจุดในสองมิติ นิยาม $q_{ij}$ ด้วยการแจกแจงทีที่มีหางหนา และเลื่อนจุดจนไดเวอร์เจนซ์ของคูลแบ็ก–ไลบ์เลอร์น้อยที่สุด
$$\mathrm{KL}(P \,\|\, Q) = \sum_{i \ne j} p_{ij} \ln \frac{p_{ij}}{q_{ij}}$$
ไดเวอร์เจนซ์นี้ลงโทษหนักเมื่อเพื่อนบ้านจริงถูกวางห่างกัน แต่ลงโทษน้อยเมื่อจุดที่ห่างกันถูกวางใกล้กัน t-SNE จึงรักษาโครงสร้างเฉพาะที่ได้ดีมาก แต่ไม่รักษาโครงสร้างระดับใหญ่

### UMAP
**UMAP** (uniform manifold approximation and projection) สร้างกราฟของเพื่อนบ้านใกล้ที่สุดในปริภูมิเดิม แล้วหาการจัดวางในมิติต่ำที่ทำให้กราฟแบบเดียวกันคล้ายกราฟเดิมที่สุด พารามิเตอร์หลักคือ `n_neighbors` (มองโครงสร้างเฉพาะที่หรือกว้างขึ้น) และ `min_dist` (จุดอัดกันแน่นเพียงใด) UMAP มักเร็วกว่า t-SNE กับข้อมูลใหญ่ และมักรักษาโครงสร้างระดับใหญ่ได้มากกว่า แต่ข้อหลังไม่ได้รับประกัน

### ลักษณะเด่นมิติสูงจาก AFM
ในงานวิจัยจริง นอกจากค่าของแต่ละพิกเซล เรามักเพิ่มลักษณะเด่นที่บอก "บริบท" รอบพิกเซล เช่น ค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐานเฉพาะที่ในหน้าต่างหลายขนาด เพราะเนื้อโครงร่างกับพาราฟินอาจต่างกันที่ความขรุขระมากกว่าที่ค่าเฉลี่ย โค้ดที่ 8.4 สร้างลักษณะเด่น 21 ตัวชุดเดียวกับส่วนที่ 2 แต่ปรับด้วย `RobustScaler` ตามบทที่ 7 แล้วสุ่มพิกเซลมา 3,000 พิกเซล เพราะ t-SNE และ UMAP ใช้เวลาเพิ่มเร็วตามจำนวนจุด

### โค้ดที่ 8.4 — ลักษณะเด่นเฉพาะที่ของแผนที่ AFM กับ PCA, t-SNE และ UMAP
⏱ เซลล์นี้ใช้เวลาราวหนึ่งถึงสองนาทีบน Colab (UMAP คอมไพล์โค้ดในการเรียกครั้งแรก)

```python
from scipy.ndimage import uniform_filter
from sklearn.preprocessing import RobustScaler
from sklearn.manifold import TSNE, trustworthiness
import umap                                   # pip install umap-learn

X_afm, y_afm, _ = make_scaffold_map()         # function from the Chapter 7 listing
n = 128
features = []
for ch in range(3):                           # H, S, A
    m = X_afm[:, ch].reshape(n, n)
    features.append(m)
    for w in (3, 7, 15):                      # local mean and local std
        mu = uniform_filter(m, w)
        sd = np.sqrt(np.maximum(uniform_filter(m * m, w) - mu * mu, 0))
        features += [mu, sd]
F = RobustScaler().fit_transform(np.column_stack([f.ravel() for f in features]))

idx = np.random.default_rng(0).choice(len(F), 3000, replace=False)
Z, labels = F[idx], y_afm[idx]
emb = {'PCA':  PCA(n_components=2).fit_transform(Z),
       't-SNE': TSNE(n_components=2, perplexity=30, random_state=42).fit_transform(Z),
       'UMAP': umap.UMAP(n_components=2, n_neighbors=15, min_dist=0.1,
                         random_state=42).fit_transform(Z)}
for name, E in emb.items():
    print(name, "trustworthiness =", round(trustworthiness(Z, E, n_neighbors=10), 3))
```

**ค่าความน่าเชื่อถือ (trustworthiness)** วัดว่าเพื่อนบ้านในภาพสองมิติเป็นเพื่อนบ้านจริงในปริภูมิเดิมหรือไม่ มีค่าสูงสุดเป็นหนึ่ง (รายละเอียดในส่วนที่ 8) เซลล์ถัดไปวาดภาพทั้งสามแบบ ระบายสีตามแผนที่อ้างอิงของบทที่ 7 (ใช้เพื่อดูเท่านั้น ไม่ได้ใช้ในการลดมิติ)

```python
_, _, glass_all = make_scaffold_map(); g_idx = glass_all[idx]
fig, axes = plt.subplots(1, 3, figsize=(14, 4.2))
for ax, (name, E) in zip(axes, emb.items()):
    ax.scatter(*E[labels == 1].T, s=3, c='tab:orange', label='scaffold')
    ax.scatter(*E[(labels == 0) & ~g_idx].T, s=3, c='tab:blue', label='paraffin')
    ax.scatter(*E[g_idx].T, s=3, c='k', label='glass'); ax.set_title(name + ' (SYNTHETIC)')
    ax.set_xticks([]); ax.set_yticks([])
axes[0].legend(markerscale=4); fig.tight_layout(); plt.show()
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier

cum_Z = np.cumsum(PCA().fit(Z).explained_variance_ratio_)
print(f"PCA of the 21 features: 2 components {cum_Z[1]:.3f}, 5 components {cum_Z[4]:.3f} of the variance")
print("5-NN accuracy (5-fold CV) predicting scaffold / substrate from each representation:")
for name, E in list(emb.items()) + [('21 features', Z), ('3 channels', RobustScaler().fit_transform(X_afm)[idx])]:
    acc = cross_val_score(KNeighborsClassifier(n_neighbors=5), E, labels, cv=5).mean()
    print(f"  {name:12s} {acc:.3f}")
print(f"  all-scaffold guess {labels.mean():.3f}")
```

PCA สองมิติได้ความน่าเชื่อถือราว 0.86 t-SNE ราว 0.98 และ UMAP ราว 0.94 สององค์ประกอบแรกของ PCA อธิบายความแปรปรวนได้เพียงราวร้อยละ 49 ข้อมูลนี้จึงไม่แบนพอจะวาดด้วย PCA สองมิติได้ครบ ในภาพ t-SNE และ UMAP พิกเซลของโครงร่างกับวัสดุรองรับแยกเป็นบริเวณค่อนข้างชัด และแก้วแยกเป็นกลุ่มเล็ก ความแม่นยำของตัวจำแนกเพื่อนบ้านใกล้ที่สุดจากพิกัดในภาพต้องเทียบกับ 0.70 ของการทายว่าเป็นโครงร่างทุกพิกเซล ตามบทเรียนของบทที่ 7

> **ความเข้าใจคลาดเคลื่อนที่พบบ่อย:** "ขนาดของกลุ่มและระยะห่างระหว่างกลุ่มในภาพ t-SNE หรือ UMAP มีความหมาย" t-SNE ปรับความกว้างของเกาส์ให้ทุกจุดมีเพื่อนบ้านเท่ากัน กลุ่มที่หนาแน่นจึงถูกขยายและกลุ่มที่เบาบางถูกบีบ ระยะห่างระหว่างกลุ่มที่ห่างกันแทบไม่มีความหมาย แกนไม่มีหน่วยและไม่มีความหมายทางฟิสิกส์ และภาพเปลี่ยนมากเมื่อเปลี่ยน `perplexity` หรือ `n_neighbors` สิ่งที่อ่านได้คือ จุดใดเป็นเพื่อนบ้านกัน และมีกลุ่มแยกออกจากกันหรือไม่ ภาพเหล่านี้เป็นเครื่องมือสำรวจ ไม่ใช่การวัด และไม่ควรจัดกลุ่มบนพิกัดของภาพ t-SNE โดยตรง

### แรงจูงใจ: ตัวแปรรวม
การจำลองพลวัตเชิงโมเลกุลของสายพอลิเมอร์ที่มีมอนอเมอร์ $N$ ตัว ให้ $3N$ พิกัดทุกเฟรม ถ้า $N = 50$ แต่ละเฟรมมี 150 มิติ แต่การเคลื่อนไหวที่สำคัญเกิดจากโหมดรวมไม่กี่โหมด โหมดช้าที่สุดคือทั้งสายโค้งเป็นรูปตัวซี ถัดไปคือรูปตัวเอส และต่อไปเรื่อย ๆ ปริมาณที่อธิบายสถานะของระบบด้วยตัวแปรไม่กี่ตัวเช่นนี้เรียกว่า **ตัวแปรรวม (collective variable)**

### แบบจำลองของเราส์
**แบบจำลองของเราส์ (Rouse model)** มองสายเป็นลูกปัด $N$ ลูกที่ต่อกันด้วยสปริงค่าคงที่ $k = 3k_BT/b^2$ ($b$ คือความยาวพันธะเฉลี่ย) และถูกแรงเสียดทานกับแรงสุ่มจากตัวทำละลาย สำหรับสายที่ปลายทั้งสองอิสระ เวกเตอร์เจาะจงของเมทริกซ์ของเราส์คือ
$$\psi_p(i) = \sqrt{\frac{2}{N}}\cos\left(\frac{p\pi(i - \tfrac12)}{N}\right), \qquad a_p = 4\sin^2\left(\frac{p\pi}{2N}\right), \quad p = 1, \ldots, N-1$$
โหมด $p = 0$ คือการเคลื่อนที่ของจุดศูนย์กลางมวลซึ่งตัดออก แอมพลิจูดของโหมดที่ $p$ ตามแกนหนึ่งแกว่งอย่างอิสระ ด้วย
$$\langle X_p^2\rangle = \frac{k_BT}{k\,a_p}, \qquad \tau_p = \frac{\zeta}{k\,a_p}$$
เมื่อ $\zeta$ คือสัมประสิทธิ์แรงเสียดทาน สำหรับ $p \ll N$ ทั้งความแปรปรวนและเวลาผ่อนคลายแปรตาม $N^2/p^2$ โหมดที่หนึ่งทั้งแกว่งกว้างที่สุดและช้าที่สุด

โค้ดที่ 8.5 ใช้หน่วยที่ $k_BT = \zeta = b = 1$ และแทนการจำลองแบบบราวน์ทีละก้าว มันใช้ข้อเท็จจริงว่าแต่ละโหมดเป็น **กระบวนการของออร์นสไตน์–อูเลนเบ็ก (Ornstein–Uhlenbeck process)** ที่อิสระต่อกัน ซึ่งปรับค่าจากเฟรมหนึ่งไปอีกเฟรมได้แม่นตรง แล้วแปลงกลับเป็นพิกัดของลูกปัด ข้อมูลที่ได้เป็นพิกัดคาร์ทีเซียนธรรมดา ซึ่ง PCA ไม่รู้ว่ามันสร้างมาจากโหมด

```python
import numpy as np
from sklearn.decomposition import PCA
from sklearn.manifold import trustworthiness
```

### โค้ดที่ 8.5 — ข้อมูลเส้นทางการเคลื่อนที่ของสายพอลิเมอร์ตามแบบจำลองของเราส์

```python
def make_rouse_trajectory(N=50, n_frames=10000, dt_frame=5.0, seed=2026):
    """Rouse chain (k_B T = friction = bond length = 1) sampled exactly in
    its normal-mode basis; returns Cartesian coordinates (n_frames, 3N)."""
    rng = np.random.default_rng(seed)
    k = 3.0                                              # 3 k_B T / b^2
    i = np.arange(N)
    p = np.arange(1, N)                                  # p = 0 (centre of mass) removed
    Q = np.sqrt(2.0 / N) * np.cos(np.outer(p, np.pi * (i + 0.5) / N))
    a = 4.0 * np.sin(p * np.pi / (2 * N))**2             # Rouse-matrix eigenvalues
    var = 1.0 / (k * a)                                  # <X_p^2> per axis
    tau = 1.0 / (k * a)                                  # relaxation times
    decay = np.exp(-dt_frame / tau)
    Xp = np.empty((n_frames, len(p), 3))
    Xp[0] = rng.normal(size=(len(p), 3)) * np.sqrt(var)[:, None]
    for t in range(1, n_frames):                         # exact Ornstein-Uhlenbeck step
        Xp[t] = (decay[:, None] * Xp[t - 1]
                 + np.sqrt(var * (1 - decay**2))[:, None] * rng.normal(size=(len(p), 3)))
    r = np.einsum('pi,tpc->tic', Q, Xp)                 # bead positions (CM at origin)
    return r.reshape(n_frames, 3 * N), tau

polymer_trajectory, tau_theory = make_rouse_trajectory()
```

```python
print(f"tau_1 = {tau_theory[0]:.1f} time units = {tau_theory[0] / 5.0:.0f} frames; "
      f"record length 10000 x 5 = {50000 / tau_theory[0]:.0f} tau_1")
```

### โค้ดที่ 8.6 — PCA ของเส้นทางการเคลื่อนที่ของสายพอลิเมอร์

```python
X = polymer_trajectory                           # (10000, 150), same units in every column
X_centered = X - X.mean(axis=0)                  # no standardisation (see text)

pca = PCA(n_components=12)
X_pca = pca.fit_transform(X_centered)
print("variance ratio:", np.round(pca.explained_variance_ratio_[:9], 3))
print("eigenvalues:  ", np.round(pca.explained_variance_[:9], 1))

N = 50
mode1 = pca.components_[0].reshape(N, 3)         # PC1 as a displacement field
profile = mode1 @ np.linalg.svd(mode1)[2][0]     # project on its dominant direction
```

### รูปร่างของโหมดและเวลาผ่อนคลาย
เมื่อนำองค์ประกอบที่หนึ่งมาจัดเป็นเวกเตอร์การกระจัดของลูกปัด 50 ลูก แล้วฉายลงบนทิศทางหลักของมัน (ตัวแปร `profile` ในโค้ดที่ 8.6) จะได้รูปโค้งที่ตรงกับ $\cos[\pi(i - \tfrac12)/N]$ คือสายโค้งเป็นรูปตัวซี และองค์ประกอบที่สี่ถึงหกตรงกับ $\cos[2\pi(i-\tfrac12)/N]$ ซึ่งเป็นรูปตัวเอส **PCA ค้นพบโหมดของเราส์จากพิกัดดิบโดยไม่ต้องรู้ทฤษฎีใดเลย**

PCA ยังให้เวลาผ่อนคลายได้ จากฟังก์ชันสหสัมพันธ์ในตัวของคะแนนองค์ประกอบ $C(t) = \langle z(0)z(t)\rangle/\langle z^2\rangle \approx e^{-t/\tau_p}$ และผลรวมของความแปรปรวนทุกองค์ประกอบเท่ากับ $N\langle R_g^2\rangle$ เมื่อ $R_g$ คือรัศมีไจเรชัน เซลล์ถัดไปคำนวณทั้งหมดและวาดภาพสามภาพ

```python
i = np.arange(N); p_all = np.arange(1, N)
a_all = 4 * np.sin(p_all * np.pi / (2 * N))**2; lam_th = 1 / (3 * a_all)
print(f"|corr| PC1 profile vs cos(pi(i+1/2)/N) = {abs(np.corrcoef(profile, np.cos(np.pi * (i + .5) / N))[0, 1]):.4f}")
for j in range(3, 6):
    m_ = pca.components_[j].reshape(N, 3); pr = m_ @ np.linalg.svd(m_)[2][0]
    print(f"|corr| PC{j + 1} profile vs cos(2 pi(i+1/2)/N) = {abs(np.corrcoef(pr, np.cos(2 * np.pi * (i + .5) / N))[0, 1]):.4f}")

z1 = X_pca[:, 0] - X_pca[:, 0].mean()
acf = np.array([1.0] + [np.mean(z1[:-l] * z1[l:]) / np.mean(z1 * z1) for l in range(1, 61)])
for last in (10, 20, 40):                                   # fit ln C(t) on lags 1..last-1 frames
    lag = np.arange(1, last)
    slope = np.polyfit(lag * 5.0, np.log(acf[1:last]), 1)[0]
    print(f"tau_1 from ln C(t), lags up to {5 * (last - 1):.0f} time units: {-1 / slope:.1f} (theory {tau_theory[0]:.1f})")
tot = X_centered.var(axis=0, ddof=1).sum()
print(f"sum of all variances {tot:.1f} = N <Rg^2>  ->  <Rg^2> = {tot / N:.2f}; Gaussian chain N b^2 / 6 = {N / 6:.2f}")

full_ev = PCA().fit(X_centered).explained_variance_
fig, axes = plt.subplots(1, 3, figsize=(14, 3.8))
axes[0].semilogy(np.arange(1, 31), full_ev[:30], 'o', label='PCA'); axes[0].semilogy(np.arange(1, 31), np.repeat(lam_th[:10], 3), 'k_', ms=12, label='1/(3 a_p)')
axes[0].set_xlabel('component'); axes[0].set_ylabel('eigenvalue'); axes[0].set_title('triplets, ~1/p^2 (SYNTHETIC)'); axes[0].legend()
sgn = np.sign(np.corrcoef(profile, np.cos(np.pi * (i + .5) / N))[0, 1])      # the sign of a PC is arbitrary
axes[1].plot(sgn * profile / np.abs(profile).max(), label='PC1 (sign aligned)'); axes[1].plot(np.cos(np.pi * (i + .5) / N), '--', label='cos, p = 1')
axes[1].set_xlabel('bead'); axes[1].legend(); axes[1].set_title('mode shape')
axes[2].plot(np.arange(61) * 5, acf, label='C(t) of PC1'); axes[2].plot(np.arange(61) * 5, np.exp(-np.arange(61) * 5 / tau_theory[0]), '--', label='exp(-t/tau_1)')
axes[2].set_xlabel('t'); axes[2].legend(); axes[2].set_title('autocorrelation')
fig.tight_layout(); plt.show()
```

### เกณฑ์ประเมินคุณภาพ
| เกณฑ์ | ความหมายและข้อจำกัด |
|---|---|
| ความคลาดเคลื่อนในการสร้างคืน | $\lVert\mathbf X - \hat{\mathbf X}\rVert^2$ เมื่อ $\hat{\mathbf X}$ สร้างคืนจาก $d$ มิติ ใช้กับวิธีเชิงเส้นที่แปลงกลับได้ |
| สัดส่วนความแปรปรวนสะสม | $\sum_{i\le d}\lambda_i/\sum\lambda_i$ เฉพาะ PCA |
| ความน่าเชื่อถือ (trustworthiness) | เพื่อนบ้านในภาพเป็นเพื่อนบ้านจริงหรือไม่ ลงโทษเพื่อนบ้านปลอม |
| ความต่อเนื่อง (continuity) | เพื่อนบ้านจริงยังเป็นเพื่อนบ้านในภาพหรือไม่ ลงโทษเพื่อนบ้านที่หายไป |
| ผลของงานขั้นต่อไป | การลดมิติทำให้การจัดกลุ่ม การจำแนก หรือการถดถอยดีขึ้นหรือไม่ |

อุปมาคือแผนที่เมือง แผนที่ที่ *น่าเชื่อถือ* ไม่วางร้านสองร้านที่อยู่คนละฝั่งเมืองไว้ติดกัน แผนที่ที่ *ต่อเนื่อง* ไม่แยกร้านที่อยู่ติดกันจริงไปไว้คนละมุม scikit-learn มี `trustworthiness` ส่วนความต่อเนื่องคำนวณด้วยฟังก์ชันเดียวกันโดยสลับบทบาทของข้อมูลเดิมกับข้อมูลที่ลดมิติแล้ว เกณฑ์สุดท้ายมักสำคัญที่สุด (แบบฝึกหัด E8.6 นอกชั้น)

### ช่วงความเชื่อมั่นของค่าเจาะจงเมื่อข้อมูลมีสหสัมพันธ์ในตัว
ค่าเฉลี่ยของสามค่าเจาะจงแรกคือราว 89.6 ขณะที่ทฤษฎีคือ 84.5 บูตสแตรปที่สุ่มเฟรมอิสระ (แบบโค้ดที่ 8.2) นับเฟรมที่ติดกัน ซึ่งเกือบเหมือนกันเพราะโหมดที่หนึ่งใช้ราว 17 เฟรมในการผ่อนคลาย เป็นตัวอย่างใหม่ วิธีแก้คือ **บูตสแตรปแบบเป็นช่วง (block bootstrap)** ตัดข้อมูลเป็นช่วงต่อเนื่องที่ยาวกว่าเวลาผ่อนคลายหลายเท่า แล้วสุ่มช่วงแทนการสุ่มเฟรม เซลล์ถัดไปเปรียบเทียบสองวิธี (ใช้เวลาราวครึ่งนาทีบน Colab)

```python
def mean_top3(Zc):
    return PCA(n_components=3).fit(Zc).explained_variance_.mean()

B = 200
print(f"mean of the first three eigenvalues: {mean_top3(X_centered):.2f} (theory {1 / (12 * np.sin(np.pi / 100)**2):.2f})")
rng_b = np.random.default_rng(2026)
iid = [mean_top3(X_centered[rng_b.integers(0, len(X_centered), len(X_centered))]) for _ in range(B)]
lo, hi = np.percentile(iid, [2.5, 97.5])
print(f"independent-frame bootstrap 95 % CI: [{lo:.1f}, {hi:.1f}]  width {hi - lo:.1f}")
L = 200                                          # block length in frames (~12 tau_1)
nb = len(X_centered) // L
rng_b = np.random.default_rng(2026)
blk = []
for _ in range(B):
    start = rng_b.integers(0, nb, nb)
    blk.append(mean_top3(X_centered[(start[:, None] * L + np.arange(L)).ravel()]))
blo, bhi = np.percentile(blk, [2.5, 97.5])
print(f"block bootstrap, blocks of {L} frames: [{blo:.1f}, {bhi:.1f}]  width {bhi - blo:.1f} ({(bhi - blo) / (hi - lo):.1f} x wider)")
```

ช่วงแบบเฟรมอิสระ $[88.1, 90.8]$ แคบจนดูแม่นยำมาก แต่ **ไม่ครอบค่าทฤษฎี** ช่วงแบบเป็นช่วง $[84.5, 94.7]$ กว้างกว่าเกือบสี่เท่าและครอบค่าทฤษฎี ผลแทบไม่เปลี่ยนเมื่อใช้ช่วงละ 100 หรือ 500 เฟรม นี่คือความคิดเดียวกับเวลาสหสัมพันธ์ในตัวในบทที่ 4 จำนวนตัวอย่างที่มีประโยชน์จริงไม่ใช่จำนวนเฟรม แต่เป็นความยาวของข้อมูลหารด้วยเวลาสหสัมพันธ์

### ✏️ E8.1 ส่วนที่เหลือ: ผลขึ้นกับการสุ่มเพียงใด (15 นาที)
ทำการคำนวณของแบบฝึก E8.1 ซ้ำด้วยเมล็ดสุ่ม 2026, 0, 1, 2 และ 42 (ลำดับการสุ่มเหมือนเซลล์ในส่วนที่ 1: สี่ลูกบาศก์ก่อน แล้วจึงจุดสองมิติและเมทริกซ์ฝัง) เก็บ `(อัตราส่วนที่ D = 3, อัตราส่วนที่ D = 100, อัตราส่วนของระนาบที่ฝัง)` ของแต่ละเมล็ดใน `my_seeds[seed]` แล้วอธิบายว่าค่าใดแปรผันมาก ค่าใดคงที่ และเพราะเหตุใด

### ✏️ E8.3 การแยกสเปกตรัม (25 นาที)
เซลล์ถัดไปสร้างแกนความยาวคลื่น 500 จุดในช่วง 400–800 nm และสเปกตรัมบริสุทธิ์สามเส้น `S_pure` จากยอดแบบเกาส์ (480 nm กว้าง 12 nm รวมกับยอดเล็กที่ 620 nm, 530 nm รวมกับยอดเล็กที่ 700 nm และ 590 nm) และฟังก์ชัน `matched(E)` ที่จับคู่องค์ประกอบกับสเปกตรัมจริงแล้วคืนขนาดของสหสัมพันธ์สามค่า

สำหรับข้อมูลสิบชุด (เมล็ดสุ่ม 0–9) ให้สุ่มความเข้มข้นของสารผสม 40 ตัวอย่างด้วย `rng.dirichlet(np.ones(3), size=40)` สร้าง `X_sp = conc @ S_pure + 0.01 * rng.normal(size=(40, 500))` แล้วแยกสเปกตรัมสามวิธี
- FastICA: `FastICA(3, whiten='unit-variance', random_state=42, max_iter=2000).fit_transform(X_sp.T).T` (ทรานสโพส เพื่อให้จุดความยาวคลื่นเป็นตัวอย่าง)
- PCA: `PCA(3).fit(X_sp).components_`
- NMF: `NMF(3, init='nndsvda', max_iter=2000, random_state=42).fit(np.clip(X_sp, 0, None)).components_`

เก็บผลของ `matched(...)` ทุกชุดลงในรายการ `res_sp['FastICA']`, `res_sp['PCA']` และ `res_sp['NMF']` แล้วอธิบายผล โดยเฉพาะว่าทำไม PCA จึงแยกได้ดีเพียงสององค์ประกอบ

```python
from sklearn.decomposition import NMF

wl = np.linspace(400, 800, 500)
G = lambda c, w: np.exp(-0.5 * ((wl - c) / w)**2)
S_pure = np.array([G(480, 12) + 0.3 * G(620, 15), G(530, 20) + 0.5 * G(700, 12), G(590, 25)])

def matched(E):
    C = np.abs(np.corrcoef(S_pure, E)[:3, 3:])
    r, c = linear_sum_assignment(-C)
    return np.sort(C[r, c])

fig, ax = plt.subplots(figsize=(6, 3))
for s_ in S_pure: ax.plot(wl, s_)
ax.set_xlabel('wavelength (nm)'); ax.set_title('pure spectra (SYNTHETIC)'); plt.show()

res_sp = {'FastICA': [], 'PCA': [], 'NMF': []}
for seed in range(10):
    rng_sp = np.random.default_rng(seed)
    # ✏️ conc = ..., X_sp = ..., then append matched(...) of each method to res_sp
    pass
```

```python
# ✅ ตรวจคำตอบ
if not res_sp['NMF']:
    print("(ยังไม่ได้ตอบ) — ตำรา: NMF 0.99-1.000, FastICA 0.66-0.97, PCA ดีเพียงสององค์ประกอบ")
else:
    _ref = {'FastICA': [], 'PCA': [], 'NMF': []}
    for _sd in range(10):
        _rg = np.random.default_rng(_sd); _c = _rg.dirichlet(np.ones(3), size=40)
        _X = _c @ S_pure + 0.01 * _rg.normal(size=(40, 500))
        _ref['FastICA'].append(matched(FastICA(3, whiten='unit-variance', random_state=42, max_iter=2000).fit_transform(_X.T).T))
        _ref['PCA'].append(matched(PCA(3).fit(_X).components_))
        _ref['NMF'].append(matched(NMF(3, init='nndsvda', max_iter=2000, random_state=42).fit(np.clip(_X, 0, None)).components_))
    for k_ in _ref:
        v, mine = np.array(_ref[k_]), np.array(res_sp[k_])
        ok = mine.shape == v.shape and np.allclose(mine, v, atol=1e-3)
        print(f"{k_:8s}", "✔" if ok else "✘", f"— |corr| {v.min():.3f}-{v.max():.3f}; median per component (worst..best) {np.median(v, axis=0).round(3)}")
```

1. คำนวณความคลาดเคลื่อนสัมพัทธ์ในการสร้างคืนของ PCA $\lVert\mathbf X - \hat{\mathbf X}\rVert^2 / \lVert\mathbf X - \bar{\mathbf X}\rVert^2$ ที่ $d = 3, 9, 30$ เก็บใน `my_rec[d]` (คำใบ้: `p = PCA(d).fit(Xs); rec = p.inverse_transform(p.transform(Xs))`)
2. ถดถอยเชิงเส้น $R_{ee}$ จากพิกัดสองมิติของ PCA ด้วย `cross_val_score(LinearRegression(), E, ree, cv=5, scoring='r2').mean()` เก็บใน `my_r2_lin` แล้วทำซ้ำโดยเพิ่มกำลังสองของพิกัด `np.c_[E, E**2]` เก็บใน `my_r2_sq`
3. อธิบายว่าทำไมการถดถอยเชิงเส้นได้ $R^2$ ราวศูนย์ และเลือกวิธีที่เหมาะที่สุดสำหรับงานที่ต้องการตัวแปรรวมที่ตีความได้

```python
from sklearn.linear_model import LinearRegression

sel = np.random.default_rng(2026).choice(len(X_centered), 2000, replace=False)
Xs = X_centered[sel]
r_sel = polymer_trajectory[sel].reshape(-1, N, 3)
ree = np.linalg.norm(r_sel[:, -1] - r_sel[:, 0], axis=1)
emb_p = {'PCA': PCA(2).fit_transform(Xs),
         't-SNE': TSNE(2, perplexity=30, random_state=42).fit_transform(Xs),
         'UMAP': umap.UMAP(n_components=2, n_neighbors=15, min_dist=0.1, random_state=42).fit_transform(Xs)}
for name, E in emb_p.items():
    print(f"{name:5s}: trustworthiness {trustworthiness(Xs, E, n_neighbors=10):.3f}  continuity {trustworthiness(E, Xs, n_neighbors=10):.3f}")

my_rec = {}        # ✏️ {3: ..., 9: ..., 30: ...}
my_r2_lin = None   # ✏️ CV R^2 of R_ee from the 2-D PCA coordinates
my_r2_sq = None    # ✏️ same with (coords, coords^2)
```

```python
# ✅ ตรวจคำตอบ
_tot = ((Xs - Xs.mean(0))**2).sum()
for _d in (3, 9, 30):
    _p = PCA(_d).fit(Xs); _e = ((Xs - _p.inverse_transform(_p.transform(Xs)))**2).sum() / _tot
    mine = my_rec.get(_d)
    print(f"d = {_d:2d}:", "(ยังไม่ได้ตอบ)" if mine is None else ("✔" if abs(mine - _e) < 1e-3 else "✘"), f"— {_e:.3f}")
_E = emb_p['PCA']
_l = cross_val_score(LinearRegression(), _E, ree, cv=5, scoring='r2').mean()
_s = cross_val_score(LinearRegression(), np.c_[_E, _E**2], ree, cv=5, scoring='r2').mean()
print("R^2 เชิงเส้น:       ", "(ยังไม่ได้ตอบ)" if my_r2_lin is None else ("✔" if abs(my_r2_lin - _l) < 1e-3 else "✘"), f"— {_l:.3f}")
print("R^2 เพิ่มกำลังสอง:  ", "(ยังไม่ได้ตอบ)" if my_r2_sq is None else ("✔" if abs(my_r2_sq - _s) < 1e-3 else "✘"), f"— {_s:.3f}")
```

### สรุปบท
ข้อมูลฟิสิกส์ที่มีมิติสูงแทบไม่เคยกระจายเต็มปริภูมิ กฎทางฟิสิกส์บังคับให้ข้อมูลอยู่บนโครงสร้างที่มีมิติต่ำกว่ามาก การลดมิติจึงวัดจำนวนองศาเสรีที่มีความหมายของระบบ PCA หาทิศทางที่ข้อมูลกระจายมากที่สุดด้วยเวกเตอร์เจาะจงของเมทริกซ์ความแปรปรวนร่วม และในระบบฮาร์มอนิกที่อยู่ในสมดุลความร้อน ทิศทางเหล่านั้นคือโหมดปกติพอดี โดย $\lambda_k = k_BT/m\omega_k^2$ ตามหลักการแบ่งเท่ากันของพลังงาน ผลนี้ต้องใช้อย่างระวัง ศักย์ต้องใกล้ฮาร์มอนิก ข้อมูลต้องยาวพอ เครื่องหมายขององค์ประกอบไม่มีความหมาย และเมื่อค่าเจาะจงซ้ำกัน องค์ประกอบแต่ละตัวในกลุ่มไม่มีเอกลักษณ์

ICA แยกแหล่งกำเนิดได้ดีเมื่อแหล่งกำเนิดเป็นอิสระจริงและไม่เป็นเกาส์ แต่กับสเปกตรัม NMF ที่ใช้ข้อจำกัดความไม่เป็นลบให้ผลดีกว่ามาก t-SNE และ UMAP คลี่โครงสร้างไม่เชิงเส้นเพื่อวาดภาพ รักษาความสัมพันธ์ของเพื่อนบ้านได้ดี แต่ขนาด ระยะห่าง และแกนของภาพไม่มีความหมายทางฟิสิกส์ สายพอลิเมอร์ตามแบบจำลองของเราส์รวมทุกความคิดไว้ด้วยกัน PCA ค้นพบโหมดของเราส์จากพิกัดดิบ ค่าเจาะจงมาเป็นกลุ่มละสามค่าและลดลงตาม $1/p^2$ และความคลาดเคลื่อนของโหมดที่ช้าที่สุดสอนว่าความไม่แน่นอนต้องประเมินด้วยวิธีที่คำนึงถึงสหสัมพันธ์ในตัว
