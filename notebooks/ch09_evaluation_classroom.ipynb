# บทที่ 9 การประเมินแบบจำลองและการระบุความไม่แน่นอน: โน้ตบุ๊กสำหรับชั้นเรียน
โน้ตบุ๊กประกอบตำรา *วิทยาศาสตร์ข้อมูลและการเรียนรู้ของเครื่องสำหรับนักฟิสิกส์*

โน้ตบุ๊กนี้ใช้ในชั้นเรียนได้ด้วยไฟล์นี้ไฟล์เดียว ไม่ต้องดึงโค้ดจาก GitHub และไม่ต้องดาวน์โหลดข้อมูล เนื้อหาเรียงตามบทที่ 9 ในตำรา แบ่งเป็นสี่คาบ รวม 240 นาทีตามแผนเวลาในคู่มือผู้สอน (คาบที่ 1 และ 2 คาบละ 65 นาที คาบที่ 3 และ 4 คาบละ 55 นาที) หัวข้อที่มีป้าย **โค้ดที่ 9.N** ตรงกับตัวอย่างโค้ดเลขเดียวกันในตำรา

## วัตถุประสงค์
เมื่อทำโน้ตบุ๊กนี้จนจบ ผู้เรียนควรสามารถ

1. แยกความไม่แน่นอนแบบเอลีอะทอริก (aleatoric uncertainty) ออกจากแบบเอพิสเทมิก (epistemic uncertainty) และบอกได้ว่าวิธีใดลดความไม่แน่นอนชนิดใด
2. เลือกรูปแบบของการตรวจสอบไขว้ (cross-validation) ให้ตรงกับโครงสร้างของข้อมูล ใช้การตรวจสอบไขว้แบบซ้อนเมื่อต้องเลือกไฮเพอร์พารามิเตอร์ และใช้เกณฑ์ AIC กับ BIC โดยอธิบายได้ว่าทั้งสองตอบคำถามต่างกันอย่างไร
3. สร้างช่วงความเชื่อมั่นด้วยบูตสแตรป (bootstrap) ทั้งกับพารามิเตอร์ของการปรับเส้นโค้งและกับค่าเฉลี่ยจากลูกโซ่มาร์คอฟ โดยแก้ผลของสหสัมพันธ์ในตัวด้วยเวลาสหสัมพันธ์แบบปริพันธ์และการสุ่มซ้ำเป็นช่วง
4. สุ่มตัวอย่างการแจกแจงภายหลัง (posterior) ด้วยวิธีเมโทรโพลิสที่เขียนเองและด้วย PyMC ตรวจการลู่เข้า และตรวจความเพียงพอของแบบจำลองด้วยการตรวจเชิงทำนายภายหลัง
5. เปรียบเทียบวิธีจำแนกสองวิธีด้วยการทดสอบแบบจับคู่ ตรวจและปรับเทียบความน่าจะเป็น และอ่านความกระจายของวิธีรวมกลุ่มอย่างถูกต้อง
6. เลือกวิธีระบุความไม่แน่นอนให้เหมาะกับคำถาม และรายงานผลในรูปแบบที่ผู้อ่านนำไปใช้ต่อได้

## ข้อมูลในโน้ตบุ๊กนี้
ข้อมูลทุกชุดเป็น **ข้อมูลจำลอง (SYNTHETIC)** ที่สร้างด้วยโค้ดในตำรา และตัวเลขตรงกับตำรา (ผลของ PyMC อาจต่างในหลักสุดท้ายตามเครื่องและรุ่นของไลบรารี)

- **อนุกรมเวลา AR(1):** ตัวแปรควบคุมที่เปลี่ยนช้าและสัญญาณรบกวนที่สัมพันธ์กันในเวลา (โค้ดที่ 9.1)
- **ออสซิลเลเตอร์แบบหน่วง:** ข้อมูล 200 จุดของบทที่ 5 (โค้ดที่ 5.6) และชุดที่แย่กว่าสำหรับเปรียบเทียบ
- **แบบจำลองไอซิง:** พลังงานจากการจำลองเมโทรโพลิสด้วยฟังก์ชันของบทที่ 4 (โค้ดที่ 4.6)
- **เหตุการณ์ฮิกส์จำลอง:** ข้อมูลและวิธีจำแนกของบทที่ 6 (โค้ดที่ 6.1–6.5) ซึ่งมีรูปร่างเพื่อการเรียนเท่านั้น
- **สายพอลิเมอร์:** เส้นทางการเคลื่อนที่ตามแบบจำลองของเราส์จากบทที่ 8 (โค้ดที่ 8.5)

หัวกราฟที่มาจากข้อมูลจำลองมีคำว่า SYNTHETIC กำกับ ตัวเลขใช้เพื่อการเรียนวิธีการ ไม่ใช่ผลการวิจัย

เซลล์ถัดไปสร้างข้อมูลออสซิลเลเตอร์แบบหน่วง 200 จุดในเวลา 10 วินาที ด้วยสัญญาณรบกวน 0.05 m แล้วปรับเส้นโค้ง โค้ดเป็นชุดเดียวกับโค้ดที่ 5.6 ของบทที่ 5 ถูกซ่อนไว้ ดับเบิลคลิกเพื่อดูได้ บทนี้ใช้ตัวแปรจากเซลล์นี้ซ้ำหลายครั้ง ได้แก่ `damped_oscillator`, `t`, `x_obs`, `sigma_x`, `bounds`, `popt` และ `pcov`

ค่าจริงที่ใช้สร้างข้อมูลคือ $A = 1.0$ m, $\gamma = 0.3$ s⁻¹, $\omega = 6.0$ rad/s และ $\phi = 0.5$ rad ตัวประกอบคุณภาพจริงจึงเป็น $Q = \omega/(2\gamma) = 10$

```python
#@title ข้อมูลออสซิลเลเตอร์แบบหน่วงจากโค้ดที่ 5.6 (ดับเบิลคลิกเพื่อดูโค้ด)
from scipy.optimize import curve_fit

def damped_oscillator(t, A, gamma, omega, phi):
    return A * np.exp(-gamma * t) * np.cos(omega * t + phi)

rng = np.random.default_rng(seed=2026)
t = np.linspace(0, 10, 200)                               # s
sigma_x = 0.05                                            # m
x_obs = damped_oscillator(t, 1.0, 0.3, 6.0, 0.5) \
        + rng.normal(scale=sigma_x, size=t.size)

p0 = [1.0, 0.2, 6.0, 0.0]                                 # initial guess
bounds = ([0, 0, 0, -np.pi], [10, 5, 20, np.pi])          # physical limits
popt, pcov = curve_fit(damped_oscillator, t, x_obs,
                       p0=p0, bounds=bounds,
                       sigma=sigma_x * np.ones(t.size),
                       absolute_sigma=True)

A, gamma, omega, phi = popt
errs = np.sqrt(np.diag(pcov))
print(f"gamma = {gamma:.4f} +/- {errs[1]:.4f} 1/s")
print(f"omega = {omega:.4f} +/- {errs[2]:.4f} rad/s")
```

เซลล์ถัดไปนิยามฟังก์ชันสามตัวของการจำลองเมโทรโพลิสสำหรับแบบจำลองไอซิงสองมิติ (`init_lattice`, `total_energy` และ `metropolis_step`) จากโค้ดที่ 4.6 ของบทที่ 4 ฟังก์ชันเหล่านี้ใช้ตัวสร้างเลขสุ่ม `rng` ส่วนกลาง โค้ดที่ 9.4 ในคาบที่ 2 ใช้ฟังก์ชันชุดนี้

เซลล์สุดท้ายของส่วนนี้นิยามแบบจำลองออสซิลเลเตอร์สี่แบบ (M1–M4) และฟังก์ชันปรับเส้นโค้งที่ใช้ในตารางเกณฑ์สารสนเทศของส่วนที่ 2 และในแบบฝึก E9.2

```python
#@title ฟังก์ชันเมโทรโพลิสจากโค้ดที่ 4.6 (ดับเบิลคลิกเพื่อดูโค้ด)
import numpy as np
rng = np.random.default_rng(seed=2026)

def init_lattice(L, hot_start=True):
    if hot_start:
        return rng.choice([-1, 1], size=(L, L))
    return np.ones((L, L), dtype=int)

def total_energy(lattice):
    # each bond counted once: one neighbour along each axis
    return -np.sum(lattice * (np.roll(lattice, 1, 0) + np.roll(lattice, 1, 1)))

def metropolis_step(lattice, beta):
    """One sweep = L*L attempted single-spin flips."""
    L = lattice.shape[0]
    for _ in range(L * L):
        i, j = rng.integers(0, L, size=2)
        s = lattice[i, j]
        nb = (lattice[(i+1) % L, j] + lattice[(i-1) % L, j]
            + lattice[i, (j+1) % L] + lattice[i, (j-1) % L])
        dE = 2 * s * nb
        if dE <= 0 or rng.random() < np.exp(-beta * dE):
            lattice[i, j] = -s
    return lattice

print('init_lattice, total_energy, metropolis_step พร้อมใช้')
```

```python
#@title แบบจำลอง M1–M4 และฟังก์ชันปรับเส้นโค้งสำหรับเกณฑ์สารสนเทศ (ดับเบิลคลิกเพื่อดูโค้ด)
from scipy import stats
from sklearn.model_selection import KFold

def M1(t, A, w, p): return A * np.cos(w * t + p)                                     # undamped
def M2(t, A, g, w, p): return A * np.exp(-g * t) * np.cos(w * t + p)                 # damped (true form)
def M3(t, A, g, w, p, c): return M2(t, A, g, w, p) + c                               # + zero offset
def M4(t, A, g, w, p, B, g2, w2, p2): return M2(t, A, g, w, p) + M2(t, B, g2, w2, p2)  # two damped terms


MODELS = {'M1': (M1, [1, 6, 0], ([0, 0, -np.pi], [10, 20, np.pi])),
          'M2': (M2, [1, .2, 6, 0], ([0, 0, 0, -np.pi], [10, 5, 20, np.pi])),
          'M3': (M3, [1, .2, 6, 0, 0], ([0, 0, 0, -np.pi, -1], [10, 5, 20, np.pi, 1])),
          'M4': (M4, None, ([0, 0, 0, -np.pi, 0, -5, 0, -np.pi], [10, 5, 20, np.pi, 10, 5, 20, np.pi]))}


def fit(f, p0, bounds, tt, xx, sig):
    return curve_fit(f, tt, xx, p0=p0, bounds=bounds, sigma=sig * np.ones(tt.size), absolute_sigma=True, maxfev=20000)


def fit_m4_multistart(tt, xx, sig):
    """M4 has many local minima: start from M2's best fit plus a small second term at ~400 starting points."""
    base = fit(M2, [1, .2, 6, 0], MODELS['M2'][2], tt, xx, sig)[0]
    found = []
    for w2 in np.arange(0.2, 20, 0.2):
        for g2 in (-0.3, -0.05, 0.05, 0.5):
            for p2 in (-2, 0, 2):
                try:
                    p, c = fit(M4, list(base) + [0.03, g2, w2, p2], MODELS['M4'][2], tt, xx, sig)
                except (RuntimeError, ValueError):
                    continue
                found.append((np.sum(((xx - M4(tt, *p)) / sig)**2), p, c))
    found.sort(key=lambda r: r[0])
    return found


def ic_table(tt, xx, sig, models=("M1", "M2", "M3", "M4"), cv_folds=True):
    """chi2_min, AIC, BIC and 10-fold CV (blocked, shuffled) for the chosen models; returns a DataFrame."""
    n = tt.size
    rows, fits = {}, {}
    for name in models:
        f, p0_, bnd = MODELS[name]
        if name == "M4":
            chi2, p, c = fit_m4_multistart(tt, xx, sig)[0]
        else:
            p, c = fit(f, p0_, bnd, tt, xx, sig)
            chi2 = np.sum(((xx - f(tt, *p)) / sig)**2)
        k = len(p)
        row = {"k": k, "chi2_min": chi2, "chi2_red": chi2 / (n - k), "AIC": chi2 + 2 * k, "BIC": chi2 + k * np.log(n)}
        if cv_folds:
            for cv_name, splitter in [("CV blocked", KFold(10)), ("CV shuffled", KFold(10, shuffle=True, random_state=0))]:
                err = []
                for tr, te in splitter.split(tt):
                    try:
                        pp, _ = curve_fit(f, tt[tr], xx[tr], p0=p, bounds=bnd, maxfev=20000)
                    except RuntimeError:
                        pp = p
                    err.append(np.mean((xx[te] - f(tt[te], *pp))**2))
                row[cv_name] = np.sqrt(np.mean(err))
        rows[name], fits[name] = row, (p, np.sqrt(np.diag(c)))
    return pd.DataFrame(rows).T, fits

print("M1, M2, M3, M4, fit, fit_m4_multistart, ic_table พร้อมใช้")
```

### แนวคิดสำคัญ
สมมติว่านักศึกษาสองคนฝึกแบบจำลองทำนายความจุความร้อนของวัสดุชนิดใหม่ คนแรกรายงาน $C = 1.43$ J/(g K) คนที่สองรายงาน $C = 1.43 \pm 0.40$ J/(g K) ตัวเลขกลางเท่ากัน แต่รายงานที่สองบอกผู้อ่านว่าตัวเลขนี้แทบแยกวัสดุนี้ออกจากวัสดุที่มีความจุความร้อน 1.1 หรือ 1.8 ไม่ได้ ถ้าจะนำผลไปออกแบบระบบระบายความร้อน รายงานที่สองมีประโยชน์กว่า แม้จะดู "แม่นน้อยกว่า"

หลักข้อที่ห้าของตำรา (P5) กล่าวว่าการวัดทุกครั้งมีความไม่แน่นอน แบบจำลองที่เรียนรู้จากข้อมูลต้องปฏิบัติตามหลักเดียวกัน และหนักกว่าเดิม เพราะผลจากแบบจำลองสะสมความไม่แน่นอนจากหลายแหล่งพร้อมกัน ได้แก่ สัญญาณรบกวนในข้อมูลฝึก การที่ข้อมูลฝึกเป็นเพียงตัวอย่างหนึ่งจากหลายตัวอย่างที่อาจเก็บได้ และรูปแบบของแบบจำลองที่อาจไม่ตรงกับกลไกจริง

> **นิยาม (ความไม่แน่นอนสองชนิด)** ความไม่แน่นอนแบบเอลีอะทอริก (aleatoric uncertainty) มาจากความสุ่มที่มีอยู่ในระบบหรือในกระบวนการวัดเอง เช่น การแกว่งเชิงความร้อน สัญญาณรบกวนแบบช็อต หรือการสลายตัวของนิวเคลียส ความไม่แน่นอนชนิดนี้ **ไม่ลดลงเมื่อเก็บข้อมูลเพิ่ม** ความไม่แน่นอนแบบเอพิสเทมิก (epistemic uncertainty) มาจากความรู้ที่จำกัดของเรา เช่น พารามิเตอร์ที่ประมาณจากข้อมูลจำนวนจำกัด หรือรูปแบบของแบบจำลองที่เลือกโดยไม่แน่ใจ ความไม่แน่นอนชนิดนี้ **ลดลงได้** เมื่อมีข้อมูลมากขึ้นหรือมีแบบจำลองที่ดีขึ้น

คำว่า aleatoric มาจากภาษาละติน *alea* แปลว่าลูกเต๋า ส่วน epistemic มาจากภาษากรีก *episteme* แปลว่าความรู้ ถ้ารู้แน่ว่าลูกเต๋าเที่ยงตรง ผลการทอยครั้งถัดไปก็ยังไม่แน่นอน (เอลีอะทอริก) แต่ถ้าไม่รู้ว่าลูกเต๋าถ่วงน้ำหนักหรือไม่ ความไม่แน่นอนส่วนนี้เป็นแบบเอพิสเทมิก และลดได้ด้วยการทอยซ้ำหลายร้อยครั้งแล้วนับ

การแยกสองชนิดนี้มีผลในทางปฏิบัติโดยตรง ถ้าความไม่แน่นอนส่วนใหญ่เป็นแบบเอพิสเทมิก การเก็บข้อมูลเพิ่มหรือปรับแบบจำลองจะช่วยได้ ถ้าส่วนใหญ่เป็นแบบเอลีอะทอริก ควรไปปรับที่การวัดแทน เช่น ลดอุณหภูมิของเครื่องตรวจวัดเพื่อลดสัญญาณรบกวนเชิงความร้อน

> **ความเข้าใจคลาดเคลื่อนที่พบบ่อย:** "เอลีอะทอริกคือความคลาดเคลื่อนเชิงสุ่ม และเอพิสเทมิกคือความคลาดเคลื่อนเชิงระบบ" สองคู่นี้แยกด้วยคำถามต่างกัน คู่แรกแยกตามว่าความรู้เพิ่มลดความไม่แน่นอนได้หรือไม่ คู่หลังแยกตามพฤติกรรมเมื่อวัดซ้ำ ความชันที่ประมาณจากข้อมูลน้อยจุดแกว่งแบบสุ่มเมื่อทดลองซ้ำ แต่เป็นแบบเอพิสเทมิก ส่วนค่าชดเชยของเครื่องมือที่ไม่เคยปรับเทียบเป็นความคลาดเคลื่อนเชิงระบบ และก็เป็นแบบเอพิสเทมิกด้วย

# ส่วนที่ 2 การตรวจสอบไขว้และการเลือกแบบจำลอง (หัวข้อ 9.2)

### รูปแบบของการตรวจสอบไขว้ต้องตรงกับโครงสร้างของข้อมูล
การตรวจสอบไขว้แบบ $k$ ส่วน (k-fold cross-validation) แบ่งข้อมูลเป็น $k$ ส่วน แต่ละรอบใช้ส่วนหนึ่งเป็นข้อมูลทดสอบและที่เหลือเป็นข้อมูลฝึก ความคิดเบื้องหลังคือการจำลองสถานการณ์ที่แบบจำลองต้องทำนายข้อมูลที่ไม่เคยเห็น **การตรวจสอบไขว้จะให้คำตอบที่ถูกต้องก็ต่อเมื่อการแบ่งข้อมูลเลียนแบบวิธีที่แบบจำลองจะถูกใช้จริง**

| โครงสร้างของข้อมูล | ตัวอย่างในฟิสิกส์ | การแบ่งที่ถูกต้อง |
|---|---|---|
| แถวเป็นอิสระและมาจากการแจกแจงเดียวกัน | เหตุการณ์ชนกันแต่ละครั้ง | สลับแล้วแบ่งแบบสุ่ม หรือแบบรักษาสัดส่วน (stratified k-fold) เมื่อประเภทไม่สมดุล |
| ข้อมูลเป็นกลุ่ม | หลายจุดวัดจากตัวอย่างชิ้นเดียวกัน | แบ่งตามกลุ่ม (group k-fold) |
| สัมพันธ์กันในเวลาหรือในปริภูมิ | อนุกรมเวลาจากเซนเซอร์ เฟรมที่ติดกันของการจำลอง | แบ่งเป็นช่วงต่อเนื่อง (blocked k-fold) โดยไม่สลับลำดับ |

ถ้าจุดหนึ่งอยู่ในข้อมูลทดสอบและจุดข้างเคียงในเวลาอยู่ในข้อมูลฝึก การ "ทำนาย" จะกลายเป็นการลอกคำตอบจากเพื่อนบ้าน ซึ่งเป็นการรั่วไหลของข้อมูล (data leakage)

### ข้อมูลของโค้ดที่ 9.1
ตัวแปรควบคุม $x(t)$ เปลี่ยนช้า ๆ ตามเวลา ค่าที่วัดได้คือ $y = 2x + \varepsilon(t)$ โดยสัญญาณรบกวน $\varepsilon$ เป็นกระบวนการ AR(1) ที่มีสหสัมพันธ์ในเวลาสูง ส่วนเบี่ยงเบนมาตรฐานของ $\varepsilon$ คือ $0.07/\sqrt{1 - 0.99^2} \approx 0.50$ ซึ่งเป็น **ขีดล่างของความคลาดเคลื่อน** ที่แบบจำลองใด ๆ ทำได้กับการวัดครั้งใหม่ เราฝึกป่าสุ่มสองแบบ แบบแรกใช้เพียง $x$ แบบที่สองใช้ทั้ง $x$ และเวลา $t$ ซึ่งผู้เริ่มต้นทำบ่อยเพราะเวลาถูกบันทึกมาพร้อมข้อมูล แล้วเทียบการตรวจสอบไขว้แบบสลับ แบบเป็นช่วง และการทดสอบกับการวัดครั้งใหม่ที่เป็นอิสระ

### โค้ดที่ 9.1 — การตรวจสอบไขว้แบบสลับกับแบบเป็นช่วงบนข้อมูลที่สัมพันธ์กันในเวลา

```python
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import KFold, cross_val_score

def ar1(n, phi, sigma, rng):
    """AR(1) process: e[i] = phi * e[i-1] + white noise."""
    e = np.empty(n)
    e[0] = rng.normal(scale=sigma / np.sqrt(1 - phi**2))
    for i in range(1, n):
        e[i] = phi * e[i - 1] + rng.normal(scale=sigma)
    return e

def make_run(n=2000, seed=0):
    rng = np.random.default_rng(seed)
    t = np.arange(n) * 1.0
    x = ar1(n, 0.995, 0.1, rng)                 # slowly varying control
    y = 2.0 * x + ar1(n, 0.99, 0.07, rng)       # response + correlated noise
    return np.c_[x, t], y

X, y = make_run(seed=2026)
X_new, y_new = make_run(seed=7)                 # an independent new run
model = RandomForestRegressor(200, min_samples_leaf=5, random_state=0, n_jobs=-1)
for cols in ([0], [0, 1]):                      # x only / x and t
    for name, cv in [('shuffled', KFold(5, shuffle=True, random_state=0)),
                     ('blocked',  KFold(5, shuffle=False))]:
        rmse = -cross_val_score(model, X[:, cols], y, cv=cv,
                                scoring='neg_root_mean_squared_error')
        print(cols, name, rmse.mean().round(3))
    model.fit(X[:, cols], y)
    rmse_new = np.sqrt(np.mean((model.predict(X_new[:, cols]) - y_new)**2))
    print(cols, 'new run', rmse_new.round(3))
```

```python
fig, ax = plt.subplots(1, 2, figsize=(11, 3.4))
ax[0].plot(X[:, 1], y, lw=0.7, label="y (measured)")
ax[0].plot(X[:, 1], 2 * X[:, 0], lw=0.9, label="2x (control)")
ax[0].set(xlabel="time t", ylabel="value", title="AR(1) run, seed 2026 (SYNTHETIC)"); ax[0].legend(fontsize=8)
res = y - 2 * X[:, 0]
ax[1].plot(X[:, 1], res, lw=0.7, color="C2")
ax[1].axhline(0.496, ls="--", c="k", lw=0.8); ax[1].axhline(-0.496, ls="--", c="k", lw=0.8)
ax[1].set(xlabel="time t", ylabel="noise y - 2x", title="correlated noise, stationary sd 0.496 (SYNTHETIC)")
fig.tight_layout(); plt.show()
print(f"stationary sd of the noise: {0.07 / np.sqrt(1 - 0.99**2):.3f}; sample sd in this run {res.std():.3f}")
```

**อ่านผล:** เมื่อใช้ $x$ อย่างเดียว แบบสลับให้ 0.42 แบบเป็นช่วงให้ 0.51 และการวัดครั้งใหม่ให้ 0.56 เมื่อเพิ่มเวลาเป็นลักษณะเด่น แบบสลับให้เพียง **0.18 ต่ำกว่าขีดล่าง 0.50 ราวสามเท่า** ซึ่งเป็นไปไม่ได้สำหรับการวัดครั้งใหม่ แบบเป็นช่วงให้ 0.58 และการวัดครั้งใหม่ให้ 0.66 ตัวเลข 0.18 ไม่ได้บอกว่าแบบจำลองดี มันบอกว่าแบบจำลองใช้เวลาเป็นดัชนีค้นหาสัญญาณรบกวนของจุดข้างเคียงในข้อมูลชุดเดิม และการเพิ่มเวลายังทำให้ผลกับข้อมูลใหม่แย่ลง (0.56 เป็น 0.66)

**หลักตรวจที่ใช้ได้ทั่วไป:** ถ้าความคลาดเคลื่อนจากการตรวจสอบไขว้ต่ำกว่าระดับสัญญาณรบกวนที่รู้จากฟิสิกส์ของการวัด ให้สงสัยการรั่วไหลก่อนเสมอ ส่วนความต่างที่เหลือระหว่างแบบเป็นช่วง 0.58 กับข้อมูลใหม่ 0.66 เกิดจากช่วงเวลาห้าช่วงยังไม่ครอบคลุมความแปรผันระหว่างการวัดคนละครั้ง ถ้ามีการวัดหลายครั้ง ควรแบ่งตามครั้งของการวัด ซึ่งก็คือการแบ่งตามกลุ่ม

### อคติกับความแปรปรวนในมุมมองของความไม่แน่นอน
การแยกความคลาดเคลื่อนเฉลี่ยกำลังสองของการทำนายในบทที่ 5
$$\mathbb{E}\big[(y - \hat f(x))^2\big] = \big(f(x) - \mathbb{E}[\hat f(x)]\big)^2 + \mathrm{Var}\big[\hat f(x)\big] + \sigma^2$$
เชื่อมกับสองชนิดของความไม่แน่นอนได้โดยตรง พจน์ $\sigma^2$ เป็นแบบเอลีอะทอริก ความแปรปรวนเป็นแบบเอพิสเทมิกที่ลดลงเมื่อข้อมูลเพิ่ม ส่วนอคติเป็นแบบเอพิสเทมิกที่ **ข้อมูลเพิ่มแก้ไม่ได้** ต้องแก้ด้วยการเปลี่ยนรูปแบบของแบบจำลอง ผลนี้บอกข้อจำกัดของเครื่องมือหลายชนิดในบทนี้ล่วงหน้า บูตสแตรปและการแจกแจงภายหลังวัดความแปรปรวน *ภายในรูปแบบที่เลือก* แต่มองไม่เห็นอคติจากรูปแบบที่ผิด เครื่องมือที่ตรวจรูปแบบได้คือการตรวจส่วนเหลือ การตรวจสอบไขว้ เกณฑ์สารสนเทศ และการตรวจเชิงทำนายภายหลัง (ส่วนที่ 5)

### เกณฑ์สารสนเทศ AIC และ BIC
ถ้าแบบจำลองมีภาวะน่าจะเป็น (likelihood) ที่ชัดเจน มีทางลัดที่ไม่ต้องฝึกแบบจำลองหลายครั้ง เรียกว่าเกณฑ์สารสนเทศ (information criteria) ซึ่งให้คะแนนความเข้ากันแล้วหักด้วยค่าปรับของความซับซ้อน

> **นิยาม (AIC และ BIC)** สำหรับแบบจำลองที่มีพารามิเตอร์ $k$ ตัว ภาวะน่าจะเป็นสูงสุด $\hat{\mathcal L}$ และข้อมูล $N$ จุด
> $$\mathrm{AIC} = -2\ln\hat{\mathcal L} + 2k, \qquad \mathrm{BIC} = -2\ln\hat{\mathcal L} + k\ln N$$
> ค่าที่ต่ำกว่าถือว่าดีกว่า สำหรับสัญญาณรบกวนแบบเกาส์ที่รู้ $\sigma_i$ จะได้ $-2\ln\hat{\mathcal L} = \chi^2_{\min} + \text{ค่าคงที่}$ เกณฑ์ทั้งสองจึงเขียนเป็น $\chi^2_{\min} + 2k$ และ $\chi^2_{\min} + k\ln N$

AIC มาจากการประมาณว่าแบบจำลองจะทำนายข้อมูลชุดใหม่ได้ดีเพียงใด จึงเป็นเกณฑ์เพื่อการทำนาย ส่วน BIC มาจากการประมาณความน่าจะเป็นภายหลังของแบบจำลองในกรอบแบบเบส์ ถ้าแบบจำลองที่ถูกต้องอยู่ในรายการ BIC จะเลือกมันได้เมื่อข้อมูลมากพอ เมื่อ $\ln N > 2$ BIC ปรับพารามิเตอร์ที่เพิ่มขึ้นหนักกว่า และเลือกแบบจำลองที่ง่ายกว่า สำหรับนักฟิสิกส์ ถ้าเป้าหมายคือทำนายให้แม่น พจน์เล็ก ๆ ที่ช่วยได้บ้างก็ควรเก็บไว้ ถ้าเป้าหมายคือระบุว่ากลไกใดมีอยู่จริง ควรเรียกร้องหลักฐานที่หนักแน่นกว่า

### ✏️ แบบฝึก E9.2 ในชั้นเรียน (5 นาที) เกณฑ์การเลือกแบบจำลองเมื่อข้อมูลสั้น
เซลล์ถัดไปสร้างข้อมูลออสซิลเลเตอร์ด้วยพารามิเตอร์และเมล็ดสุ่มเดียวกับโค้ดที่ 5.6 แต่วัดเพียง **60 จุดในช่วง 3 วินาทีแรก** แล้วปรับ M1, M2 และ M3 ให้ ($\chi^2_{\min}$ อยู่ใน `chi2_short`) คำนวณ AIC และ BIC ของทั้งสามแบบจำลองเก็บใน `my_aic[name]` และ `my_bic[name]` แล้วพิจารณาว่าแต่ละเกณฑ์เลือกแบบจำลองใด และผลต่างมากพอจะตัดสินหรือไม่ (ส่วน M4 และการตรวจสอบไขว้เป็นงานนอกชั้น)

```python
rng = np.random.default_rng(seed=2026)
t_s = np.linspace(0, 3, 60)
x_s = damped_oscillator(t_s, 1.0, 0.3, 6.0, 0.5) + rng.normal(scale=0.05, size=t_s.size)
chi2_short, k_short = {}, {}
for name in ("M1", "M2", "M3"):
    f, p0_, bnd = MODELS[name]
    p, _ = fit(f, p0_, bnd, t_s, x_s, 0.05)
    chi2_short[name], k_short[name] = np.sum(((x_s - f(t_s, *p)) / 0.05)**2), len(p)
print({k: round(v, 1) for k, v in chi2_short.items()}, " N =", t_s.size)

my_aic = {}   # ✏️ {'M1': ..., 'M2': ..., 'M3': ...}
my_bic = {}   # ✏️
```

```python
# ✅ ตรวจคำตอบ
for name in ("M1", "M2", "M3"):
    a = chi2_short[name] + 2 * k_short[name]; b = chi2_short[name] + k_short[name] * np.log(60)
    ma, mb = my_aic.get(name), my_bic.get(name)
    ca = "(ยังไม่ได้ตอบ)" if ma is None else ("✔" if abs(ma - a) < 0.05 else "✘")
    cb = "(ยังไม่ได้ตอบ)" if mb is None else ("✔" if abs(mb - b) < 0.05 else "✘")
    print(f"{name}: AIC {ca} — {a:.1f}   BIC {cb} — {b:.1f}")
```

### การตรวจสอบไขว้แบบซ้อนเมื่อต้องเลือกไฮเพอร์พารามิเตอร์
ในบทที่ 5 และ 6 เราใช้การตรวจสอบไขว้ทำสองงานพร้อมกัน คือเลือกไฮเพอร์พารามิเตอร์ แล้วรายงานคะแนนของค่าที่ดีที่สุดเป็นผลการประเมิน คะแนนที่ดีที่สุดในบรรดาหลายสิบค่ามีอคติเข้าข้างตัวเอง เหมือนการทอยลูกเต๋าสิบลูกแล้วรายงานเฉพาะหน้าที่สูงที่สุดว่าเป็น "ผลการทอย"

การตรวจสอบไขว้แบบซ้อน (nested cross-validation) มีสองชั้น ชั้นในใช้ข้อมูลฝึกของแต่ละรอบเลือกไฮเพอร์พารามิเตอร์ ชั้นนอกใช้ข้อมูลทดสอบที่ชั้นในไม่เคยเห็นประเมินทั้งกระบวนการ รวมขั้นตอนการเลือกด้วย โค้ดที่ 9.2 แสดงผลอย่างสุดขั้วด้วยข้อมูลที่ **ป้ายกำกับเป็นการสุ่มล้วน ๆ** พื้นที่ใต้เส้นโค้ง ROC ที่แท้จริงของแบบจำลองใดก็ตามจึงเป็น 0.5

### โค้ดที่ 9.2 — อคติของการรายงานคะแนนที่ดีที่สุด เทียบกับการตรวจสอบไขว้แบบซ้อน
⏱ ราวสองถึงสามนาทีบน Colab (ข้อมูล 20 ชุด ชุดละ 36 คู่ค่า × 5 ส่วน × 5 ส่วน) ระหว่างรอให้อ่านย่อหน้าหลังเซลล์

```python
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

res = []
for seed in range(20):                           # 20 independent data sets
    rng = np.random.default_rng(seed)
    X = rng.normal(size=(60, 20))
    y = rng.integers(0, 2, 60)                   # labels are pure noise
    grid = {'svc__C': np.logspace(-2, 3, 6), 'svc__gamma': np.logspace(-4, 1, 6)}
    inner = StratifiedKFold(5, shuffle=True, random_state=seed)
    outer = StratifiedKFold(5, shuffle=True, random_state=seed + 100)
    gs = GridSearchCV(make_pipeline(StandardScaler(), SVC()), grid,
                      cv=inner, scoring='roc_auc')
    gs.fit(X, y)
    nested = cross_val_score(gs, X, y, cv=outer, scoring='roc_auc').mean()
    res.append((gs.best_score_, nested))
print(np.mean(res, axis=0).round(3))             # [best inner score, nested]
```

```python
r_ = np.array(res)
fig, ax = plt.subplots(figsize=(7, 3.2))
ax.plot(range(20), r_[:, 0], "o-", label="best grid-search score")
ax.plot(range(20), r_[:, 1], "s-", label="nested CV")
ax.axhline(0.5, c="k", ls="--", lw=0.8, label="true AUC = 0.5")
ax.set(xlabel="data set (seed)", ylabel="ROC AUC", title="pure-noise labels (SYNTHETIC)", xticks=range(0, 20, 2)); ax.legend(fontsize=8)
fig.tight_layout(); plt.show()
print(f"best inner score > 0.5 in {np.sum(r_[:, 0] > 0.5)} of 20 data sets")
```

**อ่านผล:** ค่าเฉลี่ยจากข้อมูลยี่สิบชุดคือ 0.59 สำหรับคะแนนที่ดีที่สุดของการค้นหา 36 คู่ค่า และ 0.49 สำหรับการตรวจสอบไขว้แบบซ้อน คะแนนที่ดีที่สุดสูงกว่า 0.5 ใน 17 จาก 20 ชุด ตัวเลขแรกดูเหมือนแบบจำลองเรียนรู้อะไรได้ ทั้งที่ไม่มีอะไรให้เรียน ตัวเลขที่สองบอกความจริงว่าแบบจำลองเดาสุ่ม ในข้อมูลจริงที่มีสัญญาณ อคติจะเล็กกว่านี้แต่ไม่หายไป และจะมากขึ้นเมื่อข้อมูลน้อยหรือลองค่าหลายค่า

# ส่วนที่ 3 บูตสแตรปแบบไม่อิงพารามิเตอร์และแบบอิงพารามิเตอร์ (หัวข้อ 9.3.1–9.3.3)

### แนวคิดสำคัญ
ถ้าทำการทดลองซ้ำได้หลายร้อยครั้ง การหาความไม่แน่นอนของผลจะง่ายมาก แต่ในความเป็นจริงเรามีข้อมูลเพียงชุดเดียว เอฟรอน (Bradley Efron) เสนอในปี 1979 ว่า ถ้าข้อมูลที่มีเป็นตัวแทนที่ดีของประชากร ก็ใช้ข้อมูลนั้นแทนประชากรได้ แล้วจำลองการ "เก็บข้อมูลใหม่" ด้วยการสุ่มหยิบจุดจากข้อมูลเดิมแบบใส่คืน (sampling with replacement) ให้ได้จำนวนเท่าเดิม

> **นิยาม (บูตสแตรปแบบไม่อิงพารามิเตอร์)** ให้ข้อมูล $\{z_1, \dots, z_N\}$ และปริมาณ $\hat\theta = s(z_1, \dots, z_N)$ สร้างชุดบูตสแตรป $B$ ชุด แต่ละชุดได้จากการสุ่มหยิบ $N$ จุดแบบใส่คืน แล้วคำนวณ $\hat\theta^{*}_b$ ส่วนเบี่ยงเบนมาตรฐานของ $\{\hat\theta^{*}_b\}$ ประมาณความคลาดเคลื่อนมาตรฐานของ $\hat\theta$ และเปอร์เซ็นไทล์ที่ 2.5 และ 97.5 ให้ช่วงความเชื่อมั่นร้อยละ 95 แบบเปอร์เซ็นไทล์ (percentile interval)

วิธีนี้ไม่ได้สร้างข้อมูลใหม่ มันใช้ข้อมูลที่มีประมาณ *รูปร่างของความแปรปรวน* อุปมาที่ใกล้ตัวคือการชิมแกง ตักชิมหลายช้อนจากหลายตำแหน่งเพื่อรู้ว่ารสสม่ำเสมอเพียงใด ซึ่งได้ผลก็ต่อเมื่อแกงคนมาทั่วหม้อแล้ว จุดแข็งของบูตสแตรปคือไม่ต้องสมมติรูปแบบของสัญญาณรบกวน และไม่ต้องหาอนุพันธ์ของปริมาณที่สนใจ จึงใช้กับปริมาณที่ซับซ้อนได้ เช่น มัธยฐาน สัมประสิทธิ์สหสัมพันธ์ พื้นที่ใต้เส้นโค้ง ROC หรือค่าเจาะจงของเมทริกซ์

โค้ดที่ 9.3 ใช้บูตสแตรปกับออสซิลเลเตอร์แบบหน่วง แต่ละรอบสุ่มหยิบคู่ $(t_i, x_i)$ แบบใส่คืน ปรับเส้นโค้งใหม่ แล้วเก็บพารามิเตอร์ ปริมาณที่สนใจคือตัวประกอบคุณภาพ $Q = \omega/(2\gamma)$ ซึ่งแบบฝึกหัด E5.7 คำนวณความไม่แน่นอนด้วยการแพร่ความไม่แน่นอนเชิงเส้นไว้แล้ว

### โค้ดที่ 9.3 — บูตสแตรปแบบสุ่มคู่ข้อมูลสำหรับพารามิเตอร์ของการปรับเส้นโค้ง

```python
# continues from the damped-oscillator fit (popt, pcov, t, x_obs, bounds)
rng = np.random.default_rng(seed=1)
n_boot = 1000
params_boot = []
for _ in range(n_boot):
    idx = rng.integers(0, len(t), len(t))          # resample pairs (t_i, x_i)
    try:
        p_b, _ = curve_fit(damped_oscillator, t[idx], x_obs[idx],
                           p0=popt, bounds=bounds)
        params_boot.append(p_b)
    except RuntimeError:                           # rare non-convergence
        pass
params_boot = np.array(params_boot)
Q_boot = params_boot[:, 2] / (2 * params_boot[:, 1])
print("SE(gamma), SE(omega):", params_boot[:, 1:3].std(axis=0).round(4))
print("Q 95% percentile interval:", np.percentile(Q_boot, [2.5, 97.5]).round(2))
```

```python
print("curve_fit SE(gamma), SE(omega):", np.sqrt(np.diag(pcov))[1:3].round(4))
fig, ax = plt.subplots(1, 2, figsize=(10, 3.2))
ax[0].scatter(params_boot[:, 1], params_boot[:, 2], s=4, alpha=0.4)
ax[0].plot(popt[1], popt[2], "r*", ms=12, label="curve_fit")
ax[0].set(xlabel="gamma (1/s)", ylabel="omega (rad/s)", title="1000 bootstrap fits (SYNTHETIC)"); ax[0].legend(fontsize=8)
ax[1].hist(Q_boot, bins=40, alpha=0.8)
for q in np.percentile(Q_boot, [2.5, 97.5]):
    ax[1].axvline(q, c="k", ls="--", lw=0.8)
ax[1].axvline(10, c="r", lw=1, label="true Q = 10")
ax[1].set(xlabel="Q = omega / (2 gamma)", ylabel="count", title="bootstrap distribution of Q"); ax[1].legend(fontsize=8)
fig.tight_layout(); plt.show()
```

**อ่านผล:** สำหรับข้อมูล 200 จุดที่สัญญาณรบกวนต่ำ ผลจากบูตสแตรปกับผลจากเมทริกซ์ความแปรปรวนร่วมของ `curve_fit` แทบเท่ากัน ความคลาดเคลื่อนมาตรฐานของ $\gamma$ คือ 0.0083 เทียบกับ 0.0076 และของ $\omega$ คือ 0.0073 เทียบกับ 0.0074 ช่วงของ $Q$ คือ $[9.95, 11.16]$ ในบริเวณที่ความไม่แน่นอนเล็ก การประมาณเชิงเส้นใช้ได้ และไม่จำเป็นต้องใช้วิธีที่แพงกว่า

### บูตสแตรปแบบอิงพารามิเตอร์
บูตสแตรปแบบสุ่มคู่ข้อมูลสมมติว่าแต่ละจุดเป็นตัวอย่างสุ่มจากประชากร แต่ในการทดลองฟิสิกส์ ผู้ทดลองมักกำหนดค่าตัวแปรอิสระเอง เช่น วัดทุก 0.05 วินาที การสุ่มหยิบคู่ข้อมูลจึงสร้างชุดที่มีเวลาซ้ำและเวลาขาดหาย ถ้ารู้แบบจำลองของสัญญาณรบกวน มีทางเลือกที่เลียนแบบการทดลองได้ตรงกว่า เรียกว่า **บูตสแตรปแบบอิงพารามิเตอร์ (parametric bootstrap)** ใช้เส้นโค้งที่ปรับได้สร้างข้อมูลจำลองใหม่ที่ค่า $t_i$ เดิม บวกสัญญาณรบกวนสุ่ม $\mathcal N(0, \sigma^2)$ แล้วปรับเส้นโค้งซ้ำ ในภาษาของนักทดลอง นี่คือการถามว่า "ถ้าแบบจำลองที่ปรับได้คือความจริง และเราทำการทดลองเดิมซ้ำ ผลจะกระจายเพียงใด" ทางเลือกที่สามซึ่งอยู่ระหว่างกลางคือการสุ่มซ้ำส่วนเหลือ (residual bootstrap)

ฟังก์ชัน `q_intervals` ในเซลล์ถัดไปคำนวณช่วงร้อยละ 95 ของ $Q$ ด้วยสามวิธี คือการแพร่เชิงเส้น ($\hat Q \pm 1.96\sigma_Q$) บูตสแตรปแบบสุ่มคู่ และแบบอิงพารามิเตอร์ แล้วเปรียบเทียบข้อมูลสองชุด ชุดดี 200 จุดที่ $\sigma = 0.05$ m กับชุดแย่ 60 จุดในเวลาเท่าเดิมที่ $\sigma = 0.4$ m (ค่าจริง $Q = 10$ ทั้งสองชุด)

```python
def q_intervals(n, sig, n_boot=1000):
    rng_d = np.random.default_rng(seed=2026)
    tt = np.linspace(0, 10, n)
    xx = damped_oscillator(tt, 1.0, 0.3, 6.0, 0.5) + rng_d.normal(scale=sig, size=n)
    p, c = curve_fit(damped_oscillator, tt, xx, p0=[1.0, 0.2, 6.0, 0.0], bounds=bounds,
                     sigma=sig * np.ones(n), absolute_sigma=True)
    Q = p[2] / (2 * p[1])
    J = np.array([0, -p[2] / (2 * p[1]**2), 1 / (2 * p[1]), 0])
    sQ = np.sqrt(J @ c @ J)
    r = np.random.default_rng(1); pairs = []
    for _ in range(n_boot):
        i = r.integers(0, n, n)
        try:
            pairs.append(curve_fit(damped_oscillator, tt[i], xx[i], p0=p, bounds=bounds)[0])
        except RuntimeError:
            pass
    r = np.random.default_rng(1); param = []
    for _ in range(n_boot):
        xs = damped_oscillator(tt, *p) + r.normal(scale=sig, size=n)
        try:
            param.append(curve_fit(damped_oscillator, tt, xs, p0=p, bounds=bounds,
                                   sigma=sig * np.ones(n), absolute_sigma=True)[0])
        except RuntimeError:
            pass
    pairs, param = np.array(pairs), np.array(param)
    return dict(Q=Q, gamma=(p[1], np.sqrt(c[1, 1])), linear=(Q - 1.96 * sQ, Q + 1.96 * sQ),
                pairs=np.percentile(pairs[:, 2] / (2 * pairs[:, 1]), [2.5, 97.5]),
                param=np.percentile(param[:, 2] / (2 * param[:, 1]), [2.5, 97.5]), n_ok=(len(pairs), len(param)))


# ⏱ ราว 1 นาทีบน Colab (ปรับเส้นโค้ง 4,000 ครั้ง)
t0 = time.time()
q_rows = {}
for n_, sig_ in [(200, 0.05), (60, 0.4)]:
    r_ = q_intervals(n_, sig_)
    iv = lambda a: f"[{a[0]:.2f}, {a[1]:.2f}]"
    q_rows[f"{n_} pts, sigma {sig_}"] = {"best Q": f"{r_['Q']:.2f}", "gamma": f"{r_['gamma'][0]:.3f} +/- {r_['gamma'][1]:.3f}",
        "linear": iv(r_["linear"]), "pairs bootstrap": iv(r_["pairs"]), "parametric bootstrap": iv(r_["param"])}
print(pd.DataFrame(q_rows).to_string())
print(f"({time.time() - t0:.0f} s)")
```

**อ่านผล:** กรณีข้อมูลแย่แสดงสิ่งที่การแพร่เชิงเส้นมองไม่เห็น $Q$ แปรผกผันกับ $\gamma$ และเมื่อ $\gamma = 0.21 \pm 0.08$ (ความไม่แน่นอนสัมพัทธ์ราวร้อยละ 40) ชุดบูตสแตรปบางชุดได้ $\gamma$ ใกล้ศูนย์ ทำให้ $Q$ พุ่งสูงมาก การแจกแจงของ $Q$ จึงเบ้ขวาอย่างรุนแรง ช่วงเชิงเส้น $[3.3, 24.5]$ ซึ่งสมมาตรโดยโครงสร้างจึงผิดทั้งสองด้าน ด้านล่างรวมค่า $Q \approx 3$ ที่ข้อมูลแทบไม่สนับสนุน ด้านบนตัดค่าสูงที่เป็นไปได้จริงทิ้ง ส่วนเบี่ยงเบนมาตรฐานของ $Q$ จากบูตสแตรปในกรณีนี้ไม่มีความหมาย เพราะค่าไม่กี่ค่าที่ $\gamma$ แทบเป็นศูนย์ทำให้มันใหญ่ได้ไม่จำกัด สิ่งที่ใช้ได้คือช่วงแบบเปอร์เซ็นไทล์ บูตสแตรปทั้งสองแบบครอบค่าจริง 10 แต่กว้างมาก และความกว้างนี้เองคือข้อมูลที่ถูกต้อง แบบอิงพารามิเตอร์ให้ช่วงแคบกว่า ($[6.5, 50.1]$ เทียบกับ $[5.9, 78.7]$) เพราะใช้ความรู้ว่าสัญญาณรบกวนเป็นแบบเกาส์ด้วยขนาดที่รู้แน่นอน ซึ่งมีประสิทธิภาพกว่าเมื่อแบบจำลองสัญญาณรบกวนถูกต้อง แต่จะผิดตามถ้าแบบจำลองนั้นผิด

### บูตสแตรปของตัวชี้วัดและข้อจำกัด
บทที่ 6 ใช้การสุ่มซ้ำข้อมูลทดสอบหาช่วงความเชื่อมั่นของพื้นที่ใต้เส้นโค้ง ROC และบทที่ 7 ใช้การสุ่มซ้ำวัดความเสถียรของผลการจัดกลุ่ม การสุ่มซ้ำเฉพาะข้อมูลทดสอบโดยใช้แบบจำลองเดิม วัดได้เพียงความไม่แน่นอนที่มาจากการที่ข้อมูลทดสอบมีจำนวนจำกัด ไม่รวมความไม่แน่นอนจากการฝึก รายงานจึงต้องบอกว่าสุ่มซ้ำอะไร บูตสแตรปทำงานดีกับปริมาณที่ "เรียบ" เช่น ค่าเฉลี่ย แต่ไม่ดีกับปริมาณที่ขึ้นกับค่าสุดขั้ว เช่น ค่าสูงสุด และต้องการข้อมูลพอสมควร

> **ความเข้าใจคลาดเคลื่อนที่พบบ่อย:** "ยิ่งสุ่มซ้ำหลายรอบ ความไม่แน่นอนยิ่งเล็กลง" การเพิ่มจำนวนรอบ $B$ ทำให้ *ค่าประมาณ* ของความคลาดเคลื่อนมาตรฐานนิ่งขึ้น แต่ไม่ได้ทำให้ความคลาดเคลื่อนมาตรฐานเล็กลง ความคลาดเคลื่อนลดลงได้ด้วยการเพิ่มจำนวนข้อมูล $N$ เท่านั้น ในทางปฏิบัติ $B$ ราว 200 พอสำหรับความคลาดเคลื่อนมาตรฐาน และราว 1,000–2,000 พอสำหรับช่วงแบบเปอร์เซ็นไทล์

# ส่วนที่ 4 ความคลาดเคลื่อนของผลจากการจำลองมอนติคาร์โล (หัวข้อ 9.3.4–9.3.5)

### ตัวอย่างวิจัยของผู้เขียน
หัวข้อ 4.6 จำลองแบบจำลองไอซิงสองมิติด้วยวิธีเมโทรโพลิส และทิ้งคำถามไว้ว่าพลังงานเฉลี่ยและความจุความร้อนที่ได้คลาดเคลื่อนได้เท่าใด คำถามนี้เป็นงานประจำของผู้ทำการจำลองมอนติคาร์โลในฟิสิกส์เชิงสถิติ รวมถึงงานวิจัยของผู้เขียนในการจำลองระบบสปินและพอลิเมอร์ และเป็นจุดที่การใช้บูตสแตรปแบบตรงไปตรงมาให้คำตอบที่ผิดอย่างเป็นระบบ

สถานะหลังรอบกวาดหนึ่งต่างจากสถานะก่อนหน้าเพียงบางส่วน พลังงานของรอบกวาดที่ติดกันจึงสัมพันธ์กัน เหมือนการถามความเห็นคนสิบคนที่ยืนต่อแถวกันหน้าร้านเดียวกัน คำตอบมักคล้ายกัน จึงให้ข้อมูลน้อยกว่าการถามคนสิบคนที่สุ่มจากทั่วเมือง สูตร $s/\sqrt{N}$ และบูตสแตรปที่สุ่มทีละจุดสมมติความเป็นอิสระ จึงให้ค่าที่เล็กเกินจริงในแบบเดียวกัน

> **นิยาม (เวลาสหสัมพันธ์แบบปริพันธ์และขนาดตัวอย่างประสิทธิผล)** สำหรับอนุกรม $x_1, \dots, x_N$ ในสภาวะคงตัวที่มีฟังก์ชันสหสัมพันธ์ในตัว (autocorrelation function) $\rho(\tau)$ เวลาสหสัมพันธ์แบบปริพันธ์ (integrated autocorrelation time) คือ
> $$\tau_{\text{int}} = \frac{1}{2} + \sum_{\tau=1}^{\infty} \rho(\tau)$$
> และเมื่อ $N \gg \tau_{\text{int}}$
> $$\mathrm{Var}(\bar x) \approx \frac{2\tau_{\text{int}}\,\sigma_x^2}{N} = \frac{\sigma_x^2}{N_{\text{eff}}}, \qquad N_{\text{eff}} = \frac{N}{2\tau_{\text{int}}}$$
> เมื่อ $N_{\text{eff}}$ คือขนาดตัวอย่างประสิทธิผล (effective sample size) สำหรับข้อมูลอิสระ $\tau_{\text{int}} = 1/2$ และ $N_{\text{eff}} = N$

ผลรวมในนิยามไล่ไปถึงอนันต์ไม่ได้ เพราะ $\rho(\tau)$ ที่ประมาณจากข้อมูลมีสัญญาณรบกวนที่สะสมเมื่อ $\tau$ ใหญ่ โซคาล (Alan Sokal) เสนอให้หยุดที่หน้าต่าง $W$ แรกที่ $W \ge c\,\tau_{\text{int}}(W)$ โดย $c \approx 5$ วิธีที่สองซึ่งไม่ต้องคำนวณ $\rho(\tau)$ คือ **การสุ่มซ้ำเป็นช่วง (block bootstrap)** แบ่งอนุกรมเป็นช่วงต่อเนื่องยาว $b$ แล้วสุ่มหยิบ *ช่วง* แบบใส่คืน ถ้า $b \gg \tau_{\text{int}}$ สหสัมพันธ์ภายในช่วงถูกเก็บไว้ครบ และช่วงต่าง ๆ แทบเป็นอิสระต่อกัน (เท่ากับสุ่ม *แถว* ของคนแทนการสุ่มคนทีละคน) วิธีที่นิยมในฟิสิกส์คือ **การจัดกลุ่มซ้ำ (blocking)** ของฟลีฟเบิร์กและปีเตอร์เซน ซึ่งเฉลี่ยจุดที่ติดกันเป็นคู่ซ้ำหลายรอบ แล้วดูว่าความคลาดเคลื่อนมาตรฐานของค่าเฉลี่ยเพิ่มขึ้นจนคงที่เมื่อใด

โค้ดที่ 9.4 เก็บพลังงานต่อสปิน $e = E/N$ ของตาราง $16\times16$ ที่ $T = 2.3$ ซึ่งอยู่ใกล้อุณหภูมิวิกฤต ทุกรอบกวาดเป็นเวลา 20,000 รอบกวาด หลังทิ้งช่วงเข้าสู่สมดุล 2,000 รอบกวาด

```python
# uses init_lattice, total_energy and metropolis_step from the Chapter 4 listing
def energy_series(L, T, n_thermalize, n_measure):
    lattice = init_lattice(L)
    beta = 1.0 / T
    for _ in range(n_thermalize):
        metropolis_step(lattice, beta)
    E = np.empty(n_measure)
    for k in range(n_measure):
        metropolis_step(lattice, beta)
        E[k] = total_energy(lattice)
    return E

def tau_int(x, c=5.0):
    """Integrated autocorrelation time with Sokal's automatic window."""
    x = np.asarray(x) - np.mean(x)
    n = len(x)
    f = np.fft.rfft(x, 2 * n)                      # zero-padded FFT
    acf = np.fft.irfft(f * np.conj(f))[:n]
    acf /= acf[0]
    tau = 0.5
    for W in range(1, n):
        tau += acf[W]
        if W >= c * tau:
            break
    return tau, W

def block_bootstrap(x, stat, block, n_boot=2000, rng=None):
    rng = rng or np.random.default_rng(0)
    n_blocks = len(x) // block
    blocks = np.asarray(x)[:n_blocks * block].reshape(n_blocks, block)
    out = np.empty(n_boot)
    for b in range(n_boot):
        pick = rng.integers(0, n_blocks, n_blocks)   # resample whole blocks
        out[b] = stat(blocks[pick].ravel())
    return out

rng = np.random.default_rng(seed=2026)             # restart the generator
L, T = 16, 2.3
e = energy_series(L, T, n_thermalize=2000, n_measure=20000) / L**2
tau, W = tau_int(e)
se_naive = e.std(ddof=1) / np.sqrt(len(e))
print(f"<e> = {e.mean():.4f}, tau_int = {tau:.1f} sweeps (window {W})")
print(f"SE naive = {se_naive:.5f}, SE with tau = {se_naive*np.sqrt(2*tau):.5f}")
cv = lambda x: x.var() * L**2 / T**2               # C_v per spin
for stat, name in [(np.mean, '<e>'), (cv, 'C_v/N')]:
    boot = block_bootstrap(e, stat, block=200)
    print(name, "block-bootstrap SE", boot.std().round(5),
          "95% CI", np.percentile(boot, [2.5, 97.5]).round(4))
```

**อ่านผล:** ที่ $T = 2.3$ ได้ $\langle e\rangle = -1.4139$ และ $\tau_{\text{int}} = 17.5$ รอบกวาด ข้อมูล 20,000 จุดจึงเท่ากับจุดอิสระเพียงราว 570 จุด ความคลาดเคลื่อนมาตรฐานแบบไม่คิดสหสัมพันธ์คือ 0.0012 แต่ค่าที่แก้ด้วย $\tau_{\text{int}}$ คือ 0.0072 **ใหญ่กว่าเกือบหกเท่า** การสุ่มซ้ำเป็นช่วงยาว 200 รอบกวาดให้ 0.0067 ส่วนความจุความร้อนต่อสปินคือ 1.44 โดยความคลาดเคลื่อนมาตรฐานจากการสุ่มซ้ำเป็นช่วงคือ 0.041

เซลล์ถัดไปนิยามการจัดกลุ่มซ้ำ แล้ววาดฟังก์ชันสหสัมพันธ์ในตัวกับความคลาดเคลื่อนมาตรฐานของ $\langle e\rangle$ ที่แต่ละระดับของการจัดกลุ่ม

```python
def blocking(x, min_blocks=16):
    """Flyvbjerg-Petersen: SE of the mean after repeatedly averaging neighbouring pairs."""
    x = np.asarray(x, float); out = []
    while len(x) >= min_blocks:
        out.append(np.sqrt(x.var(ddof=1) / len(x)))
        m = len(x) // 2 * 2
        x = 0.5 * (x[:m:2] + x[1:m:2])
    return np.array(out)

se_blk = blocking(e)
_x = e - e.mean(); _f = np.fft.rfft(_x, 2 * len(_x)); acf = np.fft.irfft(_f * np.conj(_f))[:len(_x)]; acf /= acf[0]
fig, ax = plt.subplots(1, 3, figsize=(13, 3.2))
ax[0].plot(e[:3000], lw=0.6); ax[0].set(xlabel="sweep", ylabel="e = E/N", title="Metropolis, L=16, T=2.3 (SYNTHETIC)")
ax[1].plot(acf[:150]); ax[1].axvline(W, c="k", ls="--", lw=0.8, label=f"Sokal window W = {W}")
ax[1].axhline(0, c="k", lw=0.5); ax[1].set(xlabel="lag (sweeps)", ylabel="rho(lag)", title="autocorrelation of e"); ax[1].legend(fontsize=8)
ax[2].plot(range(len(se_blk)), se_blk, "o-", label="blocking")
ax[2].axhline(se_naive * np.sqrt(2 * tau), c="C1", ls="--", label="naive x sqrt(2 tau)")
ax[2].axhline(se_naive, c="C2", ls=":", label="naive")
ax[2].set(xlabel="blocking level (block = 2^level sweeps)", ylabel="SE of <e>", title="blocking plateau"); ax[2].legend(fontsize=8)
fig.tight_layout(); plt.show()
print(f"N_eff = {len(e) / (2 * tau):.0f} of {len(e)}")
print("blocking SE by level:", se_blk.round(5))
```

```python
#@title ฟังก์ชัน tau_int จากโค้ดที่ 9.4 (ดับเบิลคลิกเพื่อดูโค้ด)
def tau_int(x, c=5.0):
    """Integrated autocorrelation time with Sokal's automatic window."""
    x = np.asarray(x) - np.mean(x)
    n = len(x)
    f = np.fft.rfft(x, 2 * n)                      # zero-padded FFT
    acf = np.fft.irfft(f * np.conj(f))[:n]
    acf /= acf[0]
    tau = 0.5
    for W in range(1, n):
        tau += acf[W]
        if W >= c * tau:
            break
    return tau, W

print('tau_int พร้อมใช้; ข้อมูลออสซิลเลเตอร์จากส่วนที่ 0:', t.shape, x_obs.shape)
```

# ส่วนที่ 5 การระบุความไม่แน่นอนแบบเบส์ (หัวข้อ 9.4)

### ทฤษฎีบทของเบส์และความหมายของแต่ละส่วน
บูตสแตรปตอบคำถามว่า "ถ้าทำการทดลองซ้ำ ผลจะกระจายเพียงใด" กรอบแบบเบส์ถามอีกคำถามหนึ่ง "เมื่อเห็นข้อมูลชุดนี้แล้ว เราควรเชื่อว่าพารามิเตอร์มีค่าเท่าใด และเชื่อแน่นเพียงใด" คำตอบคือการแจกแจงความน่าจะเป็นทั้งการแจกแจงบนค่าของพารามิเตอร์

> **ทฤษฎีบท (เบส์)** สำหรับพารามิเตอร์ $\boldsymbol\theta$ และข้อมูล $\mathcal D$
> $$p(\boldsymbol\theta \mid \mathcal D) = \frac{p(\mathcal D \mid \boldsymbol\theta)\,p(\boldsymbol\theta)}{p(\mathcal D)}, \qquad p(\mathcal D) = \int p(\mathcal D \mid \boldsymbol\theta)\,p(\boldsymbol\theta)\,d\boldsymbol\theta$$

| ส่วน | ชื่อ | ความหมายในงานฟิสิกส์ |
|---|---|---|
| $p(\boldsymbol\theta)$ | การแจกแจงก่อนหน้า (prior) | สิ่งที่รู้ก่อนเห็นข้อมูล จากกฎทางฟิสิกส์หรือการวัดที่ทำไว้แล้ว |
| $p(\mathcal D \mid \boldsymbol\theta)$ | ภาวะน่าจะเป็น (likelihood) | แบบจำลองทางฟิสิกส์รวมกับแบบจำลองของสัญญาณรบกวน |
| $p(\boldsymbol\theta \mid \mathcal D)$ | การแจกแจงภายหลัง (posterior) | ความเชื่อที่ปรับแล้วหลังเห็นข้อมูล ค่าที่ดีที่สุด ความไม่แน่นอน และสหสัมพันธ์ระหว่างพารามิเตอร์อ่านได้จากส่วนนี้ |
| $p(\mathcal D)$ | หลักฐาน (evidence) | ค่าคงที่ที่ทำให้ผลรวมเป็นหนึ่ง |

กรอบนี้เชื่อมกับบทที่ 5 ได้สามทาง ทางแรก สำหรับสัญญาณรบกวนแบบเกาส์ที่รู้ $\sigma_i$ จะได้ $p(\mathcal D \mid \boldsymbol\theta) \propto e^{-\chi^2(\boldsymbol\theta)/2}$ การหาค่าต่ำสุดของ $\chi^2$ จึงเท่ากับการหาค่าสูงสุดของภาวะน่าจะเป็น ทางที่สอง ขอบเขตทางฟิสิกส์ เช่น $\gamma \ge 0$ คือการแจกแจงก่อนหน้าแบบสม่ำเสมอที่เป็นศูนย์นอกบริเวณที่เป็นไปได้ ทางที่สาม การถดถอยริดจ์คือค่าที่มีความน่าจะเป็นภายหลังสูงสุด (maximum a posteriori, MAP) เมื่อการแจกแจงก่อนหน้าของสัมประสิทธิ์เป็น $\mathcal N(0, \tau^2)$ โดย $\lambda = \sigma^2/\tau^2$

### การสุ่มตัวอย่างจากการแจกแจงภายหลังด้วยลูกโซ่มาร์คอฟ
ตัวส่วน $p(\mathcal D)$ เป็นปริพันธ์หลายมิติที่คำนวณตรง ๆ ไม่ได้ ปัญหาหน้าตาเดียวกันนี้เคยพบในบทที่ 4 กับฟังก์ชันแบ่งส่วน $Z$ ของการแจกแจงโบลต์ซมันน์ ถ้านิยาม "พลังงาน" ของพารามิเตอร์เป็น
$$U(\boldsymbol\theta) = -\ln\big[p(\mathcal D \mid \boldsymbol\theta)\,p(\boldsymbol\theta)\big]$$
การแจกแจงภายหลังคือ $e^{-U(\boldsymbol\theta)}/p(\mathcal D)$ ซึ่งเป็นการแจกแจงโบลต์ซมันน์ที่ $k_BT = 1$ โดย $p(\mathcal D)$ มีบทบาทเดียวกับ $Z$ สำหรับสัญญาณรบกวนแบบเกาส์และการแจกแจงก่อนหน้าแบบสม่ำเสมอ $U = \chi^2/2$ พารามิเตอร์จึง "แกว่งเชิงความร้อน" รอบก้นหุบของ $\chi^2$ และความกว้างของการแกว่งคือความไม่แน่นอน **การสุ่มตัวอย่างแบบลูกโซ่มาร์คอฟ (Markov chain Monte Carlo, MCMC)** คือการใช้วิธีเมโทรโพลิสกับ "ระบบฟิสิกส์" ที่มีพิกัดเป็นพารามิเตอร์

โค้ดที่ 9.5 ใช้ขอบเขตจากโค้ดที่ 5.6 เป็นการแจกแจงก่อนหน้าแบบสม่ำเสมอ และถือว่ารู้ $\sigma = 0.05$ m การเสนอก้าวใช้การแจกแจงแบบเกาส์หลายตัวแปรที่มีเมทริกซ์ความแปรปรวนร่วมเป็น `pcov` คูณด้วย $2.38^2/d$ เมื่อ $d = 4$ ตัวประกอบนี้ทำให้ลูกโซ่ลืมสถานะเดิมเร็วที่สุดเมื่อการแจกแจงภายหลังใกล้เคียงแบบเกาส์

### โค้ดที่ 9.5 — วิธีเมโทรโพลิสสำหรับการแจกแจงภายหลังของออสซิลเลเตอร์แบบหน่วง

```python
# continues from the damped-oscillator fit (popt, pcov, t, x_obs, sigma_x, bounds)
lo, hi = np.array(bounds[0]), np.array(bounds[1])

def log_posterior(theta):
    if np.any(theta < lo) or np.any(theta > hi):      # uniform prior inside bounds
        return -np.inf
    chi2 = np.sum(((x_obs - damped_oscillator(t, *theta)) / sigma_x)**2)
    return -0.5 * chi2                                 # ln L + const

rng = np.random.default_rng(seed=1)
step = 2.38**2 / 4 * pcov                              # proposal covariance
theta, lp = popt.copy(), log_posterior(popt)
chain, n_acc = [], 0
for k in range(50000):
    prop = rng.multivariate_normal(theta, step)
    lp_prop = log_posterior(prop)
    if np.log(rng.random()) < lp_prop - lp:            # Metropolis rule, T = 1
        theta, lp = prop, lp_prop
        n_acc += 1
    chain.append(theta)
chain = np.array(chain)[5000:]                         # drop burn-in
print("acceptance:", n_acc / 50000)
for name, c in zip(['A', 'gamma', 'omega', 'phi'], chain.T):
    lo95, med, hi95 = np.percentile(c, [2.5, 50, 97.5])
    print(f"{name:5s} median {med:.4f}  95% [{lo95:.4f}, {hi95:.4f}]  "
          f"tau_int {tau_int(c)[0]:.1f}")
```

```python
print(f"posterior sd: gamma {chain[:, 1].std():.4f}, omega {chain[:, 2].std():.4f}; "
      f"curve_fit SE {np.sqrt(pcov[1, 1]):.4f}, {np.sqrt(pcov[2, 2]):.4f}")
fig, ax = plt.subplots(1, 3, figsize=(13, 3.2))
ax[0].plot(chain[:2000, 1], lw=0.6); ax[0].set(xlabel="step (after burn-in)", ylabel="gamma", title="trace of gamma (SYNTHETIC)")
ax[1].hist(chain[:, 1], bins=60, density=True, alpha=0.8, label="MCMC")
_g = np.linspace(chain[:, 1].min(), chain[:, 1].max(), 200)
ax[1].plot(_g, np.exp(-0.5 * ((_g - popt[1]) / np.sqrt(pcov[1, 1]))**2) / np.sqrt(2 * np.pi * pcov[1, 1]), "r", label="curve_fit Gaussian")
ax[1].set(xlabel="gamma (1/s)", title="posterior of gamma"); ax[1].legend(fontsize=8)
ax[2].scatter(chain[::20, 1], chain[::20, 2], s=3, alpha=0.4)
ax[2].set(xlabel="gamma (1/s)", ylabel="omega (rad/s)", title="joint posterior (every 20th step)")
fig.tight_layout(); plt.show()
```

**อ่านผล:** โครงสร้างของโค้ดเหมือนการจำลองไอซิงในบทที่ 4 ทุกประการ เสนอก้าว คำนวณผลต่างของ "พลังงาน" ยอมรับตามกฎของเมโทรโพลิส ทิ้งช่วงเข้าสู่สมดุล แล้วเก็บตัวอย่าง อัตราการยอมรับคือ 0.30 มัธยฐานและช่วงร้อยละ 95 ของ $\gamma$ คือ 0.2861 และ $[0.2712, 0.3017]$ ส่วนเบี่ยงเบนมาตรฐานของการแจกแจงภายหลังของ $\gamma$ และ $\omega$ (0.0078 และ 0.0073) ใกล้กับความคลาดเคลื่อนมาตรฐานจาก `curve_fit` มาก เพราะเมื่อข้อมูลดีและการแจกแจงก่อนหน้าแบบสม่ำเสมอ การแจกแจงภายหลังเกือบเป็นแบบเกาส์ และความกว้างถูกกำหนดโดยความโค้งของ $\chi^2$ ซึ่งเป็นสิ่งเดียวกับที่ `curve_fit` ใช้คำนวณ `pcov`

บรรทัดสุดท้ายใช้ `tau_int` จากส่วนที่ 4 ซ้ำ ได้เวลาสหสัมพันธ์ราวเจ็ดก้าวทุกพารามิเตอร์ ลูกโซ่ 45,000 ก้าวจึงมีตัวอย่างอิสระราว 3,000 ตัว **ลูกโซ่ของ MCMC ก็เป็นลูกโซ่มาร์คอฟ** ทุกสิ่งที่เรียนเรื่องเวลาสหสัมพันธ์จึงใช้กับมันได้ตรงตัว

### มอนติคาร์โลแบบแฮมิลตันและ PyMC
เมื่อพารามิเตอร์มีหลายสิบหรือหลายร้อยตัว ก้าวแบบสุ่มส่วนใหญ่เดินออกนอกบริเวณที่ความน่าจะเป็นสูง วิธีแก้มาจากฟิสิกส์อีกครั้ง ดวน เคนเนดี เพนเดิลตัน และโรเวท (Duane, Kennedy, Pendleton and Roweth) ซึ่งทำงานด้านโครโมไดนามิกส์ควอนตัมบนแลตทิซ เสนอในปี 1987 ให้เพิ่มโมเมนตัมสมมติให้พารามิเตอร์ แล้วให้ระบบเคลื่อนที่ตามสมการของแฮมิลตันบน "ภูมิประเทศ" $U(\boldsymbol\theta)$ ก่อนตัดสินใจยอมรับ เรียกว่ามอนติคาร์โลแบบแฮมิลตัน (Hamiltonian Monte Carlo, HMC) ส่วนตัวสุ่มตัวอย่างแบบไม่เลี้ยวกลับ (No-U-Turn Sampler, NUTS) เลือกความยาวของเส้นทางให้อัตโนมัติ และเป็นวิธีมาตรฐานของ PyMC

โค้ดที่ 9.6 ทำการวิเคราะห์เดียวกันด้วย PyMC แต่ใช้การแจกแจงก่อนหน้าที่สะท้อนความรู้ทางฟิสิกส์ละเอียดขึ้น และ **ไม่สมมติว่ารู้ $\sigma$**

| พารามิเตอร์ | การแจกแจงก่อนหน้า | เหตุผล |
|---|---|---|
| $A$, $\gamma$ | แบบเกาส์ครึ่งซีก (half-normal) | ต้องไม่ติดลบ และกว้างพอที่จะไม่บังคับค่า |
| $\omega$ | แบบเกาส์รอบ 6 rad/s กว้าง 1 rad/s ตัดที่ศูนย์ | ความรู้จากการวัดมวลและค่าคงที่สปริงแยกต่างหาก (ต้องไม่ได้มาจากข้อมูลชุดเดียวกัน) |
| $\phi$ | สม่ำเสมอใน $[-\pi, \pi]$ | ไม่มีความรู้ล่วงหน้า |
| $\sigma$ | แบบเกาส์ครึ่งซีก กว้าง 0.2 m | ขนาดของสัญญาณรบกวนเป็นพารามิเตอร์ที่ต้องประมาณด้วย |

### โค้ดที่ 9.6 — การปรับเส้นโค้งแบบเบส์ด้วย PyMC
⏱ ราวหนึ่งถึงสองนาทีบน Colab (การรันครั้งแรกต้องคอมไพล์โค้ดก่อน) ข้อความเตือนหรือแถบความคืบหน้าของ PyMC ที่ปรากฏไม่ใช่ข้อผิดพลาด

```python
import pymc as pm
import arviz as az

with pm.Model() as model:
    A     = pm.HalfNormal('A', sigma=2.0)                  # A > 0
    gamma = pm.HalfNormal('gamma', sigma=1.0)              # gamma >= 0
    omega = pm.TruncatedNormal('omega', mu=6.0, sigma=1.0, lower=0.0)  # from k and m
    phi   = pm.Uniform('phi', lower=-np.pi, upper=np.pi)
    sigma = pm.HalfNormal('sigma', sigma=0.2)              # noise level, unknown
    mu = A * pm.math.exp(-gamma * t) * pm.math.cos(omega * t + phi)
    pm.Normal('x', mu=mu, sigma=sigma, observed=x_obs)
    idata = pm.sample(2000, tune=1000, chains=4, random_seed=2026,
                      idata_kwargs={'log_likelihood': True})
    pm.sample_posterior_predictive(idata, extend_inferencedata=True,
                                   random_seed=2026)
print(az.summary(idata, var_names=['A', 'gamma', 'omega', 'phi', 'sigma'],
                 hdi_prob=0.95))
```

**อ่านผล:** $\gamma = 0.286$ ช่วงความน่าจะเป็นสูงสุดร้อยละ 95 (HDI) คือ $[0.269, 0.303]$ ส่วน $\sigma$ ประมาณได้ 0.055 ช่วง $[0.049, 0.060]$ ซึ่งครอบค่าจริง 0.05 ความไม่แน่นอนของ $\gamma$ กว้างกว่าในโค้ดที่ 9.5 เล็กน้อย (0.009 เทียบกับ 0.0078) เพราะแบบนี้ไม่สมมติว่ารู้ $\sigma$ ความไม่แน่นอนของ $\sigma$ จึงแพร่ไปยังพารามิเตอร์อื่นโดยอัตโนมัติ ผ่านการอินทิเกรตทุกค่าที่เป็นไปได้ของ $\sigma$ ทิ้ง

**ตรวจการลู่เข้า:** คอลัมน์ `r_hat` เปรียบเทียบความแปรปรวนระหว่างลูกโซ่กับภายในลูกโซ่ ถ้าลูกโซ่ทั้งสี่สำรวจบริเวณเดียวกัน ค่านี้ใกล้ 1 (เกณฑ์ปัจจุบันคือไม่เกิน 1.01) คอลัมน์ `ess_bulk` คือขนาดตัวอย่างประสิทธิผล ซึ่งเป็นแนวคิดเดียวกับ $N_{\text{eff}}$ ในส่วนที่ 4 ในตัวอย่างนี้ `ess_bulk` ราว 5,600–6,900 จากตัวอย่าง 8,000 ตัว NUTS จึงให้ตัวอย่างที่เกือบอิสระ ต่างจากวิธีเมโทรโพลิสแบบเดินสุ่มที่ได้ตัวอย่างอิสระราวหนึ่งในสิบสี่ของจำนวนก้าว

> **ความเข้าใจคลาดเคลื่อนที่พบบ่อย:** "การแจกแจงก่อนหน้าทำให้ผลแบบเบส์ขึ้นกับความคิดเห็นส่วนตัว จึงไม่เป็นวิทยาศาสตร์" ทุกการวิเคราะห์มีสมมติฐาน กรอบแบบเบส์บังคับให้เขียนมันออกมาให้เห็น ขอบเขต $\gamma \ge 0$ ใน `curve_fit` ก็เป็นการแจกแจงก่อนหน้าที่ซ่อนอยู่ เมื่อข้อมูลมาก ผลแทบไม่ขึ้นกับการแจกแจงก่อนหน้าที่สมเหตุสมผล เมื่อข้อมูลน้อย ผลขึ้นกับมันจริง แต่นั่นคือข้อมูลที่ถูกต้อง เพราะข้อมูลชุดนั้นไม่พอจะตัดสินด้วยตัวเอง หลักปฏิบัติคือรายงานการแจกแจงก่อนหน้าพร้อมเหตุผลทางฟิสิกส์ และตรวจว่าผลเปลี่ยนเพียงใดเมื่อเปลี่ยนการแจกแจงก่อนหน้า

### การตรวจเชิงทำนายภายหลัง
การแจกแจงภายหลังบอกความไม่แน่นอนของพารามิเตอร์ *ภายใต้สมมติฐานว่าแบบจำลองถูกต้อง* การตรวจเชิงทำนายภายหลัง (posterior predictive check) ถามแยกว่าแบบจำลองเพียงพอหรือไม่ ถ้าแบบจำลองดี ข้อมูลที่แบบจำลองสร้างขึ้นควรมีหน้าตาเหมือนข้อมูลจริง
$$p(y^{\text{rep}} \mid \mathcal D) = \int p(y^{\text{rep}} \mid \boldsymbol\theta)\,p(\boldsymbol\theta \mid \mathcal D)\,d\boldsymbol\theta$$
โค้ดที่ 9.6 สร้างข้อมูลจำลอง $y^{\text{rep}}$ ไว้แล้ว 8,000 ชุด เซลล์ถัดไปเปรียบเทียบสถิติสามตัวของข้อมูลจริงกับข้อมูลจำลอง สัดส่วนของข้อมูลจำลองที่มีสถิติสุดขั้วกว่าข้อมูลจริงเรียกว่าค่าพีเชิงทำนายภายหลัง (posterior predictive p-value) ค่าที่ใกล้ 0 หรือ 1 บอกว่าข้อมูลจริงมีลักษณะที่แบบจำลองสร้างไม่ได้

```python
post = idata.posterior.stack(s=('chain', 'draw'))
A_, g_, w_, p_, s_ = (post[v].values for v in ('A', 'gamma', 'omega', 'phi', 'sigma'))
mu_post = A_[:, None] * np.exp(-g_[:, None] * t) * np.cos(w_[:, None] * t + p_[:, None])
yrep = idata.posterior_predictive['x'].stack(s=('chain', 'draw')).values.T

def ppc_stats(yy):
    r = (yy - mu_post) / s_[:, None]
    return (r**2).sum(1), np.abs(r).max(1), np.array([np.corrcoef(ri[:-1], ri[1:])[0, 1] for ri in r])

obs, rep = ppc_stats(np.broadcast_to(x_obs, mu_post.shape)), ppc_stats(yrep)
names = ['sum (resid/sigma)^2', 'max |resid/sigma|', 'lag-1 resid correlation']
fig, ax = plt.subplots(1, 3, figsize=(13, 3.0))
for a, name, o, r in zip(ax, names, obs, rep):
    print(f"posterior predictive p-value, {name:24s}: {(r >= o).mean():.2f}")
    a.hist(r - o, bins=50, alpha=0.8); a.axvline(0, c="r", lw=1)
    a.set(title=name, xlabel="T(y_rep) - T(y_obs)")
fig.suptitle("posterior predictive check, M2 (SYNTHETIC)", y=1.02); fig.tight_layout(); plt.show()
```

**อ่านผล:** ค่าพีเชิงทำนายคือ 0.52, 0.12 และ 0.31 ไม่มีค่าใดสุดขั้ว แบบจำลองจึงสร้างข้อมูลที่หน้าตาเหมือนข้อมูลจริงได้ในทุกด้านที่ตรวจ

> **ความเข้าใจคลาดเคลื่อนที่พบบ่อย:** "ค่าพีเชิงทำนายภายหลังของ $\chi^2$ ใกล้ 0.5 แปลว่าแบบจำลองดี" ถ้าปล่อยให้แบบจำลองประมาณ $\sigma$ เอง มันจะปรับ $\sigma$ ให้ขนาดของส่วนเหลือพอดีกับข้อมูลเสมอ ค่าพีของ $\chi^2$ จึงใกล้ 0.5 แทบทุกครั้ง ผู้เขียนลองปรับแบบจำลอง M1 ซึ่งไม่มีการหน่วงและผิดชัดเจน ค่าพีของ $\chi^2$ ยังได้ 0.50 เพราะแบบจำลองประมาณ $\sigma$ ได้ 0.18 m แทน 0.05 m ขณะที่ค่าพีของสหสัมพันธ์ของส่วนเหลือที่ติดกันต่ำกว่า 0.001 เพราะส่วนเหลือจริงมีสหสัมพันธ์ถึง 0.87 สถิติที่ใช้ตรวจได้จริงต้องวัดลักษณะที่แบบจำลองไม่ได้ปรับโดยตรง

### การเปรียบเทียบแบบจำลองแบบเบส์ด้วย WAIC และ PSIS-LOO
AIC ใช้ค่าที่ดีที่สุดเพียงจุดเดียวของพารามิเตอร์ เมื่อมีการแจกแจงภายหลังทั้งการแจกแจง เกณฑ์สารสนเทศที่ใช้ได้ทั่วไปของวาตานาเบะ (widely applicable information criterion, WAIC) และการตรวจสอบไขว้แบบเว้นทีละจุดที่ประมาณด้วยการสุ่มตามความสำคัญแบบปรับหางด้วยการแจกแจงพาเรโต (PSIS-LOO) ประมาณความสามารถในการทำนายข้อมูลใหม่โดยเฉลี่ยเหนือการแจกแจงภายหลัง โดยไม่ต้องฝึกแบบจำลองใหม่ ผลรายงานเป็นความหนาแน่นเชิงทำนายลอการิทึมที่คาดหมาย (expected log predictive density, elpd) ยิ่งสูงยิ่งดี PSIS-LOO ยังรายงานค่าวินิจฉัย $\hat k$ ของแต่ละจุด ถ้าเกินราว 0.7 การประมาณที่จุดนั้นไม่น่าเชื่อถือ

เซลล์ถัดไปปรับ M1 และ M3 ด้วย PyMC แบบเดียวกับโค้ดที่ 9.6 (ค่าชดเชยของ M3 มีการแจกแจงก่อนหน้า $\mathcal N(0, 0.5^2)$) แล้วเปรียบเทียบกับ M2 ด้วย `az.compare`

```python
# ⏱ ราวหนึ่งถึงสองนาทีบน Colab (ปรับแบบจำลองอีกสองตัว)
def fit_model(kind):
    with pm.Model():
        A = pm.HalfNormal('A', sigma=2.0)
        omega = pm.TruncatedNormal('omega', mu=6.0, sigma=1.0, lower=0.0)
        phi = pm.Uniform('phi', lower=-np.pi, upper=np.pi)
        sigma = pm.HalfNormal('sigma', sigma=0.2)
        if kind == 'M1':
            mu_ = A * pm.math.cos(omega * t + phi)
        else:
            gamma = pm.HalfNormal('gamma', sigma=1.0)
            mu_ = A * pm.math.exp(-gamma * t) * pm.math.cos(omega * t + phi)
            if kind == 'M3':
                mu_ = mu_ + pm.Normal('c', mu=0.0, sigma=0.5)       # zero offset, weak prior
        pm.Normal('x', mu=mu_, sigma=sigma, observed=x_obs)
        return pm.sample(2000, tune=1000, chains=4, random_seed=2026, progressbar=False,
                         idata_kwargs={'log_likelihood': True})

t0 = time.time()
ids = {'M1': fit_model('M1'), 'M2': idata, 'M3': fit_model('M3')}
for ic_name in ('loo', 'waic'):
    print(f"\naz.compare, {ic_name}:")
    print(az.compare(ids, ic=ic_name).iloc[:, :7].round(2).to_string())
print("\nmax Pareto k:", {k: round(float(az.loo(v, pointwise=True).pareto_k.max()), 2) for k, v in ids.items()})
print(f"({time.time() - t0:.0f} s)")
```

**อ่านผล:** elpd ของ M2 คือ 294.9 ของ M3 ต่ำกว่า 0.5 ด้วยความคลาดเคลื่อนมาตรฐานของผลต่าง (`dse`) 0.9 และของ M1 ต่ำกว่า 239 ด้วย `dse` 16.5 ทุกจุดมี $\hat k$ ต่ำกว่า 0.7 ข้อสรุปสอดคล้องกับตารางในส่วนที่ 2 M1 แย่ชัดเจน M2 กับ M3 แยกกันไม่ได้ และเมื่อแยกกันไม่ได้ ควรเลือกแบบจำลองที่ง่ายกว่า สิ่งที่ WAIC และ PSIS-LOO เพิ่มจาก AIC คือความคลาดเคลื่อนมาตรฐานของผลต่าง ซึ่งบอกว่าผลต่างเท่าใดจึงมีความหมาย

### ✏️ แบบฝึก E9.4 ในชั้นเรียน (10 นาที) การปรับเส้นโค้งแบบเบส์เมื่อข้อมูลไม่ดี
เซลล์ถัดไปสร้างข้อมูลชุดแย่ของส่วนที่ 3 (60 จุดใน 10 วินาที $\sigma = 0.4$ m เมล็ดสุ่ม 2026) ปรับด้วย `curve_fit` แล้วสุ่มตัวอย่างด้วยตัวสุ่มเดียวกับโค้ดที่ 9.5 จำนวน 100,000 ก้าว (ทิ้ง 10,000 ก้าวแรก) ผลอยู่ใน `chain_bad` (คอลัมน์ $A, \gamma, \omega, \phi$) ใช้เวลาราว 15 วินาที

คำนวณ $Q = \omega/(2\gamma)$ ของทุกตัวอย่าง แล้วเก็บเปอร์เซ็นไทล์ที่ 2.5, 50 และ 97.5 ใน `my_Q` และเปอร์เซ็นไทล์ที่ 16, 50 และ 84 ใน `my_Q68` เปรียบเทียบกับสามแถวของตารางในส่วนที่ 3 แล้วอธิบายว่าทำไมมัธยฐานภายหลังของ $\gamma$ ไม่เท่ากับค่าจาก `curve_fit`

```python
def metropolis_chain(tt, xx, sig, p_start, cov, n_steps, burn, seed=1):
    # the sampler of Listing 9.5, written as a function of the data
    lo_, hi_ = np.array(bounds[0]), np.array(bounds[1])
    def lpost(th):
        if np.any(th < lo_) or np.any(th > hi_):
            return -np.inf
        return -0.5 * np.sum(((xx - damped_oscillator(tt, *th)) / sig)**2)
    rg = np.random.default_rng(seed=seed)
    stp = 2.38**2 / 4 * cov
    th, l = p_start.copy(), lpost(p_start)
    out, acc = [], 0
    for _ in range(n_steps):
        pr = rg.multivariate_normal(th, stp); lpr = lpost(pr)
        if np.log(rg.random()) < lpr - l:
            th, l = pr, lpr; acc += 1
        out.append(th)
    return np.array(out)[burn:], acc / n_steps

rng_d = np.random.default_rng(seed=2026)
t_bad = np.linspace(0, 10, 60)
x_bad = damped_oscillator(t_bad, 1.0, 0.3, 6.0, 0.5) + rng_d.normal(scale=0.4, size=t_bad.size)
popt_bad, pcov_bad = curve_fit(damped_oscillator, t_bad, x_bad, p0=[1.0, 0.2, 6.0, 0.0], bounds=bounds,
                               sigma=0.4 * np.ones(t_bad.size), absolute_sigma=True)
print(f"curve_fit: gamma = {popt_bad[1]:.3f} +/- {np.sqrt(pcov_bad[1, 1]):.3f}, Q = {popt_bad[2] / (2 * popt_bad[1]):.1f}")
chain_bad, acc_bad = metropolis_chain(t_bad, x_bad, 0.4, popt_bad, pcov_bad, 100000, 10000)
print("acceptance:", acc_bad, "| tau_int (A, gamma, omega, phi):", [round(float(tau_int(c)[0]), 1) for c in chain_bad.T])
print(f"gamma median {np.median(chain_bad[:, 1]):.3f}, 95% {np.percentile(chain_bad[:, 1], [2.5, 97.5]).round(2)}")

my_Q = None     # ✏️ percentiles 2.5, 50, 97.5 of Q from chain_bad
my_Q68 = None   # ✏️ percentiles 16, 50, 84 of Q
```

```python
#@title ข้อมูลและวิธีจำแนกจากบทที่ 6 สายพอลิเมอร์จากบทที่ 8 และ tau_int (ดับเบิลคลิกเพื่อดูโค้ด)
# ---- Chapter 6: Listings 6.1-6.5 (without the comparison loop)
import numpy as np
import pandas as pd

def make_higgs_toy(n_sig=4000, n_bkg=16000, seed=2026):
    """Toy H->bb events vs background (GeV). Shapes are illustrative only."""
    rng = np.random.default_rng(seed)
    # signal: resonance at 125 GeV with ~10% resolution
    m_s   = rng.normal(125.0, 12.0, n_sig)
    pT_s  = rng.gamma(shape=4.0, scale=40.0, size=n_sig)     # mean 160 GeV
    dR_s  = 2 * m_s / np.maximum(pT_s, 60.0) * rng.normal(1.0, 0.15, n_sig)
    met_s = rng.exponential(20.0, n_sig)                       # semileptonic b
    # background: falling mass spectrum, softer pT, wider angles
    m_b   = 40.0 + rng.exponential(70.0, n_bkg)
    pT_b  = rng.gamma(shape=3.0, scale=40.0, size=n_bkg)     # mean 120 GeV
    dR_b  = rng.uniform(0.4, 4.0, n_bkg)
    met_b = rng.exponential(30.0, n_bkg)                       # e.g. ttbar
    df = pd.DataFrame({
        'm_bb':   np.concatenate([m_s, m_b]),
        'pT':     np.concatenate([pT_s, pT_b]),
        'dR':     np.clip(np.concatenate([dR_s, dR_b]), 0.4, 4.0),
        'ETmiss': np.concatenate([met_s, met_b]),
        'label':  np.concatenate([np.ones(n_sig, int), np.zeros(n_bkg, int)]),
    })
    return df.sample(frac=1.0, random_state=seed).reset_index(drop=True)

df = make_higgs_toy()
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

features = ['m_bb', 'pT', 'dR', 'ETmiss']
X = df[features].values
y = df['label'].values

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42)

from sklearn.linear_model import LogisticRegression

lr_pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('clf',    LogisticRegression(C=1.0, class_weight='balanced',
                                  random_state=42)),
])
lr_pipe.fit(X_train, y_train)
y_proba_lr = lr_pipe.predict_proba(X_val)[:, 1]
from sklearn.ensemble import RandomForestClassifier

rf_pipe = Pipeline([
    ('scaler', StandardScaler()),      # harmless for trees; kept for uniformity
    ('clf',    RandomForestClassifier(n_estimators=200, max_depth=10,
                                      min_samples_leaf=10,
                                      class_weight='balanced',
                                      random_state=42, n_jobs=-1)),
])
rf_pipe.fit(X_train, y_train)
importance = rf_pipe.named_steps['clf'].feature_importances_

from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score

classifiers = {
    'LogReg':     LogisticRegression(class_weight='balanced', random_state=42),
    'KNN':        KNeighborsClassifier(n_neighbors=25),
    'SVM (RBF)':  SVC(kernel='rbf', class_weight='balanced', random_state=42),
    'DecTree':    DecisionTreeClassifier(max_depth=8, class_weight='balanced',
                                         random_state=42),
    'RandForest': RandomForestClassifier(n_estimators=200, max_depth=10,
                                         min_samples_leaf=10,
                                         class_weight='balanced',
                                         random_state=42, n_jobs=-1),
}

# ---- Chapter 8: Listing 8.5
def make_rouse_trajectory(N=50, n_frames=10000, dt_frame=5.0, seed=2026):
    """Rouse chain (k_B T = friction = bond length = 1) sampled exactly in
    its normal-mode basis; returns Cartesian coordinates (n_frames, 3N)."""
    rng = np.random.default_rng(seed)
    k = 3.0                                              # 3 k_B T / b^2
    i = np.arange(N)
    p = np.arange(1, N)                                  # p = 0 (centre of mass) removed
    Q = np.sqrt(2.0 / N) * np.cos(np.outer(p, np.pi * (i + 0.5) / N))
    a = 4.0 * np.sin(p * np.pi / (2 * N))**2             # Rouse-matrix eigenvalues
    var = 1.0 / (k * a)                                  # <X_p^2> per axis
    tau = 1.0 / (k * a)                                  # relaxation times
    decay = np.exp(-dt_frame / tau)
    Xp = np.empty((n_frames, len(p), 3))
    Xp[0] = rng.normal(size=(len(p), 3)) * np.sqrt(var)[:, None]
    for t in range(1, n_frames):                         # exact Ornstein-Uhlenbeck step
        Xp[t] = (decay[:, None] * Xp[t - 1]
                 + np.sqrt(var * (1 - decay**2))[:, None] * rng.normal(size=(len(p), 3)))
    r = np.einsum('pi,tpc->tic', Q, Xp)                 # bead positions (CM at origin)
    return r.reshape(n_frames, 3 * N), tau

polymer_trajectory, tau_theory = make_rouse_trajectory()

# ---- Listing 9.4: tau_int
def tau_int(x, c=5.0):
    """Integrated autocorrelation time with Sokal's automatic window."""
    x = np.asarray(x) - np.mean(x)
    n = len(x)
    f = np.fft.rfft(x, 2 * n)                      # zero-padded FFT
    acf = np.fft.irfft(f * np.conj(f))[:n]
    acf /= acf[0]
    tau = 0.5
    for W in range(1, n):
        tau += acf[W]
        if W >= c * tau:
            break
    return tau, W

print('Higgs toy data (SYNTHETIC):', X_train.shape, X_val.shape, '| signal fraction', y.mean().round(3))
print('classifiers:', list(classifiers), '| rf_pipe fitted')
print('polymer_trajectory (SYNTHETIC):', polymer_trajectory.shape)
```

### การทดสอบแบบจับคู่สำหรับเปรียบเทียบวิธีจำแนก
บทที่ 6 รายงานพื้นที่ใต้เส้นโค้ง ROC ของ KNN, SVM และป่าสุ่มบนข้อมูลฮิกส์จำลองเป็น 0.973, 0.977 และ 0.975 และสรุปว่าความต่างเล็กเกินกว่าจะตัดสินได้ ความคิดแรกที่มักเกิดขึ้นคือคำนวณความคลาดเคลื่อนมาตรฐานของแต่ละวิธีแยกกัน แล้วดูว่าช่วงความเชื่อมั่นซ้อนทับกันหรือไม่ ความคิดนี้ละเลยว่าทั้งสองวิธีถูกทดสอบกับ **เหตุการณ์ชุดเดียวกัน** เหตุการณ์ที่ยากสำหรับวิธีหนึ่งมักยากสำหรับอีกวิธีด้วย ความแกว่งของพื้นที่ใต้เส้นโค้งทั้งสองจึงไปทางเดียวกัน และหักล้างกันเมื่อลบกัน เหมือนการเปรียบเทียบนาฬิกาสองเรือน ควรเทียบเวลาของทั้งสองเรือนในขณะเดียวกัน การทดสอบแบบจับคู่ (paired test) วิเคราะห์ *ผลต่าง* ของสองวิธีบนข้อมูลเดียวกันโดยตรง

การทดสอบของเดอลอง (DeLong test) อาศัยข้อเท็จจริงว่าพื้นที่ใต้เส้นโค้งเท่ากับความน่าจะเป็นที่เหตุการณ์สัญญาณที่สุ่มมาได้คะแนนสูงกว่าเหตุการณ์ฉากหลังที่สุ่มมา แล้วคำนวณความแปรปรวนร่วมของพื้นที่ใต้เส้นโค้งของสองวิธีจากส่วนประกอบของแต่ละเหตุการณ์
$$\mathrm{SE}(\hat A_1 - \hat A_2) = \sqrt{S_{11} + S_{22} - 2S_{12}}$$
พจน์ $-2S_{12}$ คือส่วนที่หักความแกว่งร่วมออก

### โค้ดที่ 9.7 — การทดสอบของเดอลองสำหรับผลต่างของพื้นที่ใต้เส้นโค้ง ROC

```python
from scipy import stats

def delong(y, s1, s2):
    """Paired test for AUC(s1) - AUC(s2) on the same events."""
    pos, neg = y == 1, y == 0
    def components(s):
        sp, sn = s[pos], s[neg]
        psi = (sp[:, None] > sn[None, :]) + 0.5 * (sp[:, None] == sn[None, :])
        return psi.mean(), psi.mean(axis=1), psi.mean(axis=0)
    a1, v10_1, v01_1 = components(s1)
    a2, v10_2, v01_2 = components(s2)
    S10 = np.cov(np.vstack([v10_1, v10_2]))           # over signal events
    S01 = np.cov(np.vstack([v01_1, v01_2]))           # over background events
    S = S10 / pos.sum() + S01 / neg.sum()
    diff = a1 - a2
    se = np.sqrt(S[0, 0] + S[1, 1] - 2 * S[0, 1])     # covariance term matters
    return diff, se, 2 * stats.norm.sf(abs(diff) / se)

score = {}                                        # classifiers: Chapter 6 listing
for name in ['KNN', 'SVM (RBF)', 'RandForest']:
    pipe = Pipeline([('scaler', StandardScaler()), ('clf', classifiers[name])])
    pipe.fit(X_train, y_train)
    score[name] = (pipe.decision_function(X_val) if name == 'SVM (RBF)'
                   else pipe.predict_proba(X_val)[:, 1])
d, se, p = delong(y_val, score['SVM (RBF)'], score['RandForest'])
print(f"AUC(SVM) - AUC(RF) = {d:.4f} +/- {se:.4f}, p = {p:.3f}")
```

### ✏️ แบบฝึก 9.D (5 นาที) ความคลาดเคลื่อนแบบไม่จับคู่กับแบบจับคู่
ฟังก์ชัน `auc_se` ในเซลล์ถัดไปคำนวณความคลาดเคลื่อนมาตรฐานของพื้นที่ใต้เส้นโค้งของวิธีเดียว ใช้มันคำนวณความคลาดเคลื่อนของผลต่าง AUC(SVM) − AUC(RF) **แบบไม่จับคู่** $\sqrt{\mathrm{SE}_1^2 + \mathrm{SE}_2^2}$ เก็บใน `my_se_unpaired` แล้วคำนวณอัตราส่วนระหว่างแบบไม่จับคู่กับแบบจับคู่ (`se` จากโค้ดที่ 9.7) เก็บใน `my_ratio_se` ถ้าใช้ความคลาดเคลื่อนแบบไม่จับคู่ ผลต่าง 0.0028 จะมีนัยสำคัญหรือไม่

```python
from sklearn.metrics import roc_auc_score
pos_v, neg_v = y_val == 1, y_val == 0

def auc_se(s):
    psi = (s[pos_v][:, None] > s[neg_v][None, :]) + 0.5 * (s[pos_v][:, None] == s[neg_v][None, :])
    return np.sqrt(psi.mean(1).var(ddof=1) / pos_v.sum() + psi.mean(0).var(ddof=1) / neg_v.sum())

for n_, s_v in score.items():
    print(f"{n_:11s} AUC {roc_auc_score(y_val, s_v):.4f}  SE {auc_se(s_v):.4f}")

my_se_unpaired = None   # ✏️ sqrt(SE_SVM^2 + SE_RF^2)
my_ratio_se = None      # ✏️ my_se_unpaired / se
```

```python
# ✅ ตรวจคำตอบ
_u = np.hypot(auc_se(score['SVM (RBF)']), auc_se(score['RandForest']))
print("SE unpaired:", "(ยังไม่ได้ตอบ)" if my_se_unpaired is None else ("✔" if abs(my_se_unpaired / _u - 1) < 1e-3 else "✘"), f"— {_u:.4f}")
print("ratio      :", "(ยังไม่ได้ตอบ)" if my_ratio_se is None else ("✔" if abs(my_ratio_se - _u / se) < 0.01 else "✘"), f"— {_u / se:.2f}")
print(f"unpaired z = {d / _u:.2f}, p = {2 * stats.norm.sf(d / _u):.2f}   |   paired z = {d / se:.2f}, p = {p:.3f}")
```

### การทดสอบของแม็กนีมาร์และการเปรียบเทียบที่จุดทำงาน
การทดสอบของแม็กนีมาร์ (McNemar test) ใช้กับผลการจำแนกที่เกณฑ์ตัดสินค่าหนึ่ง โดยนับเฉพาะเหตุการณ์ที่สองวิธีตัดสินต่างกัน ให้ $n_{01}$ คือจำนวนที่วิธีแรกถูกแต่วิธีที่สองผิด และ $n_{10}$ คือกรณีกลับกัน สถิติ $(|n_{01} - n_{10}| - 1)^2/(n_{01} + n_{10})$ แจกแจงแบบไคกำลังสองที่องศาเสรีหนึ่ง

```python
pred = {n_: Pipeline([('scaler', StandardScaler()), ('clf', classifiers[n_])]).fit(X_train, y_train).predict(X_val)
        for n_ in ('KNN', 'RandForest')}
ok_rf, ok_knn = pred['RandForest'] == y_val, pred['KNN'] == y_val
n01, n10 = int((ok_rf & ~ok_knn).sum()), int((~ok_rf & ok_knn).sum())
chi_mc = (abs(n01 - n10) - 1)**2 / (n01 + n10)
print(f"McNemar RF vs KNN: n01 = {n01}, n10 = {n10}, p = {stats.chi2.sf(chi_mc, 1):.3f}; "
      f"accuracy RF {ok_rf.mean():.3f}, KNN {ok_knn.mean():.3f}")
d2, se2, p2 = delong(y_val, score['RandForest'], score['KNN'])
print(f"DeLong AUC(RF) - AUC(KNN) = {d2:.4f} +/- {se2:.4f}, p = {p2:.2f}")
print(f"fraction called signal: KNN {pred['KNN'].mean():.3f}, balanced RF {pred['RandForest'].mean():.3f} (true {y_val.mean():.3f})")
```

**อ่านผล:** $n_{01} = 116$ และ $n_{10} = 173$ ค่าพี 0.001 KNN จึงแม่นกว่าป่าสุ่มอย่างมีนัยสำคัญที่ความแม่นยำ 0.925 เทียบกับ 0.915 ขณะที่การทดสอบของเดอลองบอกว่าพื้นที่ใต้เส้นโค้งของทั้งสองไม่ต่างกัน ($p = 0.32$) สองการทดสอบตอบคำถามต่างกัน พื้นที่ใต้เส้นโค้งวัดความสามารถในการเรียงลำดับทุกเกณฑ์ตัดสินรวมกัน การทดสอบของแม็กนีมาร์วัดความแม่นยำที่เกณฑ์ตัดสินค่าเดียว และป่าสุ่มที่ใช้ `class_weight='balanced'` เลื่อนเกณฑ์ไปทางทายสัญญาณบ่อยขึ้น ความแม่นยำโดยรวมจึงลดลง แม้ความสามารถในการเรียงลำดับจะเท่าเดิม **การเปรียบเทียบต้องระบุว่าเปรียบเทียบอะไรที่จุดทำงานใด** ในฟิสิกส์อนุภาค วิธีจำแนกมักใช้ที่จุดทำงานเดียว เช่น ประสิทธิภาพของฉากหลังร้อยละหนึ่ง การเปรียบเทียบที่มีความหมายจึงเป็นการตัดเส้นโค้ง ROC ที่จุดนั้น แล้วเปรียบเทียบประสิทธิภาพของสัญญาณพร้อมความไม่แน่นอนจากการสุ่มซ้ำแบบจับคู่

### การปรับเทียบความน่าจะเป็น
วิธีจำแนกที่ปรับเทียบดีแล้ว (well calibrated) คือวิธีที่ในบรรดาเหตุการณ์ที่ได้คะแนน 0.7 มีสัญญาณอยู่ราวร้อยละ 70 จริง เหมือนนักพยากรณ์อากาศที่บอกโอกาสฝนตกร้อยละ 30 แล้วฝนตกจริงราวหนึ่งในสามของวันเหล่านั้น เครื่องมือตรวจคือแผนภาพความน่าเชื่อถือ (reliability diagram) ซึ่งวาดสัดส่วนของสัญญาณจริงในแต่ละช่วงคะแนนเทียบกับคะแนนเฉลี่ยของช่วงนั้น ตัวเลขสรุปคือคะแนนของไบรเออร์ (Brier score) ค่าเฉลี่ยของ $(p_i - y_i)^2$ และความคลาดเคลื่อนของการปรับเทียบที่คาดหมาย (expected calibration error, ECE) ซึ่งเป็นค่าเฉลี่ยถ่วงน้ำหนักของระยะห่างจากเส้นทแยง การแก้ทำได้โดยเรียนรู้ฟังก์ชันที่แปลงคะแนนเป็นความน่าจะเป็นจากข้อมูลที่ไม่ได้ใช้ฝึก เช่น การถดถอยแบบไอโซโทนิก (isotonic regression) ซึ่งหาฟังก์ชันไม่ลดลงที่เข้ากับข้อมูลที่สุด

### โค้ดที่ 9.8 — การตรวจและปรับเทียบความน่าจะเป็นของป่าสุ่ม

```python
from sklearn.calibration import CalibratedClassifierCV, calibration_curve
from sklearn.metrics import brier_score_loss

def ece(y, p, bins=10):
    edges = np.linspace(0, 1, bins + 1)
    idx = np.clip(np.digitize(p, edges) - 1, 0, bins - 1)
    return sum((idx == b).mean() * abs(y[idx == b].mean() - p[idx == b].mean())
               for b in range(bins) if np.any(idx == b))

rf_cal = CalibratedClassifierCV(rf_pipe, method='isotonic', cv=5)
for name, model in [('balanced RF', rf_pipe), ('+ isotonic', rf_cal)]:
    model.fit(X_train, y_train)
    p = model.predict_proba(X_val)[:, 1]
    frac_pos, mean_p = calibration_curve(y_val, p, n_bins=10)
    print(name, "Brier", round(brier_score_loss(y_val, p), 4),
          "ECE", round(ece(y_val, p), 4))
```

```python
p_raw = rf_pipe.predict_proba(X_val)[:, 1]
p_cal = rf_cal.predict_proba(X_val)[:, 1]
fig, ax = plt.subplots(1, 2, figsize=(10, 3.6))
for pp, lab in [(p_raw, "balanced RF"), (p_cal, "+ isotonic")]:
    fp, mp = calibration_curve(y_val, pp, n_bins=10)
    ax[0].plot(mp, fp, "o-", label=lab)
    ax[1].hist(pp, bins=40, histtype="step", lw=1.3, label=lab)
ax[0].plot([0, 1], [0, 1], "k--", lw=0.8, label="perfect")
ax[0].set(xlabel="mean predicted probability", ylabel="fraction of signal", title="reliability diagram (SYNTHETIC)"); ax[0].legend(fontsize=8)
ax[1].set(xlabel="predicted probability", ylabel="events", yscale="log", title="score distribution"); ax[1].legend(fontsize=8)
fig.tight_layout(); plt.show()
for lo_, hi_ in [(0.4, 0.5), (0.5, 0.6)]:
    m_ = (p_raw >= lo_) & (p_raw < hi_)
    print(f"balanced RF, scores {lo_}-{hi_}: mean score {p_raw[m_].mean():.2f}, signal fraction {y_val[m_].mean():.2f}")
print(f"AUC before / after isotonic calibration: {roc_auc_score(y_val, p_raw):.4f} / {roc_auc_score(y_val, p_cal):.4f}")
```

**อ่านผล:** ป่าสุ่มแบบ balanced ปรับเทียบไม่ดีอย่างเห็นได้ชัด ในช่วงคะแนนราว 0.55 มีสัญญาณจริงเพียงร้อยละ 25 ECE คือ 0.059 และคะแนนของไบรเออร์คือ 0.061 เพราะการถ่วงน้ำหนักทำให้แบบจำลอง "เห็น" สัญญาณเท่ากับฉากหลัง ขณะที่ข้อมูลจริงมีสัญญาณเพียงร้อยละ 20 หลังปรับด้วยการถดถอยแบบไอโซโทนิก ECE ลดเหลือ 0.007 และคะแนนของไบรเออร์ลดเหลือ 0.050 ส่วนพื้นที่ใต้เส้นโค้งแทบไม่เปลี่ยน เพราะการปรับเทียบเปลี่ยนค่าของคะแนน แต่ไม่เปลี่ยนลำดับ

**การปรับตามสัดส่วนของประเภท:** ข้อมูลฝึกมีสัญญาณร้อยละ 20 แต่ในข้อมูลจริงสัดส่วนของสัญญาณอาจต่ำกว่านั้นหลายเท่า ถ้า $\pi_{\text{train}}$ และ $\pi_{\text{new}}$ คือสัดส่วนของสัญญาณในข้อมูลฝึกและในข้อมูลที่จะใช้จริง ทฤษฎีบทของเบส์ให้
$$p_{\text{new}} = \frac{(\pi_{\text{new}}/\pi_{\text{train}})\,p}{(\pi_{\text{new}}/\pi_{\text{train}})\,p + \big[(1-\pi_{\text{new}})/(1-\pi_{\text{train}})\big](1-p)}$$

```python
def prior_shift(p, pi_new, pi_train=0.2):
    a = pi_new / pi_train * p
    return a / (a + (1 - pi_new) / (1 - pi_train) * (1 - p))

print(f"events with calibrated p > 0.5: {100 * (p_cal > 0.5).mean():.1f} %; "
      f"after shifting to a 1 % signal fraction: {100 * (prior_shift(p_cal, 0.01) > 0.5).mean():.1f} %")
```

เมื่อสัญญาณหายากขึ้นเป็นร้อยละหนึ่ง สัดส่วนของเหตุการณ์ที่มีความน่าจะเป็นเกิน 0.5 ลดจากร้อยละ 21.5 เหลือร้อยละ 2.5 เหตุการณ์ที่ "น่าจะเป็นสัญญาณ" ในข้อมูลฝึกส่วนใหญ่กลายเป็น "น่าจะเป็นฉากหลัง" นี่คือเหตุผลที่บทที่ 6 เลือกเกณฑ์ตัดสินด้วยจำนวนสัญญาณและฉากหลังที่คาดในข้อมูลจริง ไม่ใช่ด้วยเกณฑ์ 0.5

### วิธีรวมกลุ่มและความไม่แน่นอนแบบเอพิสเทมิก
วิธีรวมกลุ่ม (ensemble methods) รวมการทำนายของแบบจำลองหลายตัว มีสามแนวทางหลัก การรวมจากบูตสแตรป (bagging) ฝึกแต่ละตัวด้วยชุดบูตสแตรปแล้วเฉลี่ยผล ป่าสุ่มคือตัวอย่างของแนวทางนี้ การเสริมกำลัง (boosting) ฝึกแบบจำลองอ่อนทีละตัวโดยตัวถัดไปเน้นแก้ความผิดพลาดของตัวก่อนหน้า เช่น XGBoost และการซ้อนชั้น (stacking) ฝึกแบบจำลองชั้นบนให้เรียนรู้วิธีรวมผลของชั้นล่างจากผลการตรวจสอบไขว้

ถ้าสมาชิกทุกตัวทำนายใกล้กัน แสดงว่าข้อมูลกำหนดคำตอบได้ชัด ถ้าเห็นต่างกันมาก แสดงว่าข้อมูลไม่พอ ความกระจายระหว่างสมาชิกจึงใช้ประมาณความไม่แน่นอนแบบเอพิสเทมิกได้ และเป็นฐานของวิธีรวมกลุ่มของโครงข่ายประสาทเชิงลึก (deep ensembles) ในบทที่ 10 แต่ข้อสรุปนี้มีเงื่อนไข ซึ่งเห็นชัดที่สุดเมื่อทำนายนอกบริเวณของข้อมูลฝึก โค้ดที่ 9.9 สร้างข้อมูล $y = \sin 2x$ บวกสัญญาณรบกวน 0.1 ที่ $x \in [0, 3]$ เท่านั้น แล้วเปรียบเทียบพหุนามดีกรีห้า 200 ตัวจากชุดบูตสแตรปกับป่าสุ่ม 200 ต้น

### โค้ดที่ 9.9 — ความกระจายของสมาชิกในวิธีรวมกลุ่ม ภายในและภายนอกช่วงของข้อมูลฝึก

```python
from sklearn.ensemble import RandomForestRegressor

rng = np.random.default_rng(seed=2026)
x = rng.uniform(0, 3, 150)                       # training data only in [0, 3]
y = np.sin(2 * x) + rng.normal(scale=0.1, size=x.size)
x_grid = np.array([0.5, 1.5, 2.5, 3.5, 4.0])

poly = np.array([np.polyval(np.polyfit(x[i], y[i], 5), x_grid)
                 for i in (rng.integers(0, x.size, x.size) for _ in range(200))])
rf = RandomForestRegressor(n_estimators=200, min_samples_leaf=5,
                           random_state=0).fit(x[:, None], y)
trees = np.array([tree.predict(x_grid[:, None]) for tree in rf.estimators_])

print("truth      ", np.sin(2 * x_grid).round(2))
print("poly mean  ", poly.mean(0).round(2), " spread", poly.std(0).round(3))
print("forest mean", trees.mean(0).round(2), " spread", trees.std(0).round(3))
```

```python
xs = np.linspace(0, 4.2, 300)
fig, ax = plt.subplots(figsize=(8, 3.6))
ax.plot(xs, np.sin(2 * xs), "k", label="truth sin(2x)")
ax.plot(x, y, ".", ms=3, alpha=0.4, label="training data")
ax.plot(xs, rf.predict(xs[:, None]), label="random forest")
ax.errorbar(x_grid, poly.mean(0), yerr=poly.std(0), fmt="s", capsize=3, label="bootstrap polynomials (mean +/- sd)")
ax.axvspan(3, 4.2, color="grey", alpha=0.15)
ax.annotate(f"polynomials at x = 4.0: {poly.mean(0)[4]:.2f} +/- {poly.std(0)[4]:.2f} (off scale)",
            xy=(4.0, -1.95), xytext=(2.3, -1.7), fontsize=8, arrowprops=dict(arrowstyle="->"))
ax.set(ylim=(-2, 2), xlabel="x", ylabel="y", title="ensembles outside the training range (grey) (SYNTHETIC)"); ax.legend(fontsize=8)
fig.tight_layout(); plt.show()
```

**อ่านผล:** ภายในช่วงของข้อมูล ทั้งสองวิธีทำนายได้ใกล้ค่าจริง ที่ $x = 3.5$ และ 4.0 ซึ่งอยู่นอกช่วง ค่าจริงคือ 0.66 และ 0.99 พหุนามทำนายได้ −0.59 และ −5.96 ผิดมาก แต่ความกระจายก็เพิ่มจากราว 0.02 ภายในช่วงเป็น 0.53 และ 2.1 นอกช่วง ความกระจายจึงเตือนได้อย่างน้อยว่าไม่ควรเชื่อการทำนายนี้ ป่าสุ่มทำนายได้ −0.54 ทั้งสองจุด เพราะต้นไม้ทุกต้นให้ค่าคงที่ของใบที่ขอบของข้อมูล และความกระจายระหว่างต้นไม้ที่สองจุดนี้คือ 0.052 **เล็กกว่าภายในช่วงของข้อมูล** (0.08–0.19) ป่าสุ่มจึงผิดไปราว 1.2 ถึง 1.5 โดยที่สมาชิกทุกตัวเห็นตรงกันอย่างมั่นใจ

> **ความเข้าใจคลาดเคลื่อนที่พบบ่อย:** "ความกระจายระหว่างสมาชิกของวิธีรวมกลุ่มวัดความไม่แน่นอนแบบเอพิสเทมิกได้เสมอ" ความกระจายวัดได้เพียงว่าสมาชิก *เห็นต่างกัน* เพียงใด ถ้าสมาชิกทุกตัวมีข้อจำกัดเดียวกัน พวกมันจะผิดไปพร้อมกัน การตรวจว่าจุดที่จะทำนายอยู่ในขอบเขตของข้อมูลฝึกหรือไม่เป็นขั้นตอนที่ต้องทำแยก และความรู้ทางฟิสิกส์ว่าปริมาณควรมีพฤติกรรมอย่างไรนอกช่วงนั้น เช่น เป็นคาบ ลดลงเข้าหาศูนย์ หรือโตตามกฎกำลัง มีค่ามากกว่าสถิติใด ๆ ของวิธีรวมกลุ่ม

```python
#@title คำถามทบทวนข้อ 7 (คลิกเพื่อดูแนวคำตอบ)
print('''คำถามทบทวนข้อ 7: ทำไมป่าสุ่มจึงให้ความกระจายระหว่างต้นไม้ที่เล็กมากนอกช่วงของข้อมูลฝึก ทั้งที่ทำนายผิดมาก

แนวคำตอบ: ต้นไม้ทุกต้นทำนายนอกช่วงของข้อมูลเป็นค่าคงที่ของใบที่ขอบ จึงเห็นตรงกันโดยไม่มีข้อมูลรองรับ
ความกระจายวัดได้เพียงความเห็นต่างของสมาชิก ไม่ได้วัดความถูกต้อง บทเรียนคือต้องตรวจว่าจุดที่จะทำนายอยู่ในขอบเขต
ของข้อมูลฝึกหรือไม่ และใช้ความรู้ทางฟิสิกส์เมื่อต้องทำนายนอกขอบเขตนั้น''')
```

### โค้ดที่ 9.10 — การแจกแจงภายหลังของค่าเจาะจงและเวกเตอร์เจาะจงของสายพอลิเมอร์

```python
from scipy.stats import invwishart
from sklearn.decomposition import PCA

X = polymer_trajectory - polymer_trajectory.mean(axis=0)   # Chapter 8 data
n, D = X.shape
S = X.T @ X / n                                   # sample covariance
pc1 = PCA(n_components=1).fit_transform(X)[:, 0]
tau_f, _ = tau_int(pc1)                           # in frames
n_eff = n / (2 * tau_f)                           # effectively independent frames
nu0, Psi0 = D + 2, 1e-3 * np.eye(D)               # weak, proper prior
post = invwishart(df=nu0 + n_eff, scale=Psi0 + n_eff * S)

N = 50
q1 = np.sqrt(2 / N) * np.cos(np.pi * (np.arange(N) + 0.5) / N)   # Rouse p = 1
P1 = np.kron(q1[:, None], np.eye(3))              # x, y, z copies, shape (150, 3)
rng = np.random.default_rng(seed=2026)
lam1, overlap = [], []
for Sig in post.rvs(size=1000, random_state=rng):
    w, V = np.linalg.eigh(Sig)
    lam1.append(w[-3:].mean())                    # degenerate p = 1 triplet
    overlap.append(np.sum((P1.T @ V[:, -3:])**2) / 3)
print("n_eff =", round(n_eff))
print("lambda_1 95% interval:", np.percentile(lam1, [2.5, 50, 97.5]).round(1))
print("overlap with p=1 mode:", np.percentile(overlap, [2.5, 50, 97.5]).round(3))
```

# ส่วนที่ 8 งานนอกชั้น: E9.1–E9.6 ส่วนที่เหลือ

ส่วนนี้ทำนอกเวลาเรียน (รวมราว 3.9 ชั่วโมงตามคู่มือผู้สอน เลือกทำตามที่ผู้สอนมอบหมาย) ต้องรันส่วนที่ 0 ก่อน (คลิกเซลล์หัวคาบที่ 1 แล้วเลือก Runtime → Run before) จากนั้นรันเซลล์เตรียมถัดไป ซึ่งนิยาม `tau_int`, `block_bootstrap` และ `blocking` จากโค้ดที่ 9.4 และเตรียมข้อมูลฮิกส์จำลองกับวิธีจำแนกจากบทที่ 6 แต่ละข้อจึงทำได้โดยไม่ต้องรันคาบที่ 1–4 ใหม่ เซลล์ที่มีเครื่องหมาย ⏱ ใช้เวลาราวหนึ่งถึงสี่นาทีบน Colab

```python
#@title เตรียมฟังก์ชันจากโค้ดที่ 9.4 และข้อมูลจากบทที่ 6 (ดับเบิลคลิกเพื่อดูโค้ด)
def tau_int(x, c=5.0):
    """Integrated autocorrelation time with Sokal's automatic window."""
    x = np.asarray(x) - np.mean(x)
    n = len(x)
    f = np.fft.rfft(x, 2 * n)                      # zero-padded FFT
    acf = np.fft.irfft(f * np.conj(f))[:n]
    acf /= acf[0]
    tau = 0.5
    for W in range(1, n):
        tau += acf[W]
        if W >= c * tau:
            break
    return tau, W

def block_bootstrap(x, stat, block, n_boot=2000, rng=None):
    rng = rng or np.random.default_rng(0)
    n_blocks = len(x) // block
    blocks = np.asarray(x)[:n_blocks * block].reshape(n_blocks, block)
    out = np.empty(n_boot)
    for b in range(n_boot):
        pick = rng.integers(0, n_blocks, n_blocks)   # resample whole blocks
        out[b] = stat(blocks[pick].ravel())
    return out

def blocking(x, min_blocks=16):
    """Flyvbjerg-Petersen: SE of the mean after repeatedly averaging neighbouring pairs."""
    x = np.asarray(x, float); out = []
    while len(x) >= min_blocks:
        out.append(np.sqrt(x.var(ddof=1) / len(x)))
        m = len(x) // 2 * 2
        x = 0.5 * (x[:m:2] + x[1:m:2])
    return np.array(out)

# ---- Chapter 6: Listings 6.1-6.5 (without the comparison loop)
import numpy as np
import pandas as pd

def make_higgs_toy(n_sig=4000, n_bkg=16000, seed=2026):
    """Toy H->bb events vs background (GeV). Shapes are illustrative only."""
    rng = np.random.default_rng(seed)
    # signal: resonance at 125 GeV with ~10% resolution
    m_s   = rng.normal(125.0, 12.0, n_sig)
    pT_s  = rng.gamma(shape=4.0, scale=40.0, size=n_sig)     # mean 160 GeV
    dR_s  = 2 * m_s / np.maximum(pT_s, 60.0) * rng.normal(1.0, 0.15, n_sig)
    met_s = rng.exponential(20.0, n_sig)                       # semileptonic b
    # background: falling mass spectrum, softer pT, wider angles
    m_b   = 40.0 + rng.exponential(70.0, n_bkg)
    pT_b  = rng.gamma(shape=3.0, scale=40.0, size=n_bkg)     # mean 120 GeV
    dR_b  = rng.uniform(0.4, 4.0, n_bkg)
    met_b = rng.exponential(30.0, n_bkg)                       # e.g. ttbar
    df = pd.DataFrame({
        'm_bb':   np.concatenate([m_s, m_b]),
        'pT':     np.concatenate([pT_s, pT_b]),
        'dR':     np.clip(np.concatenate([dR_s, dR_b]), 0.4, 4.0),
        'ETmiss': np.concatenate([met_s, met_b]),
        'label':  np.concatenate([np.ones(n_sig, int), np.zeros(n_bkg, int)]),
    })
    return df.sample(frac=1.0, random_state=seed).reset_index(drop=True)

df = make_higgs_toy()
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

features = ['m_bb', 'pT', 'dR', 'ETmiss']
X = df[features].values
y = df['label'].values

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42)

from sklearn.linear_model import LogisticRegression

lr_pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('clf',    LogisticRegression(C=1.0, class_weight='balanced',
                                  random_state=42)),
])
lr_pipe.fit(X_train, y_train)
y_proba_lr = lr_pipe.predict_proba(X_val)[:, 1]
from sklearn.ensemble import RandomForestClassifier

rf_pipe = Pipeline([
    ('scaler', StandardScaler()),      # harmless for trees; kept for uniformity
    ('clf',    RandomForestClassifier(n_estimators=200, max_depth=10,
                                      min_samples_leaf=10,
                                      class_weight='balanced',
                                      random_state=42, n_jobs=-1)),
])
rf_pipe.fit(X_train, y_train)
importance = rf_pipe.named_steps['clf'].feature_importances_

from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score

classifiers = {
    'LogReg':     LogisticRegression(class_weight='balanced', random_state=42),
    'KNN':        KNeighborsClassifier(n_neighbors=25),
    'SVM (RBF)':  SVC(kernel='rbf', class_weight='balanced', random_state=42),
    'DecTree':    DecisionTreeClassifier(max_depth=8, class_weight='balanced',
                                         random_state=42),
    'RandForest': RandomForestClassifier(n_estimators=200, max_depth=10,
                                         min_samples_leaf=10,
                                         class_weight='balanced',
                                         random_state=42, n_jobs=-1),
}

print('tau_int, block_bootstrap, blocking พร้อมใช้ | Higgs toy data (SYNTHETIC):', df.shape)
```

### ✏️ E9.3 ส่วนที่เหลือ: เวลาสหสัมพันธ์ของแมกเนไทเซชัน (45 นาที)
ดัดแปลงฟังก์ชัน `energy_series` ของโค้ดที่ 9.4 ให้เก็บแมกเนไทเซชันรวม $M$ (ผลรวมของสปินทั้งตาราง) ด้วย โดยเติมบรรทัดที่ขาดในเซลล์ถัดไป แล้วรันที่ $T = 2.3$ ด้วยเมล็ดสุ่มเดิม (⏱ ราวสองนาทีบน Colab) จากนั้นใช้ `m_abs` $= |M|/N$ คำนวณ

- `my_tau_m` เวลาสหสัมพันธ์ของ $|m|$ (ค่าแรกที่ `tau_int` คืนมา)
- `my_se_naive` และ `my_se_tau` ความคลาดเคลื่อนมาตรฐานของ $\langle|m|\rangle$ แบบไม่คิดสหสัมพันธ์และแบบแก้ด้วย $\tau_{\text{int}}$
- `my_se_blk_m[b]` ความคลาดเคลื่อนจากการสุ่มซ้ำเป็นช่วงยาว $b = 100, 200, 500, 1000$

แล้วเทียบกับที่ราบของ `blocking(m_abs)` และอธิบายว่าทำไมสองปริมาณจากการจำลองเดียวกันจึงมีเวลาสหสัมพันธ์ต่างกัน

```python
def energy_magnetisation_series(L, T, n_thermalize, n_measure):      # energy_series, also storing M
    lattice = init_lattice(L)
    beta = 1.0 / T
    for _ in range(n_thermalize):
        metropolis_step(lattice, beta)
    E, M = np.empty(n_measure), np.empty(n_measure)
    for k in range(n_measure):
        metropolis_step(lattice, beta)
        E[k] = total_energy(lattice)
        M[k] = np.nan                                              # ✏️ total magnetisation of the lattice
    return E, M

# ⏱ ราวสองนาทีบน Colab
rng = np.random.default_rng(seed=2026)                             # same restart as Listing 9.4
L, T = 16, 2.3
E_run, M_run = energy_magnetisation_series(L, T, 2000, 20000)
e_run, m_abs = E_run / L**2, np.abs(M_run) / L**2
print(f"<e> = {e_run.mean():.4f}, tau_int(e) = {tau_int(e_run)[0]:.1f};  <|m|> = {np.nanmean(m_abs):.3f}")

my_tau_m = None      # ✏️
my_se_naive = None   # ✏️
my_se_tau = None     # ✏️
my_se_blk_m = {}     # ✏️ {100: ..., 200: ..., 500: ..., 1000: ...}
```

```python
# ✅ ตรวจคำตอบ
if np.isnan(m_abs).any() or abs(m_abs.mean() - 0.683) > 5e-4:
    print("(ยังไม่ได้ตอบ) — m_abs ยังไม่ถูกต้อง: เติมบรรทัด M[k] = ... แล้วรันเซลล์ก่อนหน้าใหม่ (ควรได้ <|m|> = 0.683)")
else:
    _tm = tau_int(m_abs)[0]; _nv = m_abs.std(ddof=1) / np.sqrt(len(m_abs))
    for nm, mine, ref in [("tau_int(|m|)", my_tau_m, _tm), ("SE naive", my_se_naive, _nv), ("SE with tau", my_se_tau, _nv * np.sqrt(2 * _tm))]:
        print(f"{nm:12s}:", "(ยังไม่ได้ตอบ)" if mine is None else ("✔" if abs(mine / ref - 1) < 1e-3 else "✘"), f"— {ref:.4g}")
    for b in (100, 200, 500, 1000):
        ref = block_bootstrap(m_abs, np.mean, block=b).std(); mine = my_se_blk_m.get(b)
        print(f"block {b:4d}  :", "(ยังไม่ได้ตอบ)" if mine is None else ("✔" if abs(mine / ref - 1) < 1e-3 else "✘"), f"— {ref:.4f}")
    print("blocking:", blocking(m_abs).round(4))
```

### ✏️ E9.4 ส่วนที่เหลือ: การแจกแจงก่อนหน้ามีผลต่อตัวประกอบคุณภาพอย่างไร (60 นาที)
ในคาบที่ 3 เราใช้การแจกแจงก่อนหน้าแบบสม่ำเสมอบน $\gamma$ ในช่วง $[0, 5]$ ซึ่งให้น้ำหนักกับ $\gamma$ ค่าสูงมาก ทางเลือกที่นิยมสำหรับพารามิเตอร์ที่เป็นมาตราส่วนคือการแจกแจงที่สม่ำเสมอบน $\ln\gamma$ ซึ่งเท่ากับ $p(\gamma) \propto 1/\gamma$ หรือ $\ln p(\gamma) = -\ln\gamma + \text{ค่าคงที่}$

ฟังก์ชัน `chain_with_prior` ในเซลล์ถัดไปคือตัวสุ่มของโค้ดที่ 9.5 ที่รับลอการิทึมของการแจกแจงก่อนหน้าเป็นอาร์กิวเมนต์ เติม `my_log_prior` ให้เป็นการแจกแจงสม่ำเสมอบน $\ln\gamma$ (รับ `theta` = $(A, \gamma, \omega, \phi)$) แล้วรันเซลล์ (⏱ ราวหนึ่งนาทีบน Colab สำหรับสองลูกโซ่) เปรียบเทียบมัธยฐานและช่วงร้อยละ 95 ของ $Q$ ทั้งสองแบบ แล้วอธิบายว่าผลบอกอะไรเกี่ยวกับข้อมูลชุดนี้ ถ้ามีเวลา ให้ทำซ้ำด้วยโค้ดที่ 9.6 กับข้อมูลชุดนี้

```python
rng_d = np.random.default_rng(seed=2026)
t_bad = np.linspace(0, 10, 60)
x_bad = damped_oscillator(t_bad, 1.0, 0.3, 6.0, 0.5) + rng_d.normal(scale=0.4, size=t_bad.size)
popt_bad, pcov_bad = curve_fit(damped_oscillator, t_bad, x_bad, p0=[1.0, 0.2, 6.0, 0.0], bounds=bounds,
                               sigma=0.4 * np.ones(t_bad.size), absolute_sigma=True)

def chain_with_prior(log_prior, n=100000, burn=10000, seed=1):
    lo_, hi_ = np.array(bounds[0]), np.array(bounds[1])
    def lp(th):
        if np.any(th < lo_) or np.any(th > hi_) or th[1] <= 0:
            return -np.inf
        return -0.5 * np.sum(((x_bad - damped_oscillator(t_bad, *th)) / 0.4)**2) + log_prior(th)
    rg = np.random.default_rng(seed)
    stp = 2.38**2 / 4 * pcov_bad
    th, l = popt_bad.copy(), lp(popt_bad); ch = []
    for _ in range(n):
        pr = rg.multivariate_normal(th, stp); lpr = lp(pr)
        if np.log(rg.random()) < lpr - l:
            th, l = pr, lpr
        ch.append(th)
    return np.array(ch)[burn:]

my_log_prior = None   # ✏️ a function of theta: uniform in ln(gamma)

# ⏱ ราวหนึ่งนาทีบน Colab
if my_log_prior is not None:
    Q_prior = {}
    for name, prior in [("uniform in gamma", lambda th: 0.0), ("uniform in ln gamma", my_log_prior)]:
        ch_ = chain_with_prior(prior)
        Q_prior[name] = ch_[:, 2] / (2 * ch_[:, 1])
        print(f"{name:20s}: gamma median {np.median(ch_[:, 1]):.2f}, Q median {np.median(Q_prior[name]):.1f}, "
              f"95% {np.percentile(Q_prior[name], [2.5, 97.5]).round(1)}")
```

```python
# ✅ ตรวจคำตอบ
if my_log_prior is None:
    print("(ยังไม่ได้ตอบ)")
else:
    _th = np.array([1.0, 0.3, 6.0, 0.5]); _th2 = np.array([1.0, 0.03, 6.0, 0.5])
    ok = abs((my_log_prior(_th2) - my_log_prior(_th)) - np.log(10)) < 1e-9
    print("log prior:", "✔" if ok else "✘", "— ln p(gamma = 0.03) - ln p(gamma = 0.3) ต้องเท่ากับ ln 10 = 2.303")
    _q = np.percentile(Q_prior["uniform in ln gamma"], [2.5, 50, 97.5])
    print(f"ln-gamma prior: Q median {_q[1]:.1f}, 95% [{_q[0]:.1f}, {_q[2]:.0f}]  (textbook: 13.5, [5.4, about 1000])")
    fig, ax = plt.subplots(figsize=(7, 3.2))
    for name, Qv in Q_prior.items():
        ax.hist(np.log10(Qv), bins=80, histtype="step", lw=1.3, density=True, label=name)
    ax.axvline(1, c="r", lw=1, label="true Q = 10")
    ax.set(xlabel="log10 Q", ylabel="density", title="posterior of Q, poor data, two priors (SYNTHETIC)"); ax.legend(fontsize=8)
    fig.tight_layout(); plt.show()
```

### ✏️ E9.5 เปรียบเทียบวิธีจำแนกด้วยการตรวจสอบไขว้ซ้ำ (30 นาที)
เซลล์ถัดไปคำนวณพื้นที่ใต้เส้นโค้ง ROC ของ SVM, ป่าสุ่ม และ KNN ($k = 25$) บนข้อมูลฮิกส์จำลองทั้งชุด ด้วยการตรวจสอบไขว้ห้าส่วนแบบรักษาสัดส่วน ทำซ้ำห้ารอบ (`RepeatedStratifiedKFold`, `random_state=2026`) โดยใช้การแบ่งเดียวกันกับทุกวิธี ได้ $J = 25$ คะแนนต่อวิธีใน `auc[name]` (⏱ ราวสามถึงสี่นาทีบน Colab)

สำหรับผลต่างของ SVM กับป่าสุ่ม $d_j$ ($j = 1, \dots, J$) คำนวณ
- สถิติทีแบบจับคู่ธรรมดา $t = \bar d / (s_d/\sqrt{J})$ เก็บใน `my_t_naive`
- สถิติทีที่แก้ตามนาโดและเบนจิโอ $t = \bar d \big/ \sqrt{(1/J + n_{\text{test}}/n_{\text{train}})\,s_d^2}$ โดย $n_{\text{test}}/n_{\text{train}} = 1/4$ เก็บใน `my_t_nb`

เมื่อ $s_d$ คือส่วนเบี่ยงเบนมาตรฐานของ $d_j$ (`ddof=1`) แล้วเปรียบเทียบค่าพีของทั้งสองแบบ (องศาเสรี $J - 1$)

```python
from sklearn.model_selection import RepeatedStratifiedKFold, cross_val_score

# ⏱ ราวสามถึงสี่นาทีบน Colab
t0 = time.time()
X_all, y_all = df[features].to_numpy(), df['label'].to_numpy()
cv_rep = RepeatedStratifiedKFold(n_splits=5, n_repeats=5, random_state=2026)
auc = {}
for name in ('SVM (RBF)', 'RandForest', 'KNN'):
    pipe = Pipeline([('scaler', StandardScaler()), ('clf', classifiers[name])])
    auc[name] = cross_val_score(pipe, X_all, y_all, cv=cv_rep, scoring='roc_auc', n_jobs=-1)
    print(f"{name:11s} mean AUC {auc[name].mean():.4f}")
print(f"({time.time() - t0:.0f} s)")
J, ratio = 25, 1 / 4
d_svm_rf = auc['SVM (RBF)'] - auc['RandForest']

my_t_naive = None   # ✏️
my_t_nb = None      # ✏️
```

```python
# ✅ ตรวจคำตอบ
from scipy import stats
_tn = d_svm_rf.mean() / (d_svm_rf.std(ddof=1) / np.sqrt(J))
_tb = d_svm_rf.mean() / np.sqrt((1 / J + ratio) * d_svm_rf.var(ddof=1))
for nm, mine, ref in [("naive t", my_t_naive, _tn), ("corrected t", my_t_nb, _tb)]:
    print(f"{nm:11s}:", "(ยังไม่ได้ตอบ)" if mine is None else ("✔" if abs(mine / ref - 1) < 1e-3 else "✘"),
          f"— {ref:.1f}, p = {2 * stats.t.sf(abs(ref), J - 1):.1g}")
print(f"correction factor sqrt(1 + J n_test/n_train) = {np.sqrt(1 + J * ratio):.2f}")
for a, b in [('SVM (RBF)', 'KNN'), ('RandForest', 'KNN')]:
    dd = auc[a] - auc[b]
    tn = dd.mean() / (dd.std(ddof=1) / np.sqrt(J)); tb = dd.mean() / np.sqrt((1 / J + ratio) * dd.var(ddof=1))
    print(f"{a} - {b}: {dd.mean():+.4f}; naive t = {tn:.1f}, p = {2 * stats.t.sf(abs(tn), J - 1):.1g}; "
          f"corrected t = {tb:.1f}, p = {2 * stats.t.sf(abs(tb), J - 1):.1g}")
```

### สรุปบท
ความไม่แน่นอนแบบเอลีอะทอริกมาจากความสุ่มของระบบและการวัด และข้อมูลเพิ่มแก้ไม่ได้ ความไม่แน่นอนแบบเอพิสเทมิกมาจากความรู้ที่จำกัด และลดลงได้ด้วยข้อมูลหรือแบบจำลองที่ดีขึ้น การรู้ว่าความไม่แน่นอนส่วนใหญ่เป็นชนิดใดบอกได้ว่าควรลงทุนแก้ที่จุดใด การตรวจสอบไขว้ให้คำตอบที่ถูกต้องก็ต่อเมื่อการแบ่งข้อมูลเลียนแบบการใช้งานจริง ข้อมูลที่สัมพันธ์กันในเวลาหรือเป็นกลุ่มต้องแบ่งเป็นช่วงหรือตามกลุ่ม และการเลือกไฮเพอร์พารามิเตอร์ต้องอยู่ในชั้นในของการตรวจสอบไขว้แบบซ้อน AIC มุ่งเลือกแบบจำลองที่ทำนายดี BIC มุ่งเลือกแบบจำลองที่ถูกต้อง และทั้งสองต้องอ่านคู่กับความรู้ทางฟิสิกส์

บูตสแตรปประมาณความแปรปรวนจากการสุ่มตัวอย่างโดยไม่ต้องหาอนุพันธ์ และจับความเบ้ของปริมาณที่ไม่เชิงเส้นได้ แต่เมื่อข้อมูลสัมพันธ์กันในเวลา การสุ่มทีละจุดให้ความไม่แน่นอนที่เล็กเกินจริง ใกล้จุดวิกฤตของแบบจำลองไอซิงเล็กไปเกือบหกเท่า การแก้ต้องใช้เวลาสหสัมพันธ์แบบปริพันธ์ การสุ่มซ้ำเป็นช่วง หรือการจัดกลุ่มซ้ำ กรอบแบบเบส์เขียนสิ่งที่รู้ก่อนเห็นข้อมูลเป็นการแจกแจงก่อนหน้า การสุ่มตัวอย่างการแจกแจงภายหลังคือการใช้วิธีเมโทรโพลิสกับพารามิเตอร์ที่มี "พลังงาน" เป็นลบของลอการิทึมของการแจกแจงภายหลัง และการตรวจเชิงทำนายภายหลังตรวจว่าแบบจำลองเพียงพอหรือไม่

การเปรียบเทียบวิธีจำแนกบนข้อมูลเดียวกันต้องใช้การทดสอบแบบจับคู่ ความน่าจะเป็นจากวิธีจำแนกต้องตรวจการปรับเทียบก่อนนำไปใช้เป็นตัวเลข และความกระจายของวิธีรวมกลุ่มเตือนเรื่องความไม่แน่นอนแบบเอพิสเทมิกได้เฉพาะเมื่อสมาชิกมีโอกาสเห็นต่างกันจริง เมื่อทำนายไกลออกไปนอกข้อมูล ความรู้ทางฟิสิกส์เป็นเครื่องมือเดียวที่เชื่อถือได้ บทที่ 10 จะเริ่มเรื่องโครงข่ายประสาทเทียม ซึ่งการระบุความไม่แน่นอนต้องอาศัยวิธีรวมกลุ่มและการประมาณแบบเบส์ที่พัฒนาจากความคิดในบทนี้
