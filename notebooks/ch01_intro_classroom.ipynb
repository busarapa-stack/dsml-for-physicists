### การล็อกแนวคำตอบ
เซลล์ ✅ และเซลล์แนวคำตอบในโน้ตบุ๊กนี้แสดงคำตอบอ้างอิงเมื่อผู้เรียนเติมคำตอบในเซลล์ ✏️ ของข้อนั้นครบแล้วเท่านั้น ถ้ายังไม่ได้ตอบ เซลล์จะแสดงเพียงข้อความ 🔒 บอกว่ายังขาดคำตอบใด การคำนวณในเซลล์ยังทำงานตามปกติ ดังนั้น Run all จึงรันได้ครบโดยไม่เปิดเผยคำตอบ แนวคำตอบของงานเขียนและงานนอกชั้นซึ่งตรวจด้วยโค้ดไม่ได้จะเปิดเมื่อผู้สอนเลือกเท่านั้น

**สำหรับผู้สอน:** ติ๊กช่อง `SHOW_ANSWERS` ในเซลล์ถัดไป (หรือแก้เป็น `True`) แล้วรันเซลล์นั้นอีกครั้ง เพื่อแสดงแนวคำตอบทุกข้อ เช่น เมื่อเฉลยในชั้นหรือเตรียมสอน

```python
#@title ⚙️ ตัวเลือกของผู้สอน: แสดงแนวคำตอบ
SHOW_ANSWERS = False  #@param {type:"boolean"}

def _answered(v):
    """True when an exercise variable holds an answer (not None, "?", "", empty or containing blanks)."""
    if v is None or v is Ellipsis:
        return False
    if isinstance(v, str):                       # placeholders such as "?", "?-?", "..." count as blank
        return v.replace("?", "").replace("-", "").replace(".", "").strip() != ""
    if isinstance(v, dict):
        return len(v) > 0 and all(_answered(x) for x in v.values())
    if isinstance(v, (list, tuple, set, frozenset)):
        return len(v) > 0 and all(_answered(x) for x in v)
    if type(v).__name__ == "ndarray":
        return v.size > 0
    return True

def _gate(names=()):
    """Return print if the answers may be shown, otherwise print one lock message and return a no-op."""
    if SHOW_ANSWERS:
        return print
    if names:
        missing = [n for n in names if not _answered(globals().get(n))]
        if not missing:
            return print
        print("🔒 ยังไม่ได้ตอบ:", ", ".join(missing),
              "— เติมคำตอบในเซลล์ ✏️ ด้านบนให้ครบก่อน แล้วรันเซลล์นี้อีกครั้ง")
    else:
        print("🔒 แนวคำตอบข้อนี้ผู้สอนเป็นผู้เปิด (ตั้ง SHOW_ANSWERS = True ในส่วนที่ 0)")
    return lambda *args, **kwargs: None

print("แสดงแนวคำตอบทั้งหมด:", "ใช่ (โหมดผู้สอน)" if SHOW_ANSWERS else "ไม่ (แสดงเมื่อผู้เรียนตอบข้อนั้นครบแล้ว)")
```

เซลล์ถัดไปสร้างฟังก์ชันช่วยสองตัว ตัวแรก `download()` ดาวน์โหลดไฟล์ไปเก็บใน `data/real/` และข้ามการดาวน์โหลดถ้ามีไฟล์อยู่แล้ว ฟังก์ชันนี้ลองส่งคำขอหลายแบบ เพราะบางเซิร์ฟเวอร์ปฏิเสธคำขอบางรูปแบบด้วยรหัส 403 ตัวที่สอง `banner()` บอกว่ากำลังใช้ข้อมูลจริงหรือไฟล์จำลอง ผู้เรียนไม่จำเป็นต้องเข้าใจรายละเอียดของเซลล์นี้

```python
os.makedirs("data/real", exist_ok=True)

def download(url, fname, timeout=60):
    # ดาวน์โหลด url ไปเก็บที่ fname; คืนค่า True ถ้าสำเร็จ
    if os.path.exists(fname) and os.path.getsize(fname) > 0:
        return True
    last_err = None
    for ua in ["python-requests/2.32", "Mozilla/5.0 (Macintosh; Intel Mac OS X 10_15_7)", None]:
        try:
            req = urllib.request.Request(url, headers={"User-Agent": ua} if ua else {})
            with urllib.request.urlopen(req, timeout=timeout) as r, open(fname, "wb") as f:
                while True:
                    chunk = r.read(1 << 20)
                    if not chunk:
                        break
                    f.write(chunk)
            return True
        except Exception as e:
            last_err = e
            if os.path.exists(fname):
                os.remove(fname)
    print(f"  ดาวน์โหลดไม่สำเร็จ: {type(last_err).__name__}: {last_err}")
    return False

def banner(is_real, source):
    if is_real:
        print(f"✔ ใช้ข้อมูลจริงจาก {source}")
    else:
        print(f"⚠ เชื่อมต่อ {source} ไม่ได้ จึงใช้ไฟล์จำลองใน data/ แทน (ตีความเชิงคุณภาพเท่านั้น)")
```

เซลล์ถัดไปสร้างไฟล์จำลองสำรองห้าไฟล์ในโฟลเดอร์ `data/` ในรูปแบบ ROOT, HDF5, FITS และ JSON โค้ดของเซลล์เป็นชุดเดียวกับสคริปต์ `01_make_sample_files.py` ในโค้ดประกอบตำรา (เมล็ดสุ่ม 2026) ไฟล์จำลองจึงตรงกับที่ตำราใช้ทุกประการ ส่วนที่ 2–6 จะใช้ไฟล์เหล่านี้แทนเมื่อโหลดข้อมูลจริงไม่ได้ โค้ดถูกซ่อนไว้เพื่อให้โน้ตบุ๊กอ่านง่าย ผู้ที่สนใจดับเบิลคลิกที่เซลล์เพื่อดูโค้ดได้

```python
#@title สร้างไฟล์จำลองสำรอง 5 ไฟล์ใน data/ (ดับเบิลคลิกเพื่อดูโค้ด)
# โค้ดชุดเดียวกับ ch01_intro/01_make_sample_files.py ในโค้ดประกอบตำรา
SRC = r'''"""Ch.1  Create small SYNTHETIC sample files in the four formats of Section 1.2.2.

Book references
    ssec:example, code:root-tree, code:ligo-hdf5, code:fits-header, code:matproj-json

The files mimic the *structure* of the real open datasets (CMS dimuon ROOT tree,
LIGO strain HDF5, SDSS FITS image/spectrum, Materials Project JSON) but the
numbers are generated here. They are small (a few MB) so every reader gets the
same result without downloading gigabytes. To open the real data, see optional/.

Output (in data/)
    dimuon_sample.root                     ROOT tree "Events" with jagged muon branches
    H-H1_SYNTH_4_V1-1126259446-32.hdf5     32 s of strain at 4096 Hz
    sdss_like_image_r.fits                 256 x 256 galaxy image, r band
    sdss_like_spectrum.fits                1D stellar spectrum (loglam, flux, ivar)
    mp_like_mp-149.json                    Materials-Project-style record for Si
"""
import json
from pathlib import Path

import awkward as ak
import h5py
import numpy as np
import uproot
from astropy.io import fits

SEED = 2026
HERE = Path(__file__).resolve().parent
DATA = HERE / "data"
DATA.mkdir(exist_ok=True)
rng = np.random.default_rng(SEED)

MUON_MASS = 0.1056584  # GeV/c^2
Z_MASS, Z_WIDTH = 91.19, 2.50  # GeV/c^2


# ---------------------------------------------------------------------------
# 1) ROOT: event-level data with a variable number of muons per event
# ---------------------------------------------------------------------------
def z_pair(n):
    """Two muons whose invariant mass follows a Breit-Wigner around the Z."""
    pts, etas, phis = [], [], []
    while len(pts) < n:
        m = Z_MASS + Z_WIDTH / 2 * np.tan(np.pi * (rng.random() - 0.5))
        m += rng.normal(0, 1.5)                       # detector resolution
        if not 60 < m < 120:
            continue
        eta = rng.uniform(-2.4, 2.4, 2)
        phi = rng.uniform(-np.pi, np.pi, 2)
        pt1 = rng.uniform(25, 60)
        # massless approximation: m^2 = 2 pt1 pt2 (cosh(d_eta) - cos(d_phi))
        pt2 = m**2 / (2 * pt1 * (np.cosh(eta[0] - eta[1]) - np.cos(phi[0] - phi[1])))
        if not 10 < pt2 < 150:
            continue
        pts.append([pt1, pt2]); etas.append(list(eta)); phis.append(list(phi))
    return pts, etas, phis


def make_root(n_events=5000, frac_signal=0.6):
    n_sig = int(frac_signal * n_events)
    pt, eta, phi, charge = [], [], [], []

    s_pt, s_eta, s_phi = z_pair(n_sig)
    for i in range(n_sig):
        q = rng.choice([-1, 1])
        p, e, f, c = s_pt[i], s_eta[i], s_phi[i], [q, -q]   # opposite charge
        # about 10 % of events carry one extra soft muon
        if rng.random() < 0.10:
            p = p + [rng.exponential(8) + 3]
            e = e + [rng.uniform(-2.4, 2.4)]
            f = f + [rng.uniform(-np.pi, np.pi)]
            c = c + [int(rng.choice([-1, 1]))]
        pt.append(p); eta.append(e); phi.append(f); charge.append(c)

    for _ in range(n_events - n_sig):                     # continuum background
        k = rng.choice([1, 2, 3], p=[0.2, 0.7, 0.1])
        pt.append(list(rng.exponential(12, k) + 3))
        eta.append(list(rng.uniform(-2.4, 2.4, k)))
        phi.append(list(rng.uniform(-np.pi, np.pi, k)))
        charge.append([int(x) for x in rng.choice([-1, 1], k)])

    order = rng.permutation(n_events)                     # mix signal and background
    pick = lambda a: [a[i] for i in order]
    f32 = lambda a: ak.values_astype(ak.Array(a), np.float32)   # real file uses float
    muon = ak.zip({
        "pt": f32(pick(pt)),
        "eta": f32(pick(eta)),
        "phi": f32(pick(phi)),
        "mass": ak.full_like(f32(pick(pt)), MUON_MASS),
        "charge": ak.values_astype(ak.Array(pick(charge)), np.int32),
    })

    path = DATA / "dimuon_sample.root"
    with uproot.recreate(path) as f:
        f.mktree("Events", {"Muon": muon.type.content},
                 counter_name=lambda c: "n" + c,
                 field_name=lambda outer, inner: f"{outer}_{inner}",
                 title="SYNTHETIC dimuon events (structure of CMS open data)")
        f["Events"].extend({"Muon": muon})
    return path


# ---------------------------------------------------------------------------
# 2) HDF5: 32 s strain time series at 4096 Hz with meta and quality groups
# ---------------------------------------------------------------------------
def make_hdf5():
    fs, duration, gps0 = 4096, 32, 1126259446
    n = fs * duration
    t = np.arange(n) / fs
    strain = rng.normal(0, 1e-21, n)                      # white noise, toy level

    # toy "chirp": frequency sweeps 35 -> 250 Hz during 0.2 s before t = 16.4 s
    tc = 16.4
    m = (t > tc - 0.2) & (t < tc)
    tau = tc - t[m]
    f_inst = 35 + (250 - 35) * (1 - tau / 0.2) ** 3
    phase = 2 * np.pi * np.cumsum(f_inst) / fs
    strain[m] += 1.0e-21 * (1 - tau / 0.2) * np.sin(phase)

    path = DATA / f"H-H1_SYNTH_4_V1-{gps0}-{duration}.hdf5"
    # Same layout as a real GWOSC/LOSC file (checked against
    # H-H1_LOSC_4_V1-1126256640-4096.hdf5). Note: there is no "SampleRate"
    # entry in meta/; the sampling interval is the attribute Xspacing of Strain.
    with h5py.File(path, "w") as h:
        meta = h.create_group("meta")
        meta["Description"] = b"SYNTHETIC strain for teaching (layout of a GWOSC file)"
        meta["DescriptionURL"] = b"https://gwosc.org/"
        meta["Detector"] = b"H1"
        meta["Duration"] = duration
        meta["GPSstart"] = gps0
        meta["Observatory"] = b"H"
        meta["Type"] = b"StrainTimeSeries"
        meta["UTCstart"] = b"2015-09-14T09:50:29"
        ds = h.create_group("strain").create_dataset("Strain", data=strain)
        ds.attrs.update({"Npoints": n, "Xlabel": "GPS time", "Xspacing": 1 / fs,
                         "Xstart": gps0, "Xunits": "second", "Ylabel": "Strain",
                         "Yunits": ""})
        q = h.create_group("quality")
        simple = q.create_group("simple")
        simple["DQShortnames"] = np.array([b"DATA", b"CBC_CAT1", b"CBC_CAT2", b"CBC_CAT3",
                                           b"BURST_CAT1", b"BURST_CAT2", b"BURST_CAT3"])
        simple["DQmask"] = np.full(duration, 127, dtype=np.uint32)   # all 7 bits = good
        inj = q.create_group("injections")
        inj["InjShortnames"] = np.array([b"NO_CBC_HW_INJ", b"NO_BURST_HW_INJ",
                                         b"NO_DETCHAR_HW_INJ", b"NO_CW_HW_INJ",
                                         b"NO_STOCH_HW_INJ"])
        inj["Injmask"] = np.full(duration, 31, dtype=np.uint32)      # 1 = no injection
    return path


# ---------------------------------------------------------------------------
# 3) FITS image: header (metadata) + 2D pixel array
# ---------------------------------------------------------------------------
def make_fits_image():
    ny = nx = 256
    y, x = np.mgrid[0:ny, 0:nx]
    x0, y0, q, pa = 128.0, 128.0, 0.6, np.deg2rad(30)
    xr = (x - x0) * np.cos(pa) + (y - y0) * np.sin(pa)
    yr = -(x - x0) * np.sin(pa) + (y - y0) * np.cos(pa)
    r = np.hypot(xr, yr / q)
    galaxy = 40.0 * np.exp(-r / 12.0)                     # exponential disk
    sky = 5.0
    img = rng.poisson(galaxy + sky).astype(np.float32) - sky

    hdr = fits.Header()
    hdr["RA"] = (179.689453125, "Right Ascension [deg]")
    hdr["DEC"] = (0.28563112068, "Declination [deg]")
    hdr["FILTER"] = ("r", "r-band")
    hdr["EXPTIME"] = (53.91, "exposure time [s]")
    hdr["BUNIT"] = ("counts", "pixel unit (sky subtracted)")
    hdr["COMMENT"] = "SYNTHETIC image for teaching; header keywords follow SDSS style"
    path = DATA / "sdss_like_image_r.fits"
    fits.PrimaryHDU(data=img, header=hdr).writeto(path, overwrite=True)
    return path


# ---------------------------------------------------------------------------
# 4) FITS spectrum: binary table with log10(wavelength), flux, inverse variance
# ---------------------------------------------------------------------------
def make_fits_spectrum():
    loglam = np.arange(np.log10(3800), np.log10(9200), 1e-4)   # SDSS-like sampling
    lam = 10**loglam                                            # Angstrom
    h, c, k, T = 6.626e-34, 2.998e8, 1.381e-23, 6000.0
    lam_m = lam * 1e-10
    bb = 1 / lam_m**5 / (np.exp(h * c / (lam_m * k * T)) - 1)
    flux = 50 * bb / bb.max()
    for line, depth, width in [(6563, 0.35, 6), (4861, 0.30, 5), (4340, 0.25, 5),
                               (5892, 0.20, 3)]:                # H-alpha/beta/gamma, Na D
        flux *= 1 - depth * np.exp(-0.5 * ((lam - line) / width) ** 2)
    sigma = 0.02 * flux.max() * np.ones_like(flux)
    flux_obs = flux + rng.normal(0, sigma)

    cols = fits.ColDefs([
        fits.Column(name="loglam", format="E", array=loglam),
        fits.Column(name="flux", format="E", array=flux_obs, unit="arbitrary"),
        fits.Column(name="ivar", format="E", array=1 / sigma**2),
    ])
    hdr = fits.Header()
    hdr["OBJTYPE"] = ("STAR", "object class")
    hdr["TEFF"] = (T, "temperature used to generate the continuum [K]")
    hdr["COMMENT"] = "SYNTHETIC spectrum for teaching (blackbody + 4 absorption lines)"
    path = DATA / "sdss_like_spectrum.fits"
    fits.HDUList([fits.PrimaryHDU(header=hdr),
                  fits.BinTableHDU.from_columns(cols, name="COADD")]).writeto(path, overwrite=True)
    return path


# ---------------------------------------------------------------------------
# 5) JSON: nested key-value record, as returned by the Materials Project API
# ---------------------------------------------------------------------------
def make_json():
    rec = {
        "_note": "Structure of an MP summary record. Fields set to null must be "
                 "fetched from the API (see optional/fetch_real_data.py).",
        "material_id": "mp-149",
        "formula_pretty": "Si",
        "symmetry": {"crystal_system": "Cubic", "symbol": "Fd-3m", "number": 227},
        "formation_energy_per_atom": 0.0,   # elemental reference, zero by definition
        "energy_above_hull": 0.0,           # ground state
        "is_stable": True,
        "band_gap": None,                   # eV, from DFT -> fetch from API
        "density": None,                    # g/cm^3       -> fetch from API
    }
    path = DATA / "mp_like_mp-149.json"
    path.write_text(json.dumps(rec, indent=2, ensure_ascii=False), encoding="utf-8")
    return path


if __name__ == "__main__":
    for fn in (make_root, make_hdf5, make_fits_image, make_fits_spectrum, make_json):
        p = fn()
        print(f"{p.name:42s} {p.stat().st_size / 1024:9.1f} kB")
'''
if not os.path.exists(f"{SYNTH}/dimuon_sample.root"):
    exec(compile(SRC, "make_sample_files.py", "exec"),
         {"__name__": "__main__", "__file__": os.path.abspath("make_sample_files.py")})
print("ไฟล์จำลอง:", sorted(x for x in os.listdir(SYNTH) if x != "real"))
```

# ส่วนที่ 1 กรอบ 4V ด้วยตัวเลข (หัวข้อ 1.2.1)

ตำรานิยามปริมาณของข้อมูล (volume) ว่าเป็นขนาดของข้อมูลที่ต้องจัดเก็บและประมวลผล และนิยามความเร็วของข้อมูล (velocity) ว่าเป็นอัตราที่ข้อมูลถูกสร้าง สองมิตินี้คำนวณได้โดยตรงจากคุณสมบัติของเครื่องวัด

ยกตัวอย่างเครื่องตรวจวัด LIGO หนึ่งเครื่อง ซึ่งบันทึกค่า strain ด้วยอัตราการสุ่มตัวอย่าง (sampling rate) $f_s$ ตัวอย่างต่อวินาที โดยแต่ละตัวอย่างเก็บเป็นเลขทศนิยมแบบ 64 บิต หรือ 8 ไบต์ ขนาดข้อมูลที่สะสมในเวลา $T$ วินาทีจึงเท่ากับ

$$\text{Volume} = f_s \times T \times 8~\text{bytes}$$

```python
fs = 4096            # sampling rate [samples/s]  -> Velocity
bytes_per_sample = 8 # float64
T_event = 32         # ความยาวไฟล์ข้อมูลหนึ่งเหตุการณ์ [s]
T_year = 365.25 * 24 * 3600

vol_event = fs * T_event * bytes_per_sample
vol_year = fs * T_year * bytes_per_sample

print(f"Velocity             : {fs:,} samples/s")
print(f"จำนวนจุดข้อมูลใน 32 s : {fs * T_event:,} จุด")
print(f"Volume ต่อไฟล์ 32 s   : {vol_event / 1e6:.2f} MB")
print(f"Volume ต่อปี (1 ช่อง) : {vol_year / 1e12:.2f} TB")
```

ผลข้างต้นคิดเพียงช่องสัญญาณ strain ช่องเดียว ในความเป็นจริง เครื่อง LIGO บันทึกช่องสัญญาณตรวจสภาพแวดล้อมและสถานะของเครื่องมืออีกนับหมื่นช่องพร้อมกัน ปริมาณข้อมูลจริงจึงสูงกว่าตัวเลขนี้หลายอันดับ

### ✏️ แบบฝึก 1.1
LIGO ยังเผยแพร่ข้อมูลที่อัตราการสุ่มตัวอย่าง 16384 Hz ด้วย ให้แก้ค่า `fs` ในเซลล์ด้านบนเป็น 16384 แล้วรันใหม่ ขนาดข้อมูลต่อปีเปลี่ยนไปกี่เท่า และเหตุใดจึงเป็นเช่นนั้น

# ส่วนที่ 2 CERN Open Data: ข้อมูลรายเหตุการณ์ในไฟล์ ROOT (หัวข้อ 1.2.2)

### ที่มาของข้อมูล
ชุดข้อมูลในส่วนนี้คือ *DoubleMuParked dataset from 2012 in NanoAOD format reduced on muons* (Wunsch, 2019; DOI: [10.7483/OPENDATA.CMS.LVG5.QT81](https://doi.org/10.7483/OPENDATA.CMS.LVG5.QT81)) ซึ่งลดรูปจากข้อมูลต้นฉบับของการทดลอง CMS ที่ LHC ให้เหลือเฉพาะตัวแปรของมิวออน เผยแพร่ภายใต้สัญญาอนุญาต CC0 1.0 ไฟล์มีขนาดประมาณ 2.1 GB และบรรจุเหตุการณ์การชนประมาณ 61.5 ล้านเหตุการณ์

### แนวคิดสำคัญ
ไลบรารี `uproot` อ่านไฟล์ ROOT ผ่านเครือข่ายได้โดยดึงเฉพาะส่วนของไฟล์ที่ต้องการ เราจึงไม่ต้องดาวน์โหลดไฟล์ทั้ง 2.1 GB แต่อ่านเฉพาะ 200,000 เหตุการณ์แรกของตัวแปรที่ใช้ ซึ่งเป็นแนวปฏิบัติจริงเมื่อข้อมูลมีปริมาณมาก (volume)

```python
import uproot
import awkward as ak

CERN_URLS = [
    "https://opendata.cern.ch/eos/opendata/cms/derived-data/AOD2NanoAODOutreachTool/Run2012BC_DoubleMuParked_Muons.root",
    "root://eospublic.cern.ch//eos/opendata/cms/derived-data/AOD2NanoAODOutreachTool/Run2012BC_DoubleMuParked_Muons.root",
]

cern_tree = None
for url in CERN_URLS:
    try:
        cern_tree = uproot.open(url, timeout=60)["Events"]
        break
    except Exception as e:
        print(f"  เปิด {url.split(':')[0]} ไม่สำเร็จ: {type(e).__name__}")
CERN_REAL = cern_tree is not None
if not CERN_REAL:
    cern_tree = uproot.open(f"{SYNTH}/dimuon_sample.root")["Events"]
banner(CERN_REAL, "CERN Open Data Portal")
```

### โค้ดที่ 1.1 — ดูโครงสร้าง Tree ก่อนอ่านข้อมูล
ไฟล์ ROOT จัดเก็บข้อมูลเป็น Tree และแต่ละ Tree มี Branch ซึ่งเทียบได้กับช่องข้อมูลที่ทุกเหตุการณ์มีเหมือนกัน เซลล์ถัดไปแสดงชื่อ Branch ชนิดข้อมูล และจำนวนเหตุการณ์ทั้งหมด ให้เปรียบเทียบผลกับโค้ดที่ 1.1 ในตำรา

```python
print(f"จำนวนเหตุการณ์ทั้งหมด: {cern_tree.num_entries:,}" + ("" if CERN_REAL else "  (SYNTHETIC)") + "\n")
cern_tree.show()
```

สังเกตว่า Branch ส่วนใหญ่มีชนิดข้อมูลเป็น `var * float32` ซึ่งหมายความว่า **แต่ละเหตุการณ์มีจำนวนมิวออนไม่เท่ากัน** ข้อมูลลักษณะนี้เรียกว่า jagged array และจัดลงตารางสี่เหลี่ยมธรรมดาโดยตรงไม่ได้ นี่เป็นตัวอย่างหนึ่งของความหลากหลายของข้อมูล (variety) ในระดับโครงสร้าง

เซลล์ถัดไปอ่าน 200,000 เหตุการณ์แรก และแสดงห้าเหตุการณ์แรกให้เห็นเป็นตัวอย่าง

```python
N_EVENTS = 200_000
branches = ["nMuon", "Muon_pt", "Muon_eta", "Muon_phi", "Muon_charge"]

ev = cern_tree.arrays(branches, entry_stop=min(N_EVENTS, cern_tree.num_entries))
print(f"อ่าน {len(ev):,} เหตุการณ์" + ("" if CERN_REAL else " จากไฟล์จำลอง (มีเพียง Z และพื้นหลังต่อเนื่อง)"))

for i in range(5):
    print(f"event {i}: nMuon={ev['nMuon'][i]}, pt={np.round(ak.to_list(ev['Muon_pt'][i]), 1)} GeV/c, "
          f"charge={ak.to_list(ev['Muon_charge'][i])}")
```

### จากข้อมูลรายเหตุการณ์สู่ปริมาณทางฟิสิกส์: มวลยืนยง (invariant mass)
หากอนุภาคที่ไม่เสถียรตัวหนึ่งสลายตัวเป็นมิวออนสองตัว มวลยืนยงของคู่มิวออนจะเท่ากับมวลของอนุภาคต้นกำเนิด เมื่อมิวออนมีพลังงานสูงมากเมื่อเทียบกับมวลนิ่งของมันเอง ($m_\mu \approx 0.106~\text{GeV}/c^2$) มวลยืนยงประมาณได้จากโมเมนตัมตามขวาง $p_T$ ค่า pseudorapidity $\eta$ และมุม $\phi$ ดังนี้

$$M_{\mu\mu} \approx \sqrt{2\,p_{T,1}\,p_{T,2}\,\left[\cosh(\eta_1-\eta_2) - \cos(\phi_1-\phi_2)\right]}$$

ตามหลัก P1 เราเลือกปริมาณนี้เพราะฟิสิกส์บอกว่ามันเก็บข้อมูลของอนุภาคต้นกำเนิด ไม่ใช่เพราะเป็นคอลัมน์ที่มีอยู่ในไฟล์ ขั้นแรกเราคัดเลือกเฉพาะเหตุการณ์ที่มีมิวออนสองตัวซึ่งมีประจุตรงข้ามกัน ตามกฎอนุรักษ์ประจุ (P5) เพราะเมื่ออนุภาคที่เป็นกลางทางไฟฟ้าสลายตัวเป็นมิวออนสองตัว ประจุของมิวออนทั้งสองต้องตรงข้ามกัน

```python
sel = (ev["nMuon"] == 2)
sel = sel & (ak.sum(ev["Muon_charge"], axis=1) == 0)   # ประจุรวม = 0
two = ev[sel]
print(f"เหตุการณ์ที่ผ่านการคัดเลือก: {len(two):,} จาก {len(ev):,} ({100*len(two)/len(ev):.1f}%)")
```

```python
pt, eta, phi = two["Muon_pt"], two["Muon_eta"], two["Muon_phi"]
M = np.sqrt(2 * pt[:, 0] * pt[:, 1] * (np.cosh(eta[:, 0] - eta[:, 1]) - np.cos(phi[:, 0] - phi[:, 1])))
M = ak.to_numpy(M)
print(f"ตัวอย่างค่ามวลยืนยง 5 ค่าแรก [GeV/c²]: {np.round(M[:5], 2)}")
```

กราฟถัดไปเป็นฮิสโทแกรมของมวลยืนยง โดยใช้สเกลลอการิทึมทั้งสองแกน เนื่องจากมวลที่สนใจกระจายตั้งแต่ประมาณ 1 ถึง 100 GeV/$c^2$ และจำนวนเหตุการณ์ต่างกันหลายอันดับ เส้นประแนวตั้งแสดงมวลของอนุภาคที่รู้จักแล้ว

```python
bins = np.logspace(np.log10(0.3), np.log10(200), 300)
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.hist(M, bins=bins, histtype="step", color="k")
ax.set_xscale("log"); ax.set_yscale("log")
ax.set_xlabel(r"Dimuon invariant mass $M_{\mu\mu}$ [GeV/$c^2$]")
ax.set_ylabel("Events / bin")
for m, lab in [(0.782, r"$\omega$"), (1.019, r"$\phi$"), (3.097, r"J/$\psi$"),
               (3.686, r"$\psi$(2S)"), (9.46, r"$\Upsilon$"), (91.19, "Z")]:
    ax.axvline(m, ls="--", lw=0.8, color="tab:red")
    ax.text(m, ax.get_ylim()[1] * 0.5, lab, rotation=90, color="tab:red", ha="right", va="top")
ax.set_title(("CMS Open Data 2012" if CERN_REAL else "SYNTHETIC") + f" — first {len(ev):,} events")
plt.tight_layout(); plt.show()
```

### การตีความเชิงฟิสิกส์ (P4)
ยอดแหลม (peak) แต่ละยอดในกราฟคืออนุภาคที่มีอยู่จริง เช่น J/ψ ที่ประมาณ 3.1 GeV/$c^2$ และ Υ ที่ประมาณ 9.5 GeV/$c^2$ ส่วนพื้นหลังที่ลาดลงอย่างต่อเนื่องมาจากกระบวนการที่ไม่เกิดเรโซแนนซ์ (resonance) เช่น กระบวนการเดรล–ยาน (Drell–Yan) ซึ่งให้คู่มิวออนได้ทุกค่ามวล ภาพนี้คือจุดเริ่มต้นของการค้นหาอนุภาคใหม่ในฟิสิกส์อนุภาค ปัญหาแยกสัญญาณ (signal) ออกจากพื้นหลัง (background) จะกลับมาเป็นโจทย์การจำแนกประเภท (classification) ในบทที่ 6

### โครงสร้างของสัญญาณรบกวน: Amplitude Spectral Density (P2a)
เพื่อเข้าใจสัญญาณรบกวน เราคำนวณ amplitude spectral density (ASD) ซึ่งบอกว่าสัญญาณรบกวนมีขนาดเท่าใดในแต่ละความถี่ หากสัญญาณรบกวนเป็น white noise กราฟ ASD จะเป็นเส้นราบ

```python
fs = 1 / dt
freqs, psd = signal.welch(strain, fs=fs, nperseg=int(4 * fs))
asd = np.sqrt(psd)

fig, ax = plt.subplots(figsize=(9, 4))
ax.loglog(freqs, asd, lw=0.8)
ax.axvspan(35, 350, color="tab:green", alpha=0.12, label="35–350 Hz band")
ax.set_xlim(10, 2000); ax.set_xlabel("Frequency [Hz]"); ax.set_ylabel(r"ASD [strain/$\sqrt{Hz}$]")
ax.legend(); ax.set_title("Noise is strongly colored, not white")
plt.tight_layout(); plt.show()
```

> ถ้าใช้ไฟล์จำลอง กราฟ ASD จะเป็นเส้นราบ เพราะไฟล์จำลองใส่สัญญาณรบกวนแบบ white noise ไว้อย่างง่าย ความต่างนี้เองแสดงว่าข้อมูลจำลองขาดสมบัติสำคัญของข้อมูลจริง ซึ่งเป็นเหตุผลของหลัก P3

สำหรับข้อมูลจริง ASD สูงมากที่ความถี่ต่ำ (จากแรงสั่นสะเทือนของพื้นดิน) สูงขึ้นอีกที่ความถี่สูง (จาก shot noise ของเลเซอร์) และมียอดแหลมแคบหลายยอด (จากระบบไฟฟ้า 60 Hz และการสั่นของสายแขวนกระจก) สัญญาณรบกวนจึงเป็น **colored noise** ซึ่งละเมิดสมมติฐาน i.i.d. ที่วิธีทางสถิติจำนวนมากใช้ ตามที่ตำราอธิบายไว้ใน P2a

### ทำให้สัญญาณปรากฏ: whitening และ bandpass
เนื่องจากสัญญาณรบกวนมีขนาดแตกต่างกันมากในแต่ละความถี่ เราจึงหารสเปกตรัมของข้อมูลด้วย ASD เพื่อให้ทุกความถี่มีน้ำหนักเท่ากัน กระบวนการนี้เรียกว่าการทำให้ข้อมูลขาว (whitening) จากนั้นกรองให้เหลือเฉพาะช่วง 35–350 Hz ซึ่งเป็นช่วงที่ทั้งสัญญาณรบกวนต่ำและทฤษฎีสัมพัทธภาพทั่วไปทำนายว่าสัญญาณจากการควบรวมของหลุมดำมวลขนาดนี้อยู่ (P2b)

```python
def whiten(x, asd_f, asd_v, dt):
    n = len(x)
    win = signal.windows.tukey(n, alpha=0.1)
    xf = np.fft.rfft(x * win)
    f = np.fft.rfftfreq(n, dt)
    return np.fft.irfft(xf / np.interp(f, asd_f, asd_v), n)

white = whiten(strain, freqs, asd, dt)
b, a = signal.butter(4, [35, 350], btype="band", fs=fs)
white_bp = signal.filtfilt(b, a, white)

m = np.abs(time - T_EVENT) < 0.2
fig, ax = plt.subplots(figsize=(9, 3.5))
ax.plot(time[m] - T_EVENT, white_bp[m] / np.std(white_bp), lw=1)
ax.set_xlabel("Time since GPS 1126259462.4 [s]"); ax.set_ylabel("Whitened strain [σ]")
ax.set_title("After whitening + 35–350 Hz bandpass" + ("" if LIGO_REAL else " (SYNTHETIC)"))
plt.tight_layout(); plt.show()
```

ใกล้เวลา 0 วินาทีจะเห็นรูปคลื่นที่ความถี่และแอมพลิจูดเพิ่มขึ้นอย่างรวดเร็ว แล้วหายไป รูปแบบนี้เรียกว่า **chirp** ซึ่งเป็นลักษณะเฉพาะของหลุมดำสองดวงที่โคจรเข้าหากันเร็วขึ้นเรื่อย ๆ จนควบรวม กระบวนการที่ทำในส่วนนี้เป็นเพียงรูปแบบอย่างง่ายของขั้นตอนจริง การยืนยันการค้นพบของ LIGO ใช้ matched filtering กับแม่แบบรูปคลื่น (template) จากทฤษฎีสัมพัทธภาพทั่วไป และต้องพบสัญญาณในทั้งสองเครื่องตรวจวัด ซึ่งจะกล่าวถึงในบทที่ 4 และบทที่ 9

### แผนภาพเวลา–ความถี่ (spectrogram)
Spectrogram แสดงว่าพลังงานของสัญญาณอยู่ที่ความถี่ใดในแต่ละช่วงเวลา ให้สังเกตแถบที่โค้งขึ้นใกล้เวลา 0

```python
m2 = np.abs(time - T_EVENT) < 1.0
f_s, t_s, Sxx = signal.spectrogram(white[m2], fs=fs, nperseg=256, noverlap=240)
fig, ax = plt.subplots(figsize=(9, 3.5))
ax.pcolormesh(t_s - 1.0, f_s, Sxx, shading="auto", cmap="viridis")
ax.set_ylim(20, 400); ax.set_xlim(-0.5, 0.2)
ax.set_xlabel("Time since merger [s]"); ax.set_ylabel("Frequency [Hz]")
ax.set_title("Spectrogram of whitened strain (chirp)")
plt.tight_layout(); plt.show()
```

1. เปลี่ยนช่วง bandpass จาก `[35, 350]` เป็น `[10, 1000]` แล้วรันเซลล์ whitening ใหม่ สัญญาณยังเห็นชัดเท่าเดิมหรือไม่ เพราะเหตุใด
2. ข้ามขั้น whitening โดยแทน `white` ด้วย `strain` ในบรรทัด `filtfilt` แล้วเปรียบเทียบผล ข้อสังเกตนี้บอกอะไรเกี่ยวกับ P2a

### สเปกตรัม: อีกโมดาลิตีหนึ่งของวัตถุเดียวกัน
สเปกตรัมคือความเข้มของแสงเทียบกับความยาวคลื่น ใน HDU ที่ 1 ของไฟล์สเปกตรัม SDSS เก็บค่า `flux` และ `loglam` ซึ่งคือ $\log_{10}$ ของความยาวคลื่นในหน่วยอังสตรอม ส่วน HDU ที่ 2 เก็บค่าการเลื่อนไปทางแดง (redshift) $z$ เส้นประในกราฟแสดงตำแหน่งเส้นสเปกตรัมของไฮโดรเจนเมื่อเลื่อนไปด้วยปัจจัย $(1+z)$

```python
sp = spec_hdul[1].data
wl, flux = 10 ** sp["loglam"], sp["flux"]
z = float(spec_hdul[2].data["Z"][0]) if SDSS_REAL else 0.0   # ไฟล์จำลองเป็นดาวฤกษ์ z = 0

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(wl, flux, lw=0.6, color="k")
for L, lab in [(6563, r"H$\alpha$"), (4861, r"H$\beta$"), (4340, r"H$\gamma$")]:
    ax.axvline(L * (1 + z), ls="--", color="tab:red", lw=0.8)
    ax.text(L * (1 + z), np.nanpercentile(flux, 99.5), lab, color="tab:red", ha="right")
ax.set_xlabel(r"Observed wavelength [$\AA$]"); ax.set_ylabel(r"Flux [$10^{-17}$ erg/s/cm$^2$/$\AA$]" if SDSS_REAL else "Flux (arbitrary units)")
ax.set_title(f"SDSS spectrum, z = {z:.4f}" + ("" if SDSS_REAL else " (SYNTHETIC)"))
plt.tight_layout(); plt.show()
```

### ตรวจความสมเหตุสมผลเชิงฟิสิกส์ (P4 และ P5)
ก่อนนำข้อมูลไปใช้ ควรตรวจว่าค่าต่าง ๆ ไม่ขัดกับหลักฟิสิกส์ พลังงานการก่อตัว (formation energy) ของธาตุในสถานะอ้างอิงต้องเป็นศูนย์ตามนิยาม ช่องว่างแถบพลังงาน (band gap) ต้องไม่ติดลบ และวัสดุที่ `is_stable` เป็นจริงต้องมี `energy_above_hull` เท่ากับศูนย์ ฟังก์ชันถัดไปรวบรวมการตรวจเหล่านี้ไว้ และข้ามค่าที่ยังเป็น `None`

```python
def physics_check(entry):
    problems = []
    if entry.get("band_gap") is not None and entry["band_gap"] < 0:
        problems.append("band gap ติดลบ")
    if entry["is_stable"] != (entry["energy_above_hull"] == 0):
        problems.append("is_stable ไม่สอดคล้องกับ energy_above_hull")
    f = entry["formula_pretty"]
    is_element = f.isalpha() and sum(ch.isupper() for ch in f) == 1
    if is_element and abs(entry["formation_energy_per_atom"]) > 1e-6:
        problems.append("ธาตุบริสุทธิ์ต้องมี formation energy = 0")
    return problems or ["ผ่านการตรวจทุกข้อ"]

print(physics_check(si))
bad = dict(si, formation_energy_per_atom=-0.465, energy_above_hull=0.02)
print(physics_check(bad))
```

```python
ASK_MP_KEY = False  #@param {type:"boolean"}
# ติ๊กช่องนี้เมื่อมีคีย์ของ Materials Project แล้วต้องการดึงข้อมูลจริง (คีย์ไม่ถูกบันทึกในโน้ตบุ๊ก)
from getpass import getpass
MP_KEY = getpass("Materials Project API key: ") if ASK_MP_KEY else os.environ.get("MP_API_KEY", "")
if MP_KEY:
    if importlib.util.find_spec("mp_api") is None:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", "mp-api"], check=True)
    from mp_api.client import MPRester
    fields = ["material_id", "formula_pretty", "band_gap", "formation_energy_per_atom",
              "energy_above_hull", "density", "is_stable"]
    with MPRester(MP_KEY) as mpr:
        docs = mpr.materials.summary.search(material_ids=["mp-149"], fields=fields)
    real = docs[0].model_dump(include=set(fields))
    display(pd.Series(real, name="mp-149 (real)"))
    print(physics_check(real))
else:
    print("ข้ามการดึงข้อมูลจริง")
```

## 8.1 P1 (หลักคำถามทางฟิสิกส์นำ) และ P2b (สมมติฐานของแบบจำลอง): สปริงที่ถูกดึงเกินขีดจำกัดของ Hooke

ตำรายกตัวอย่างนิสิตที่ใช้การถดถอยเชิงเส้น (linear regression) กับข้อมูลสปริงเพียงเพราะเป็นเครื่องมือที่คุ้นเคย ในการจำลองนี้ สปริงมีค่าคงตัว $k = 50$ N/m และเป็นไปตามกฎของ Hooke เมื่อแรงไม่เกิน $F_{\text{lim}} = 10$ N หากแรงมากกว่านั้น สปริงจะยืดเพิ่มขึ้นแบบไม่เชิงเส้น

ให้เลื่อน `F_max` (แรงสูงสุดในการทดลอง) แล้วสังเกตกราฟส่วนตกค้าง (residual) ทางขวา ถ้าแบบจำลองเหมาะสม ส่วนตกค้างควรกระจายแบบสุ่มรอบศูนย์ ถ้าเห็นรูปแบบโค้งอย่างเป็นระบบ แสดงว่าสมมติฐานเชิงโครงสร้างของแบบจำลองผิด (P2b)

```python
def spring_demo(F_max=8.0):
    k, F_lim = 50.0, 10.0
    F = np.linspace(0.5, F_max, 30)
    x_true = F / k + np.where(F > F_lim, 0.002 * (F - F_lim) ** 2, 0.0)   # เกินขีดจำกัด: สปริงอ่อนตัว ยืดมากขึ้น
    x = x_true + np.random.default_rng(1).normal(0, 0.002, F.size)
    w, b = np.polyfit(F, x, 1)
    pred = w * F + b
    r2 = 1 - np.sum((x - pred) ** 2) / np.sum((x - x.mean()) ** 2)
    fig, ax = plt.subplots(1, 2, figsize=(10, 3.8))
    ax[0].plot(F, x, "o", ms=4, label="data"); ax[0].plot(F, pred, "-", label=f"linear fit, R²={r2:.3f}")
    ax[0].axvline(F_lim, ls=":", color="gray"); ax[0].set_xlabel("Force F [N]"); ax[0].set_ylabel("Extension x [m]"); ax[0].legend()
    ax[1].plot(F, x - pred, "o", ms=4); ax[1].axhline(0, color="k", lw=0.8)
    ax[1].set_xlabel("Force F [N]"); ax[1].set_ylabel("Residual [m]"); ax[1].set_title("Residuals")
    plt.tight_layout(); plt.show()
    print(f"k จากความชัน = 1/w = {1/w:.1f} N/m  (ค่าจริงในช่วง Hooke = 50 N/m)")

interact(spring_demo, F_max=FloatSlider(value=8, min=3, max=25, step=0.5, description="F_max [N]"));
```

เมื่อ `F_max` เกิน 10 N ค่า $R^2$ อาจยังสูง แต่ส่วนตกค้างแสดงรูปแบบโค้งชัดเจน และค่า $k$ ที่คำนวณได้คลาดจากค่าจริง บทเรียนคือ $R^2$ ที่ดูดีไม่ได้ยืนยันว่าแบบจำลองถูกต้อง ต้องถามก่อนเสมอว่าฟิสิกส์ของระบบบอกว่าความสัมพันธ์ควรมีรูปแบบใด (P1)

## 8.2 P2a (สมมติฐานของสัญญาณรบกวน) และ P3 (หลักข้อมูลฟิสิกส์จริง): สัญญาณรบกวนแบบปัวซองกับแบบเกาส์

เมื่อนับจำนวนโฟตอนหรือจำนวนเหตุการณ์ สัญญาณรบกวนเป็นไปตามการแจกแจงปัวซอง (Poisson distribution) ซึ่งมีค่าเฉลี่ยและความแปรปรวนเท่ากับ $\lambda$ หากประมาณด้วยการแจกแจงเกาส์ (Gaussian) ที่มีค่าเฉลี่ย $\lambda$ และส่วนเบี่ยงเบนมาตรฐาน $\sqrt{\lambda}$ จะใช้ได้ดีเฉพาะเมื่อ $\lambda$ มีค่ามาก ตามทฤษฎีบทขีดจำกัดกลาง

ให้เลื่อนค่า $\lambda$ จากน้อยไปมาก แล้วสังเกตว่าเมื่อ $\lambda$ น้อย การแจกแจงเกาส์ให้ความน่าจะเป็นแก่จำนวนนับที่ติดลบ ซึ่งเป็นไปไม่ได้ทางฟิสิกส์

```python
from scipy import stats

def poisson_vs_gauss(lam=3.0):
    kmax = int(lam + 5 * np.sqrt(lam) + 5)
    k = np.arange(0, kmax + 1)
    xg = np.linspace(min(-2.0, lam - 4 * np.sqrt(lam)), kmax, 400)
    fig, ax = plt.subplots(figsize=(8, 3.8))
    ax.bar(k, stats.poisson.pmf(k, lam), width=0.8, alpha=0.6, label="Poisson (physical)")
    ax.plot(xg, stats.norm.pdf(xg, lam, np.sqrt(lam)), "r-", label="Gaussian approx.")
    ax.axvspan(xg.min(), -0.5, color="gray", alpha=0.2, label="negative counts (impossible)")
    ax.set_xlabel("Photon count k"); ax.set_ylabel("Probability"); ax.legend()
    p_neg = stats.norm.cdf(-0.5, lam, np.sqrt(lam))
    ax.set_title(f"λ = {lam:.1f};  Gaussian assigns P(k<0) = {p_neg:.3f}")
    plt.tight_layout(); plt.show()

interact(poisson_vs_gauss, lam=FloatSlider(value=2, min=0.5, max=100, step=0.5, description="λ"));
```

## 8.3 P4 (หลักการตีความเชิงฟิสิกส์): ความชันของเส้นตรงมีหน่วยและความหมายทางฟิสิกส์

ตำราอธิบายว่า เมื่อปรับเส้นตรงให้เข้ากับ (fit) ความสัมพันธ์ $F = w\,a$ ค่าความชัน $w$ ต้องมีหน่วยกิโลกรัม และต้องเท่ากับมวลของวัตถุ เซลล์ถัดไปจำลองการทดลองกับวัตถุมวล 2.28 kg แล้ว fit พร้อมคำนวณความไม่แน่นอนของความชันจากเมทริกซ์ความแปรปรวนร่วม (covariance matrix)

```python
m_true = 2.28                                   # kg (ชั่งได้)
a = np.linspace(0.5, 5, 12)                     # m/s^2
F = m_true * a + rng.normal(0, 0.15, a.size)    # N, ความคลาดเคลื่อนของเครื่องวัดแรง 0.15 N

(w, b), cov = np.polyfit(a, F, 1, cov=True)
sw = np.sqrt(cov[0, 0])
print(f"w = {w:.2f} ± {sw:.2f} kg     (มวลที่ชั่งได้ = {m_true} kg)")
print(f"ต่างกัน {abs(w - m_true) / sw:.1f} σ  ->  {'สอดคล้อง' if abs(w - m_true) < 2 * sw else 'ต้องตรวจสอบ'}")
print(f"b = {b:.2f} N  (ฟิสิกส์คาดว่า ≈ 0: ถ้าไม่ใกล้ศูนย์ อาจมีแรงเสียดทานหรือ offset ของเครื่องวัด)")
```

สังเกตว่าเรารายงาน $w$ พร้อมความไม่แน่นอน และใช้ความไม่แน่นอนนั้นตัดสินว่าผลสอดคล้องกับค่าที่ชั่งได้หรือไม่ นอกจากนี้ จุดตัดแกน $b$ ก็มีความหมายทางฟิสิกส์เช่นกัน แบบจำลอง ML ที่ซับซ้อนขึ้นในบทหลัง ๆ ต้องผ่านการตีความในลักษณะเดียวกันนี้

## 8.4 P5 (หลักความไม่แน่นอนและกฎอนุรักษ์): ความไม่แน่นอนเชิงสถิติกับเชิงระบบ

จำลองการวัดความเร่งโน้มถ่วง $g$ ซ้ำ $N$ ครั้ง แต่ละครั้งมีความคลาดเคลื่อนสุ่ม $\sigma = 0.20$ m/s² และเครื่องมือมีความเอนเอียงเชิงระบบ (bias) คงที่ค่าหนึ่ง ให้เลื่อนค่า bias แล้วดูว่าเมื่อเพิ่ม $N$ ความคลาดเคลื่อนรวมลดลงได้ถึงระดับใด

```python
def stat_vs_sys(bias=0.05):
    g_true, sigma = 9.80, 0.20
    Ns = np.unique(np.logspace(0, 5, 40).astype(int))
    r = np.random.default_rng(7)
    err = [abs(np.mean(g_true + bias + r.normal(0, sigma, n)) - g_true) for n in Ns]
    fig, ax = plt.subplots(figsize=(8, 4))
    ax.loglog(Ns, err, "o", ms=3, label="|mean − g_true| (one experiment)")
    ax.loglog(Ns, sigma / np.sqrt(Ns), "--", label=r"$\sigma_{stat}=\sigma/\sqrt{N}$")
    if bias > 0:
        ax.axhline(bias, color="tab:red", label=r"$\sigma_{sys}$ = bias")
    ax.set_xlabel("Number of measurements N"); ax.set_ylabel(r"Error [m/s$^2$]"); ax.legend()
    plt.tight_layout(); plt.show()
    N = 10_000
    print(f"ที่ N = {N:,}:  g = {g_true + bias:.3f} ± {sigma/np.sqrt(N):.3f} (stat) ± {bias:.3f} (sys) m/s²")

interact(stat_vs_sys, bias=FloatSlider(value=0.05, min=0, max=0.2, step=0.01, description="bias"));
```

เส้นประลดลงตาม $1/\sqrt{N}$ แต่เส้นสีแดงคงที่ เมื่อ $N$ มากพอ ความคลาดเคลื่อนจะหยุดลดลงที่ระดับของ bias การเก็บข้อมูลเพิ่มจึงแก้ความไม่แน่นอนเชิงระบบไม่ได้ ต้องปรับปรุงการสอบเทียบเครื่องมือ ผลนี้ตรงกับตารางเปรียบเทียบความไม่แน่นอนสองประเภทในตำรา และบรรทัดสุดท้ายแสดงรูปแบบการรายงานผลที่แยกความไม่แน่นอนทั้งสองประเภท

## 8.5 P5: กฎอนุรักษ์เป็นเครื่องตรวจแบบจำลอง

สมมติว่ามี "แบบจำลอง ML" ที่ทำนายพลังงานจลน์ของอนุภาคสองตัวหลังการชนแบบยืดหยุ่น (elastic collision) ในที่นี้จำลองแบบจำลองที่ให้ค่าทำนายใกล้เคียงค่าจริงเกือบทุกกรณี แต่มีบางกรณีที่ผิดพลาดมาก เราตรวจแต่ละกรณีด้วยกฎอนุรักษ์พลังงาน ซึ่งกำหนดให้พลังงานจลน์รวมก่อนและหลังการชนต้องเท่ากัน

```python
n = 200
KE_before = rng.uniform(1, 10, n)                               # J
KE_after_pred = KE_before * (1 + rng.normal(0, 0.01, n))        # แบบจำลองที่ดีส่วนใหญ่
bad_idx = rng.choice(n, 6, replace=False)
KE_after_pred[bad_idx] *= rng.uniform(1.10, 1.25, 6)            # บางกรณีสร้างพลังงานขึ้นเอง

rel = (KE_after_pred - KE_before) / KE_before
mae = np.mean(np.abs(KE_after_pred - KE_before))
violations = np.where(np.abs(rel) > 0.05)[0]
print(f"ค่าคลาดเคลื่อนสัมบูรณ์เฉลี่ย = {mae:.3f} J  (ดูดีเมื่อมองค่าเฉลี่ยอย่างเดียว)")
print(f"กรณีที่ละเมิดการอนุรักษ์พลังงานเกิน 5%: {len(violations)} กรณี -> index {violations.tolist()}")

plt.figure(figsize=(8, 3.5))
plt.plot(rel * 100, "o", ms=3); plt.axhspan(-5, 5, color="tab:green", alpha=0.15, label="±5% tolerance")
plt.xlabel("Test case"); plt.ylabel(r"$\Delta KE / KE$ [%]"); plt.legend(); plt.tight_layout(); plt.show()
```

ค่าคลาดเคลื่อนเฉลี่ยต่ำทำให้แบบจำลองดูน่าเชื่อถือ แต่การตรวจรายกรณีด้วยกฎอนุรักษ์พบกรณีที่แบบจำลองสร้างพลังงานขึ้นจากความว่างเปล่า ตำราชี้ว่าการตัดกรณีเหล่านี้ทิ้งแล้วรายงานเฉพาะผลที่ "สมเหตุสมผล" เป็นแนวปฏิบัติที่ผิด (anti-pattern) ทางที่ถูกคือหาสาเหตุ หรือฝังกฎอนุรักษ์ไว้ในโครงสร้างของแบบจำลอง ซึ่งจะเรียนในบทที่ 5 และบทที่ 10
