## วัตถุประสงค์
เมื่อทำโน้ตบุ๊กนี้จนจบ ผู้เรียนควรสามารถ

1. ระบุแบบจำลองสัญญาณรบกวน (noise model) ที่เหมาะกับการวัดแต่ละชนิดจากกลไกทางฟิสิกส์ของการวัด
2. ใช้สถิติพรรณนาและภาพสำรวจข้อมูลอย่างมีจุดมุ่งหมาย และปรับการแจกแจงให้เข้ากับข้อมูลพร้อมตรวจความเหมาะสมทั้งด้วยภาพและด้วยการทดสอบ
3. อธิบายความหมายที่ถูกต้องของค่าพี (p-value) และช่วงความเชื่อมั่น (confidence interval) และแยกนัยสำคัญทางสถิติออกจากนัยสำคัญทางฟิสิกส์
4. เขียนและตรวจการจำลองมอนติคาร์โลด้วยขั้นตอนวิธีเมโทรโพลิสและวิธีหวัง–แลนเดาสำหรับแบบจำลองไอซิงสองมิติ และตีความผลในเชิงการเปลี่ยนสถานะ
5. รายงานผลการวัดพร้อมความไม่แน่นอนด้วยจำนวนเลขนัยสำคัญที่ถูกต้อง และใช้การแพร่ของความคลาดเคลื่อนออกแบบการทดลอง

### การล็อกแนวคำตอบ
เซลล์ ✅ และเซลล์แนวคำตอบในโน้ตบุ๊กนี้แสดงคำตอบอ้างอิงเมื่อผู้เรียนเติมคำตอบในเซลล์ ✏️ ของข้อนั้นครบแล้วเท่านั้น ถ้ายังไม่ได้ตอบ เซลล์จะแสดงเพียงข้อความ 🔒 บอกว่ายังขาดคำตอบใด การคำนวณในเซลล์ยังทำงานตามปกติ ดังนั้น Run all จึงรันได้ครบโดยไม่เปิดเผยคำตอบ แนวคำตอบของงานเขียนและงานนอกชั้นซึ่งตรวจด้วยโค้ดไม่ได้จะเปิดเมื่อผู้สอนเลือกเท่านั้น

**สำหรับผู้สอน:** ติ๊กช่อง `SHOW_ANSWERS` ในเซลล์ถัดไป (หรือแก้เป็น `True`) แล้วรันเซลล์นั้นอีกครั้ง เพื่อแสดงแนวคำตอบทุกข้อ เช่น เมื่อเฉลยในชั้นหรือเตรียมสอน

```python
#@title ⚙️ ตัวเลือกของผู้สอน: แสดงแนวคำตอบ
SHOW_ANSWERS = False  #@param {type:"boolean"}

def _answered(v):
    """True when an exercise variable holds an answer (not None, "?", "", empty or containing blanks)."""
    if v is None or v is Ellipsis:
        return False
    if isinstance(v, str):                       # placeholders such as "?", "?-?", "..." count as blank
        return v.replace("?", "").replace("-", "").replace(".", "").strip() != ""
    if isinstance(v, dict):
        return len(v) > 0 and all(_answered(x) for x in v.values())
    if isinstance(v, (list, tuple, set, frozenset)):
        return len(v) > 0 and all(_answered(x) for x in v)
    if type(v).__name__ == "ndarray":
        return v.size > 0
    return True

def _gate(names=()):
    """Return print if the answers may be shown, otherwise print one lock message and return a no-op."""
    if SHOW_ANSWERS:
        return print
    if names:
        missing = [n for n in names if not _answered(globals().get(n))]
        if not missing:
            return print
        print("🔒 ยังไม่ได้ตอบ:", ", ".join(missing),
              "— เติมคำตอบในเซลล์ ✏️ ด้านบนให้ครบก่อน แล้วรันเซลล์นี้อีกครั้ง")
    else:
        print("🔒 แนวคำตอบข้อนี้ผู้สอนเป็นผู้เปิด (ตั้ง SHOW_ANSWERS = True ในส่วนที่ 0)")
    return lambda *args, **kwargs: None

print("แสดงแนวคำตอบทั้งหมด:", "ใช่ (โหมดผู้สอน)" if SHOW_ANSWERS else "ไม่ (แสดงเมื่อผู้เรียนตอบข้อนั้นครบแล้ว)")
```

```python
#@title สร้างข้อมูลจำลองของบทที่ 3 และ 4 (ดับเบิลคลิกเพื่อดูโค้ด)
# โค้ดชุดเดียวกับ ch03_cleaning/01_make_raw_data.py และ ch04_statistics/01_make_data.py
SRC3 = r'''"""Ch.3  Create the SYNTHETIC raw data used by every script of this chapter.

The book refers to files such as data/raw/galaxy10_metadata.csv and data/raw/afm_spectrum.csv.
They are teaching files built here with a fixed seed; the numbers are NOT measurements.
  * galaxy table: Galaxy10 DECaLS class names, but redshift and magnitudes are simulated.
  * AFM files: simulated curves (the author's own AFM measurements are not published here).
  * 4-lepton events: simulated H -> ZZ* -> 4l signal plus a smooth background.

Output (data/raw/)
  galaxy10_metadata.csv      2,000 galaxies: image_id, galaxy_class, redshift, r_mag, g_mag
  galaxy10_image_index.csv   image_id -> image file (some ids missing, some extra)
  galaxy10_truth.csv         the true redshift of every galaxy (answer key, for E3.4 only)
  galaxy10_messy.csv         ~500 rows with NaN, -999 sentinels and outliers (for E3.2)
  galaxy_images.npy          4 synthetic images, shape (4, 256, 256, 3), 2 hot pixels in image 0
  afm_timeseries.csv         force signal while scanning: drift + spikes + noise
  afm_spectrum.csv           one force-distance curve (approach + retract) with baseline shift
  afm_curves.csv             15 force-distance curves with different stiffness/adhesion (E3.5)
  afm_curves_truth.csv       the parameters used to make them (answer key)
  higgs_4l_events.csv        E, px, py, pz of four leptons per event [GeV]
"""
from pathlib import Path

import numpy as np
import pandas as pd

SEED = 2026
rng = np.random.default_rng(SEED)
RAW = Path(__file__).resolve().parent / "data" / "raw"
RAW.mkdir(parents=True, exist_ok=True)

CLASSES = ["Disturbed", "Merging", "Round Smooth", "In-between Round Smooth",
           "Cigar Shaped Smooth", "Barred Spiral", "Unbarred Tight Spiral",
           "Unbarred Loose Spiral", "Edge-on without Bulge", "Edge-on with Bulge"]


# --------------------------------------------------------------- galaxies
def make_galaxies(n=2000):
    cls = rng.choice(len(CLASSES), n, p=[.06, .10, .15, .15, .06, .11, .10, .15, .06, .06])
    z_true = np.clip(rng.gamma(3.0, 0.035, n), 0.005, 0.40)
    # toy magnitude-redshift relation (inverse of photo_z_estimate in the book) + scatter
    r_mag = 15.0 + z_true / 0.05 + rng.normal(0, 0.6, n)
    color = np.array([0.7, 0.7, 0.9, 0.85, 0.8, 0.75, 0.6, 0.55, 0.65, 0.8])[cls]
    g_mag = r_mag + color + rng.normal(0, 0.1, n)
    # redshift missing more often for faint galaxies (MAR on r_mag), ~15 % overall
    p_miss = 1 / (1 + np.exp(-(r_mag - 18.9) / 0.6))
    missing = rng.random(n) < p_miss
    df = pd.DataFrame({
        "image_id": [f"galaxy_{i:05d}" for i in range(n)],
        "galaxy_class": [CLASSES[c] for c in cls],
        "redshift": np.where(missing, np.nan, np.round(z_true, 4)),
        "r_mag": np.round(r_mag, 3),
        "g_mag": np.round(g_mag, 3).astype(object),
    })
    # five bright foreground stars that slipped into the catalogue (r < 8)
    stars = rng.choice(n, 5, replace=False)
    df.loc[stars, "r_mag"] = np.round(rng.uniform(6.0, 7.8, 5), 3)
    df.loc[stars, "redshift"] = 0.0
    # three g_mag values typed as text in the source file -> the column is read as text
    bad = rng.choice(np.setdiff1d(np.arange(n), stars), 3, replace=False)
    df.loc[bad, "g_mag"] = "--"
    df.to_csv(RAW / "galaxy10_metadata.csv", index=False)
    pd.DataFrame({"image_id": df["image_id"], "z_true": np.round(z_true, 4),
                  "is_star": np.isin(np.arange(n), stars)}).to_csv(RAW / "galaxy10_truth.csv",
                                                                   index=False)

    # image index: 20 galaxies have no image file, 15 image files have no metadata row
    keep = np.sort(rng.choice(n, n - 20, replace=False))
    ids = [f"galaxy_{i:05d}" for i in keep] + [f"galaxy_{i:05d}" for i in range(n, n + 15)]
    pd.DataFrame({"image_id": ids,
                  "image_file": [f"images/{i}.png" for i in ids]}).to_csv(
        RAW / "galaxy10_image_index.csv", index=False)
    return df, z_true


def make_messy(df, n=500):
    m = df.sample(n, random_state=SEED).reset_index(drop=True).copy()
    m["g_mag"] = pd.to_numeric(m["g_mag"], errors="coerce")
    idx = rng.permutation(n)
    m.loc[idx[:20], "redshift"] = -999          # sentinel: "could not be measured"
    m.loc[idx[20:25], "r_mag"] = -999
    m.loc[idx[25:31], "r_mag"] = np.round(rng.uniform(25.5, 28.0, 6), 3)   # too faint
    m.loc[idx[31:33], "redshift"] = np.round(-rng.uniform(0.01, 0.05, 2), 4)  # sign typo
    m.to_csv(RAW / "galaxy10_messy.csv", index=False)


def make_images(n=4, size=256):
    y, x = np.mgrid[0:size, 0:size]
    imgs = np.zeros((n, size, size, 3), dtype=np.float32)
    for k in range(n):
        x0, y0 = 128 + rng.normal(0, 3, 2)
        q, pa = rng.uniform(0.5, 1.0), rng.uniform(0, np.pi)
        xr = (x - x0) * np.cos(pa) + (y - y0) * np.sin(pa)
        yr = -(x - x0) * np.sin(pa) + (y - y0) * np.cos(pa)
        r = np.hypot(xr, yr / q)
        for b, amp in enumerate((0.6, 1.0, 1.3)):                  # g, r, z bands
            bulge = amp * 60 * np.exp(-7.67 * ((r / 6) ** 0.25 - 1)) / np.exp(7.67)
            disk = amp * 8 * np.exp(-r / 25)
            imgs[k, :, :, b] = bulge + disk + rng.normal(0, 0.5, (size, size))
    imgs[0, 40, 200, 0] = 60.0        # two hot pixels in image 0, band 0
    imgs[0, 210, 35, 0] = 45.0
    np.save(RAW / "galaxy_images.npy", imgs)


# --------------------------------------------------------------- AFM
def afm_curve(k_sample, adhesion_nN, z_contact=20.0, snap=15.0, offset=0.8, noise=0.05,
              n=301):
    """One approach + retract force-distance curve [nN vs nm].
    distance_nm is the piezo height above the contact point + z_contact."""
    z = np.linspace(300.0, 0.0, n)
    appr = np.where(z < z_contact, k_sample * (z_contact - z), 0.0)
    z_r = z[::-1]
    retr = np.where(z_r < z_contact, k_sample * (z_contact - z_r), 0.0)
    # adhesion: the tip stays stuck and is pulled down to -adhesion before snapping off
    stick = (z_r >= z_contact) & (z_r <= z_contact + snap)
    retr = np.where(stick, -adhesion_nN * (z_r - z_contact) / snap, retr)
    d = np.concatenate([z, z_r])
    f = np.concatenate([appr, retr]) + offset + rng.normal(0, noise, 2 * n)
    seg = ["approach"] * n + ["retract"] * n
    return pd.DataFrame({"segment": seg, "distance_nm": np.round(d, 3),
                         "force_nN": np.round(f, 4)})


def make_afm():
    # time series: 2 s at 1 kHz while the tip scans a surface
    t = np.arange(0, 2.0, 0.001)
    topo = 0.3 * np.sin(2 * np.pi * 1.5 * t)
    drift = 0.4 * t                                  # slow thermal drift [nN/s]
    f = 2.0 + topo + drift + rng.normal(0, 0.03, t.size)
    spikes = rng.choice(t.size, 15, replace=False)
    f[spikes] += rng.choice([-1, 1], 15) * rng.uniform(1.0, 2.5, 15)
    pd.DataFrame({"time_s": np.round(t, 4), "force_nN": np.round(f, 4)}).to_csv(
        RAW / "afm_timeseries.csv", index=False)
    pd.DataFrame({"spike_index": np.sort(spikes)}).to_csv(RAW / "afm_timeseries_spikes.csv",
                                                           index=False)

    afm_curve(k_sample=0.25, adhesion_nN=3.0).to_csv(RAW / "afm_spectrum.csv", index=False)

    rows, truth = [], []
    for cid in range(15):
        k_s = rng.uniform(0.1, 0.5)                  # nN/nm
        adh = rng.uniform(1.0, 5.0)                  # nN
        off = rng.normal(0.5, 0.3)
        c = afm_curve(k_s, adh, offset=off)
        c.insert(0, "curve_id", cid)
        rows.append(c)
        truth.append({"curve_id": cid, "slope_nN_per_nm": k_s, "adhesion_nN": adh,
                      "work_aJ": 0.5 * adh * 15.0, "baseline_nN": off})
    pd.concat(rows).to_csv(RAW / "afm_curves.csv", index=False)
    pd.DataFrame(truth).round(4).to_csv(RAW / "afm_curves_truth.csv", index=False)


# --------------------------------------------------------------- 4-lepton events
def two_body(M, m1, m2, n):
    """Momenta of two daughters in the parent rest frame, isotropic."""
    p = np.sqrt((M**2 - (m1 + m2)**2) * (M**2 - (m1 - m2)**2)) / (2 * M)
    cth = rng.uniform(-1, 1, n); phi = rng.uniform(0, 2 * np.pi, n)
    sth = np.sqrt(1 - cth**2)
    v = np.stack([p * sth * np.cos(phi), p * sth * np.sin(phi), p * cth], axis=1)
    E1 = np.sqrt(p**2 + m1**2); E2 = np.sqrt(p**2 + m2**2)
    return np.column_stack([E1, v]), np.column_stack([E2, -v])


def boost(P, beta):
    """Lorentz boost of four-vectors P (n,4) by velocity vectors beta (n,3)."""
    b2 = np.sum(beta**2, axis=1)
    g = 1 / np.sqrt(1 - b2)
    bp = np.sum(beta * P[:, 1:], axis=1)
    E = g * (P[:, 0] + bp)
    coef = np.where(b2 > 0, (g - 1) * bp / np.where(b2 > 0, b2, 1), 0.0) + g * P[:, 0]
    return np.column_stack([E, P[:, 1:] + coef[:, None] * beta])


def make_4l(n_sig=400, n_bkg=1600):
    M = np.concatenate([rng.normal(125.0, 1.8, n_sig),              # Higgs + resolution
                        70 + rng.gamma(2.5, 45.0, n_bkg)])            # smooth background
    n = M.size
    mZ1 = np.minimum(rng.normal(91.19, 2.5, n), M - 13)
    mZ2 = np.clip(rng.uniform(12, 60, n), 12, M - mZ1 - 1)
    Z1, Z2 = two_body(M, mZ1, mZ2, n)
    leps = []
    for Z, mZ in ((Z1, mZ1), (Z2, mZ2)):
        l1, l2 = two_body(mZ, 0.0, 0.0, n)
        beta = Z[:, 1:] / Z[:, [0]]
        leps += [boost(l1, beta), boost(l2, beta)]
    # boost the whole system along the beam (z) and give it a little transverse kick
    beta_lab = np.column_stack([rng.normal(0, 0.05, n), rng.normal(0, 0.05, n),
                                rng.uniform(-0.7, 0.7, n)])
    leps = [boost(l, beta_lab) for l in leps]
    order = rng.permutation(n)
    cols = {}
    for i, l in enumerate(leps, start=1):
        for j, c in enumerate(("E", "px", "py", "pz")):
            cols[f"{c}{i}"] = np.round(l[order, j], 4)
    ev = pd.DataFrame(cols)
    ev.insert(0, "is_signal", (np.arange(n) < n_sig)[order])
    ev.to_csv(RAW / "higgs_4l_events.csv", index=False)


if __name__ == "__main__":
    df, _ = make_galaxies()
    make_messy(df)
    make_images()
    make_afm()
    make_4l()
    for p in sorted(RAW.iterdir()):
        print(f"{p.name:30s} {p.stat().st_size / 1024:8.1f} kB")
'''
SRC4 = r'''"""Ch.4  Build the data files used in this chapter (all SYNTHETIC, SEED = 2026).

  data/raw/co60_counts.csv            1,000 one-second counts, Poisson with mean 5
                                      (exactly the recipe of exercise E4.2)
  data/raw/halpha_spectrum.csv        an H-alpha line: Gaussian, sigma = 0.020 nm
                                      (Doppler width at T = 1e4 K) + baseline + noise
  data/processed/galaxy10_features.csv
                                      the Chapter 3 galaxy table (stars removed, g_mag
                                      numeric) plus a light 'concentration' feature C
                                      (tab:invariant), drawn per class: smooth galaxies
                                      C ~ 4, spirals C ~ 2.8
Chapter 3 must have been run once (its 01_make_raw_data.py); if not, it is run here.
"""
import os
import subprocess
import sys
from pathlib import Path

import numpy as np
import pandas as pd
import scipy.constants as const

SEED = 2026
HERE = Path(__file__).resolve().parent
os.chdir(HERE)
rng = np.random.default_rng(SEED)
Path('data/raw').mkdir(parents=True, exist_ok=True)
Path('data/processed').mkdir(parents=True, exist_ok=True)

# --- Co-60 counts: exactly the recipe of E4.2 ---------------------------------
counts = rng.poisson(lam=5, size=1000)
pd.DataFrame({'interval': np.arange(1000), 'counts': counts}).to_csv(
    'data/raw/co60_counts.csv', index=False)

# --- H-alpha line with Doppler width at 1e4 K ---------------------------------
lam0 = 656.28                                    # nm
m_H = const.m_p + const.m_e
sigma_true = lam0 * np.sqrt(const.k * 1e4 / (m_H * const.c**2))
wavelength = np.linspace(lam0 - 0.25, lam0 + 0.25, 201)
intensity = (100 * np.exp(-(wavelength - lam0)**2 / (2 * sigma_true**2)) + 10
             + rng.normal(0, 2.0, wavelength.size))
pd.DataFrame({'wavelength_nm': np.round(wavelength, 5),
              'intensity': np.round(intensity, 3)}).to_csv('data/raw/halpha_spectrum.csv',
                                                           index=False)

# --- Galaxy10-style features ----------------------------------------------------
ch3 = HERE.parent / 'ch03_cleaning'
meta = ch3 / 'data' / 'raw' / 'galaxy10_metadata.csv'
if not meta.exists():
    subprocess.run([sys.executable, str(ch3 / '01_make_raw_data.py')], check=True)
df = pd.read_csv(meta)
df['g_mag'] = pd.to_numeric(df['g_mag'], errors='coerce')
df = df[df['r_mag'] >= 8].copy()                 # the five foreground stars (Ch.3)
C_mean = {'Round Smooth': 4.1, 'In-between Round Smooth': 3.8, 'Cigar Shaped Smooth': 3.6,
          'Edge-on with Bulge': 3.3, 'Edge-on without Bulge': 2.8, 'Barred Spiral': 3.0,
          'Unbarred Tight Spiral': 2.9, 'Unbarred Loose Spiral': 2.7,
          'Disturbed': 2.8, 'Merging': 2.9}
df['concentration'] = np.round(df['galaxy_class'].map(C_mean) + rng.normal(0, 0.3, len(df)), 3)
df.to_csv('data/processed/galaxy10_features.csv', index=False)

print(f"co60_counts.csv        : {counts.size} counts, mean {counts.mean():.3f}")
print(f"halpha_spectrum.csv    : sigma used {sigma_true:.4f} nm (T = 1e4 K)")
print(f"galaxy10_features.csv  : {len(df)} galaxies, columns {list(df.columns)}")
'''
HERE4 = os.getcwd()
ch3 = os.path.abspath(os.path.join(HERE4, "..", "ch03_cleaning"))
if not os.path.exists(os.path.join(ch3, "data", "raw", "galaxy10_metadata.csv")):
    os.makedirs(ch3, exist_ok=True)
    exec(compile(SRC3, "make_raw_data.py", "exec"),
         {"__name__": "__main__", "__file__": os.path.join(ch3, "make_raw_data.py")})
if not os.path.exists("data/processed/galaxy10_features.csv"):
    exec(compile(SRC4, "make_data.py", "exec"),
         {"__name__": "__main__", "__file__": os.path.join(HERE4, "make_data.py")})
    os.chdir(HERE4)
else:
    print("มีข้อมูลจำลองของบทที่ 4 อยู่แล้ว ไม่สร้างใหม่")
print("data/raw:", sorted(os.listdir("data/raw")), "| data/processed:", sorted(os.listdir("data/processed")))
```

# ส่วนที่ 1 ทฤษฎีบทการแกว่งและการกระจาย และแบบจำลองสัญญาณรบกวน (หัวข้อ 4.1.1–4.1.3)

### แนวคิดสำคัญ
อุณหภูมิเป็นพารามิเตอร์ของการแจกแจงความเร็วของโมเลกุล การวัดทุกครั้งจึงเป็นการได้ตัวอย่างหนึ่งจากการแจกแจงของค่าที่เป็นไปได้ งานของนักฟิสิกส์คือการอนุมานย้อนกลับจากตัวอย่างไปหาพารามิเตอร์ของการแจกแจง

ทฤษฎีบทการแกว่งและการกระจาย (fluctuation–dissipation theorem) บอกว่าระบบที่กระจายพลังงานได้ต้องมีการแกว่งเชิงความร้อนที่สอดคล้องกับการกระจายนั้น เหมือนฝูงชนที่ทั้งต้านคนที่พยายามเดินฝ่าและผลักคนที่ยืนนิ่งไปมา ผลที่ตามมาสองตัวอย่างคือ

- **ความสัมพันธ์ของไอน์สไตน์ (Einstein relation):** $D = k_BT/\zeta$ และสำหรับทรงกลมรัศมี $a$ ในของไหลความหนืด $\eta$ มี $\zeta = 6\pi\eta a$
- **สัญญาณรบกวนจอห์นสัน–ไนควิสต์ (Johnson–Nyquist noise):** $\langle V^2\rangle = 4k_BTR\,\Delta f$

เซลล์ถัดไปคำนวณขนาดของทั้งสองปริมาณในสถานการณ์ที่นิสิตพบได้ในห้องปฏิบัติการ

```python
k_B = const.k
# Johnson-Nyquist noise of a 100 kOhm resistor at 300 K, bandwidth 10 kHz
R, T, df_ = 100e3, 300.0, 10e3
V_rms = np.sqrt(4 * k_B * T * R * df_)
print(f"Johnson noise: V_rms = {V_rms * 1e6:.2f} microvolt  (R = 100 kOhm, T = 300 K, bandwidth 10 kHz)")
print(f"  cooled to 77 K (liquid nitrogen): {np.sqrt(4 * k_B * 77 * R * df_) * 1e6:.2f} microvolt")

# Einstein relation for a 1 micrometre sphere in water at 293 K
eta, a = 1.0e-3, 0.5e-6                       # Pa s, radius [m]
D = k_B * 293 / (6 * np.pi * eta * a)
print(f"\nEinstein: D = {D:.2e} m^2/s -> rms displacement in 1 s along one axis = {np.sqrt(2 * D * 1) * 1e6:.2f} micrometre")
```

แรงดันรบกวนระดับไมโครโวลต์ดูเล็ก แต่เทียบได้กับสัญญาณจากเซนเซอร์หลายชนิด และลดได้ด้วยการลดอุณหภูมิหรือแบนด์วิดท์เท่านั้น ไม่ใช่ด้วยการซื้อเครื่องมือที่ดีกว่า การเคลื่อนที่แบบบราวน์ของอนุภาคขนาดไมโครเมตรในหนึ่งวินาทีก็มีขนาดใกล้เคียงกับตัวอนุภาคเอง ซึ่งเห็นได้ด้วยกล้องจุลทรรศน์ธรรมดา

### แบบจำลองสัญญาณรบกวนสามแบบ
| แบบจำลอง | กลไกทางฟิสิกส์ | สมบัติ |
|---|---|---|
| เกาส์ (Gaussian) | ผลรวมของการรบกวนเล็ก ๆ อิสระจำนวนมาก | ค่าต่อเนื่อง $\mu$ และ $\sigma^2$ อิสระต่อกัน |
| ปัวซง (Poisson) | การนับเหตุการณ์อิสระที่อัตราคงที่ | ค่าเฉลี่ยเท่ากับความแปรปรวน ค่าเป็นจำนวนเต็ม |
| สี ($1/f$) | ความสัมพันธ์ระยะยาวตามเวลา | กำลังขึ้นกับความถี่ ค่าที่ใกล้กันในเวลาสัมพันธ์กัน |

เซลล์ถัดไปสร้างสัญญาณรบกวนทั้งสามแบบ แล้ววาดทั้งฮิสโทแกรม (การแจกแจงของค่าแต่ละค่า) และสเปกตรัมกำลัง (ความสัมพันธ์ตามเวลา) ให้สังเกตว่าสองภาพนี้ตอบคำถามต่างกัน

```python
from scipy.signal import welch
rng_n = np.random.default_rng(41)
n = 2**14
white = rng_n.normal(0, 1, n)                                  # Gaussian, white
counts_demo = rng_n.poisson(3, n).astype(float)                # Poisson counts, mean 3
spec = np.fft.rfft(rng_n.normal(0, 1, n))
f_ = np.fft.rfftfreq(n); f_[0] = f_[1]
pink = np.fft.irfft(spec / np.sqrt(f_), n); pink = (pink - pink.mean()) / pink.std()   # 1/f, still Gaussian values

fig, ax = plt.subplots(2, 3, figsize=(12, 6))
for k, (name, x) in enumerate((("Gaussian white", white), ("Poisson counts (mean 3)", counts_demo),
                               ("1/f (pink)", pink))):
    bins = np.arange(-0.5, 13.5) if "Poisson" in name else 50
    ax[0, k].hist(x, bins=bins, density=True, alpha=0.7); ax[0, k].set_title(name)
    fr, P = welch(x - x.mean(), nperseg=1024)
    ax[1, k].loglog(fr[1:], P[1:]); ax[1, k].set_xlabel("frequency (1/sample)")
ax[0, 0].set_ylabel("histogram"); ax[1, 0].set_ylabel("power spectrum")
fig.suptitle("three noise models (SYNTHETIC)"); plt.tight_layout(); plt.show()
print(f"Poisson counts: mean {counts_demo.mean():.3f}, variance {counts_demo.var(ddof=1):.3f}")
print(f"1/f noise: skewness {stats.skew(pink):+.3f}, excess kurtosis {stats.kurtosis(pink):+.3f} (both ~0 -> Gaussian values)")
```

สัญญาณรบกวน $1/f$ มีฮิสโทแกรมเป็นรูประฆังเหมือนแบบเกาส์ แต่สเปกตรัมกำลังลดลงตามความถี่ แสดงว่าคำว่า "สี" กับคำว่า "เกาส์" ตอบคำถามคนละข้อ ความเป็นสีดูจากสเปกตรัมกำลัง ส่วนความเป็นเกาส์ดูจากฮิสโทแกรม การตรวจสัญญาณรบกวนจึงต้องดูทั้งสองด้าน ส่วนจำนวนนับแบบปัวซงเป็นค่าจำนวนเต็ม มีความแปรปรวนเท่ากับค่าเฉลี่ย และสเปกตรัมกำลังราบเหมือนสัญญาณรบกวนขาว

# ส่วนที่ 3 สถิติพรรณนาและกับดักของ ddof (หัวข้อ 4.2.1)

สถิติพรรณนา (descriptive statistics) สรุปข้อมูลจำนวนมากด้วยตัวเลขไม่กี่ตัว

| ตัววัด | สูตร | ความหมายและข้อควรรู้ |
|---|---|---|
| ค่าเฉลี่ย $\bar x$ | $\frac{1}{N}\sum_i x_i$ | ไวต่อค่าผิดปกติ |
| มัธยฐาน | ค่ากลางหลังเรียงลำดับ | ทนต่อค่าผิดปกติ |
| ความแปรปรวน $s^2$ | $\frac{1}{N-1}\sum_i(x_i-\bar x)^2$ | ในแก๊สอุดมคติ ความแปรปรวนของความเร็วแต่ละแกนคือ $k_BT/m$ |
| ความเบ้ (skewness) | $\frac{1}{Ns^3}\sum(x_i-\bar x)^3$ | กระบวนการที่มีขีดเริ่มต้นมักเบ้ขวา |
| ความโด่งส่วนเกิน (excess kurtosis) | $\frac{1}{Ns^4}\sum(x_i-\bar x)^4 - 3$ | ค่าบวกแปลว่าค่าสุดโต่งเกิดบ่อยกว่าแบบเกาส์ |

### กับดักในทางปฏิบัติ: `numpy.var` กับ `pandas.Series.var`
`numpy.var()` หารด้วย $N$ (ค่าตั้งต้น `ddof=0`) ขณะที่ `pandas.Series.var()` หารด้วย $N-1$ (ค่าตั้งต้น `ddof=1`) ข้อมูลชุดเดียวกันจึงให้ความแปรปรวนต่างกันเมื่อใช้คนละไลบรารี

### ✏️ แบบฝึก 4.A ทายก่อนรัน
สำหรับข้อมูล 5 ค่า อัตราส่วน `np.var(x) / pd.Series(x).var()` เท่ากับเท่าใด (เขียนเป็นเศษส่วนหรือทศนิยม)

```python
my_ratio = None
```

```python
x5 = np.array([2.01, 2.03, 1.98, 2.02, 1.99])          # five pendulum periods [s]
ratio = np.var(x5) / pd.Series(x5).var()
print(f"np.var = {np.var(x5):.6f}, pandas var = {pd.Series(x5).var():.6f}, ratio = {ratio:.3f}")
print("✅", "(ยังไม่ได้ทาย)" if my_ratio is None else ("✔" if abs(my_ratio - ratio) < 1e-3 else f"✘ ค่าจริงคือ (N-1)/N = {ratio:.1f}"))

# ทำไมต้องหารด้วย N-1: ทดลองซ้ำ 20,000 ครั้งจากประชากรที่รู้ sigma = 1
samples = np.random.default_rng(3).normal(0, 1, (20000, 5))
print(f"\naverage of np.var (ddof=0) over 20,000 samples of N = 5: {samples.var(axis=1, ddof=0).mean():.3f}")
print(f"average of ddof=1 variance                             : {samples.var(axis=1, ddof=1).mean():.3f}  (true value 1)")
```

การหารด้วย $N$ ให้ความแปรปรวนต่ำกว่าความจริงอย่างเป็นระบบ เพราะจุดข้อมูลอยู่ใกล้ค่าเฉลี่ยของตัวอย่างมากกว่าค่าเฉลี่ยจริงเสมอ การหารด้วย $N-1$ แก้ความเอนเอียงนี้ ความต่างมีผลมากเมื่อ $N$ น้อย วิธีป้องกันคือระบุ `ddof=1` ใน NumPy ทุกครั้ง

# ส่วนที่ 5 ปรับการแจกแจงปัวซงกับจำนวนนับ (หัวข้อ 4.2.3)

โคบอลต์-60 มีครึ่งชีวิตราว 5.27 ปี อัตราการสลายจึงถือว่าคงที่ตลอดการทดลอง ถ้านับการสลายช่วงละหนึ่งวินาที ทำซ้ำหนึ่งพันครั้ง ทฤษฎีบอกว่าจำนวนนับต้องแจกแจงแบบปัวซง ตัวอย่างโค้ดตรวจคำทำนายนี้สามทาง คือ (1) อัตราส่วนความแปรปรวนต่อค่าเฉลี่ยควรใกล้หนึ่ง (2) ฮิสโทแกรมซ้อนกับการแจกแจงปัวซง และ (3) การทดสอบไคกำลังสอง (chi-squared goodness-of-fit test)

### โค้ดที่ 4.2 — ปรับการแจกแจงปัวซงและทดสอบความเหมาะสม

```python
import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt

counts = pd.read_csv('data/raw/co60_counts.csv')['counts'].values
lam_hat = counts.mean()
print(f"variance/mean = {counts.var(ddof=1)/lam_hat:.3f}")   # near 1?

# histogram with the Poisson pmf on top
n_max = counts.max() + 1
bins = np.arange(-0.5, n_max + 0.5, 1)
fig, ax = plt.subplots()
ax.hist(counts, bins=bins, density=True, alpha=0.6, label='data')
n_range = np.arange(0, n_max)
ax.plot(n_range, stats.poisson.pmf(n_range, mu=lam_hat), 'ro-', label='Poisson')
ax.set_xlabel('counts per 1 s'); ax.set_ylabel('probability'); ax.legend()

# chi-squared goodness of fit, keep bins with expected count >= 5
observed, _ = np.histogram(counts, bins=bins)
expected = len(counts) * stats.poisson.pmf(n_range, mu=lam_hat)
mask = expected >= 5
obs, exp = observed[mask], expected[mask]
exp = exp * obs.sum() / exp.sum()          # same total in both
chi2_stat, p = stats.chisquare(obs, f_exp=exp, ddof=1)   # 1 fitted parameter
print(f"chi^2 = {chi2_stat:.2f}, dof = {mask.sum() - 2}, p = {p:.3f}")
```

รายละเอียดสองจุด จุดแรก ช่องที่มีจำนวนคาดหวังต่ำกว่าห้าถูกตัดออก เพราะการประมาณของการทดสอบไคกำลังสองใช้ไม่ได้เมื่อจำนวนในช่องน้อยเกินไป แล้วปรับผลรวมของจำนวนคาดหวังให้เท่ากับจำนวนที่สังเกตได้ จุดที่สอง `ddof=1` ลดองศาอิสระลงหนึ่ง เพราะประมาณ $\lambda$ จากข้อมูลชุดเดียวกัน

อัตราส่วน 0.991 "ใกล้หนึ่ง" เพียงใดจึงพอ เซลล์ถัดไปตอบด้วยการจำลองข้อมูลปัวซงจริงขนาดเดียวกัน 2,000 ชุด

```python
rng1 = np.random.default_rng(1)
ratios = [(x := rng1.poisson(5, 1000)).var(ddof=1) / x.mean() for _ in range(2000)]
lo, hi = np.percentile(ratios, [2.5, 97.5])
print(f"mean = {lam_hat:.3f} counts/s from {counts.size} intervals; kept {mask.sum()} of {n_range.size} bins")
print(f"for true Poisson data (N = 1000) variance/mean falls in [{lo:.3f}, {hi:.3f}] 95 % of the time")
```

# ส่วนที่ 6 ปรับฟังก์ชันเกาส์กับเส้นสเปกตรัม (หัวข้อ 4.2.4–4.2.5)

อะตอมในแก๊สที่อุณหภูมิ $T$ เคลื่อนที่ด้วยความเร็วแจกแจงแบบแมกซ์เวลล์–โบลต์ซมันน์ ปรากฏการณ์ดอปเพลอร์ทำให้เส้นสเปกตรัมถ่างเป็นรูปเกาส์ที่มีความกว้าง

$$\sigma_\lambda = \lambda_0 \sqrt{\frac{k_B T}{m c^2}}$$

สูตรนี้เชื่อมพารามิเตอร์ทางสถิติ $\sigma_\lambda$ กับพารามิเตอร์ทางฟิสิกส์ $T$ โดยตรง การวัดความกว้างของเส้นจึงบอกอุณหภูมิของแก๊สได้ เซลล์แรกประเมินขนาดของ $\sigma_\lambda$ สำหรับเส้น H-alpha ที่ $10^4$ K แล้วโหลดสเปกตรัมจำลอง

```python
lam0, T_gas = 656.28, 1e4                      # nm, K
m_H = const.m_p + const.m_e
sig_expected = lam0 * np.sqrt(const.k * T_gas / (m_H * const.c**2))
print(f"expected at T = 1e4 K: sigma = {sig_expected:.4f} nm, FWHM = {2.355 * sig_expected:.4f} nm")

spec = pd.read_csv('data/raw/halpha_spectrum.csv')
wavelength, intensity = spec['wavelength_nm'].values, spec['intensity'].values
```

### โค้ดที่ 4.3 — ปรับฟังก์ชันเกาส์กับเส้นสเปกตรัม

```python
from scipy.optimize import curve_fit

def gaussian(x, A, mu, sigma, b):
    return A * np.exp(-(x - mu)**2 / (2 * sigma**2)) + b

p0 = [intensity.max() - intensity.min(),
      wavelength[intensity.argmax()], 0.02, intensity.min()]
popt, pcov = curve_fit(gaussian, wavelength, intensity, p0=p0)
A, mu, sigma, b = popt
print(f"line centre  mu    = {mu:.4f} nm")
print(f"line width   sigma = {sigma:.4f} +/- {np.sqrt(pcov[2, 2]):.4f} nm")
print(f"FWHM = {2.355 * sigma:.4f} nm")
```

```python
T_fit = m_H * const.c**2 / const.k * (sigma / mu)**2
dT = 2 * T_fit * np.sqrt(pcov[2, 2]) / sigma
print(f"temperature from the width: T = {T_fit:.0f} +/- {dT:.0f} K (generated with 10000 K)")

plt.figure(figsize=(7, 3.2))
plt.plot(wavelength, intensity, ".", ms=3, label="data (SYNTHETIC)")
plt.plot(wavelength, gaussian(wavelength, *popt), "r", label="Gaussian + baseline fit")
plt.xlabel("wavelength (nm)"); plt.ylabel("intensity"); plt.legend(fontsize=8); plt.tight_layout(); plt.show()
```

การปรับฟังก์ชันเกาส์สมมติว่าการถ่างของเส้นมาจากปรากฏการณ์ดอปเพลอร์เพียงอย่างเดียว ถ้ามีความกว้างธรรมชาติ รูปร่างรวมจะเป็นแบบฟอยก์ (Voigt) ซึ่งหางกว้างกว่า และถ้าอุณหภูมิที่ได้ไม่ตรงกับเงื่อนไขการทดลอง อาจมีกลไกการถ่างอื่น เช่น ปรากฏการณ์สตาร์ก ปรากฏการณ์ซีมัน หรือความกว้างจากเครื่องมือ นี่คือหลัก P2b การเลือกรูปแบบของแบบจำลองคือการเลือกสมมติฐานทางฟิสิกส์

# ส่วนที่ 8 การทดสอบสมมติฐานและความหมายของค่าพี (หัวข้อ 4.3.1)

การทดสอบสมมติฐานมีตรรกะคล้ายการพิจารณาคดีในศาล จำเลยได้รับการสันนิษฐานว่าบริสุทธิ์จนกว่าหลักฐานจะหนักแน่นพอ คำตัดสินว่า "ไม่มีความผิด" ไม่ได้แปลว่า "พิสูจน์แล้วว่าบริสุทธิ์"

- **สมมติฐานว่าง (null hypothesis, $H_0$):** ข้อสันนิษฐานตั้งต้นว่าไม่มีผลใหม่ เช่น "จำนวนนับแจกแจงแบบปัวซง"
- **ตัวสถิติทดสอบ (test statistic):** ค่าที่คำนวณจากข้อมูล เช่น $\chi^2$ หรือ $t$ ซึ่งรู้การแจกแจงเมื่อ $H_0$ เป็นจริง
- **ค่าพี:** ความน่าจะเป็นที่จะได้ตัวสถิติทดสอบสุดโต่งเท่ากับหรือมากกว่าค่าที่สังเกตได้ **ถ้า** $H_0$ เป็นจริง
- **ระดับนัยสำคัญ $\alpha$:** เกณฑ์ที่ต้องกำหนด **ก่อน** ดูข้อมูล

การตัดสินผิดมีสองแบบ ความผิดพลาดแบบที่หนึ่ง (type I error) คือปฏิเสธ $H_0$ ทั้งที่เป็นจริง หรือการประกาศการค้นพบที่ไม่มีอยู่จริง ความผิดพลาดแบบที่สอง (type II error) คือไม่ปฏิเสธ $H_0$ ทั้งที่ผิด หรือการพลาดสิ่งที่มีอยู่จริง

ค่าพีไม่ใช่ความน่าจะเป็นที่ $H_0$ เป็นจริง เปรียบกับคำถามสองข้อ "ถ้าเป็นพระสงฆ์ ความน่าจะเป็นที่โกนศีรษะคือเท่าใด" กับ "ถ้าโกนศีรษะ ความน่าจะเป็นที่เป็นพระสงฆ์คือเท่าใด" ค่าพีตอบคำถามแบบแรก

```python
rng7 = np.random.default_rng(7)
p_sim = np.array([stats.ttest_1samp(rng7.poisson(5, 1000), 5.0).pvalue for _ in range(4000)])
print(f"4,000 simulated experiments with true mean 5.0: fraction with p < 0.05 = {np.mean(p_sim < 0.05):.3f}")

plt.figure(figsize=(6, 3))
plt.hist(p_sim, bins=20, range=(0, 1), alpha=0.7)
plt.axvline(p_t, color="r", label=f"this data set, p = {p_t:.3f}")
plt.xlabel("p-value"); plt.ylabel("experiments"); plt.title("p-values when H0 is true are uniform")
plt.legend(fontsize=8); plt.tight_layout(); plt.show()
```

เมื่อ $H_0$ เป็นจริง ค่าพีกระจายสม่ำเสมอระหว่าง 0 ถึง 1 การทดลองราวร้อยละ 5 จึงได้ $p < 0.05$ โดยบังเอิญ ตามนิยามของ $\alpha$ ข้อมูลของเราเป็นหนึ่งในนั้น ผู้เขียนตำราไม่ได้เปลี่ยนเมล็ดสุ่มเพื่อเลี่ยงผลนี้ เพราะเป็นตัวอย่างที่เห็นได้ชัดว่าการทดสอบครั้งเดียวกับข้อมูลชุดเดียวยังไม่ใช่การค้นพบ

# ส่วนที่ 10 เกณฑ์ 5σ และการทดสอบหลายครั้ง (หัวข้อ 4.3.4)

ฟิสิกส์อนุภาคใช้เกณฑ์ $5\sigma$ สำหรับการประกาศการค้นพบ เพราะการประกาศอนุภาคใหม่ที่ผิดมีต้นทุนสูงมาก เกณฑ์นี้เทียบเท่า $\alpha$ เท่าใด และถ้าทดสอบ 20 สมมติฐานที่ทุกข้อเป็นจริงตาม $H_0$ ความน่าจะเป็นที่จะมีอย่างน้อยหนึ่งข้อได้ $p < 0.05$ เป็นเท่าใด

```python
print(f"5 sigma, one-sided: alpha = {stats.norm.sf(5):.2e}")
print(f"20 independent tests at 0.05: P(at least one false positive) = {1 - 0.95**20:.2f}")

# ตรวจด้วยการจำลอง: 20 การทดสอบที่ H0 เป็นจริงทุกข้อ ทำซ้ำ 10,000 รอบ
p20 = np.random.default_rng(20).uniform(0, 1, (10000, 20))      # p-values are uniform under H0
print(f"simulation: fraction of rounds with at least one p < 0.05 = {np.mean((p20 < 0.05).any(axis=1)):.3f}")
print(f"with Bonferroni (0.05/20 = {0.05 / 20}): {np.mean((p20 < 0.05 / 20).any(axis=1)):.3f}")
```

ในฟิสิกส์อนุภาค ปรากฏการณ์เดียวกันนี้เรียกว่าผลจากการมองหาทุกที่ (look-elsewhere effect) เมื่อค้นหายอดของสัญญาณในช่วงมวลกว้าง ๆ ย่อมมีโอกาสพบยอดปลอมที่ไหนสักแห่ง วิธีแก้ที่ง่ายที่สุดคือการปรับแบบบอนเฟอร์โรนี (Bonferroni correction) ซึ่งใช้เกณฑ์ $\alpha/k$ แทน $\alpha$

### แบบฝึก E4.3 การทดสอบไคกำลังสองและการตีความทางฟิสิกส์
แบบฝึกนี้ถามสามข้อ (ก) ค่าพีของข้อมูลโคบอลต์-60 เป็นเท่าใด และควรปฏิเสธ $H_0$ ที่ $\alpha = 0.05$ หรือไม่ (ข) ถ้าได้ค่าพีต่ำกว่า 0.05 จะหมายถึงอะไรได้บ้างในทางฟิสิกส์ (ค) ถ้าได้ค่าพีสูงกว่า 0.05 สรุปได้หรือไม่ว่าระบบเป็นปัวซง "แน่นอน" ผู้เรียนเขียนคำตอบในเซลล์ ✏️ ท้ายส่วนนี้ หลังจากดูผลของเซลล์ต่อไปนี้

เซลล์ถัดไปใช้การทดสอบเดียวกับโค้ดที่ 4.2 กับข้อมูลสามกรณี คือ ข้อมูลของบท เครื่องนับที่มีเวลาตาย (dead time) และรังสีพื้นหลังที่เลื่อนจาก 3 เป็น 7 ครั้งต่อวินาทีระหว่างการวัด **ก่อนรัน** ให้ทายว่าอัตราส่วนความแปรปรวนต่อค่าเฉลี่ยของสองกรณีหลังจะ `"น้อยกว่า 1"` หรือ `"มากกว่า 1"`

```python
my_direction = {"dead time": "?", "drifting background": "?"}
```

```python
rng43 = np.random.default_rng(43)

def chi2_poisson(c):
    lam = c.mean()
    b = np.arange(-0.5, c.max() + 1.5)
    k = np.arange(c.max() + 1)
    o_, _ = np.histogram(c, b)
    e_ = len(c) * stats.poisson.pmf(k, lam)
    m = e_ >= 5
    o, e = o_[m], e_[m] * o_[m].sum() / e_[m].sum()
    return stats.chisquare(o, e, ddof=1).pvalue

def dead_time_counts(rate, tau, n, T=1.0):
    # non-paralyzable counter: events closer than tau to the last counted one are lost
    out = []
    for _ in range(n):
        t = np.sort(rng43.uniform(0, T, rng43.poisson(rate * T)))
        kept, last = 0, -np.inf
        for ti in t:
            if ti - last >= tau:
                kept += 1; last = ti
        out.append(kept)
    return np.array(out)

cases = {
    'Co-60 data (this chapter)': counts,
    'dead time: rate 12/s, tau 0.05 s': dead_time_counts(12, 0.05, 1000),
    'background drifting 3 -> 7 /s': rng43.poisson(np.linspace(3, 7, 1000)),
}
print(f"{'case':36s} {'mean':>6s} {'var/mean':>9s} {'chi2 p':>8s}")
for name, c in cases.items():
    print(f"{name:36s} {c.mean():6.2f} {c.var(ddof=1) / c.mean():9.3f} {chi2_poisson(c):8.3g}")

# ✅ ตรวจการทาย
_dir = {"dead time": "น้อยกว่า 1", "drifting background": "มากกว่า 1"}
for k_, v in _dir.items():
    mine = my_direction[k_]
    print(f"✅ {k_}:", "(ยังไม่ได้ทาย)" if mine == "?" else ("✔" if mine == v else f"✘ ผลจริง {v}"))
```

เวลาตายทำให้เหตุการณ์ที่มาติดกันถูกนับเป็นครั้งเดียว จำนวนนับจึงสม่ำเสมอกว่าปัวซง ความแปรปรวนต่ำกว่าค่าเฉลี่ย ส่วนรังสีพื้นหลังที่เปลี่ยนตามเวลาเพิ่มการกระจายนอกเหนือจากความสุ่มของการสลาย ความแปรปรวนจึงสูงกว่าค่าเฉลี่ย ทั้งสองกรณีถูกปฏิเสธด้วยค่าพีต่ำมาก และทิศทางของการเบี่ยงบอกสาเหตุทางฟิสิกส์ได้

เซลล์ถัดไปสร้างช่วงความเชื่อมั่นของอัตราส่วนความแปรปรวนต่อค่าเฉลี่ยของข้อมูลในบทด้วยการสุ่มซ้ำแบบบูตสแตรป (bootstrap) ให้ใช้ผลนี้ประกอบการตอบข้อ (ค)

```python
boot = []
for _ in range(4000):
    s_ = rng43.choice(counts, counts.size)
    boot.append(s_.var(ddof=1) / s_.mean())
lo_, hi_ = np.percentile(boot, [2.5, 97.5])
print(f"Co-60 data: variance/mean 95 % bootstrap CI [{lo_:.3f}, {hi_:.3f}]")
print("-> deviations from Poisson of up to ~8 % either way are not excluded by these data")
```

สูตรนี้คำนวณตรง ๆ ไม่ได้ ลองนับจำนวนสถานะของแม่เหล็กจำลองขนาด $100\times100$ ช่อง

ทางออกคือการสุ่มตัวอย่างตามความสำคัญ (importance sampling) สุ่มสถานะมาโดยให้โอกาสถูกเลือกเป็นสัดส่วนกับ $e^{-\beta E_i}$ แล้วประมาณค่าเฉลี่ยเชิงความร้อนด้วยค่าเฉลี่ยธรรมดา $\langle A\rangle \approx \frac{1}{M}\sum_n A_{i_n}$ เหมือนการสำรวจความเห็นก่อนเลือกตั้งที่ถามกลุ่มตัวอย่างไม่กี่พันคนที่เลือกมาอย่างถูกวิธี คำถามคือจะสุ่มให้ได้ตามการแจกแจงโบลต์ซมันน์อย่างไรทั้งที่ยังไม่รู้ $Z$

# ส่วนที่ 14 ขั้นตอนวิธีเมโทรโพลิสและช่วงเข้าสู่สมดุล (หัวข้อ 4.4.2)

**ขั้นตอนวิธีเมโทรโพลิส (Metropolis algorithm):** เสนอสถานะใหม่ที่ต่างจากเดิมเล็กน้อย เช่น พลิกสปินหนึ่งตัว แล้วคำนวณ $\Delta E$ ถ้า $\Delta E \le 0$ ยอมรับเสมอ ถ้า $\Delta E > 0$ ยอมรับด้วยความน่าจะเป็น $e^{-\beta\Delta E}$ ถ้าไม่ยอมรับ ให้นับสถานะเดิมซ้ำ ค่า $Z$ ตัดกันหายไปในอัตราส่วน $P_{i'}/P_i = e^{-\beta\Delta E}$ กฎนี้จึงต้องการเพียงผลต่างพลังงาน เหมือนลูกบอลที่กลิ้งบนภูมิประเทศขรุขระและถูกเขย่าตลอดเวลา การลงเนินเกิดเสมอ การขึ้นเนินเกิดได้บ้างตามความแรงของการเขย่า ซึ่งก็คืออุณหภูมิ

**แบบจำลองไอซิงสองมิติ (two-dimensional Ising model):** สปิน $s_i = \pm1$ บนตาราง $L\times L$ มีพลังงาน $H = -J\sum_{\langle i,j\rangle} s_is_j$ (ใช้ $J = k_B = 1$) และมีการเปลี่ยนสถานะที่ $T_c = 2/\ln(1+\sqrt2) \approx 2.269$ ตามผลเฉลยของออนซาเกอร์ การพลิกสปินหนึ่งตัวเปลี่ยนพลังงาน $\Delta E = 2s_{ij}\sum_{\text{nb}} s_{\text{nb}}$ และใช้เงื่อนไขขอบแบบคาบ (periodic boundary condition)

### โค้ดที่ 4.6 (ส่วนแรก) — นิยามฟังก์ชันของการจำลองเมโทรโพลิส
โค้ดที่ 4.6 ในตำราแบ่งเป็นสองเซลล์ในโน้ตบุ๊กนี้ เซลล์นี้นิยามฟังก์ชัน ส่วนการกวาดอุณหภูมิอยู่ในส่วนที่ 15

```python
import numpy as np
rng = np.random.default_rng(seed=2026)

def init_lattice(L, hot_start=True):
    if hot_start:
        return rng.choice([-1, 1], size=(L, L))
    return np.ones((L, L), dtype=int)

def total_energy(lattice):
    # each bond counted once: one neighbour along each axis
    return -np.sum(lattice * (np.roll(lattice, 1, 0) + np.roll(lattice, 1, 1)))

def metropolis_step(lattice, beta):
    """One sweep = L*L attempted single-spin flips."""
    L = lattice.shape[0]
    for _ in range(L * L):
        i, j = rng.integers(0, L, size=2)
        s = lattice[i, j]
        nb = (lattice[(i+1) % L, j] + lattice[(i-1) % L, j]
            + lattice[i, (j+1) % L] + lattice[i, (j-1) % L])
        dE = 2 * s * nb
        if dE <= 0 or rng.random() < np.exp(-beta * dE):
            lattice[i, j] = -s
    return lattice

def simulate_T(L, T, n_thermalize, n_measure):
    lattice = init_lattice(L)
    beta = 1.0 / T
    for _ in range(n_thermalize):          # discard: not yet in equilibrium
        metropolis_step(lattice, beta)
    energies = []
    for _ in range(n_measure):
        metropolis_step(lattice, beta)
        energies.append(total_energy(lattice))
    energies = np.asarray(energies)
    E_mean = energies.mean() / (L * L)
    Cv = energies.var() / (T**2 * L * L)   # fluctuation formula
    return E_mean, Cv
```

```python
from scipy.special import ellipk
print(f"run time: {time.perf_counter() - t_start:.0f} s for 21 temperatures")

def onsager_energy(T):
    """Exact energy per spin of the infinite 2D Ising model (J = k_B = 1)."""
    b = 1.0 / T
    k = 2 * np.sinh(2 * b) / np.cosh(2 * b)**2
    return -(1 / np.tanh(2 * b)) * (1 + 2 / np.pi * (2 * np.tanh(2 * b)**2 - 1) * ellipk(k**2))

Tc = 2 / np.log(1 + np.sqrt(2))
res16 = pd.DataFrame({'T': T_values, 'E_per_spin': E_means, 'Cv_per_spin': Cvs,
                      'E_onsager': [onsager_energy(T) for T in T_values]})
print(res16.round(3).iloc[::2].to_string(index=False))
i_pk = int(np.argmax(Cvs))
print(f"\nT_c (Onsager) = {Tc:.4f}; exact E/N at T_c = -sqrt(2) = {-np.sqrt(2):.4f}")
print(f"C_v peak of this L = 16 run at T = {T_values[i_pk]:.2f} (grid step 0.1), C_v/N = {Cvs[i_pk]:.2f}")

fig, ax = plt.subplots(1, 2, figsize=(10, 3.8))
Tf = np.linspace(1.5, 3.5, 200)
ax[0].plot(T_values, E_means, 'o', label='Metropolis, L = 16')
ax[0].plot(Tf, [onsager_energy(T) for T in Tf], '-', label='Onsager, L = infinity')
ax[0].set_xlabel('T'); ax[0].set_ylabel('<E>/N'); ax[0].legend(fontsize=8)
ax[1].plot(T_values, Cvs, 'o-'); ax[1].axvline(Tc, ls=':', color='gray')
ax[1].set_xlabel('T'); ax[1].set_ylabel('C_v/N')
fig.suptitle("2D Ising model, Metropolis"); plt.tight_layout(); plt.show()
```

# ส่วนที่ 17 วิธีหวัง–แลนเดาและการตรวจผลรวมของ g(E) (หัวข้อ 4.4.3)

วิธีเมโทรโพลิสให้ผลทีละอุณหภูมิ และที่อุณหภูมิต่ำระบบอาจติดอยู่ในหุบพลังงานเดียว หวัง (Fugao Wang) และแลนเดา (David P. Landau) จึงเปลี่ยนสิ่งที่ต้องการหา ให้หาความหนาแน่นสถานะ (density of states) $g(E)$ ซึ่งไม่ขึ้นกับอุณหภูมิ เมื่อรู้ $g(E)$ แล้ว $Z(T) = \sum_E g(E)\,e^{-E/k_BT}$ คำนวณได้ทุกอุณหภูมิ

**วิธีหวัง–แลนเดา:** เริ่มจาก $\ln g(E) = 0$ และ $\ln f = 1$ เสนอการพลิกสปินจาก $E$ ไป $E'$ แล้วยอมรับด้วยความน่าจะเป็น $\min[1, \exp(\ln g(E) - \ln g(E'))]$ หลังตัดสินใจ เพิ่ม $\ln g$ ของพลังงานปัจจุบันขึ้น $\ln f$ และเพิ่มฮิสโทแกรม $H$ เมื่อฮิสโทแกรม "แบนพอ" ให้ลด $\ln f$ ลงครึ่งหนึ่งแล้วเริ่มรอบใหม่ เหมือนคนเดินสำรวจเมืองที่ขีดเครื่องหมายทุกย่านที่ไปและเลี่ยงย่านที่มีเครื่องหมายมาก

โค้ดทำงานกับ $\ln g$ เพื่อป้องกันการล้นของตัวเลข (overflow) ใช้เฉพาะช่องพลังงานที่ไปเยือนจริง และยึดค่าด้วย $g(E_{\min}) = 2$ เพราะสถานะพื้นมีสองสถานะ เซลล์แรกตั้งเมล็ดสุ่มใหม่เป็น 2026 เพื่อให้ผลตรงกับตำรา และนิยาม `total_energy` ซ้ำในกรณีที่ข้ามคาบที่ 3 มา

### โค้ดที่ 4.7 — การประมาณความหนาแน่นสถานะด้วยวิธีหวัง–แลนเดา

```python
rng = np.random.default_rng(seed=2026)

def total_energy(lattice):
    # each bond counted once: one neighbour along each axis
    return -np.sum(lattice * (np.roll(lattice, 1, 0) + np.roll(lattice, 1, 1)))
```

```python
def wang_landau_2d_ising(L, ln_f_final=1e-8, flatness=0.8,
                         check_every=1000, max_sweeps=10**7):
    lattice = rng.choice([-1, 1], size=(L, L))
    N = L * L
    E_min = -2 * N
    E_bins = np.arange(E_min, 2 * N + 1, 4)
    log_g = np.zeros(len(E_bins))
    H = np.zeros(len(E_bins), dtype=int)
    visited = np.zeros(len(E_bins), dtype=bool)   # some bins never occur
    ln_f = 1.0
    idx = lambda E: (E - E_min) // 4
    E_curr = total_energy(lattice)
    sweep = 0
    while ln_f > ln_f_final and sweep < max_sweeps:
        for _ in range(N):
            i, j = rng.integers(0, L, size=2)
            s = lattice[i, j]
            nb = (lattice[(i+1) % L, j] + lattice[(i-1) % L, j]
                + lattice[i, (j+1) % L] + lattice[i, (j-1) % L])
            E_new = E_curr + 2 * s * nb
            if rng.random() < np.exp(log_g[idx(E_curr)] - log_g[idx(E_new)]):
                lattice[i, j] = -s
                E_curr = E_new
            k = idx(E_curr)
            log_g[k] += ln_f
            H[k] += 1
            visited[k] = True
        sweep += 1
        if sweep % check_every == 0:
            h = H[visited]
            if h.min() >= flatness * h.mean():    # histogram is flat enough
                ln_f /= 2.0
                H[:] = 0
    E_bins, log_g = E_bins[visited], log_g[visited]
    log_g += np.log(2.0) - log_g[0]               # anchor: g(E_min) = 2
    return E_bins, log_g, sweep

def thermodynamic(E_bins, log_g, T_values):
    res = {'T': T_values, 'E_mean': [], 'Cv': [], 'S': []}
    for T in T_values:
        beta = 1.0 / T
        log_w = log_g - beta * E_bins
        log_w_max = log_w.max()                   # log-sum-exp trick
        w = np.exp(log_w - log_w_max)
        Z = w.sum()
        E_mean = (E_bins * w).sum() / Z
        E2_mean = (E_bins**2 * w).sum() / Z
        res['E_mean'].append(E_mean)
        res['Cv'].append(beta**2 * (E2_mean - E_mean**2))
        res['S'].append(np.log(Z) + log_w_max + beta * E_mean)   # S/k_B
    return res

E_bins, log_g, n_sweeps = wang_landau_2d_ising(L=4, ln_f_final=1e-6)
print(np.exp(log_g).sum(), 2**16)               # sanity check
```

ผลรวมของ $g(E)$ ต้องเท่ากับจำนวนสถานะทั้งหมด $2^{16} = 65{,}536$ ตาราง $4\times4$ เล็กพอที่เราไล่ทุกสถานะได้จริง เซลล์ถัดไปคำนวณ $g(E)$ แม่นตรงด้วยการนับทุกสถานะ แล้วเทียบกับผลของวิธีหวัง–แลนเดาทีละระดับพลังงาน

```python
import itertools
states = np.array(list(itertools.product([-1, 1], repeat=16))).reshape(-1, 4, 4)
E_all = -np.sum(states * (np.roll(states, 1, 1) + np.roll(states, 1, 2)), axis=(1, 2))
vals, cnt = np.unique(E_all, return_counts=True)
exact = dict(zip(vals.tolist(), cnt.tolist()))
print(f"this run (seed 2026): sum g = {np.exp(log_g).sum():.0f} ({100 * (np.exp(log_g).sum() / 2**16 - 1):+.1f} %), {n_sweeps} sweeps\n")
cmp_g = pd.DataFrame({'E': E_bins, 'g_WL': np.exp(log_g).round(1),
                      'g_exact': [exact[e] for e in E_bins]})
cmp_g['error_%'] = (100 * (cmp_g['g_WL'] / cmp_g['g_exact'] - 1)).round(1)
print(cmp_g.to_string(index=False))
```

ถ้าลด $\ln f$ สุดท้ายจาก $10^{-6}$ เป็น $10^{-8}$ ผลจะแม่นขึ้นหรือไม่ เซลล์ถัดไปรันด้วยเมล็ดสุ่ม 1 ถึง 4 ทั้งสองค่า (ราวสองนาทีบน Colab) ตำรารายงานผลของเมล็ดสุ่ม 1 ถึง 8 ถ้าต้องการครบ ให้เปลี่ยนเป็น `SEEDS = range(1, 9)` ซึ่งใช้ราวสี่นาทีครึ่ง

```python
SEEDS = range(1, 5)          # book: seeds 1-8 -> range(1, 9), about twice as long
t0_ = time.perf_counter()
print("seed  ln_f=1e-6: sum error  g(-24)   g(-20) | ln_f=1e-8: sum error")
for seed in SEEDS:
    out = []
    for lnf in (1e-6, 1e-8):
        rng = np.random.default_rng(seed)
        Eb, lg, _ = wang_landau_2d_ising(L=4, ln_f_final=lnf)
        out.append((Eb, np.exp(lg)))
    (Eb, g6), (_, g8) = out
    d = dict(zip(Eb.tolist(), g6))
    print(f"{seed:4d}       {100 * (g6.sum() / 2**16 - 1):+6.1f} %  {d[-24]:6.1f}  {d[-20]:6.1f}"
          f"  |        {100 * (g8.sum() / 2**16 - 1):+6.1f} %")
print(f"({time.perf_counter() - t0_:.0f} s)")
```

ผลรวมคลาดได้ตั้งแต่ไม่ถึงร้อยละหนึ่งจนถึงราวร้อยละ 16 ขึ้นกับเมล็ดสุ่ม (ในตำรา เมล็ดสุ่ม 3 คลาดมากที่สุด) และการลด $\ln f$ จาก $10^{-6}$ เป็น $10^{-8}$ แทบไม่เปลี่ยนผล เพราะเมื่อ $\ln f$ เล็กมากแล้ว การปรับแต่ละครั้งเล็กเกินกว่าจะแก้ความคลาดเคลื่อนที่สะสมไว้ตั้งแต่รอบแรก ๆ เบลาร์ดิเนลลีและเปเรย์รา (Belardinelli and Pereyra) จึงเสนอให้ลด $\ln f$ ตาม $1/t$ ซึ่งให้ความคลาดเคลื่อนลดลงต่อเนื่องตามเวลา (โค้ดอยู่ใน `exercises/E4_5c_wang_landau_1t.py` ของโค้ดประกอบตำรา)

# ส่วนที่ 19 แถบความคลาดเคลื่อนและช่วงความเชื่อมั่นเมื่อ N น้อย (หัวข้อ 4.5.1–4.5.2)

| ชนิด | ที่มา | สมบัติ | วิธีลด |
|---|---|---|---|
| แบบสุ่ม (random) | การแกว่งเชิงความร้อน สัญญาณรบกวนในวงจร | กระจายรอบค่าเฉลี่ย ลดลงตาม $1/\sqrt N$ | เพิ่มจำนวนการวัด |
| เชิงระบบ (systematic) | การสอบเทียบ การเลื่อนของเครื่องมือ แบบจำลองที่ไม่ตรงความจริง | เลื่อนผลทั้งชุดไปทางเดียว ไม่ลดลงเมื่อเพิ่ม $N$ | ปรับปรุงเครื่องมือ ตรวจไขว้ด้วยวิธีอื่น |

ต้องแยก $s$ ซึ่งบอกว่าการวัดแต่ละครั้งกระจายเพียงใด ออกจาก $\mathrm{SE}(\bar x) = s/\sqrt N$ ซึ่งบอกว่าค่าเฉลี่ยแน่นอนเพียงใด รายงานที่เขียนเพียง "± ค่าหนึ่ง" โดยไม่บอกว่าเป็นตัวใด ทำให้ผู้อ่านตีความผิดได้ถึง $\sqrt N$ เท่า

เมื่อไม่รู้ $\sigma$ จริงและต้องใช้ $s$ แทน ช่วงความเชื่อมั่นร้อยละ 95 ของค่าเฉลี่ยคือ $\bar x \pm t_{0.025,N-1}\,s/\sqrt N$ เซลล์ถัดไปสร้างตารางค่า $t$ และตรวจความครอบคลุมจริงเมื่อ $N = 5$

```python
print("  N   t_0.025,N-1   wider than z = 1.96 by")
for n_ in (5, 10, 30, 100):
    t_ = stats.t.ppf(0.975, df=n_ - 1)
    print(f"{n_:4d}   {t_:8.3f}      {100 * (t_ / 1.96 - 1):5.1f} %")

rng19 = np.random.default_rng(19)
x = rng19.normal(0, 1, (20000, 5))
h = x.std(axis=1, ddof=1) / np.sqrt(5)
print(f"\nN = 5: '95 %' interval with 1.96 covers {np.mean(np.abs(x.mean(axis=1)) < 1.96 * h):.3f}, "
      f"with t covers {np.mean(np.abs(x.mean(axis=1)) < stats.t.ppf(0.975, 4) * h):.3f}")
```

เมื่อวัดซ้ำห้าครั้งตามปกติในห้องปฏิบัติการระดับปริญญาตรี ช่วงที่ใช้ 1.96 ครอบค่าจริงเพียงราวร้อยละ 88 ไม่ใช่ร้อยละ 95 ส่วนช่วงที่ใช้ค่า $t$ ครอบร้อยละ 95 ตามที่ควร ในโค้ด ค่า $t$ หาได้จาก `stats.t.ppf(0.975, df=N-1)` และทั้งช่วงหาได้จาก `stats.t.interval`

อีกข้อที่มักลืม ถ้าแถบ $1\sigma$ ครอบค่าจริงราวร้อยละ 68 ราวหนึ่งในสามของจุดข้อมูลที่ดีควร **ไม่** ครอบเส้นทฤษฎี กราฟที่แถบทุกแถบครอบเส้นทฤษฎีได้หมด อาจประเมินความคลาดเคลื่อนไว้สูงเกินจริง

# ส่วนที่ 20 การรายงานผลการวัด (หัวข้อ 4.5.3 และแบบฝึก E4.6)

แนวปฏิบัติของเทย์เลอร์ (Taylor)
1. เขียนในรูป "ค่ากลาง ± ความไม่แน่นอน หน่วย"
2. ปัดความไม่แน่นอนเหลือหนึ่งหลัก ยกเว้นเมื่อหลักแรกเป็น 1 ให้เก็บสองหลัก (การปัด 0.14 เป็น 0.1 เปลี่ยนค่าไปเกือบร้อยละ 30)
3. ปัดค่ากลางให้หลักสุดท้ายอยู่ตำแหน่งเดียวกับหลักสุดท้ายของความไม่แน่นอน
4. เมื่อรายงานช่วงความเชื่อมั่นแทน $\pm1\sigma$ ต้องระบุระดับความเชื่อมั่นเสมอ

เซลล์ถัดไปเขียนกฎนี้เป็นฟังก์ชัน (ชุดเดียวกับ `09_reporting.py` ในโค้ดประกอบตำรา) แล้วแก้ตัวอย่างที่ผิดในตารางของตำรา

```python
def report(value, unc, unit=""):
    """Format value +/- uncertainty following the rule in the book."""
    exp10 = math.floor(math.log10(abs(unc)))
    first = int(unc / 10**exp10)
    digits = 2 if first == 1 else 1
    dec = -(exp10 - digits + 1)                         # decimal places to keep
    u = round(unc, dec); v = round(value, dec)
    if abs(v) >= 1e4 or abs(v) < 1e-3:                  # scientific notation
        e = math.floor(math.log10(abs(v)))
        s = 10**e
        d = max(dec + e, 0)
        return f"({v / s:.{d}f} ± {u / s:.{d}f}) × 10^{e} {unit}".strip()
    d = max(dec, 0)
    return f"({v:.{d}f} ± {u:.{d}f}) {unit}".strip()

for v, u, unit in [(656.28135, 0.00521, "nm"), (9.81254, 0.02, "m/s^2"), (342156.7, 15234.2, "m/s")]:
    print(f"  {v} ± {u}  ->  {report(v, u, unit)}")
```

# ส่วนที่ 21 การแพร่ของความคลาดเคลื่อน (หัวข้อ 4.5.4)

ความไม่แน่นอนบางส่วนเป็นขีดจำกัดพื้นฐาน เช่น หลักความไม่แน่นอนของไฮเซนเบิร์ก การแจกแจงแมกซ์เวลล์–โบลต์ซมันน์ และความคลาดเคลื่อนของการนับ $1/\sqrt N$ เมื่อ $f$ เป็นฟังก์ชันของตัวแปรที่คลาดเคลื่อนเล็กและไม่สัมพันธ์กัน

$$\sigma_f^2 \approx \sum_i \left(\frac{\partial f}{\partial x_i}\right)^2 \sigma_{x_i}^2$$

กรณี $f \propto x^n$ ให้ $\sigma_f/f = |n|\,\sigma_x/x$ ลูกตุ้มอย่างง่ายมี $g = 4\pi^2\ell/T^2$ คาบที่คลาดร้อยละหนึ่งจึงทำให้ $g$ คลาดร้อยละสอง เซลล์ถัดไปตรวจสูตรนี้ด้วยการจำลองมอนติคาร์โล

```python
l_, sl = 1.000, 0.001                  # m (0.1 %)
Tp, sT = 2.007, 0.01 * 2.007           # s (1 %)
g_ = 4 * np.pi**2 * l_ / Tp**2
sg = g_ * np.hypot(sl / l_, 2 * sT / Tp)
print(f"pendulum: T known to 1 %, l to 0.1 % -> g = {report(g_, sg, 'm/s^2')}; relative error of g = {100 * sg / g_:.2f} %")
gs = 4 * np.pi**2 * np.random.default_rng(5).normal(l_, sl, 200000) / np.random.default_rng(6).normal(Tp, sT, 200000)**2
print(f"Monte Carlo check: std(g)/g = {100 * gs.std() / gs.mean():.2f} %")
```

### E4.7 การวิเคราะห์ข้อมูลเชิงสำรวจของ Galaxy10 ฉบับสมบูรณ์
เลือกตัวแปรสองคู่ที่มีสหสัมพันธ์สูงที่สุด คำนวณสัมประสิทธิ์สหสัมพันธ์พร้อมช่วงความเชื่อมั่นร้อยละ 95 ด้วยการแปลงของฟิชเชอร์ (Fisher $z$-transformation) และด้วยการสุ่มซ้ำแบบบูตสแตรป แล้วเขียนรายงานหนึ่งหน้า เซลล์ถัดไปเป็นจุดเริ่มต้น (ชุดเดียวกับ `exercises/E4_7_eda_correlations.py`)

```python
from itertools import combinations
df7 = pd.read_csv('data/processed/galaxy10_features.csv')
num = df7[['redshift', 'r_mag', 'g_mag', 'concentration']]
pairs = sorted(combinations(num.columns, 2), key=lambda p: -abs(num[list(p)].dropna().corr().iloc[0, 1]))
rng47 = np.random.default_rng(47)
for a, b in pairs[:2]:
    d = num[[a, b]].dropna()
    r, n = d.corr().iloc[0, 1], len(d)
    z, se_ = np.arctanh(r), 1 / np.sqrt(n - 3)
    lo_, hi_ = np.tanh(z - 1.96 * se_), np.tanh(z + 1.96 * se_)
    boot = [d.iloc[rng47.integers(0, n, n)].corr().iloc[0, 1] for _ in range(1000)]
    blo, bhi = np.percentile(boot, [2.5, 97.5])
    print(f"{a} vs {b}: r = {r:.3f} (n = {n}); Fisher 95 % CI [{lo_:.3f}, {hi_:.3f}]; bootstrap [{blo:.3f}, {bhi:.3f}]")

d = df7.dropna(subset=['redshift', 'r_mag', 'g_mag'])
res_r = d['r_mag'] - np.polyval(np.polyfit(d['redshift'], d['r_mag'], 1), d['redshift'])
res_g = d['g_mag'] - np.polyval(np.polyfit(d['redshift'], d['g_mag'], 1), d['redshift'])
print(f"\nr_mag vs g_mag after removing the common dependence on redshift: r = {np.corrcoef(res_r, res_g)[0, 1]:.3f}")
```

ระยะทาง (แทนด้วยเรดชิฟต์) ขับทั้งสองโชติมาตร สหสัมพันธ์ที่เหลือหลังตัดผลของเรดชิฟต์มาจากความสว่างและสีของกาแล็กซีแต่ละดวง และเนื่องจากเรดชิฟต์หายไปอย่างไม่สุ่ม (บทที่ 3) สหสัมพันธ์ที่เกี่ยวกับเรดชิฟต์จึงบรรยายเฉพาะส่วนที่สว่างของตัวอย่าง รายงาน E4.7 ต้องมีความไม่แน่นอนกำกับทุกตัวเลข อภิปรายผลของการคัดเลือกตัวอย่าง และแยกสหสัมพันธ์ออกจากเหตุและผล
