โน้ตบุ๊กประกอบตำรา *วิทยาศาสตร์ข้อมูลและการเรียนรู้ของเครื่องสำหรับนักฟิสิกส์*

โน้ตบุ๊กนี้ใช้ในชั้นเรียนได้ด้วยไฟล์นี้ไฟล์เดียว ไม่ต้องดึงโค้ดจาก GitHub และไม่ต้องดาวน์โหลดข้อมูล เนื้อหาเรียงตามบทที่ 3 ในตำรา แบ่งเป็นสี่คาบ คาบละประมาณ 60 นาที หัวข้อที่มีป้าย **โค้ดที่ 3.N** ตรงกับตัวอย่างโค้ดเลขเดียวกันในตำรา

## วัตถุประสงค์
เมื่อทำโน้ตบุ๊กนี้จนจบ ผู้เรียนควรสามารถ

1. แยกการทำความสะอาดข้อมูล (data cleaning) ที่มีเหตุผลรองรับออกจากการบิดเบือนข้อมูล (data massaging) และบันทึกการตัดสินใจด้วยการทำเครื่องหมายแทนการลบ
2. ระบุปัญหาคุณภาพของข้อมูลตาราง อนุกรมเวลา ภาพ และสเปกตรัม แล้วเลือกวิธีแก้ที่สอดคล้องกับกลไกทางฟิสิกส์ของปัญหา พร้อมบอกข้อแลกเปลี่ยน (trade-off) ของแต่ละวิธี
3. ใช้ pandas โหลด ตรวจชนิดข้อมูล กรอง สรุปตามกลุ่ม เติมค่าที่ขาดหาย และรวมตาราง โดยอธิบายได้ว่าการเติมค่าแต่ละวิธีใส่สมมติฐานใดลงไป
4. สร้างลักษณะเด่น (feature) จากการวิเคราะห์มิติ สมมาตร และกฎอนุรักษ์ และปรับค่าโดยรักษาสเกลทางกายภาพ
5. ประเมินคุณภาพของชุดข้อมูลด้วยแบบประเมินคุณภาพข้อมูล (data quality scorecard) ก่อนส่งต่อให้บทที่ 4

### การล็อกแนวคำตอบ
เซลล์ ✅ และเซลล์แนวคำตอบในโน้ตบุ๊กนี้แสดงคำตอบอ้างอิงเมื่อผู้เรียนเติมคำตอบในเซลล์ ✏️ ของข้อนั้นครบแล้วเท่านั้น ถ้ายังไม่ได้ตอบ เซลล์จะแสดงเพียงข้อความ 🔒 บอกว่ายังขาดคำตอบใด การคำนวณในเซลล์ยังทำงานตามปกติ ดังนั้น Run all จึงรันได้ครบโดยไม่เปิดเผยคำตอบ แนวคำตอบของงานเขียนและงานนอกชั้นซึ่งตรวจด้วยโค้ดไม่ได้จะเปิดเมื่อผู้สอนเลือกเท่านั้น

**สำหรับผู้สอน:** ติ๊กช่อง `SHOW_ANSWERS` ในเซลล์ถัดไป (หรือแก้เป็น `True`) แล้วรันเซลล์นั้นอีกครั้ง เพื่อแสดงแนวคำตอบทุกข้อ เช่น เมื่อเฉลยในชั้นหรือเตรียมสอน

```python
#@title ⚙️ ตัวเลือกของผู้สอน: แสดงแนวคำตอบ
SHOW_ANSWERS = False  #@param {type:"boolean"}

def _answered(v):
    """True when an exercise variable holds an answer (not None, "?", "", empty or containing blanks)."""
    if v is None or v is Ellipsis:
        return False
    if isinstance(v, str):                       # placeholders such as "?", "?-?", "..." count as blank
        return v.replace("?", "").replace("-", "").replace(".", "").strip() != ""
    if isinstance(v, dict):
        return len(v) > 0 and all(_answered(x) for x in v.values())
    if isinstance(v, (list, tuple, set, frozenset)):
        return len(v) > 0 and all(_answered(x) for x in v)
    if type(v).__name__ == "ndarray":
        return v.size > 0
    return True

def _gate(names=()):
    """Return print if the answers may be shown, otherwise print one lock message and return a no-op."""
    if SHOW_ANSWERS:
        return print
    if names:
        missing = [n for n in names if not _answered(globals().get(n))]
        if not missing:
            return print
        print("🔒 ยังไม่ได้ตอบ:", ", ".join(missing),
              "— เติมคำตอบในเซลล์ ✏️ ด้านบนให้ครบก่อน แล้วรันเซลล์นี้อีกครั้ง")
    else:
        print("🔒 แนวคำตอบข้อนี้ผู้สอนเป็นผู้เปิด (ตั้ง SHOW_ANSWERS = True ในส่วนที่ 0)")
    return lambda *args, **kwargs: None

print("แสดงแนวคำตอบทั้งหมด:", "ใช่ (โหมดผู้สอน)" if SHOW_ANSWERS else "ไม่ (แสดงเมื่อผู้เรียนตอบข้อนั้นครบแล้ว)")
```

```python
#@title สร้างข้อมูลจำลองใน data/raw/ (ดับเบิลคลิกเพื่อดูโค้ด)
# โค้ดชุดเดียวกับ ch03_cleaning/01_make_raw_data.py ในโค้ดประกอบตำรา
SRC = r'''"""Ch.3  Create the SYNTHETIC raw data used by every script of this chapter.

The book refers to files such as data/raw/galaxy10_metadata.csv and data/raw/afm_spectrum.csv.
They are teaching files built here with a fixed seed; the numbers are NOT measurements.
  * galaxy table: Galaxy10 DECaLS class names, but redshift and magnitudes are simulated.
  * AFM files: simulated curves (the author's own AFM measurements are not published here).
  * 4-lepton events: simulated H -> ZZ* -> 4l signal plus a smooth background.

Output (data/raw/)
  galaxy10_metadata.csv      2,000 galaxies: image_id, galaxy_class, redshift, r_mag, g_mag
  galaxy10_image_index.csv   image_id -> image file (some ids missing, some extra)
  galaxy10_truth.csv         the true redshift of every galaxy (answer key, for E3.4 only)
  galaxy10_messy.csv         ~500 rows with NaN, -999 sentinels and outliers (for E3.2)
  galaxy_images.npy          4 synthetic images, shape (4, 256, 256, 3), 2 hot pixels in image 0
  afm_timeseries.csv         force signal while scanning: drift + spikes + noise
  afm_spectrum.csv           one force-distance curve (approach + retract) with baseline shift
  afm_curves.csv             15 force-distance curves with different stiffness/adhesion (E3.5)
  afm_curves_truth.csv       the parameters used to make them (answer key)
  higgs_4l_events.csv        E, px, py, pz of four leptons per event [GeV]
"""
from pathlib import Path

import numpy as np
import pandas as pd

SEED = 2026
rng = np.random.default_rng(SEED)
RAW = Path(__file__).resolve().parent / "data" / "raw"
RAW.mkdir(parents=True, exist_ok=True)

CLASSES = ["Disturbed", "Merging", "Round Smooth", "In-between Round Smooth",
           "Cigar Shaped Smooth", "Barred Spiral", "Unbarred Tight Spiral",
           "Unbarred Loose Spiral", "Edge-on without Bulge", "Edge-on with Bulge"]


# --------------------------------------------------------------- galaxies
def make_galaxies(n=2000):
    cls = rng.choice(len(CLASSES), n, p=[.06, .10, .15, .15, .06, .11, .10, .15, .06, .06])
    z_true = np.clip(rng.gamma(3.0, 0.035, n), 0.005, 0.40)
    # toy magnitude-redshift relation (inverse of photo_z_estimate in the book) + scatter
    r_mag = 15.0 + z_true / 0.05 + rng.normal(0, 0.6, n)
    color = np.array([0.7, 0.7, 0.9, 0.85, 0.8, 0.75, 0.6, 0.55, 0.65, 0.8])[cls]
    g_mag = r_mag + color + rng.normal(0, 0.1, n)
    # redshift missing more often for faint galaxies (MAR on r_mag), ~15 % overall
    p_miss = 1 / (1 + np.exp(-(r_mag - 18.9) / 0.6))
    missing = rng.random(n) < p_miss
    df = pd.DataFrame({
        "image_id": [f"galaxy_{i:05d}" for i in range(n)],
        "galaxy_class": [CLASSES[c] for c in cls],
        "redshift": np.where(missing, np.nan, np.round(z_true, 4)),
        "r_mag": np.round(r_mag, 3),
        "g_mag": np.round(g_mag, 3).astype(object),
    })
    # five bright foreground stars that slipped into the catalogue (r < 8)
    stars = rng.choice(n, 5, replace=False)
    df.loc[stars, "r_mag"] = np.round(rng.uniform(6.0, 7.8, 5), 3)
    df.loc[stars, "redshift"] = 0.0
    # three g_mag values typed as text in the source file -> the column is read as text
    bad = rng.choice(np.setdiff1d(np.arange(n), stars), 3, replace=False)
    df.loc[bad, "g_mag"] = "--"
    df.to_csv(RAW / "galaxy10_metadata.csv", index=False)
    pd.DataFrame({"image_id": df["image_id"], "z_true": np.round(z_true, 4),
                  "is_star": np.isin(np.arange(n), stars)}).to_csv(RAW / "galaxy10_truth.csv",
                                                                   index=False)

    # image index: 20 galaxies have no image file, 15 image files have no metadata row
    keep = np.sort(rng.choice(n, n - 20, replace=False))
    ids = [f"galaxy_{i:05d}" for i in keep] + [f"galaxy_{i:05d}" for i in range(n, n + 15)]
    pd.DataFrame({"image_id": ids,
                  "image_file": [f"images/{i}.png" for i in ids]}).to_csv(
        RAW / "galaxy10_image_index.csv", index=False)
    return df, z_true


def make_messy(df, n=500):
    m = df.sample(n, random_state=SEED).reset_index(drop=True).copy()
    m["g_mag"] = pd.to_numeric(m["g_mag"], errors="coerce")
    idx = rng.permutation(n)
    m.loc[idx[:20], "redshift"] = -999          # sentinel: "could not be measured"
    m.loc[idx[20:25], "r_mag"] = -999
    m.loc[idx[25:31], "r_mag"] = np.round(rng.uniform(25.5, 28.0, 6), 3)   # too faint
    m.loc[idx[31:33], "redshift"] = np.round(-rng.uniform(0.01, 0.05, 2), 4)  # sign typo
    m.to_csv(RAW / "galaxy10_messy.csv", index=False)


def make_images(n=4, size=256):
    y, x = np.mgrid[0:size, 0:size]
    imgs = np.zeros((n, size, size, 3), dtype=np.float32)
    for k in range(n):
        x0, y0 = 128 + rng.normal(0, 3, 2)
        q, pa = rng.uniform(0.5, 1.0), rng.uniform(0, np.pi)
        xr = (x - x0) * np.cos(pa) + (y - y0) * np.sin(pa)
        yr = -(x - x0) * np.sin(pa) + (y - y0) * np.cos(pa)
        r = np.hypot(xr, yr / q)
        for b, amp in enumerate((0.6, 1.0, 1.3)):                  # g, r, z bands
            bulge = amp * 60 * np.exp(-7.67 * ((r / 6) ** 0.25 - 1)) / np.exp(7.67)
            disk = amp * 8 * np.exp(-r / 25)
            imgs[k, :, :, b] = bulge + disk + rng.normal(0, 0.5, (size, size))
    imgs[0, 40, 200, 0] = 60.0        # two hot pixels in image 0, band 0
    imgs[0, 210, 35, 0] = 45.0
    np.save(RAW / "galaxy_images.npy", imgs)


# --------------------------------------------------------------- AFM
def afm_curve(k_sample, adhesion_nN, z_contact=20.0, snap=15.0, offset=0.8, noise=0.05,
              n=301):
    """One approach + retract force-distance curve [nN vs nm].
    distance_nm is the piezo height above the contact point + z_contact."""
    z = np.linspace(300.0, 0.0, n)
    appr = np.where(z < z_contact, k_sample * (z_contact - z), 0.0)
    z_r = z[::-1]
    retr = np.where(z_r < z_contact, k_sample * (z_contact - z_r), 0.0)
    # adhesion: the tip stays stuck and is pulled down to -adhesion before snapping off
    stick = (z_r >= z_contact) & (z_r <= z_contact + snap)
    retr = np.where(stick, -adhesion_nN * (z_r - z_contact) / snap, retr)
    d = np.concatenate([z, z_r])
    f = np.concatenate([appr, retr]) + offset + rng.normal(0, noise, 2 * n)
    seg = ["approach"] * n + ["retract"] * n
    return pd.DataFrame({"segment": seg, "distance_nm": np.round(d, 3),
                         "force_nN": np.round(f, 4)})


def make_afm():
    # time series: 2 s at 1 kHz while the tip scans a surface
    t = np.arange(0, 2.0, 0.001)
    topo = 0.3 * np.sin(2 * np.pi * 1.5 * t)
    drift = 0.4 * t                                  # slow thermal drift [nN/s]
    f = 2.0 + topo + drift + rng.normal(0, 0.03, t.size)
    spikes = rng.choice(t.size, 15, replace=False)
    f[spikes] += rng.choice([-1, 1], 15) * rng.uniform(1.0, 2.5, 15)
    pd.DataFrame({"time_s": np.round(t, 4), "force_nN": np.round(f, 4)}).to_csv(
        RAW / "afm_timeseries.csv", index=False)
    pd.DataFrame({"spike_index": np.sort(spikes)}).to_csv(RAW / "afm_timeseries_spikes.csv",
                                                           index=False)

    afm_curve(k_sample=0.25, adhesion_nN=3.0).to_csv(RAW / "afm_spectrum.csv", index=False)

    rows, truth = [], []
    for cid in range(15):
        k_s = rng.uniform(0.1, 0.5)                  # nN/nm
        adh = rng.uniform(1.0, 5.0)                  # nN
        off = rng.normal(0.5, 0.3)
        c = afm_curve(k_s, adh, offset=off)
        c.insert(0, "curve_id", cid)
        rows.append(c)
        truth.append({"curve_id": cid, "slope_nN_per_nm": k_s, "adhesion_nN": adh,
                      "work_aJ": 0.5 * adh * 15.0, "baseline_nN": off})
    pd.concat(rows).to_csv(RAW / "afm_curves.csv", index=False)
    pd.DataFrame(truth).round(4).to_csv(RAW / "afm_curves_truth.csv", index=False)


# --------------------------------------------------------------- 4-lepton events
def two_body(M, m1, m2, n):
    """Momenta of two daughters in the parent rest frame, isotropic."""
    p = np.sqrt((M**2 - (m1 + m2)**2) * (M**2 - (m1 - m2)**2)) / (2 * M)
    cth = rng.uniform(-1, 1, n); phi = rng.uniform(0, 2 * np.pi, n)
    sth = np.sqrt(1 - cth**2)
    v = np.stack([p * sth * np.cos(phi), p * sth * np.sin(phi), p * cth], axis=1)
    E1 = np.sqrt(p**2 + m1**2); E2 = np.sqrt(p**2 + m2**2)
    return np.column_stack([E1, v]), np.column_stack([E2, -v])


def boost(P, beta):
    """Lorentz boost of four-vectors P (n,4) by velocity vectors beta (n,3)."""
    b2 = np.sum(beta**2, axis=1)
    g = 1 / np.sqrt(1 - b2)
    bp = np.sum(beta * P[:, 1:], axis=1)
    E = g * (P[:, 0] + bp)
    coef = np.where(b2 > 0, (g - 1) * bp / np.where(b2 > 0, b2, 1), 0.0) + g * P[:, 0]
    return np.column_stack([E, P[:, 1:] + coef[:, None] * beta])


def make_4l(n_sig=400, n_bkg=1600):
    M = np.concatenate([rng.normal(125.0, 1.8, n_sig),              # Higgs + resolution
                        70 + rng.gamma(2.5, 45.0, n_bkg)])            # smooth background
    n = M.size
    mZ1 = np.minimum(rng.normal(91.19, 2.5, n), M - 13)
    mZ2 = np.clip(rng.uniform(12, 60, n), 12, M - mZ1 - 1)
    Z1, Z2 = two_body(M, mZ1, mZ2, n)
    leps = []
    for Z, mZ in ((Z1, mZ1), (Z2, mZ2)):
        l1, l2 = two_body(mZ, 0.0, 0.0, n)
        beta = Z[:, 1:] / Z[:, [0]]
        leps += [boost(l1, beta), boost(l2, beta)]
    # boost the whole system along the beam (z) and give it a little transverse kick
    beta_lab = np.column_stack([rng.normal(0, 0.05, n), rng.normal(0, 0.05, n),
                                rng.uniform(-0.7, 0.7, n)])
    leps = [boost(l, beta_lab) for l in leps]
    order = rng.permutation(n)
    cols = {}
    for i, l in enumerate(leps, start=1):
        for j, c in enumerate(("E", "px", "py", "pz")):
            cols[f"{c}{i}"] = np.round(l[order, j], 4)
    ev = pd.DataFrame(cols)
    ev.insert(0, "is_signal", (np.arange(n) < n_sig)[order])
    ev.to_csv(RAW / "higgs_4l_events.csv", index=False)


if __name__ == "__main__":
    df, _ = make_galaxies()
    make_messy(df)
    make_images()
    make_afm()
    make_4l()
    for p in sorted(RAW.iterdir()):
        print(f"{p.name:30s} {p.stat().st_size / 1024:8.1f} kB")
'''
if not os.path.exists("data/raw/higgs_4l_events.csv"):
    exec(compile(SRC, "make_raw_data.py", "exec"),
         {"__name__": "__main__", "__file__": os.path.abspath("make_raw_data.py")})
else:
    print("มีข้อมูลจำลองอยู่แล้ว ไม่สร้างใหม่")
print("ไฟล์ใน data/raw:", sorted(os.listdir("data/raw")))
```

ตารางกาแล็กซีในบทนี้มีคอลัมน์ `redshift` (เรดชิฟต์) `r_mag` และ `g_mag` (โชติมาตรปรากฏในแถบสี r และ g) และ `galaxy_class` (ประเภทของกาแล็กซี) โชติมาตรยิ่งมาก วัตถุยิ่งจาง ขั้นแรกของการทำงานกับข้อมูลตารางเสมอคือนับว่าแต่ละคอลัมน์มีค่าขาดหาย (missing value) เท่าใด

### โค้ดที่ 3.1 — นับค่าขาดหายในแต่ละคอลัมน์

```python
import pandas as pd
import numpy as np

df = pd.read_csv('data/raw/galaxy10_metadata.csv')
print(df.shape)              # (rows, columns)
print(df.isna().sum())       # number of NaN in each column
```

### โค้ดที่ 3.2 — ทำเครื่องหมายแทนการลบ

```python
df['has_redshift'] = df['redshift'].notna()
print(f"with redshift: {df['has_redshift'].sum()} / {len(df)}")
print(f"missing fraction: {100 * (1 - df['has_redshift'].mean()):.1f} %")
```

### ค่าหายไปอย่างสุ่มหรือไม่
ก่อนตัดสินใจทำอะไรกับค่าที่หาย ต้องถามก่อนว่า **ทำไมค่าจึงหาย** นักสถิติแบ่งกลไกการขาดหายเป็นสามแบบ

| กลไก | ความหมาย | ตัวอย่าง |
|---|---|---|
| ขาดหายอย่างสุ่มสมบูรณ์ (MCAR) | การหายไม่เกี่ยวกับสิ่งใดเลย | ไฟล์บางส่วนเสียระหว่างส่ง |
| ขาดหายอย่างสุ่มเมื่อทราบตัวแปรอื่น (MAR) | การหายขึ้นกับตัวแปรอื่นที่เรามีข้อมูล | เรดชิฟต์หายบ่อยในกาแล็กซีที่จาง (รู้ค่า `r_mag`) |
| ขาดหายอย่างไม่สุ่ม (MNAR) | การหายขึ้นกับค่าที่หายไปเอง | กาแล็กซีเรดชิฟต์สูงวัดเรดชิฟต์ไม่ได้ |

เซลล์ถัดไปตรวจว่าการหายของเรดชิฟต์สัมพันธ์กับความสว่างหรือไม่ โดยเปรียบเทียบสัดส่วนที่หายในแต่ละช่วงของ `r_mag`

```python
r = df['r_mag']
print("median r_mag with redshift   :", round(r[df['has_redshift']].median(), 2))
print("median r_mag without redshift:", round(r[~df['has_redshift']].median(), 2))

bins = pd.cut(r, [0, 8, 16, 17, 18, 19, 20, 30])
frac_missing = 1 - df.groupby(bins, observed=True)['has_redshift'].mean()
print("\nmissing fraction by r_mag bin (larger r_mag = fainter):")
print(frac_missing.round(3).to_string())

fig, ax = plt.subplots(figsize=(7, 3.2))
ax.bar(range(len(frac_missing)), frac_missing.values, color="tab:blue")
ax.set_xticks(range(len(frac_missing))); ax.set_xticklabels([str(b) for b in frac_missing.index])
ax.set_xlabel("r_mag bin"); ax.set_ylabel("fraction without redshift")
ax.set_title("SYNTHETIC: redshift is missing mostly for faint galaxies")
plt.tight_layout(); plt.show()
```

ค่าเรดชิฟต์หายเกือบทั้งหมดในกาแล็กซีที่จาง การขาดหายจึง **ไม่ใช่** แบบสุ่มสมบูรณ์ กาแล็กซีที่จางส่วนใหญ่อยู่ไกล ถ้าลบแถวที่ไม่มีเรดชิฟต์ทิ้ง ตัวอย่างที่เหลือจะเอียงไปทางกาแล็กซีที่สว่างและอยู่ใกล้ นักดาราศาสตร์เรียกอคติจากการเลือกตัวอย่างตามความสว่างแบบนี้ว่าอคติจากขีดจำกัดฟลักซ์ (flux-limited selection) ซึ่งเกี่ยวข้องใกล้ชิดกับอคติแบบมาล์มควิสต์ (Malmquist bias) ส่วนที่ 10 จะวัดขนาดของอคตินี้เทียบกับค่าจริง

# ส่วนที่ 3 อนุกรมเวลา: ค่ากระโดด การเลื่อน และการแฝงนาม (หัวข้อ 3.2.3)

ตัวอย่างนี้คือสัญญาณแรงที่ปลายเข็มของกล้องจุลทรรศน์แรงอะตอม (AFM) ขณะเคลื่อนผ่านพื้นผิว บันทึก 2 วินาทีที่อัตรา 1,000 จุดต่อวินาที ข้อมูลจำลองนี้มีปัญหาสามชนิดปนอยู่

| ปัญหา | ลักษณะ | วิธีแก้ในตำรา |
|---|---|---|
| ค่ากระโดด (spike) | ค่าผิดแผกจากข้างเคียงเพียงจุดเดียว | ตัวกรองมัธยฐาน (median filter) |
| การเลื่อน (drift) | ค่ากลางค่อย ๆ เปลี่ยนตามเวลา เช่น จากอุณหภูมิ | การลบแนวโน้ม (detrending) |
| สัญญาณรบกวน | การแกว่งแบบสุ่มเล็ก ๆ | ปรับให้เรียบ (ส่วนที่ 5) |

ตัวกรองมัธยฐานแทนค่าแต่ละจุดด้วยค่ามัธยฐานของจุดในหน้าต่างรอบตัว ค่ามัธยฐานไม่หวั่นไหวต่อค่าสุดโต่งเพียงค่าเดียว เหมือนการหาเงินเดือน "ทั่วไป" ของคนในห้อง ถ้ามีมหาเศรษฐีเดินเข้ามาหนึ่งคน ค่าเฉลี่ยจะพุ่งขึ้นทันที แต่ค่ามัธยฐานแทบไม่ขยับ

### โค้ดที่ 3.3 — กรองค่ากระโดดด้วยตัวกรองมัธยฐานแล้วลบแนวโน้ม

```python
afm = pd.read_csv('data/raw/afm_timeseries.csv')

from scipy.signal import medfilt, detrend

afm['force_clean'] = medfilt(afm['force_nN'].values, kernel_size=5)
afm['force_detrended'] = detrend(afm['force_clean'].values, type='linear')
```

### ตรวจผลเทียบกับความจริง
ข้อมูลจำลองเก็บตำแหน่งของค่ากระโดด 15 จุดที่ใส่ไว้ในไฟล์ `afm_timeseries_spikes.csv` เราจึงตรวจได้ว่าตัวกรองกำจัดค่ากระโดดได้จริงหรือไม่ และเปรียบเทียบกับค่าเฉลี่ยเคลื่อนที่ (moving average) ที่ความกว้างหน้าต่างเท่ากัน

```python
spikes = pd.read_csv('data/raw/afm_timeseries_spikes.csv')['spike_index'].values
raw, clean = afm['force_nN'].values, afm['force_clean'].values
resid = raw - clean
print(f"{len(afm)} samples; {len(spikes)} spikes were injected")
print(f"|raw - median-filtered| at the spikes    : {np.abs(resid[spikes]).min():.2f} to {np.abs(resid[spikes]).max():.2f} nN")
print(f"|raw - median-filtered| elsewhere (95th %): {np.percentile(np.abs(np.delete(resid, spikes)), 95):.3f} nN")

mov = pd.Series(raw).rolling(5, center=True, min_periods=1).mean().values
worst = spikes[np.argmax(np.abs(resid[spikes]))]
print(f"\nat the largest spike: raw {raw[worst]:.2f}, median {clean[worst]:.2f}, "
      f"moving average {mov[worst]:.2f} nN")

slope = np.polyfit(afm['time_s'], afm['force_clean'], 1)[0]
print(f"\nstraight line removed by detrend: slope {slope:.3f} nN/s (the drift put in was 0.40 nN/s)")
```

ผลข้างบนบอกสามเรื่อง เรื่องแรก ตัวกรองมัธยฐานแยกค่ากระโดดออกจากสัญญาณได้ชัด ส่วนต่างที่จุดค่ากระโดดใหญ่กว่าส่วนต่างที่จุดอื่นหลายสิบเท่า เรื่องที่สอง ค่าเฉลี่ยเคลื่อนที่ไม่ได้กำจัดค่ากระโดด แต่กระจายความผิดปกติไปยังจุดข้างเคียง เรื่องที่สาม `detrend` ลบเส้นตรงที่มีความชันต่างจากการเลื่อนจริง เพราะในสองวินาทีมีสัญญาณภูมิประเทศเพียงสามรอบ เส้นตรงที่ปรับได้จึงเอียงตามสัญญาณจริงด้วย `detrend` จึงลบสัญญาณจริงไปบางส่วน ทุกวิธีมีข้อแลกเปลี่ยน และต้องวาดกราฟก่อนและหลังเสมอ

### การแฝงนาม (aliasing): ปัญหาที่แก้ภายหลังไม่ได้
เมื่อสัญญาณมีองค์ประกอบความถี่สูงกว่าครึ่งหนึ่งของอัตราสุ่มตัวอย่าง ซึ่งเรียกว่าความถี่ไนควิสต์ (Nyquist frequency) องค์ประกอบนั้นจะปรากฏในข้อมูลเป็นความถี่ต่ำปลอม เหมือนล้อรถในภาพยนตร์ที่ดูเหมือนหมุนช้าหรือหมุนถอยหลัง เซลล์ถัดไปสุ่มตัวอย่างโคไซน์ความถี่ 90 Hz และ 10 Hz ด้วยอัตรา 100 จุดต่อวินาที

```python
fs = 100.0                                   # samples per second -> Nyquist frequency 50 Hz
t = np.arange(0, 0.2, 1 / fs)
t_fine = np.linspace(0, 0.2, 2000)
f_high, f_low = 90.0, 10.0

fig, ax = plt.subplots(figsize=(9, 3.2))
ax.plot(t_fine, np.cos(2 * np.pi * f_high * t_fine), lw=0.6, color="0.6", label=f"{f_high:.0f} Hz (true signal)")
ax.plot(t_fine, np.cos(2 * np.pi * f_low * t_fine), lw=1.2, color="tab:blue", label=f"{f_low:.0f} Hz")
ax.plot(t, np.cos(2 * np.pi * f_high * t), "o", color="tab:red", label=f"samples of {f_high:.0f} Hz at {fs:.0f} Hz")
ax.set_xlabel("time (s)"); ax.legend(loc="lower right", fontsize=8)
ax.set_title("aliasing: the samples cannot tell 90 Hz from 10 Hz")
plt.tight_layout(); plt.show()

x90 = np.cos(2 * np.pi * 90 * np.arange(0, 1, 1 / fs))
x10 = np.cos(2 * np.pi * 10 * np.arange(0, 1, 1 / fs))
print(f"largest difference between the two sampled signals: {np.abs(x90 - x10).max():.1e}")
```

จุดสีแดงที่ได้จากสัญญาณ 90 Hz ตกบนเส้นโคไซน์ 10 Hz ทุกจุด ผลต่างระหว่างสองชุดเป็นเพียงความคลาดเคลื่อนจากการปัดเศษ เมื่อสุ่มตัวอย่างแล้ว ไม่มีตัวกรองใดแยกสองสัญญาณนี้ออกจากกันได้ การแฝงนามจึงต้องป้องกันตั้งแต่ขั้นการวัด ด้วยการสุ่มตัวอย่างเร็วพอ หรือใส่ตัวกรองแอนะล็อกตัดความถี่สูงก่อนแปลงสัญญาณเป็นดิจิทัล

### ✏️ แบบฝึก 3.B คำนวณความถี่แฝงนาม
ความถี่ $f$ ที่สุ่มด้วยอัตรา $f_s$ จะปรากฏที่ความถี่ $|f - n f_s|$ โดยเลือกจำนวนเต็ม $n$ ที่ทำให้ผลอยู่ระหว่าง 0 ถึง $f_s/2$ จงคำนวณด้วยมือว่า การสั่น **60 Hz** และ **130 Hz** ที่บันทึกด้วยอัตรา 100 จุดต่อวินาที จะปรากฏที่ความถี่ใด แล้วเติมคำตอบในเซลล์ถัดไป เซลล์ ✅ จะตรวจด้วยการสุ่มตัวอย่างจริง

```python
my_alias_60 = None     # Hz
my_alias_130 = None    # Hz
```

```python
#@title ✅ ตรวจคำตอบด้วยการหาความถี่ของยอดในสเปกตรัมของสัญญาณที่สุ่มแล้ว
_say = _gate(['my_alias_130', 'my_alias_60'])
def apparent_frequency(f, fs=100.0, T=10.0):
    x = np.cos(2 * np.pi * f * np.arange(0, T, 1 / fs))
    spec = np.abs(np.fft.rfft(x))
    return np.fft.rfftfreq(x.size, 1 / fs)[np.argmax(spec)]

for f, mine in ((60, my_alias_60), (130, my_alias_130)):
    true = apparent_frequency(f)
    if mine is None:
        _say(f"{f} Hz: ยังไม่ได้เติมคำตอบ")
    else:
        _say(f"{f} Hz -> {true:.0f} Hz;", "✔" if abs(mine - true) < 0.5 else f"✘ ได้ {mine} Hz")
```

```python
images = np.load('data/raw/galaxy_images.npy')
print("images:", images.shape, images.dtype)

from scipy.ndimage import gaussian_filter

img = images[0, :, :, 0].astype(float)      # first image, first band
threshold = img.mean() + 5 * img.std()
hot_mask = img > threshold

img_smooth = gaussian_filter(img, sigma=1.0)
```

เกณฑ์ "ค่าเฉลี่ยบวกห้าเท่าของส่วนเบี่ยงเบนมาตรฐาน" ต้องใช้อย่างระวัง เซลล์ถัดไปตรวจว่าพิกเซลที่ผ่านเกณฑ์อยู่ที่ใด

```python
from scipy.ndimage import median_filter

ys, xs = np.nonzero(hot_mask)
near_centre = np.hypot(xs - 128, ys - 128) < 10
print(f"threshold = {threshold:.2f}; pixels above it: {hot_mask.sum()}")
print(f"  in the galaxy core (real light): {near_centre.sum()}, elsewhere: {(~near_centre).sum()}")

# better test: bright AND very different from the median of its neighbours
local = median_filter(img, size=3)
noise = 1.4826 * np.median(np.abs(img - np.median(img)))       # robust sigma
hot = (img - local) > 10 * noise
print(f"neighbour test: {hot.sum()} pixels at {[tuple(int(v) for v in p) for p in zip(*np.nonzero(hot))]}")
fixed = np.where(hot, local, img)

fig, ax = plt.subplots(1, 3, figsize=(11, 3.8))
ax[0].imshow(np.arcsinh(img), cmap="gray"); ax[0].set_title("band 0 (asinh stretch)")
hy, hx = np.nonzero(hot)
ax[0].scatter(hx, hy, s=120, facecolors="none", edgecolors="tab:red", lw=1.5)   # circle the hot pixels
ax[1].imshow(hot_mask, cmap="Reds"); ax[1].set_title(f"5-sigma cut: {hot_mask.sum()} px")
ax[2].imshow(hot, cmap="Reds"); ax[2].set_title(f"neighbour test: {hot.sum()} px")
for a in ax: a.set_xticks([]); a.set_yticks([]); a.grid(False)
fig.suptitle("SYNTHETIC galaxy image"); plt.tight_layout(); plt.show()
```

เกณฑ์ 5σ จับใจกลางกาแล็กซีซึ่งเป็นแสงจริงมาเกือบทั้งหมด พิกเซลร้อนต่างจากใจกลางกาแล็กซีตรงที่มันสว่างอยู่เพียงพิกเซลเดียว ขณะที่แสงจริงจากวัตถุท้องฟ้ากระจายไปยังพิกเซลข้างเคียงตามฟังก์ชันการกระจายของจุด (point spread function) การทดสอบที่เทียบกับค่ามัธยฐานของพิกเซลข้างเคียงจึงจับพิกเซลร้อนทั้งสองจุดได้โดยไม่แตะใจกลางกาแล็กซี

ต่อไปดูราคาของการปรับให้เรียบด้วยตัวกรองเกาส์ และการแก้ขอบภาพมืดด้วยการหารด้วยภาพสนามแบน (flat field) ที่ถ่ายจากแหล่งแสงสม่ำเสมอ

```python
print(f"Gaussian smoothing (sigma = 1 px): noise std {np.std(img[:30, :30]):.3f} -> "
      f"{np.std(img_smooth[:30, :30]):.3f}, peak {img[118:138, 118:138].max():.1f} -> "
      f"{img_smooth[118:138, 118:138].max():.1f}")

yy, xx = np.mgrid[0:256, 0:256]
rr = np.hypot(xx - 127.5, yy - 127.5) / 181.0            # 0 at centre, 1 at the corners
vignette = 1 - 0.4 * rr**2                                # corners 40 % darker
observed = fixed * vignette
flat = vignette * (1 + np.random.default_rng(4).normal(0, 0.002, vignette.shape))  # flat-field frame
corrected = observed / flat
corner, centre = (slice(0, 30), slice(0, 30)), (slice(113, 143), slice(113, 143))
print(f"vignetting: corner/centre ratio {vignette[corner].mean() / vignette[centre].mean():.2f}; "
      f"after dividing by the flat field the image differs from the original by "
      f"{np.median(np.abs(corrected - fixed)[fixed > 2] / fixed[fixed > 2]):.1%} (median)")
```

ตัวกรองเกาส์ลดสัญญาณรบกวนได้ แต่ความสว่างสูงสุดที่ใจกลางก็ลดลงด้วย เพราะรายละเอียดที่เล็กกว่าความกว้างของตัวกรองถูกเกลี่ยออกไป ส่วนการหารด้วยภาพสนามแบนแก้ขอบภาพมืดได้เกือบสมบูรณ์ เพราะวิธีนี้วัดผลของเครื่องมือโดยตรง ไม่ได้เดาจากภาพของวัตถุ

# ส่วนที่ 5 สเปกตรัม: เส้นฐานและตัวกรองซาวิตซกี–โกเลย์ (หัวข้อ 3.2.5)

เส้นโค้งแรงกับระยะของ AFM (force–distance curve) บันทึกแรงที่ปลายเข็มรับรู้ขณะเคลื่อนเข้าหาและถอยออกจากพื้นผิว เมื่อปลายเข็มยังอยู่ไกล แรงควรเป็นศูนย์ แต่ข้อมูลจริงมักมีเส้นฐานเลื่อน (baseline shift) เพราะแสงเลเซอร์ไม่ได้ตกตรงกลางตัวรับพอดี ขณะถอยออก ปลายเข็มติดอยู่กับพื้นผิวแล้วหลุดออกฉับพลัน เหตุการณ์หลุดติด (snap-off) นี้บอกแรงยึดติดระหว่างปลายเข็มกับพื้นผิว ซึ่งเป็นฟิสิกส์ที่เราต้องการ ในข้อมูลจำลอง แรงยึดติดจริงคือ 3.0 nN และเส้นฐานคือ 0.8 nN

ตัวกรองซาวิตซกี–โกเลย์ (Savitzky–Golay filter) ปรับพหุนามดีกรีต่ำให้เข้ากับจุดในหน้าต่างเล็ก ๆ ทีละหน้าต่าง แล้วใช้ค่าของพหุนามที่จุดกลางแทน จึงรักษาความสูงและตำแหน่งของยอดได้ดีกว่าค่าเฉลี่ยเคลื่อนที่

### โค้ดที่ 3.5 — ลบเส้นฐานแล้วปรับให้เรียบด้วยตัวกรองซาวิตซกี–โกเลย์

```python
afm_spec = pd.read_csv('data/raw/afm_spectrum.csv')

from scipy.signal import savgol_filter

# baseline: mean force where the tip is far from the surface
far_mask = afm_spec['distance_nm'] > 200
baseline = afm_spec.loc[far_mask, 'force_nN'].mean()
afm_spec['force_corrected'] = afm_spec['force_nN'] - baseline

afm_spec['force_smooth'] = savgol_filter(
    afm_spec['force_corrected'].values,
    window_length=11, polyorder=3
)
```

```python
print(f"baseline from distance > 200 nm: {baseline:.3f} nN (0.8 nN was put in)")
ret = afm_spec['segment'] == 'retract'
print(f"adhesion (minimum of the retract curve), true value 3.0 nN")
print(f"  corrected, not smoothed : {-afm_spec.loc[ret, 'force_corrected'].min():.2f} nN")
for w in (11, 31, 61):
    ma = afm_spec.loc[ret, 'force_corrected'].rolling(w, center=True, min_periods=1).mean()
    sg = savgol_filter(afm_spec.loc[ret, 'force_corrected'].values, w, 3)
    print(f"  window {w:2d}: moving average {-ma.min():.2f} nN, Savitzky-Golay {-sg.min():.2f} nN")

r_ = afm_spec[ret]
fig, ax = plt.subplots(figsize=(8, 3.8))
ax.plot(r_['distance_nm'], r_['force_corrected'], '.', ms=2, color="0.5", label='corrected')
ax.plot(r_['distance_nm'], r_['force_smooth'], lw=1.2, label='Savitzky-Golay (11, 3)')
ax.plot(r_['distance_nm'], savgol_filter(r_['force_corrected'].values, 61, 3), lw=1.2,
        label='Savitzky-Golay (61, 3): too wide')
ax.set_xlim(0, 80); ax.set_xlabel('distance (nm)'); ax.set_ylabel('force (nN)')
ax.set_title('retract curve before and after smoothing (SYNTHETIC)'); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
```

เส้นฐานที่ประมาณได้ใกล้ค่าที่ใส่ไว้มาก เพราะใช้เฉพาะช่วงที่รู้แน่ว่าปลายเข็มยังไม่สัมผัสพื้นผิว ส่วนแรงยึดติดลดลงเรื่อย ๆ เมื่อหน้าต่างของตัวกรองกว้างขึ้น และค่าเฉลี่ยเคลื่อนที่ลดมากกว่าซาวิตซกี–โกเลย์ที่หน้าต่างเท่ากัน ทุกครั้งที่ปรับให้เรียบ จึงควรวาดเส้นโค้งก่อนและหลังซ้อนกัน เพื่อยืนยันว่าลักษณะทางฟิสิกส์ที่สำคัญยังอยู่ครบ

### (เสริม) สอบเทียบค่าคงที่สปริงด้วยสัญญาณรบกวนเชิงความร้อน
การแปลงการโก่งของคานเข็มเป็นแรงต้องใช้ค่าคงที่สปริง $k$ วิธีสัญญาณรบกวนเชิงความร้อน (thermal noise method) อาศัยทฤษฎีบทการแบ่งพลังงานเท่า ๆ กัน $\tfrac{1}{2}k\langle x^2\rangle = \tfrac{1}{2}k_BT$ เซลล์ถัดไปจำลองการสั่นด้วยความร้อนของคานเข็มที่ $k = 0.1$ N/m แล้วประมาณ $k$ กลับจากความแปรปรวนของการสั่น

```python
from scipy.constants import k as k_B
T_K, k_true_Nm = 295.0, 0.1
x_m = np.random.default_rng(7).normal(0, np.sqrt(k_B * T_K / k_true_Nm), 20_000)   # thermal motion [m]
k_est = k_B * T_K / np.var(x_m)
print(f"rms thermal motion: {1e9 * x_m.std():.3f} nm;  k estimated = {k_est:.4f} N/m (true {k_true_Nm})")
```

### โค้ดที่ 3.7 — ตรวจว่าคอลัมน์ที่ควรเป็นตัวเลขเป็นตัวเลขจริง

คอลัมน์ `g_mag` ควรเป็นตัวเลข แต่ pandas อ่านเป็นข้อความ (`object` หรือ `str` ใน pandas รุ่น 3) การโหลดไฟล์ผ่านไปโดยไม่มีข้อความเตือนใด เซลล์ถัดไปหาต้นเหตุ แล้วแปลงด้วย `pd.to_numeric(..., errors='coerce')` ซึ่งเปลี่ยนค่าที่แปลงไม่ได้ให้เป็นค่าว่าง และนับว่ามีกี่ค่าที่กลายเป็นค่าว่างจากขั้นนี้

### ✏️ แบบฝึก E3.4 ส่วนที่ 1: คาดการณ์ก่อนคำนวณ
เราทราบแล้วว่าเรดชิฟต์หายไปเพราะกาแล็กซีจาง จงคาดการณ์ว่า **ค่าเฉลี่ย** ของเรดชิฟต์ที่ได้จากแต่ละวิธีจะ `"ต่ำกว่า"`, `"สูงกว่า"` หรือ `"ใกล้เคียง"` ค่าเฉลี่ยจริงของกาแล็กซีทุกดวง และส่วนเบี่ยงเบนมาตรฐานเมื่อเติมค่ามัธยฐานจะ `"ต่ำกว่า"`, `"สูงกว่า"` หรือ `"ใกล้เคียง"` ค่าจริง

```python
my_prediction = {
    "mean: drop rows": "?",
    "mean: fill median": "?",
    "mean: fill from r_mag": "?",
    "std: fill median": "?",
}
```

### โค้ดที่ 3.11 — สามวิธีเติมค่าเรดชิฟต์เพื่อเปรียบเทียบผล
เซลล์แรกเตรียมข้อมูลและโหลดเฉลย เซลล์ที่สองคือโค้ดในตำรา ความสัมพันธ์ระหว่างโชติมาตรกับเรดชิฟต์ในฟังก์ชัน `photo_z_estimate` เขียนขึ้นเพื่อการสอนเท่านั้น การประมาณเรดชิฟต์จากโฟโตเมตรี (photometric redshift) ในงานวิจัยจริงใช้ข้อมูลหลายแถบสีและแบบจำลองที่ซับซ้อนกว่านี้มาก

```python
truth = pd.read_csv('data/raw/galaxy10_truth.csv', index_col='image_id')
df = df[df['r_mag'] >= 8].copy()             # set the 5 foreground stars aside first
z_true = truth.loc[df.index, 'z_true']
print(f"{len(df)} galaxies, {df['redshift'].isna().sum()} missing redshifts ({100 * df['redshift'].isna().mean():.1f} %)")
```

```python
df['z_imp_mean']   = df['redshift'].fillna(df['redshift'].mean())
df['z_imp_median'] = df['redshift'].fillna(df['redshift'].median())

# toy photometric-redshift relation (for teaching only)
def photo_z_estimate(r_mag):
    return 0.05 * (r_mag - 15.0)

df['z_imp_photo'] = df['redshift'].fillna(photo_z_estimate(df['r_mag']))
```

### แบบฝึก E3.4 ส่วนที่ 2: เทียบกับค่าจริง
ข้อมูลจำลองให้เรารู้เรดชิฟต์จริงของทุกดวง จึงวัดอคติของแต่ละวิธีได้โดยตรง แถวสุดท้ายเพิ่มกรณีที่ความสัมพันธ์ระหว่างโชติมาตรกับเรดชิฟต์ **ผิด** โดยความชันคลาดไปร้อยละ 20 เพื่อดูว่าถ้าแบบจำลองที่ใช้เติมค่าผิด ผลจะเป็นอย่างไร

```python
wrong = lambda r: 0.04 * (r - 15.0)          # slope 20 % too small
methods = {
    'truth (all galaxies)':      z_true,
    'drop rows':                 df['redshift'].dropna(),
    'fill mean':                 df['z_imp_mean'],
    'fill median':               df['z_imp_median'],
    'fill from r_mag (right)':   df['z_imp_photo'],
    'fill from r_mag (wrong)':   df['redshift'].fillna(wrong(df['r_mag'])),
}
cmp_ = pd.DataFrame({k: {'N': z.size, 'mean z': z.mean(), 'std z': z.std(),
                         'bias of mean': z.mean() - z_true.mean()} for k, z in methods.items()}).T
cmp_['N'] = cmp_['N'].astype(int)
cmp_.round(4)
```

```python
#@title ✅ ตรวจการคาดการณ์ (เกณฑ์ "ใกล้เคียง" = คลาดไม่เกิน 0.004)
_say = _gate(['my_prediction'])
def _direction(x, ref, tol=0.004):
    return "ใกล้เคียง" if abs(x - ref) <= tol else ("สูงกว่า" if x > ref else "ต่ำกว่า")

_actual = {
    "mean: drop rows":       _direction(cmp_.loc['drop rows', 'mean z'], z_true.mean()),
    "mean: fill median":     _direction(cmp_.loc['fill median', 'mean z'], z_true.mean()),
    "mean: fill from r_mag": _direction(cmp_.loc['fill from r_mag (right)', 'mean z'], z_true.mean()),
    "std: fill median":      _direction(cmp_.loc['fill median', 'std z'], z_true.std()),
}
for k, v in _actual.items():
    mine = my_prediction[k].strip()
    _say(f"{k:22s} คาดว่า {mine:9s} ผลจริง {v:9s}", "✔" if mine == v else "✘")
```

**สิ่งที่ควรสังเกต** การลบแถวทำให้ค่าเฉลี่ยต่ำกว่าความจริง เพราะกาแล็กซีที่ถูกลบคือกาแล็กซีจางซึ่งส่วนใหญ่อยู่ไกลและมีเรดชิฟต์สูง การเติมค่ามัธยฐานไม่ได้แก้อคตินี้ และยังทำให้ส่วนเบี่ยงเบนมาตรฐานลดลงชัดเจน เพราะค่าเดียวถูกใส่ซ้ำหลายร้อยแถว การเติมจากโชติมาตรให้ผลใกล้ความจริงที่สุดเมื่อความสัมพันธ์ถูกต้อง (ในโลกจำลองนี้ถูกต้องโดยการสร้าง) แต่ถ้าความสัมพันธ์ผิด ความผิดนั้นจะเข้าไปอยู่ในทุกค่าที่เติม

การรายงานผลของหลายวิธีประกอบกันเรียกว่าการวิเคราะห์ความไว (sensitivity analysis) ถ้าข้อสรุปเปลี่ยนไปตามวิธีเติมค่ามากกว่าความไม่แน่นอนของการวัด แสดงว่าข้อสรุปนั้นขึ้นกับสมมติฐานของเรามากกว่าข้อมูล

**ทฤษฎีบทพายของบักกิงแฮม (Buckingham π theorem)** ถ้าปรากฏการณ์หนึ่งเกี่ยวข้องกับตัวแปรทางกายภาพ $n$ ตัว ซึ่งมีมิติพื้นฐานที่อิสระต่อกัน $k$ มิติ เราสร้างกลุ่มไร้มิติ (dimensionless group) ที่อิสระต่อกันได้ $n-k$ กลุ่ม และความสัมพันธ์ทางฟิสิกส์ทั้งหมดเขียนใหม่ได้ด้วยกลุ่มเหล่านี้

ของไหลที่ไหลผ่านท่อมีตัวแปรหลักสามตัว คือ อัตราเร็วเฉลี่ย $v$ (m/s) ขนาดของท่อ $L$ (m) และความหนืดจลน์ $\nu$ (m²/s) ซึ่งมีมิติพื้นฐานสองมิติ คือความยาวและเวลา จึงสร้างกลุ่มไร้มิติได้ $3-2=1$ กลุ่ม คือจำนวนเรย์โนลด์ส $\mathrm{Re} = vL/\nu$ ซึ่งบอกอัตราส่วนระหว่างแรงเฉื่อยกับแรงหนืด

### โค้ดที่ 3.14 — สร้างลักษณะเด่นจำนวนเรย์โนลด์สจากข้อมูลการไหลของน้ำ

```python
flow_df = pd.DataFrame({
    'v_ms':       [0.005, 0.10, 0.02, 0.015, 0.08],
    'L_m':        [0.10,  0.01, 0.50, 0.02,  0.08],
    'nu_m2s':     [1e-6] * 5,
    'flow_class': ['laminar', 'laminar', 'turbulent', 'laminar', 'turbulent'],
})
flow_df['Re'] = flow_df['v_ms'] * flow_df['L_m'] / flow_df['nu_m2s']
# Re = 500, 1000, 10000, 300, 6400
flow_df
```

```python
def separable(col):
    lam = flow_df.loc[flow_df.flow_class == 'laminar', col]
    tur = flow_df.loc[flow_df.flow_class == 'turbulent', col]
    return lam.max() < tur.min() or tur.max() < lam.min()

fig, ax = plt.subplots(1, 3, figsize=(10, 2.6))
for a, col in zip(ax, ('v_ms', 'L_m', 'Re')):
    for cls_, c, y in (('laminar', 'tab:blue', 0), ('turbulent', 'tab:red', 1)):
        vals = flow_df.loc[flow_df.flow_class == cls_, col]
        a.plot(vals, [y] * len(vals), 'o', color=c, ms=8, label=cls_)
    a.set_xscale('log'); a.set_yticks([0, 1]); a.set_yticklabels(['laminar', 'turbulent'])
    a.set_ylim(-0.6, 1.6); a.set_title(f"{col}: one threshold works? {separable(col)}", fontsize=9)
plt.tight_layout(); plt.show()
```

เมื่อเรียงข้อมูลตาม $\mathrm{Re}$ ประเภทของการไหลแยกกันได้ด้วยเกณฑ์เดียว ขณะที่ $v$ หรือ $L$ อย่างเดียวแยกไม่ได้ ลักษณะเด่นนี้ไม่ขึ้นกับระบบหน่วย มีความหมายทางทฤษฎี และลดจำนวนตัวแปรจากสามเหลือหนึ่ง อย่างไรก็ตาม ค่าวิกฤตที่มักอ้างราว 2,300 ใช้ได้กับท่อกลมที่เรียบเมื่อใช้เส้นผ่านศูนย์กลางเป็น $L$ เท่านั้น ตัวเลขไร้มิติไม่มีหน่วยก็จริง แต่การตีความค่าของมันยังผูกกับนิยามและเงื่อนไขที่ใช้สร้างมันเสมอ

| ลักษณะเด่น | นิยาม | ไม่แปรเปลี่ยนต่อ |
|---|---|---|

| ความเข้มข้น | $C = 5\log_{10}(r_{80}/r_{20})$ | การหมุน การย่อขยาย |
| ความไม่สมมาตร | $A = \sum\lvert I - I_{180^\circ}\rvert / \sum\lvert I\rvert$ วัดรอบใจกลาง | การหมุน |

### โค้ดที่ 3.15 — ลักษณะเด่นที่ไม่แปรเปลี่ยนต่อการหมุน

```python
def total_flux(image):
    return image.sum()

def radial_profile(image, center=None):
    if center is None:
        center = ((image.shape[0] - 1) / 2, (image.shape[1] - 1) / 2)
    y, x = np.indices(image.shape)
    r = np.sqrt((x - center[1])**2 + (y - center[0])**2).astype(int)
    profile = np.bincount(r.ravel(), weights=image.ravel())
    counts  = np.bincount(r.ravel())
    return profile / np.maximum(counts, 1)

img     = np.random.default_rng(2026).normal(size=(64, 64))
img_rot = np.rot90(img)
print(np.isclose(total_flux(img), total_flux(img_rot)))   # True
print(np.allclose(radial_profile(img), radial_profile(img_rot)))   # True
```

### ✏️ แบบฝึก 3.G ปริมาณใดไม่แปรเปลี่ยนเมื่อระบบเคลื่อนที่ตามแนวลำอนุภาค
เซลล์ ✅ ถัดไปจะแปลงลอเรนตซ์ (Lorentz boost) เหตุการณ์ทั้งหมดไปตามแกน $z$ ด้วยความเร็ว $0.5c$ แล้วเปรียบเทียบปริมาณสี่ตัวก่อนและหลัง **ก่อนรัน** ให้ทายว่าปริมาณใด `"เปลี่ยน"` หรือ `"ไม่เปลี่ยน"` ได้แก่ พลังงานรวม $E$, โมเมนตัมตามแนวลำ $p_z$, โมเมนตัมตามขวางของเลปตอนตัวแรก $p_T = \sqrt{p_x^2 + p_y^2}$ และมวลยืนยง $M_{\text{inv}}$

```python
my_boost = {"E": "?", "pz": "?", "pT_lepton1": "?", "M_inv": "?"}
```

```python
#@title ✅ ตรวจคำตอบด้วยการแปลงลอเรนตซ์จริง
_say = _gate(['my_boost'])
beta = 0.5
gamma = 1 / np.sqrt(1 - beta**2)
boosted = events.copy()
for i in range(1, 5):
    Ei, pzi = events[f'E{i}'], events[f'pz{i}']
    boosted[f'E{i}'] = gamma * (Ei + beta * pzi)
    boosted[f'pz{i}'] = gamma * (pzi + beta * Ei)

def quantities(ev):
    E_ = ev[['E1', 'E2', 'E3', 'E4']].sum(axis=1)
    px_ = ev[['px1', 'px2', 'px3', 'px4']].sum(axis=1)
    py_ = ev[['py1', 'py2', 'py3', 'py4']].sum(axis=1)
    pz_ = ev[['pz1', 'pz2', 'pz3', 'pz4']].sum(axis=1)
    return {"E": E_, "pz": pz_, "pT_lepton1": np.hypot(ev['px1'], ev['py1']),
            "M_inv": np.sqrt(np.clip(E_**2 - px_**2 - py_**2 - pz_**2, 0, None))}

before, after = quantities(events), quantities(boosted)
for k in before:
    changed = not np.allclose(before[k], after[k], rtol=1e-6, atol=1e-6)
    truth_ = "เปลี่ยน" if changed else "ไม่เปลี่ยน"
    _say(f"{k:11s} first event {before[k].iloc[0]:9.3f} -> {after[k].iloc[0]:9.3f}  ผลจริง: {truth_:9s}",
          "(ยังไม่ได้ทาย)" if my_boost[k] == "?" else ("✔" if my_boost[k] == truth_ else "✘"))
```

$p_T$ และ $M_{\text{inv}}$ ไม่เปลี่ยน จึงเป็นลักษณะเด่นที่ดีสำหรับการวิเคราะห์ข้อมูลจากเครื่องชนอนุภาค ซึ่งอนุภาคที่ชนกันมีโมเมนตัมตามแนวลำไม่เท่ากันในแต่ละเหตุการณ์ ส่วนพลังงานตามขวางที่หายไป (missing transverse energy) ใช้หลักเดียวกัน เพราะโมเมนตัมตามขวางรวมก่อนชนเป็นศูนย์ ส่วนที่ขาดไปจึงบ่งชี้อนุภาคที่ตัวตรวจวัดมองไม่เห็น เช่น นิวทริโน

```python
#@title ฟังก์ชันอ้างอิง (ชุดเดียวกับ ch03_cleaning/src/features.py) ดับเบิลคลิกเพื่อดูเฉลย
from scipy.signal import savgol_filter
_trapz = getattr(np, "trapezoid", None) or np.trapz

def remove_baseline(curve, far_nm=200.0):
    base = curve.loc[curve["distance_nm"] > far_nm, "force_nN"].mean()
    out = curve.copy()
    out["force_corrected"] = out["force_nN"] - base
    return out, base

def _ref_contact_slope(curve, f_min=0.5):
    a = curve[(curve["segment"] == "approach") & (curve["force_corrected"] > f_min)]
    return -np.polyfit(a["distance_nm"], a["force_corrected"], 1)[0]

def _ref_max_adhesion(curve, window=None):
    f = curve.loc[curve["segment"] == "retract", "force_corrected"].values
    if window:
        f = savgol_filter(f, window, 3)
    return -f.min()

def _ref_adhesion_work(curve, noise, n_sigma=3.0):
    r = curve[curve["segment"] == "retract"].sort_values("distance_nm")
    f = r["force_corrected"].values
    f = np.where(f < -n_sigma * noise, f, 0.0)
    return -_trapz(f, r["distance_nm"].values)
```

```python
curves = pd.read_csv('data/raw/afm_curves.csv')
truth_afm = pd.read_csv('data/raw/afm_curves_truth.csv', index_col='curve_id')

def _use(mine, ref, name):
    try:
        test = mine(*ref_args)
    except Exception as e:
        print(f"⚠ {name}: เกิดข้อผิดพลาด {type(e).__name__}: {e} จึงใช้ฟังก์ชันอ้างอิงแทน"); return ref
    if test is None:
        print(f"⚠ {name}: ยังไม่ได้เขียน จึงใช้ฟังก์ชันอ้างอิงแทน"); return ref
    print(f"✔ {name}: ใช้ฟังก์ชันของผู้เรียน"); return mine

c0, _ = remove_baseline(curves[curves.curve_id == 0])
noise0 = c0.loc[c0["distance_nm"] > 200, "force_corrected"].std()
ref_args = (c0,);        f_slope = _use(contact_slope, _ref_contact_slope, "contact_slope")
ref_args = (c0,);        f_adh   = _use(max_adhesion, _ref_max_adhesion, "max_adhesion")
ref_args = (c0, noise0); f_work  = _use(adhesion_work, _ref_adhesion_work, "adhesion_work")

rows = []
for cid, cv in curves.groupby('curve_id'):
    cv, base = remove_baseline(cv)
    noise = cv.loc[cv["distance_nm"] > 200, "force_corrected"].std()
    rows.append({'curve_id': cid, 'baseline_nN': base,
                 'slope_nN_per_nm': f_slope(cv), 'adhesion_nN': f_adh(cv),
                 'adhesion_nN_sg31': _ref_max_adhesion(cv, window=31),
                 'work_aJ': f_work(cv, noise)})
feat = pd.DataFrame(rows).set_index('curve_id')
feat.round(3).head()
```

```python
# ✅ ตรวจกับค่าที่ใช้สร้างข้อมูล
_who = {'baseline_nN': None, 'slope_nN_per_nm': (f_slope, _ref_contact_slope),
        'adhesion_nN': (f_adh, _ref_max_adhesion), 'work_aJ': (f_work, _ref_adhesion_work)}
for col, tol in (('baseline_nN', 0.02), ('slope_nN_per_nm', 0.02), ('adhesion_nN', 0.2), ('work_aJ', 3.0)):
    err = feat[col] - truth_afm[col]
    _fb = _who[col] is not None and _who[col][0] is _who[col][1]     # reference function used instead of the student's
    print(f"{col:16s}: mean error {err.mean():+.4f}, rms {np.sqrt((err**2).mean()):.4f}",
          "(ค่าจากฟังก์ชันอ้างอิง ไม่ได้ตรวจฟังก์ชันของผู้เรียน)" if _fb else
          ("✔" if np.sqrt((err**2).mean()) < tol else "✘ ตรวจฟังก์ชันอีกครั้ง"))
err_sg = feat['adhesion_nN_sg31'] - truth_afm['adhesion_nN']
print(f"\nadhesion after Savitzky-Golay (31): mean error {err_sg.mean():+.3f} nN -> smoothing biases the peak low")
extra = 0.5 * truth_afm['adhesion_nN'] * 1.0
print(f"work: the trapezoid rule adds about 0.5 x adhesion x 1 nm = {extra.mean():.2f} aJ on average at the snap-off")

feat.round(4).to_csv('data/processed/afm_features.csv')
print("saved data/processed/afm_features.csv")
```

แบบประเมินคุณภาพข้อมูล (data quality scorecard) แปลงสี่มิตินี้เป็นรายการคำถาม และเพิ่มมิติที่ห้าคือความสามารถในการทำซ้ำ แต่ละข้อให้คะแนน 0 ถึง 3 ตั้งแต่ไม่ได้ทำเลย ไปจนถึงทำครบพร้อมหลักฐาน

### โค้ดที่ 3.18 — แบบประเมินคุณภาพข้อมูล
ตำราพิมพ์แบบประเมินเป็นข้อความ เซลล์ถัดไปเขียนแบบประเมินเดียวกันเป็นคลาสของ Python (ชุดเดียวกับ `ch03_cleaning/src/scorecard.py`) เพื่อให้ใช้ซ้ำกับข้อมูลชุดใหม่ได้ และพิมพ์แบบประเมินเปล่าออกมา

```python
from dataclasses import dataclass, field
from datetime import date

ITEMS = {
    "completeness": ["missing values: type and cause identified",
                     "sample selection bias assessed",
                     "coverage of the parameter space"],
    "accuracy": ["calibration source and date",
                 "systematic uncertainty estimated",
                 "at least one sanity check against known physics"],
    "consistency": ["units stated and consistent for every column",
                    "dtypes and naming consistent"],
    "timeliness": ["dataset version and last-update date"],
    "reproducibility": ["cleaning script and seed committed to git",
                        "requirements.txt and environment documented",
                        "notebook passes Restart & Run All"],
}

@dataclass
class Scorecard:
    dataset: str
    reviewer: str = "<name>"
    commit: str = "<hash>"
    scores: dict = field(default_factory=dict)          # (section, item) -> (score, evidence)

    def score(self, section, item, value, evidence=""):
        assert item in ITEMS[section], f"unknown item: {item}"
        assert value in (0, 1, 2, 3), "each item is scored 0-3"
        self.scores[(section, item)] = (value, evidence)

    def total(self):
        return sum(v for v, _ in self.scores.values())

    def report(self):
        lines = [f"DATA QUALITY SCORECARD - {self.dataset}",
                 f"Reviewer: {self.reviewer}     Date: {date.today()}     git commit: {self.commit}", ""]
        for i, (sec, items) in enumerate(ITEMS.items(), 1):
            got = sum(self.scores.get((sec, it), (0, ""))[0] for it in items)
            lines.append(f"[{i}] {sec.upper():58s}{got:2d}/{3 * len(items)}")
            for it in items:
                v, ev = self.scores.get((sec, it), (None, ""))
                box = f"[{v}]" if v is not None else "[ ]"
                lines.append(f"  {box} {it:54s}(0-3)" + (f"  {ev}" if ev else ""))
        maxi = 3 * sum(len(v) for v in ITEMS.values())
        lines += ["", f"TOTAL SCORE: {self.total():44d} / {maxi}"]
        return "\n".join(lines)

print(Scorecard("<dataset name>").report())
```

| ตัดค่าผิดปกติ | ความแปรปรวนลดลงอย่างไม่สมมาตร ถ้าตัดด้านเดียวมากกว่า ค่าเฉลี่ยจะเลื่อน |
| แปลงหน่วยด้วยค่าที่วัดได้ เช่น $k$ | ความไม่แน่นอนของค่านั้นส่งต่อไปยังทุกจุดข้อมูลพร้อมกัน (ส่วนที่ 16) |

### การเติมค่าเฉลี่ยทำให้มั่นใจเกินจริง
เซลล์ถัดไปคำนวณค่าเฉลี่ยของเรดชิฟต์พร้อมความคลาดเคลื่อนมาตรฐาน (SEM $= s/\sqrt{N}$) จากข้อมูลที่เติมค่าเฉลี่ยแล้ว แล้วเทียบกับความคลาดจริงจากค่าเฉลี่ยจริงของกาแล็กซีทุกดวง ซึ่งรู้ได้เพราะเป็นข้อมูลจำลอง

```python
for name, z in (("drop rows", df['redshift'].dropna()), ("fill mean", df['z_imp_mean']),
                ("fill from r_mag", df['z_imp_photo'])):
    sem = z.std() / np.sqrt(z.size)
    err = z.mean() - z_true.mean()
    print(f"{name:16s} N = {z.size:4d}  mean = {z.mean():.4f} ± {sem:.4f} (SEM)   "
          f"actual error = {err:+.4f}  ->  {abs(err) / sem:4.1f} SEM")
```

เมื่อลบแถวหรือเติมค่าเฉลี่ย ค่าเฉลี่ยคลาดจากความจริงหลายเท่าของ SEM ที่คำนวณได้ ผู้วิเคราะห์ที่ดูเพียงค่า SEM จะเชื่อว่าผลแม่นกว่าความเป็นจริงมาก การเติมค่าเฉลี่ยยังทำให้ SEM เล็กลงอีก เพราะจำนวนแถวเพิ่มขึ้นและความแปรปรวนลดลง ทั้งที่ไม่ได้มีข้อมูลใหม่เข้ามาเลย

### การปรับให้เรียบทำให้จุดข้างเคียงสัมพันธ์กัน
สูตรความคลาดเคลื่อนมาตรฐานทั่วไปสมมติว่าแต่ละจุดเป็นอิสระต่อกัน หลังปรับให้เรียบ สมมติฐานนี้ไม่เป็นจริง เซลล์ถัดไปวัดสหสัมพันธ์ระหว่างจุดที่อยู่ติดกัน (lag-1 autocorrelation) ของสัญญาณรบกวนล้วน ก่อนและหลังตัวกรองซาวิตซกี–โกเลย์

```python
noise_only = np.random.default_rng(11).normal(0, 1, 5000)
def lag1(x):
    return np.corrcoef(x[:-1], x[1:])[0, 1]
for w in (None, 11, 31):
    y_ = noise_only if w is None else savgol_filter(noise_only, w, 3)
    label = "raw noise" if w is None else f"Savitzky-Golay ({w}, 3)"
    print(f"{label:24s} std {y_.std():.3f}   lag-1 correlation {lag1(y_):+.3f}")
```

ส่วนเบี่ยงเบนมาตรฐานลดลงหลังปรับให้เรียบ แต่จุดข้างเคียงสัมพันธ์กันสูง ถ้านำจุดเหล่านี้ไปคำนวณความไม่แน่นอนโดยถือว่าเป็นอิสระต่อกัน ความไม่แน่นอนที่ได้จะเล็กเกินจริง

### กรอบทั่วไป
เมื่อ $y$ เป็นฟังก์ชันของตัวแปร $x_i$ การส่งต่อความไม่แน่นอนเขียนได้เป็น

$$\sigma_y^2 = \sum_i \left(\frac{\partial y}{\partial x_i}\right)^2 \sigma_{x_i}^2 + \text{พจน์ความแปรปรวนร่วม}$$

พจน์ความแปรปรวนร่วม (covariance) จับผลของความคลาดเคลื่อนที่เคลื่อนไปด้วยกัน เช่น ความคลาดของ $k$ ที่เลื่อนทุกจุดไปทางเดียวกัน หรือสัญญาณรบกวนของจุดข้างเคียงที่สัมพันธ์กันหลังปรับให้เรียบ บทนี้ขอเพียงให้ตระหนักว่าลำดับงานทำความสะอาดเป็นแหล่งของความไม่แน่นอนที่ต้องรายงาน การคำนวณอย่างเป็นระบบอยู่ในบทที่ 9
