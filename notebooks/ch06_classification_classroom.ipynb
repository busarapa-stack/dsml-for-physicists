## วัตถุประสงค์
เมื่อทำโน้ตบุ๊กนี้จนจบ ผู้เรียนควรสามารถ

1. อธิบายลักษณะเฉพาะของการจำแนกประเภท (classification) ในฟิสิกส์ ได้แก่ ความไม่สมดุลของประเภทและต้นทุนของความผิดพลาดสองชนิดที่ไม่เท่ากัน
2. สร้างลักษณะเด่นที่มีความหมายทางฟิสิกส์ เช่น มวลยืนยงและดัชนีสี และอธิบายว่าลักษณะเด่นที่ดีทำให้วิธีจำแนกที่ง่ายใช้ได้อย่างไร
3. อธิบายหลักการและรูปร่างของขอบเขตการตัดสินใจ (decision boundary) ของวิธีจำแนกห้าวิธี และใช้ในไปป์ไลน์ของ scikit-learn
4. คำนวณและตีความความบริสุทธิ์ ประสิทธิภาพ คะแนน F1 เส้นโค้ง ROC และพื้นที่ใต้เส้นโค้ง (AUC)
5. เลือกเกณฑ์ตัดสินจากนัยสำคัญทางสถิติ $Z_A$ ที่คำนวณด้วยจำนวนเหตุการณ์ที่คาดว่าจะพบจริง ไม่ใช่จำนวนในข้อมูลตรวจสอบ
6. ระบุการรั่วไหลของข้อมูลในความหมายกว้าง เช่น การใช้เรดชิฟต์จากสเปกตรัมจำแนกวัตถุท้องฟ้า

## ข้อมูลในโน้ตบุ๊กนี้
- **ฟิสิกส์อนุภาค:** ข้อมูลจำลองแบบง่าย (toy data) ของ $H \to b\bar b$ กับฉากหลัง สร้างด้วยโค้ดที่ 6.1 (เมล็ดสุ่ม 2026) ไม่ใช่ข้อมูลจริงจากเครื่องเร่งอนุภาค ตัวเลขทุกตัวในส่วนนี้ตรงกับตำรา
- **SDSS:** ตำราใช้ตาราง 100,000 วัตถุที่คัดจาก SDSS DR17 และเผยแพร่เพื่อการเรียนบน Kaggle (Stellar Classification Dataset – SDSS17) โน้ตบุ๊กนี้ไม่ดาวน์โหลดไฟล์นั้น แต่สร้าง **ข้อมูลแทนสังเคราะห์ (synthetic stand-in)** ที่มีหลักและสัดส่วนประเภทเดียวกัน ด้วยโค้ดชุดเดียวกับสคริปต์ `05_sdss_data.py` สีของวัตถุเลียนแบบฟิสิกส์ในตำราอย่างคร่าว ๆ เท่านั้น ตัวเลขจากข้อมูลแทนจึง **ไม่ใช่** ผลเกี่ยวกับท้องฟ้าจริง และต่างจากตัวเลขในตำรา (ซึ่งมาจากไฟล์จริง) ส่วนที่ 13 บอกวิธีใช้ไฟล์จริงถ้าผู้สอนต้องการ

หัวกราฟที่มาจากข้อมูลจำลองหรือข้อมูลแทนมีคำว่า SYNTHETIC กำกับ ตัวเลขใช้เพื่อการเรียนวิธีการ ไม่ใช่ผลการวิจัย

### การล็อกแนวคำตอบ
เซลล์ ✅ และเซลล์แนวคำตอบในโน้ตบุ๊กนี้แสดงคำตอบอ้างอิงเมื่อผู้เรียนเติมคำตอบในเซลล์ ✏️ ของข้อนั้นครบแล้วเท่านั้น ถ้ายังไม่ได้ตอบ เซลล์จะแสดงเพียงข้อความ 🔒 บอกว่ายังขาดคำตอบใด การคำนวณในเซลล์ยังทำงานตามปกติ ดังนั้น Run all จึงรันได้ครบโดยไม่เปิดเผยคำตอบ แนวคำตอบของงานเขียนและงานนอกชั้นซึ่งตรวจด้วยโค้ดไม่ได้จะเปิดเมื่อผู้สอนเลือกเท่านั้น

**สำหรับผู้สอน:** ติ๊กช่อง `SHOW_ANSWERS` ในเซลล์ถัดไป (หรือแก้เป็น `True`) แล้วรันเซลล์นั้นอีกครั้ง เพื่อแสดงแนวคำตอบทุกข้อ เช่น เมื่อเฉลยในชั้นหรือเตรียมสอน

```python
#@title ⚙️ ตัวเลือกของผู้สอน: แสดงแนวคำตอบ
SHOW_ANSWERS = False  #@param {type:"boolean"}

def _answered(v):
    """True when an exercise variable holds an answer (not None, "?", "", empty or containing blanks)."""
    if v is None or v is Ellipsis:
        return False
    if isinstance(v, str):                       # placeholders such as "?", "?-?", "..." count as blank
        return v.replace("?", "").replace("-", "").replace(".", "").strip() != ""
    if isinstance(v, dict):
        return len(v) > 0 and all(_answered(x) for x in v.values())
    if isinstance(v, (list, tuple, set, frozenset)):
        return len(v) > 0 and all(_answered(x) for x in v)
    if type(v).__name__ == "ndarray":
        return v.size > 0
    return True

def _gate(names=()):
    """Return print if the answers may be shown, otherwise print one lock message and return a no-op."""
    if SHOW_ANSWERS:
        return print
    if names:
        missing = [n for n in names if not _answered(globals().get(n))]
        if not missing:
            return print
        print("🔒 ยังไม่ได้ตอบ:", ", ".join(missing),
              "— เติมคำตอบในเซลล์ ✏️ ด้านบนให้ครบก่อน แล้วรันเซลล์นี้อีกครั้ง")
    else:
        print("🔒 แนวคำตอบข้อนี้ผู้สอนเป็นผู้เปิด (ตั้ง SHOW_ANSWERS = True ในส่วนที่ 0)")
    return lambda *args, **kwargs: None

print("แสดงแนวคำตอบทั้งหมด:", "ใช่ (โหมดผู้สอน)" if SHOW_ANSWERS else "ไม่ (แสดงเมื่อผู้เรียนตอบข้อนั้นครบแล้ว)")
```

```python
#@title เตรียมตาราง SDSS: ใช้ไฟล์จริงถ้ามี มิฉะนั้นสร้างข้อมูลแทน (ดับเบิลคลิกเพื่อดูโค้ด)
# โค้ดชุดเดียวกับ ch06_classification/05_sdss_data.py
SRC_SDSS = r'''"""Ch.6  The SDSS table used by code:sdss-cc and code:sdss-comp.

The book uses the 100,000-object table selected from SDSS DR17 and published for teaching as the
Kaggle "Stellar Classification Dataset - SDSS17" (fedesoriano).  Put the real file at
    data/raw/star_classification.csv
(see optional/README_sdss.md or run optional/fetch_sdss_skyserver.py) and this script only checks it.

If the real file is absent, this script writes a SYNTHETIC STAND-IN with the same columns and the same
class counts (59,445 galaxies, 21,594 stars, 18,961 quasars), so that every script of the chapter runs.
Its colours follow the physics described in the book only schematically (stellar locus, redder
galaxies, blue low-z quasars, z ~ 2.5-3 quasars on the stellar locus).  Numbers from the stand-in are
NOT results about the real sky; the book quotes no numbers from this table except the class fractions.
"""
import os
from pathlib import Path

import numpy as np
import pandas as pd

os.chdir(Path(__file__).resolve().parent)
RAW = Path('data/raw/star_classification.csv')
SOURCE = Path('data/raw/SOURCE.txt')
RAW.parent.mkdir(parents=True, exist_ok=True)


def make_standin(seed=2026):
    rng = np.random.default_rng(seed)
    n_gal, n_star, n_qso = 59445, 21594, 18961
    sc = lambda n, s: rng.normal(0, s, n)

    # stars: one temperature-like parameter t along the stellar locus (hot/blue t=0 -> cool/red t=1)
    t = rng.beta(2.0, 2.5, n_star)
    s_gr = -0.25 + 1.65 * t + sc(n_star, 0.05)
    # SDSS stellar locus: u-g stays near 1.0-1.2 for hot A/F stars (g-r < ~0.3), then climbs to ~2.6 for K/M
    s_ug = 1.05 + 1.6 * np.clip((s_gr - 0.25) / 0.95, 0, 1) + sc(n_star, 0.08)
    s_ri = 0.02 + 1.3 * t**2.2 + sc(n_star, 0.04)
    s_iz = 0.00 + 0.65 * t**2.2 + sc(n_star, 0.04)
    s_r = rng.uniform(14.0, 20.5, n_star)
    s_zred = rng.normal(0, 3e-4, n_star)

    # galaxies: red sequence and blue cloud, colours redden with redshift
    zg = np.clip(rng.gamma(3.0, 0.07, n_gal), 0.005, 1.0)
    red = rng.random(n_gal) < 0.55
    g_gr = np.where(red, 0.72 + 1.3 * zg, 0.42 + 0.9 * zg) + sc(n_gal, 0.10)
    g_ug = np.where(red, 1.75 + 0.8 * zg, 1.25 + 0.5 * zg) + sc(n_gal, 0.25)
    g_ri = 0.33 + 0.9 * zg + sc(n_gal, 0.08)
    g_iz = 0.22 + 0.35 * zg + sc(n_gal, 0.10)
    g_r = rng.uniform(15.0, 20.5, n_gal) + 2.0 * zg

    # quasars: blue in u-g until Lyman-alpha enters u (z > ~2.2), then u-g rises steeply
    zq = np.clip(np.concatenate([rng.normal(1.4, 0.6, int(0.7 * n_qso)),
                                 rng.normal(2.7, 0.35, n_qso - int(0.7 * n_qso))]), 0.1, 5.0)
    q_ug = 0.15 + 1.4 * np.clip(zq - 2.2, 0, None) + sc(n_qso, 0.18)
    q_gr = 0.15 + 0.25 * np.clip(zq - 2.4, 0, None) + 0.1 * np.sin(2.5 * zq) + sc(n_qso, 0.12)
    q_ri = 0.12 + 0.08 * np.sin(1.7 * zq) + sc(n_qso, 0.10)
    q_iz = 0.05 + 0.10 * np.cos(1.3 * zq) + sc(n_qso, 0.12)
    q_r = rng.uniform(17.0, 21.5, n_qso)

    def mags(ug, gr, ri, iz, r):
        g = r + gr; u = g + ug; i = r - ri; z = i - iz
        return u, g, r, i, z

    parts = []
    for cls, cols, zred in (('GALAXY', (g_ug, g_gr, g_ri, g_iz, g_r), zg),
                            ('STAR', (s_ug, s_gr, s_ri, s_iz, s_r), s_zred),
                            ('QSO', (q_ug, q_gr, q_ri, q_iz, q_r), zq)):
        u, g, r, i, z = mags(*cols)
        n = len(r)
        parts.append(pd.DataFrame({'alpha': rng.uniform(0, 360, n), 'delta': rng.uniform(-10, 70, n),
                                   'u': u, 'g': g, 'r': r, 'i': i, 'z': z,
                                   'class': cls, 'redshift': zred}))
    df = pd.concat(parts, ignore_index=True).sample(frac=1.0, random_state=seed).reset_index(drop=True)
    df.insert(0, 'obj_ID', np.arange(len(df)) + 1_000_000_000)
    df.loc[df.sample(1, random_state=seed).index, ['u', 'g', 'z']] = -9999.0   # one sentinel row, as in the real table
    return df.round({'u': 5, 'g': 5, 'r': 5, 'i': 5, 'z': 5, 'redshift': 6})


def is_synthetic(path):
    """The stand-in has no spec_obj_ID column; the Kaggle file and the SkyServer fetch both have it."""
    return 'spec_obj_ID' not in pd.read_csv(path, nrows=1).columns


if __name__ == '__main__':
    if not RAW.exists() or is_synthetic(RAW):          # a real file is never overwritten
        make_standin().to_csv(RAW, index=False)
        SOURCE.write_text("SYNTHETIC stand-in written by 05_sdss_data.py (not real SDSS data)\n")
    else:
        SOURCE.write_text("REAL SDSS table supplied by the user (Kaggle SDSS17 or optional/fetch_sdss_skyserver.py)\n")
    df = pd.read_csv(RAW)
    print(f"source: {SOURCE.read_text().strip()}")
    print(f"{len(df)} rows; columns: {list(df.columns)}")
    frac = df['class'].value_counts(normalize=True)
    print("class fractions: " + ", ".join(f"{k} {v:.3f}" for k, v in frac.items()))
    bad = (df[['u', 'g', 'r', 'i', 'z']] < -100).any(axis=1).sum()
    print(f"rows with a sentinel magnitude (< -100): {bad}")
'''
HERE6 = os.getcwd()
if os.path.exists("data/raw/star_classification.csv") and os.path.exists("data/raw/SOURCE.txt"):
    print("ใช้ตาราง SDSS ที่มีอยู่แล้ว:", open("data/raw/SOURCE.txt").read().strip())
else:                     # writes the stand-in, or only records the source of a real file (never overwrites it)
    exec(compile(SRC_SDSS, "05_sdss_data.py", "exec"),
         {"__name__": "__main__", "__file__": os.path.join(HERE6, "05_sdss_data.py")})
    os.chdir(HERE6)
```

# ส่วนที่ 1 แรงจูงใจจากฮิกส์ กรอบของการจำแนก และความไม่สมดุล (หัวข้อ 6.1.1–6.1.3)

### แนวคิดสำคัญ
เครื่องเร่งอนุภาคขนาดใหญ่ (Large Hadron Collider, LHC) ทำให้โปรตอนชนกันราวพันล้านครั้งต่อวินาที การชนที่ผลิตฮิกส์โบซอนเกิดเพียงราวหนึ่งครั้งในพันล้านครั้ง ที่เหลือเกือบทั้งหมดเป็นกระบวนการที่รู้จักแล้ว เรียกรวมว่าฉากหลัง (background) งานของการจำแนกประเภท (classification) คือเรียนรู้ฟังก์ชันที่ทำนายป้ายกำกับ $y \in \{0, 1\}$ (ฉากหลังหรือสัญญาณ) จากเวกเตอร์ลักษณะเด่น $\mathbf{x}$ โดยป้ายกำกับของข้อมูลฝึกได้จากการจำลองการชน

การจำแนกในฟิสิกส์มีลักษณะสามข้อที่ต่างจากงานทั่วไป

1. **ประเภทไม่สมดุลอย่างรุนแรง** แม้หลังการคัดเลือกเบื้องต้น สัญญาณต่อฉากหลังอาจยังเป็น 1 ต่อ 50 หรือต่ำกว่า
2. **ความผิดพลาดสองชนิดมีต้นทุนไม่เท่ากัน** ผลบวกลวงทำให้ตัวอย่างปนเปื้อน ผลลบลวงทำให้สัญญาณที่หายากยิ่งน้อยลง
3. **ผลต้องตีความได้** วิธีจำแนกที่ฝึกจากข้อมูลจำลองอาจเรียนรู้จุดบกพร่องของการจำลองแทนฟิสิกส์จริง

หลักของบทนี้คือ วิธีจำแนกที่ดีไม่ใช่วิธีที่ได้ความแม่นยำสูงที่สุด แต่เป็นวิธีที่ตอบคำถามทางฟิสิกส์ได้ตรง และตีความได้ว่าตอบเพราะอะไร

### การจำแนกสามแบบ
| แบบ | ลักษณะของป้ายกำกับ | ตัวอย่างทางฟิสิกส์ |
|---|---|---|
| สองประเภท (binary) | $y \in \{0,1\}$ | สัญญาณฮิกส์กับฉากหลัง |
| หลายประเภท (multi-class) | $y \in \{1,\ldots,K\}$ อยู่ได้ประเภทเดียว | ดาวฤกษ์ กาแล็กซี และเควซาร์ |
| หลายป้ายกำกับ (multi-label) | มีได้หลายป้ายกำกับพร้อมกัน | กาแล็กซีที่มีทั้งแขนกังหันและแกนกลางแบบคาน |

วิธีจำแนกส่วนใหญ่ใน scikit-learn ให้คะแนนหรือความน่าจะเป็นผ่าน `predict_proba()` การตัดสินว่าเหตุการณ์ใดเป็นสัญญาณขึ้นกับเกณฑ์ (threshold) ที่ตั้งบนคะแนน และเกณฑ์ที่เหมาะสมขึ้นกับเป้าหมายทางฟิสิกส์ ไม่ใช่ค่า 0.5 ที่ไลบรารีตั้งไว้ (ส่วนที่ 12)

### ความไม่สมดุลเป็นสมบัติของธรรมชาติ
วิธีจัดการที่ใช้บ่อยมีสี่วิธี ได้แก่ ถ่วงน้ำหนักประเภทด้วย `class_weight='balanced'` แบ่งข้อมูลแบบรักษาสัดส่วน ใช้ตัวชี้วัดที่ไม่ถูกประเภทส่วนใหญ่ครอบงำ และสร้างตัวอย่างสังเคราะห์ของประเภทส่วนน้อย เช่น SMOTE ข้อที่ต้องจำคือ ฮิกส์ถูกผลิตน้อยเพราะภาคตัดขวางมีค่าน้อยตามทฤษฎี การ "แก้" ความไม่สมดุลในขั้นฝึกจึงต้องทำโดยไม่ลืมว่า สัดส่วนในข้อมูลจริงยังเหมือนเดิม

### ลำดับขั้นหกขั้นสำหรับการจำแนก

# ส่วนที่ 3 ลักษณะเด่นเชิงจลนศาสตร์และข้อมูลจำลอง (หัวข้อ 6.2.1–6.2.2)

เมื่อฮิกส์สลายเป็นควาร์กบอตทอมคู่ ($H \to b\bar b$) ควาร์กแต่ละตัวกลายเป็นลำอนุภาค (jet) จากข้อมูลของลำอนุภาคสองลำ เราสร้างลักษณะเด่นที่มีความหมายทางฟิสิกส์ได้ (หน่วยที่ $c = 1$)

| ลักษณะเด่น | นิยาม | ความหมาย | สมบัติเชิงสมมาตร |
|---|---|---|---|
| $m_{bb}$ | $\sqrt{(E_1+E_2)^2 - \lvert\vec p_1+\vec p_2\rvert^2}$ | มวลยืนยง ถ้ามาจากฮิกส์จะใกล้ 125 GeV | ไม่แปรเปลี่ยนภายใต้การแปลงลอเรนตซ์ |
| $p_T$ | $\sqrt{p_x^2+p_y^2}$ ของระบบสองลำ | โมเมนตัมในระนาบตั้งฉากกับลำโปรตอน | ไม่เปลี่ยนเมื่อเปลี่ยนกรอบตามแนวลำ |
| $E_T^{\text{miss}}$ | ขนาดของ $-\sum \vec p_T$ ของอนุภาคที่มองเห็น | โมเมนตัมตามขวางที่ "หายไป" กับนิวทริโน | มาจากการอนุรักษ์โมเมนตัมตามขวาง |
| $\Delta R$ | $\sqrt{\Delta\eta^2 + \Delta\phi^2}$ | ระยะเชิงมุมระหว่างลำอนุภาคสองลำ | ไม่เปลี่ยนเมื่อหมุนรอบแนวลำและเปลี่ยนกรอบตามแนวลำโดยประมาณ |

$m_{bb}$ แยกสัญญาณได้ดีที่สุด เพราะถ้าลำทั้งสองมาจากอนุภาคแม่ตัวเดียว มวลยืนยงจะเท่ากับมวลของอนุภาคแม่เสมอ ไม่ว่าอนุภาคแม่เคลื่อนที่เร็วเพียงใด ฮิกส์ที่มี $p_T$ สูงให้ลำอนุภาคที่อยู่ใกล้กัน โดยประมาณ $\Delta R \approx 2m_H/p_T$ (เหมือนไฟหน้ารถสองดวงที่ดูรวมเป็นดวงเดียวเมื่อรถแล่นมาเร็ว) ส่วน $E_T^{\text{miss}}$ ของ $H \to b\bar b$ มีไม่มาก ขณะที่ฉากหลังอย่างควาร์กท็อปคู่มีนิวทริโนพลังงานสูงจากการสลายของบอซอน $W$

โค้ดที่ 6.1 สร้างข้อมูลที่มีสัญญาณ 4,000 และฉากหลัง 16,000 เหตุการณ์ สัดส่วน 1 ต่อ 4 นี้สูงกว่าธรรมชาติมาก เพราะชุดข้อมูลฝึกมักจำลองสัญญาณมากเป็นพิเศษ ข้อนี้สำคัญมากในส่วนที่ 12

### โค้ดที่ 6.1 — ข้อมูลจำลองแบบง่ายของ H → bb̄ และฉากหลัง

```python
import numpy as np
import pandas as pd

def make_higgs_toy(n_sig=4000, n_bkg=16000, seed=2026):
    """Toy H->bb events vs background (GeV). Shapes are illustrative only."""
    rng = np.random.default_rng(seed)
    # signal: resonance at 125 GeV with ~10% resolution
    m_s   = rng.normal(125.0, 12.0, n_sig)
    pT_s  = rng.gamma(shape=4.0, scale=40.0, size=n_sig)     # mean 160 GeV
    dR_s  = 2 * m_s / np.maximum(pT_s, 60.0) * rng.normal(1.0, 0.15, n_sig)
    met_s = rng.exponential(20.0, n_sig)                       # semileptonic b
    # background: falling mass spectrum, softer pT, wider angles
    m_b   = 40.0 + rng.exponential(70.0, n_bkg)
    pT_b  = rng.gamma(shape=3.0, scale=40.0, size=n_bkg)     # mean 120 GeV
    dR_b  = rng.uniform(0.4, 4.0, n_bkg)
    met_b = rng.exponential(30.0, n_bkg)                       # e.g. ttbar
    df = pd.DataFrame({
        'm_bb':   np.concatenate([m_s, m_b]),
        'pT':     np.concatenate([pT_s, pT_b]),
        'dR':     np.clip(np.concatenate([dR_s, dR_b]), 0.4, 4.0),
        'ETmiss': np.concatenate([met_s, met_b]),
        'label':  np.concatenate([np.ones(n_sig, int), np.zeros(n_bkg, int)]),
    })
    return df.sample(frac=1.0, random_state=seed).reset_index(drop=True)

df = make_higgs_toy()
```

ขั้นถัดไปคือแบ่งข้อมูลฝึกกับข้อมูลตรวจสอบ อาร์กิวเมนต์ `stratify=y` บังคับให้ทั้งสองชุดมีสัดส่วนสัญญาณเท่ากับข้อมูลทั้งหมด ถ้าไม่ระบุ การสุ่มอาจทำให้ชุดตรวจสอบมีสัญญาณน้อยกว่าที่ควรโดยบังเอิญ

### โค้ดที่ 6.2 — การแบ่งข้อมูลแบบรักษาสัดส่วนของประเภท

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

features = ['m_bb', 'pT', 'dR', 'ETmiss']
X = df[features].values
y = df['label'].values

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42)
```

```python
print(f"{len(df)} events, signal fraction: all {y.mean():.3f}, train {y_train.mean():.3f}, "
      f"validation {y_val.mean():.3f}; validation size {len(y_val)} "
      f"({int(y_val.sum())} signal, {int((1 - y_val).sum())} background)")
print("\nmean of each feature (GeV, dR dimensionless):")
print(df.groupby('label')[features].mean().round(2).rename(index={0: 'background', 1: 'signal'}))
print(f"\ntoy mass resolution 12/125 = {12 / 125:.3f} (about 10 %)")
```

# ส่วนที่ 6 การถดถอยโลจิสติก (หัวข้อ 6.3.2)

การถดถอยโลจิสติก (logistic regression) เป็นวิธีจำแนก ไม่ใช่วิธีถดถอย ชื่อมาจากการถดถอย "ลอการิทึมของอัตราต่อรอง" ด้วยฟังก์ชันเชิงเส้น แล้วแปลงเป็นความน่าจะเป็นด้วยฟังก์ชันซิกมอยด์ (sigmoid)

$$P(y=1 \mid \mathbf{x}) = \sigma(\boldsymbol\theta^\top\mathbf{x}) = \frac{1}{1+e^{-\boldsymbol\theta^\top\mathbf{x}}}$$

นักฟิสิกส์จะจำรูปนี้ได้ มันคือการแจกแจงของแฟร์มี–ดิแรกเมื่อกลับด้าน ซึ่งเปลี่ยนจาก 1 เป็น 0 อย่างนุ่มนวลรอบพลังงานแฟร์มี ในการถดถอยโลจิสติก จุดเปลี่ยนคือระนาบ $\boldsymbol\theta^\top\mathbf{x} = 0$ ขอบเขตจึงเป็นเชิงเส้นเสมอ

พารามิเตอร์หาได้ด้วยการประมาณภาวะน่าจะเป็นสูงสุด แต่ข้อมูลเป็น 0 กับ 1 ภาวะน่าจะเป็นจึงมาจากการแจกแจงแบร์นูลลี ฟังก์ชันที่ทำให้น้อยที่สุดคือครอสเอนโทรปี (cross-entropy)

$$J(\boldsymbol\theta) = -\sum_i \big[\,y_i \ln P_i + (1-y_i)\ln(1-P_i)\,\big]$$

scikit-learn ใช้การปรับลงโทษแบบริดจ์จากบทที่ 5 ด้วยพารามิเตอร์ $C = 1/\lambda$ ค่า $C$ ที่เล็กจึงหมายถึงการลงโทษที่แรง

```python
E = np.linspace(-6, 6, 300)
fig, ax = plt.subplots(1, 2, figsize=(10, 3.2))
ax[0].plot(E, 1 / (1 + np.exp(-E)), label=r"sigmoid $1/(1+e^{-z})$")
ax[0].plot(E, 1 / (1 + np.exp(E)), "--", label=r"Fermi-Dirac $1/(e^{(E-\mu)/kT}+1)$")
ax[0].set_xlabel(r"$z$ or $(E-\mu)/kT$"); ax[0].legend(fontsize=8); ax[0].set_title("same shape, mirrored")
P = np.linspace(0.005, 1, 300)
ax[1].plot(P, -np.log(P)); ax[1].set_xlabel(r"predicted $P_i$ for a true signal event"); ax[1].set_ylabel(r"$-\ln P_i$")
ax[1].set_title("cross-entropy punishes confident mistakes"); plt.tight_layout(); plt.show()
```

### เพื่อนบ้านใกล้ที่สุด k ตัว (KNN)
ถ้าอยากรู้ว่าบ้านหลังหนึ่งอยู่ในหมู่บ้านใด ให้ถามเพื่อนบ้านที่ใกล้ที่สุด $k$ หลัง แล้วตอบตามเสียงส่วนใหญ่ วิธีนี้ไม่มีขั้นฝึกในความหมายปกติ เพียงเก็บข้อมูลฝึกไว้ แล้วคำนวณระยะทางเมื่อต้องทำนาย จึงเป็นวิธีไม่อิงพารามิเตอร์ (non-parametric)

| $k$ | อคติ | ความแปรปรวน | ลักษณะของขอบเขต |
|---|---|---|---|
| เล็ก เช่น 1 | ต่ำ | สูง ไวต่อสัญญาณรบกวนทุกจุด | ขรุขระ มีเกาะเล็ก ๆ |
| กลาง เช่น 10–50 | ปานกลาง | ปานกลาง | เรียบพอสมควร |
| ใหญ่ เช่น $N/10$ | สูง | ต่ำ | เรียบเกินไป |

ข้อจำกัดคือ ทำนายช้าเมื่อข้อมูลมาก ต้องเก็บข้อมูลฝึกทั้งหมด และเมื่อจำนวนมิติสูง ระยะระหว่างจุดทุกคู่ใกล้เคียงกันจนคำว่า "ใกล้" แทบไม่มีความหมาย เรียกว่าคำสาปของมิติ (curse of dimensionality) และ **ต้องปรับข้อมูลเป็นมาตรฐานก่อนเสมอ**

### ซัพพอร์ตเวกเตอร์แมชชีน (SVM)
ถ้ามีเส้นหลายเส้นที่แยกสองประเภทได้ SVM เลือกเส้นที่อยู่ห่างจากจุดที่ใกล้ที่สุดของทั้งสองประเภทมากที่สุด เหมือนการขีดเส้นแบ่งเลนให้อยู่กลางระหว่างรถสองฝั่ง ระยะว่างนี้เรียกว่าขอบ (margin) และจุดที่อยู่บนขอบหรือล้ำเข้ามาเรียกว่าซัพพอร์ตเวกเตอร์ (support vector) ในทางฟิสิกส์ ซัพพอร์ตเวกเตอร์คือเหตุการณ์ก้ำกึ่งที่คล้ายทั้งสัญญาณและฉากหลัง

ขอบเขตโค้งมาจากกลเม็ดเคอร์เนล (kernel trick) ข้อมูลที่แยกด้วยเส้นตรงไม่ได้ในปริภูมิเดิมอาจแยกได้ในปริภูมิที่มีมิติสูงกว่า เซลล์ถัดไปแสดงตัวอย่างในตำรา: จุดที่เรียงเป็นวงแหวนแยกด้วยเส้นตรงไม่ได้ แต่ถ้าเพิ่มมิติที่สาม $x^2 + y^2$ ก็แยกได้ด้วยระนาบ

| เคอร์เนล | ฟังก์ชัน | เหมาะกับ |
|---|---|---|
| เชิงเส้น | $K(\mathbf x,\mathbf x') = \mathbf x^\top\mathbf x'$ | ข้อมูลที่แยกได้ด้วยระนาบหลังสร้างลักษณะเด่นแล้ว |
| ฟังก์ชันฐานแนวรัศมี (RBF) | $K = \exp(-\gamma\lVert\mathbf x-\mathbf x'\rVert^2)$ | ขอบเขตโค้งเรียบทั่วไป |
| พหุนาม | $K = (\gamma\,\mathbf x^\top\mathbf x' + r)^d$ | ขอบเขตที่รู้ว่ามีรูปพหุนาม |

```python
from sklearn.datasets import make_circles
Xc, yc = make_circles(n_samples=400, noise=0.08, factor=0.45, random_state=2026)
fig = plt.figure(figsize=(10, 3.8))
a1 = fig.add_subplot(1, 2, 1); a1.scatter(Xc[:, 0], Xc[:, 1], c=yc, cmap="coolwarm", s=8)
a1.set_title("2-D: no straight line separates the classes"); a1.set_aspect("equal")
a2 = fig.add_subplot(1, 2, 2, projection="3d")
a2.scatter(Xc[:, 0], Xc[:, 1], (Xc**2).sum(1), c=yc, cmap="coolwarm", s=6)
a2.set_zlabel(r"$x^2+y^2$"); a2.set_title("3-D: a plane separates them")
plt.tight_layout(); plt.show()
```

ข้อจำกัดของ SVM คือเวลาฝึกเพิ่มระหว่าง $N^2$ ถึง $N^3$ ไม่ได้ให้ความน่าจะเป็นโดยตรง และต้องปรับไฮเพอร์พารามิเตอร์ $C$ กับ $\gamma$ ร่วมกัน

# ส่วนที่ 8 ต้นไม้ตัดสินใจและป่าสุ่ม (หัวข้อ 6.3.4)

ต้นไม้ตัดสินใจ (decision tree) ทำงานเหมือนเกมยี่สิบคำถาม แต่ละขั้นถามคำถามใช่หรือไม่เกี่ยวกับลักษณะเด่นตัวเดียว เช่น "$m_{bb}$ มากกว่า 102 GeV หรือไม่" ขอบเขตจึงเป็นขั้นบันไดที่ขนานกับแกน คล้ายการคัดเลือกเหตุการณ์ด้วยเกณฑ์ตัดที่นักฟิสิกส์อนุภาคทำมาหลายสิบปี ต่างกันที่ต้นไม้เลือกเกณฑ์ให้เองจากข้อมูล โดยเลือกคำถามที่ทำให้กิ่งลูก "บริสุทธิ์" ที่สุด วัดด้วยดัชนีจีนี (Gini impurity) $G = 1 - \sum_k p_k^2$ หรือเอนโทรปี $H = -\sum_k p_k\ln p_k$ ซึ่งคือเอนโทรปีแบบกิบส์ของการแจกแจงประเภทในโหนด

ต้นไม้เดี่ยวมีความแปรปรวนสูง ป่าสุ่ม (random forest) ฝึกต้นไม้หลายร้อยต้น แต่ละต้นฝึกจากตัวอย่างบูตสแตรป และที่แต่ละโหนดเลือกคำถามจากลักษณะเด่นที่สุ่มมาบางตัว เมื่อเฉลี่ยคำตอบ ความผิดพลาดที่ไม่สัมพันธ์กันจะหักล้างกัน วิธีนี้เรียกว่าแบ็กกิง (bagging) ส่วนวิธีที่รวมต้นไม้แบบต่อเนื่อง เรียกว่าต้นไม้ตัดสินใจแบบบูสต์ (boosted decision tree) ใช้แพร่หลายในฟิสิกส์อนุภาค

### โค้ดที่ 6.4 — ป่าสุ่มสำหรับการจำแนกสัญญาณ

```python
from sklearn.ensemble import RandomForestClassifier

rf_pipe = Pipeline([
    ('scaler', StandardScaler()),      # harmless for trees; kept for uniformity
    ('clf',    RandomForestClassifier(n_estimators=200, max_depth=10,
                                      min_samples_leaf=10,
                                      class_weight='balanced',
                                      random_state=42, n_jobs=-1)),
])
rf_pipe.fit(X_train, y_train)
importance = rf_pipe.named_steps['clf'].feature_importances_
```

ป่าสุ่มให้ความสำคัญของลักษณะเด่น (feature importance) จากการลดความไม่บริสุทธิ์ เซลล์ถัดไปเทียบกับความสำคัญแบบสับเปลี่ยน (permutation importance) ซึ่งวัดว่า AUC ของข้อมูลตรวจสอบลดลงเท่าใดเมื่อสลับค่าของลักษณะเด่นนั้นแบบสุ่ม (ใช้เวลาราว 20–40 วินาที)

```python
from sklearn.inspection import permutation_importance

perm = permutation_importance(rf_pipe, X_val, y_val, scoring='roc_auc',
                              n_repeats=10, random_state=42, n_jobs=-1)
print("feature   impurity   permutation (drop in validation AUC)")
for f, a, b in sorted(zip(features, importance, perm.importances_mean), key=lambda t: -t[1]):
    print(f"{f:7s}   {a:8.3f}   {b:8.3f}")
print(f"sums: impurity {importance.sum():.2f}, permutation {perm.importances_mean.sum():.2f}")
```

ลำดับ $m_{bb} > \Delta R > p_T > E_T^{\text{miss}}$ สอดคล้องกับฟิสิกส์ที่ใส่ไว้ในการจำลอง ซึ่งเป็นการตรวจความสมเหตุสมผลที่ควรทำทุกครั้ง ถ้าลักษณะเด่นที่ไม่ควรมีความหมาย เช่น หมายเลขรอบการเก็บข้อมูล ขึ้นมาอยู่อันดับต้น เป็นสัญญาณว่าข้อมูลมีสิ่งแปลกปลอม

### โค้ดที่ 6.5 — เปรียบเทียบวิธีจำแนกห้าวิธีบนข้อมูลฮิกส์จำลอง

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score

classifiers = {
    'LogReg':     LogisticRegression(class_weight='balanced', random_state=42),
    'KNN':        KNeighborsClassifier(n_neighbors=25),
    'SVM (RBF)':  SVC(kernel='rbf', class_weight='balanced', random_state=42),
    'DecTree':    DecisionTreeClassifier(max_depth=8, class_weight='balanced',
                                         random_state=42),
    'RandForest': RandomForestClassifier(n_estimators=200, max_depth=10,
                                         min_samples_leaf=10,
                                         class_weight='balanced',
                                         random_state=42, n_jobs=-1),
}
for name, clf in classifiers.items():
    pipe = Pipeline([('scaler', StandardScaler()), ('clf', clf)])
    pipe.fit(X_train, y_train)
    score = (pipe.decision_function(X_val) if name == 'SVM (RBF)'
             else pipe.predict_proba(X_val)[:, 1])
    y_pred = pipe.predict(X_val)
    print(f"{name:11s} acc={accuracy_score(y_val, y_pred):.3f} "
          f"F1={f1_score(y_val, y_pred):.3f} AUC={roc_auc_score(y_val, score):.3f}")
```

# ส่วนที่ 10 เมทริกซ์ความสับสนและตัวชี้วัดพื้นฐาน (หัวข้อ 6.4.1)

เมื่อวิธีจำแนกตัดสินแล้ว ผลแต่ละเหตุการณ์ตกอยู่ในหนึ่งในสี่กรณี ได้แก่ ผลบวกจริง (true positive, TP) ผลบวกลวง (false positive, FP) ผลลบลวง (false negative, FN) และผลลบจริง (true negative, TN) ตารางที่นับจำนวนในแต่ละกรณีเรียกว่าเมทริกซ์ความสับสน (confusion matrix)

| ตัวชี้วัด | สูตร | ความหมายในการค้นหาฮิกส์ |
|---|---|---|
| ความแม่นยำ (accuracy) | $(\text{TP}+\text{TN})/N$ | ชวนให้เข้าใจผิดเมื่อประเภทไม่สมดุล |
| ความแม่นยำเชิงบวก (precision) | $\text{TP}/(\text{TP}+\text{FP})$ | ความบริสุทธิ์ (purity) ของตัวอย่างที่คัดได้ |
| ค่าการเรียกคืน (recall) | $\text{TP}/(\text{TP}+\text{FN})$ | ประสิทธิภาพของการเก็บสัญญาณ (signal efficiency) |
| ความจำเพาะ (specificity) | $\text{TN}/(\text{TN}+\text{FP})$ | การปฏิเสธฉากหลัง (background rejection) |
| F1 | $2PR/(P+R)$ | ค่าเฉลี่ยฮาร์มอนิกของความบริสุทธิ์และประสิทธิภาพ |

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
cm_rf = confusion_matrix(y_val, rf_pipe.predict(X_val))
tn, fp, fn, tp = cm_rf.ravel()
print(f"random forest (code 6.4), threshold 0.5: TN = {tn}, FP = {fp}, FN = {fn}, TP = {tp}")
fig, ax = plt.subplots(figsize=(4.5, 3.8))
ConfusionMatrixDisplay(cm_rf, display_labels=["background", "signal"]).plot(ax=ax, colorbar=False)
plt.title("random forest, validation set (SYNTHETIC)"); plt.tight_layout(); plt.show()
```

# ส่วนที่ 12 เส้นโค้ง ROC และการเลือกเกณฑ์จากนัยสำคัญ (หัวข้อ 6.4.3–6.4.4)

เส้นโค้งลักษณะเฉพาะการดำเนินงานของเครื่องรับ (receiver operating characteristic curve, ROC curve) พล็อตอัตราผลบวกจริง (ประสิทธิภาพของสัญญาณ) เทียบกับอัตราผลบวกลวง (สัดส่วนฉากหลังที่ผ่านเกณฑ์) ทุกเกณฑ์พร้อมกัน ชื่อมาจากงานเรดาร์ในสงครามโลกครั้งที่สอง พื้นที่ใต้เส้นโค้ง (area under the curve, AUC) คือความน่าจะเป็นที่เหตุการณ์สัญญาณที่สุ่มมาได้คะแนนสูงกว่าเหตุการณ์ฉากหลังที่สุ่มมา AUC ไม่ขึ้นกับเกณฑ์และสัดส่วนของประเภท จึงเหมาะสำหรับเปรียบเทียบวิธีจำแนก แต่ด้วยเหตุผลเดียวกัน AUC ไม่ได้บอกว่าควรตั้งเกณฑ์ที่ใด

การเลือกเกณฑ์ในการค้นหาอนุภาคใหม่ใช้นัยสำคัญของการค้นพบ ถ้าหลังตัดแล้วคาดว่าจะมีสัญญาณ $S$ และฉากหลัง $B$ เหตุการณ์ นัยสำคัญโดยประมาณ (Cowan et al., 2011) คือ

$$Z_A = \sqrt{2\left[(S+B)\ln\!\left(1+\frac{S}{B}\right) - S\right]} \;\xrightarrow{S \ll B}\; \frac{S}{\sqrt{B}}$$

**ประเด็นที่สำคัญที่สุด:** $S$ และ $B$ ต้องเป็นจำนวนที่ **คาดว่าจะพบในข้อมูลจริง** ซึ่งคำนวณจากภาคตัดขวาง ความส่องสว่างสะสม และประสิทธิภาพของการคัดเลือก ไม่ใช่จำนวนในชุดข้อมูลตรวจสอบ โค้ดที่ 6.6 สมมติว่าหลังการคัดเลือกเบื้องต้น ข้อมูลจริงมีสัญญาณที่คาดไว้ 50 และฉากหลัง 5,000 เหตุการณ์

### โค้ดที่ 6.6 — เส้นโค้ง ROC พื้นที่ใต้เส้นโค้ง และการเลือกเกณฑ์จากนัยสำคัญ

```python
from sklearn.metrics import roc_curve, roc_auc_score

y_score = rf_pipe.predict_proba(X_val)[:, 1]
fpr, tpr, thresholds = roc_curve(y_val, y_score)
auc = roc_auc_score(y_val, y_score)

# expected yields in real data after preselection (NOT validation-set counts)
s_exp, b_exp = 50.0, 5000.0
S = s_exp * tpr                         # expected signal passing each cut
B = b_exp * fpr                         # expected background passing each cut

ok = B >= 10.0                          # avoid cuts that leave almost no background
Z_A = np.zeros_like(S)
Z_A[ok] = np.sqrt(2 * ((S[ok] + B[ok]) * np.log1p(S[ok] / B[ok]) - S[ok]))
i_opt = np.argmax(Z_A)
print(f"AUC = {auc:.3f}")
print(f"best cut {thresholds[i_opt]:.3f}: eff_S = {tpr[i_opt]:.3f}, "
      f"eff_B = {fpr[i_opt]:.4f}, S = {S[i_opt]:.1f}, B = {B[i_opt]:.1f}, "
      f"Z = {Z_A[i_opt]:.2f}")
```

```python
def z_asimov(S, B):
    return np.sqrt(2 * ((S + B) * np.log1p(S / B) - S))

print(f"no cut: S/sqrt(B) = {s_exp / np.sqrt(b_exp):.2f}, Z_A = {z_asimov(s_exp, b_exp):.2f}")
print(f"gain in significance from the classifier: x{Z_A[i_opt] / z_asimov(s_exp, b_exp):.1f}")
z_simple = np.zeros_like(S); z_simple[ok] = S[ok] / np.sqrt(B[ok])
j = np.argmax(z_simple)
print(f"maximising S/sqrt(B) instead: cut {thresholds[j]:.3f}, B = {B[j]:.1f}, S/sqrt(B) = {z_simple[j]:.2f}, "
      f"but Z_A there = {Z_A[j]:.2f} (< {Z_A[i_opt]:.2f})")

purity_real = S / np.where(S + B > 0, S + B, 1)
sel = thresholds <= 1
fig, ax = plt.subplots(1, 3, figsize=(15, 3.8))
ax[0].plot(fpr, tpr); ax[0].plot([0, 1], [0, 1], "k:"); ax[0].plot(fpr[i_opt], tpr[i_opt], "ro")
ax[0].set_xlabel("background efficiency (FPR)"); ax[0].set_ylabel("signal efficiency (TPR)"); ax[0].set_title(f"ROC, AUC = {auc:.3f}")
ax[1].plot(thresholds[sel], Z_A[sel]); ax[1].axvline(thresholds[i_opt], ls=":")
ax[1].set_xlabel("threshold"); ax[1].set_ylabel(r"$Z_A$ (50 signal, 5000 background expected)")
ax[2].plot(thresholds[sel], purity_real[sel], label="purity (expected real data)")
ax[2].plot(thresholds[sel], tpr[sel], label="signal efficiency"); ax[2].axvline(thresholds[i_opt], ls=":")
ax[2].set_xlabel("threshold"); ax[2].legend(fontsize=8)
fig.suptitle("random forest on toy Higgs data (SYNTHETIC)"); plt.tight_layout(); plt.show()
```

ถ้าไม่ตัดเลย นัยสำคัญคือราว 0.7 ซึ่งอ้างการค้นพบไม่ได้ เกณฑ์ที่ดีที่สุดอยู่ที่คะแนนราว 0.89 เก็บสัญญาณไว้ร้อยละ 64.5 ปล่อยฉากหลังผ่านเพียงร้อยละ 2 ให้ $Z_A \approx 3.1$ วิธีจำแนกจึงเพิ่มนัยสำคัญได้ราวสี่เท่า เกณฑ์ที่ดีที่สุดไม่ใช่ 0.5 และไม่ได้อยู่ที่จุดที่ F1 สูงสุด เพราะเป้าหมายทางฟิสิกส์ต่างจากสิ่งที่ F1 วัด เงื่อนไข `B >= 10.0` มีเหตุผลทางฟิสิกส์ เพราะเมื่อเกณฑ์เข้มมาก การประมาณ $B$ มาจากเหตุการณ์จำลองเพียงไม่กี่เหตุการณ์และไม่แน่นอนมาก

**การตั้งค่าเพื่อให้ทันเวลาในชั้นเรียน:** ตำราใช้ข้อมูลทุกแถว (ราว 100,000 แถว) และแนะนำให้ SVM ฝึกด้วยข้อมูลย่อย 10,000 แถวเพราะช้า เซลล์ถัดไปทำสองอย่าง คือ (1) สุ่มตาราง `N_SDSS` แถวมาใช้ (ค่าตั้งต้น 20,000 แถว เปลี่ยนเป็น `None` เพื่อใช้ทุกแถวตามตำรา ซึ่งบน Colab ใช้ราว 5 นาที) และ (2) ห่อ `SVC` ให้ฝึกด้วยข้อมูลย่อยไม่เกิน 10,000 แถว แบบเดียวกับสคริปต์ `07_sdss_compare.py` โค้ดที่ 6.8 จึงคงเดิมทุกบรรทัด

```python
N_SDSS = 20_000          # book: all rows -> None (about 5 minutes on Colab)
if N_SDSS:
    df = df.sample(N_SDSS, random_state=0)
print(f"SDSS rows used: {len(df)}")

from sklearn.svm import SVC as _SVC

class SVC(_SVC):
    # the book's advice: train the SVM on a random subset of at most 10,000 rows
    def fit(self, X, y, sample_weight=None):
        idx = np.random.default_rng(0).permutation(len(X))[:10_000]
        return super().fit(X[idx], y[idx])
```

### โค้ดที่ 6.8 — เปรียบเทียบวิธีจำแนกห้าวิธีสำหรับปัญหาหลายประเภท
⏱ ใช้เวลาราว 20–40 วินาทีบน Colab ด้วยค่าตั้งต้น

```python
from sklearn.model_selection import cross_val_score, StratifiedKFold

features = ['u_g', 'g_r', 'r_i', 'i_z', 'r', 'redshift']
X = df[features].values
y = df['class'].to_numpy()          # a plain NumPy array of labels
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=42)

classifiers = {
    'LogReg':     LogisticRegression(max_iter=1000),
    'KNN':        KNeighborsClassifier(n_neighbors=15),
    'SVM (RBF)':  SVC(kernel='rbf', random_state=42),
    'DecTree':    DecisionTreeClassifier(max_depth=8, random_state=42),
    'RandForest': RandomForestClassifier(n_estimators=200, random_state=42,
                                         n_jobs=-1),
}
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
results = {}
for name, clf in classifiers.items():
    pipe = Pipeline([('scaler', StandardScaler()), ('clf', clf)])
    scores = cross_val_score(pipe, X_train, y_train, cv=cv, scoring='f1_macro')
    results[name] = (scores.mean(), scores.std())
    print(f"{name:11s} macro-F1 = {scores.mean():.3f} +/- {scores.std():.3f}")
```

โค้ดใช้ `to_numpy()` แทน `.values` เพราะ pandas รุ่น 3 ที่ติดตั้ง pyarrow ไว้จะคืนอาร์เรย์ของ Arrow สำหรับหลักข้อความ ซึ่ง `train_test_split` แบ่งไม่ได้

ผลที่ได้ต้องอ่านอย่างระมัดระวัง เรดชิฟต์ในตารางได้จากการวัดสเปกตรัม และสเปกตรัมก็คือสิ่งที่ SDSS ใช้กำหนดประเภทของวัตถุตั้งแต่แรก ดาวฤกษ์มีเรดชิฟต์ใกล้ศูนย์เสมอ เรดชิฟต์จึงเกือบบอกคำตอบไว้แล้ว เซลล์ถัดไปฝึกป่าสุ่มบนข้อมูลฝึก แล้วประเมินบนข้อมูลทดสอบ ทั้งแบบใช้และไม่ใช้เรดชิฟต์ พร้อมเมทริกซ์ความสับสนและความสำคัญของลักษณะเด่น (ใช้เวลาราว 10–20 วินาที)

```python
from sklearn.metrics import f1_score
labels = ['GALAXY', 'STAR', 'QSO']
sdss_res = {}
fig, axes = plt.subplots(1, 2, figsize=(11, 4.3))
for ax, feats, tag in [(axes[0], features, 'with redshift'),
                       (axes[1], [f for f in features if f != 'redshift'], 'without redshift')]:
    Xa = df[feats].values
    Xtr, Xte, ytr, yte = train_test_split(Xa, y, test_size=0.25, stratify=y, random_state=42)
    rf = Pipeline([('scaler', StandardScaler()),
                   ('clf', RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1))]).fit(Xtr, ytr)
    yp = rf.predict(Xte)
    cm = confusion_matrix(yte, yp, labels=labels)
    ConfusionMatrixDisplay(cm, display_labels=labels).plot(ax=ax, colorbar=False)
    f1m = f1_score(yte, yp, average='macro')
    ax.set_title(f"{tag}: macro-F1 = {f1m:.3f}")
    imp = rf.named_steps['clf'].feature_importances_
    off = {f"{a}->{b}": cm[i, j] for i, a in enumerate(labels) for j, b in enumerate(labels) if i != j}
    sdss_res[tag] = dict(f1=f1m, cm=cm, imp=dict(zip(feats, imp)), worst=max(off, key=off.get))
    print(f"[{tag}] macro-F1 = {f1m:.3f}; largest confusion {sdss_res[tag]['worst']} ({off[sdss_res[tag]['worst']]}); importance: "
          + ", ".join(f"{f} {v:.2f}" for f, v in sorted(zip(feats, imp), key=lambda t: -t[1])))
fig.suptitle("random forest, SDSS test set (" + ("SYNTHETIC stand-in" if "SYNTHETIC" in _src else "real table") + ")")
plt.tight_layout(); plt.show()
```

> **ความเข้าใจคลาดเคลื่อนที่พบบ่อย:** "ความแม่นยำสูงจากการใช้เรดชิฟต์แปลว่าจำแนกจากภาพถ่ายได้ดี" ถ้ามีสเปกตรัมแล้ว เราแทบไม่ต้องใช้วิธีจำแนกเลย ปัญหาที่มีประโยชน์ทางวิทยาศาสตร์คือการจำแนกวัตถุหลายร้อยล้านชิ้นที่ **ยังไม่มี** สเปกตรัม การใส่เรดชิฟต์จากสเปกตรัมจึงเป็นการรั่วไหลของข้อมูลในความหมายกว้าง คือใช้ข้อมูลที่จะไม่มีเมื่อนำไปใช้จริง

**เทียบกับตำรา:** กับตาราง Kaggle จริง (ข้อมูลทุกแถว) ป่าสุ่มได้ macro-F1 ราว 0.976 เมื่อใช้เรดชิฟต์ โดยเรดชิฟต์มีความสำคัญ 0.56 และความสับสนที่เหลือเกือบทั้งหมดอยู่ระหว่างกาแล็กซีกับเควซาร์ เมื่อไม่ใช้เรดชิฟต์ macro-F1 ลดเหลือราว 0.854 ความสับสนกระจายทุกคู่ คู่ที่มากที่สุดคือดาวฤกษ์ที่ถูกจัดเป็นกาแล็กซี และ $r - i$ สำคัญกว่า $u - g$ ข้อมูลแทนสังเคราะห์ให้ภาพเชิงคุณภาพเดียวกัน (macro-F1 ลดลงเมื่อตัดเรดชิฟต์ และคู่ดาวฤกษ์–กาแล็กซีสับสนมากที่สุด) แต่ตัวเลขสูงกว่า เพราะสีของข้อมูลแทนสร้างให้แยกกันง่ายกว่าท้องฟ้าจริง

### ✏️ E6.6 SMOTE เทียบกับการถ่วงน้ำหนักประเภท (20 นาที)
เซลล์ถัดไปสร้างข้อมูลฮิกส์จำลองที่ไม่สมดุลมากขึ้น (สัญญาณ 400 ฉากหลัง 40,000) ฝึกป่าสุ่มสามแบบ คือไม่แก้ความไม่สมดุล ใช้ `class_weight='balanced'` และใช้ SMOTE ในไปป์ไลน์ของ imbalanced-learn (ซึ่งสร้างตัวอย่างเฉพาะในข้อมูลฝึก) แล้วเทียบ AUC ประสิทธิภาพ และความบริสุทธิ์ที่เกณฑ์ 0.5 พร้อมวาด $\Delta R$ เทียบกับ $p_T$ ของตัวอย่างสังเคราะห์ ⏱ ใช้เวลาราว 1 นาทีบน Colab อ่านผลแล้วตอบคำถามในเซลล์ ✏️

```python
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline
from sklearn.metrics import recall_score

dfi = make_higgs_toy(n_sig=400, n_bkg=40000)
Xi, yi = dfi[features].values, dfi['label'].values
Xi_tr, Xi_va, yi_tr, yi_va = train_test_split(Xi, yi, test_size=0.3, stratify=yi, random_state=42)
rfp = dict(n_estimators=200, max_depth=10, min_samples_leaf=10, random_state=42, n_jobs=-1)
imb_models = {
    'no correction': Pipeline([('scaler', StandardScaler()), ('clf', RandomForestClassifier(**rfp))]),
    'class_weight':  Pipeline([('scaler', StandardScaler()), ('clf', RandomForestClassifier(class_weight='balanced', **rfp))]),
    'SMOTE':         ImbPipeline([('scaler', StandardScaler()), ('smote', SMOTE(random_state=42)),
                                  ('clf', RandomForestClassifier(**rfp))]),
}
imb_res = {}
print("model           AUC    efficiency  purity   (threshold 0.5, validation set)")
for name, m_ in imb_models.items():
    p_ = m_.fit(Xi_tr, yi_tr).predict_proba(Xi_va)[:, 1]; yp_ = (p_ >= 0.5).astype(int)
    imb_res[name] = (roc_auc_score(yi_va, p_), recall_score(yi_va, yp_), precision_score(yi_va, yp_, zero_division=0))
    print(f"{name:14s}  {imb_res[name][0]:.3f}   {imb_res[name][1]:.3f}      {imb_res[name][2]:.3f}")

sc = StandardScaler().fit(Xi_tr)
Xr, yr = SMOTE(random_state=42).fit_resample(sc.transform(Xi_tr), yi_tr)
synth = sc.inverse_transform(Xr[len(Xi_tr):]); real = Xi_tr[yi_tr == 1]
rel = lambda A: A[:, 2] * A[:, 1] / (2 * A[:, 0])
lo_, hi_ = np.percentile(rel(real), [1, 99])
print(f"\n{len(synth)} synthetic signal events; fraction outside the 1-99 % band of dR*pT/(2 m_bb) of real signal: "
      f"{np.mean((rel(synth) < lo_) | (rel(synth) > hi_)):.3f}")
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(synth[:3000, 1], synth[:3000, 2], s=3, alpha=0.3, label='SMOTE synthetic')
ax.scatter(real[:, 1], real[:, 2], s=6, c='k', label='real signal')
pt = np.linspace(60, 600, 200); ax.plot(pt, 250 / pt, 'r--', label=r'$2 m_H / p_T$')
ax.set_xlabel(r'$p_T$ (GeV)'); ax.set_ylabel(r'$\Delta R$'); ax.set_ylim(0, 4.2); ax.legend()
ax.set_title("SMOTE events vs real signal (SYNTHETIC)"); plt.tight_layout(); plt.show()
```

```python
my_same_auc = None       # ✏️ True/False: AUC ของสามแบบใกล้เคียงกัน (ต่างกันไม่เกิน 0.01)
my_best_eff = "?"        # ✏️ แบบที่ได้ประสิทธิภาพสูงสุดที่เกณฑ์ 0.5: 'no correction', 'class_weight' หรือ 'SMOTE'
```

```python
#@title ✅ ตรวจคำตอบ
_say = _gate(['my_best_eff', 'my_same_auc'])
_aucs = [v[0] for v in imb_res.values()]
_same = (max(_aucs) - min(_aucs)) <= 0.01
_be = max(imb_res, key=lambda k: imb_res[k][1])
_say("AUC ใกล้เคียงกัน:", "(ยังไม่ได้ตอบ)" if my_same_auc is None else ("✔" if my_same_auc == _same else "✘"),
      f"— ช่วง {min(_aucs):.3f}-{max(_aucs):.3f}; AUC วัดการเรียงลำดับ ซึ่งการแก้ความไม่สมดุลเปลี่ยนไม่มาก")
_say("ประสิทธิภาพสูงสุด:", "(ยังไม่ได้ตอบ)" if my_best_eff == "?" else ("✔" if my_best_eff == _be else "✘"), f"— {_be}")
_say("สองแบบหลังเพียงเลื่อนเกณฑ์โดยนัย ซึ่งทำได้เท่ากันด้วยการเลือกเกณฑ์เองจากนัยสำคัญกับแบบแรก")
```

การจำแนกในฟิสิกส์มีลักษณะเฉพาะ คือประเภทไม่สมดุลตามธรรมชาติ ความผิดพลาดสองชนิดมีต้นทุนต่างกัน และผลต้องตีความได้ ลักษณะเด่นจากฟิสิกส์ เช่น มวลยืนยงซึ่งไม่แปรเปลี่ยนภายใต้การแปลงลอเรนตซ์ และดัชนีสีซึ่งตัดผลของระยะทาง เป็นหัวใจของการจำแนกที่ดี และทำให้วิธีเชิงเส้นซึ่งตีความง่ายกลับมาใช้ได้ (AUC 0.70 → 0.93) วิธีจำแนกทั้งห้าต่างกันที่รูปร่างของขอบเขตการตัดสินใจ ความแม่นยำต้องเทียบกับเส้นฐานเสมอ เกณฑ์ตัดสินต้องเลือกจากเป้าหมายทางฟิสิกส์ เช่น $Z_A$ ที่คำนวณจากจำนวนที่คาดในข้อมูลจริง และลักษณะเด่นที่ทำให้ผลดีที่สุดอาจเป็นข้อมูลที่จะไม่มีเมื่อนำไปใช้จริง
