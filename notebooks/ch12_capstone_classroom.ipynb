## วัตถุประสงค์
เมื่อทำโน้ตบุ๊กนี้จนจบ ผู้เรียนควรสามารถ

1. ระบุความเสี่ยงเชิงจริยธรรมของข้อมูลทางฟิสิกส์ได้ ทั้งอคติจากการเลือกตัวอย่าง ความเป็นส่วนตัว การใช้งานสองทาง เครดิต และต้นทุนของการคำนวณ
2. อ่านสัญญาอนุญาตของข้อมูลและซอฟต์แวร์ และตรวจชุดข้อมูลตามหลักการ FAIR ได้

6. รายงานผลโดยแยกสิ่งที่ข้อมูลบอกได้ออกจากข้อจำกัด

### การล็อกแนวคำตอบ
เซลล์ ✅ และเซลล์แนวคำตอบในโน้ตบุ๊กนี้แสดงคำตอบอ้างอิงเมื่อผู้เรียนเติมคำตอบในเซลล์ ✏️ ของข้อนั้นครบแล้วเท่านั้น ถ้ายังไม่ได้ตอบ เซลล์จะแสดงเพียงข้อความ 🔒 บอกว่ายังขาดคำตอบใด การคำนวณในเซลล์ยังทำงานตามปกติ ดังนั้น Run all จึงรันได้ครบโดยไม่เปิดเผยคำตอบ แนวคำตอบของงานเขียนและงานนอกชั้นซึ่งตรวจด้วยโค้ดไม่ได้จะเปิดเมื่อผู้สอนเลือกเท่านั้น

**สำหรับผู้สอน:** ติ๊กช่อง `SHOW_ANSWERS` ในเซลล์ถัดไป (หรือแก้เป็น `True`) แล้วรันเซลล์นั้นอีกครั้ง เพื่อแสดงแนวคำตอบทุกข้อ เช่น เมื่อเฉลยในชั้นหรือเตรียมสอน

```python
#@title ⚙️ ตัวเลือกของผู้สอน: แสดงแนวคำตอบ
SHOW_ANSWERS = False  #@param {type:"boolean"}

def _answered(v):
    """True when an exercise variable holds an answer (not None, "?", "", empty or containing blanks)."""
    if v is None or v is Ellipsis:
        return False
    if isinstance(v, str):                       # placeholders such as "?", "?-?", "..." count as blank
        return v.replace("?", "").replace("-", "").replace(".", "").strip() != ""
    if isinstance(v, dict):
        return len(v) > 0 and all(_answered(x) for x in v.values())
    if isinstance(v, (list, tuple, set, frozenset)):
        return len(v) > 0 and all(_answered(x) for x in v)
    if type(v).__name__ == "ndarray":
        return v.size > 0
    return True

def _gate(names=()):
    """Return print if the answers may be shown, otherwise print one lock message and return a no-op."""
    if SHOW_ANSWERS:
        return print
    if names:
        missing = [n for n in names if not _answered(globals().get(n))]
        if not missing:
            return print
        print("🔒 ยังไม่ได้ตอบ:", ", ".join(missing),
              "— เติมคำตอบในเซลล์ ✏️ ด้านบนให้ครบก่อน แล้วรันเซลล์นี้อีกครั้ง")
    else:
        print("🔒 แนวคำตอบข้อนี้ผู้สอนเป็นผู้เปิด (ตั้ง SHOW_ANSWERS = True ในส่วนที่ 0)")
    return lambda *args, **kwargs: None

print("แสดงแนวคำตอบทั้งหมด:", "ใช่ (โหมดผู้สอน)" if SHOW_ANSWERS else "ไม่ (แสดงเมื่อผู้เรียนตอบข้อนั้นครบแล้ว)")
```

เซลล์ซ่อนถัดไปสร้างข้อมูลฮิกส์จำลองของบทที่ 6 (โค้ดที่ 6.1) เป็นตาราง `df` ขนาด 20,000 เหตุการณ์ ซึ่งโค้ดที่ 12.1 และ 12.2 ใช้ ดับเบิลคลิกเพื่อดูโค้ดได้

```python
#@title ข้อมูลฮิกส์จำลองจากโค้ดที่ 6.1 (ดับเบิลคลิกเพื่อดูโค้ด)
import numpy as np
import pandas as pd

def make_higgs_toy(n_sig=4000, n_bkg=16000, seed=2026):
    """Toy H->bb events vs background (GeV). Shapes are illustrative only."""
    rng = np.random.default_rng(seed)
    # signal: resonance at 125 GeV with ~10% resolution
    m_s   = rng.normal(125.0, 12.0, n_sig)
    pT_s  = rng.gamma(shape=4.0, scale=40.0, size=n_sig)     # mean 160 GeV
    dR_s  = 2 * m_s / np.maximum(pT_s, 60.0) * rng.normal(1.0, 0.15, n_sig)
    met_s = rng.exponential(20.0, n_sig)                       # semileptonic b
    # background: falling mass spectrum, softer pT, wider angles
    m_b   = 40.0 + rng.exponential(70.0, n_bkg)
    pT_b  = rng.gamma(shape=3.0, scale=40.0, size=n_bkg)     # mean 120 GeV
    dR_b  = rng.uniform(0.4, 4.0, n_bkg)
    met_b = rng.exponential(30.0, n_bkg)                       # e.g. ttbar
    df = pd.DataFrame({
        'm_bb':   np.concatenate([m_s, m_b]),
        'pT':     np.concatenate([pT_s, pT_b]),
        'dR':     np.clip(np.concatenate([dR_s, dR_b]), 0.4, 4.0),
        'ETmiss': np.concatenate([met_s, met_b]),
        'label':  np.concatenate([np.ones(n_sig, int), np.zeros(n_bkg, int)]),
    })
    return df.sample(frac=1.0, random_state=seed).reset_index(drop=True)

df = make_higgs_toy()

print('Higgs toy data (SYNTHETIC):', df.shape, '| signal fraction', df.label.mean().round(3))
```

| เครดิตและความรับผิดชอบ | ทีมสร้างเครื่องมือ ทีมเก็บข้อมูล อาสาสมัครที่ให้ป้ายกำกับ โปรแกรมช่วยเขียนโค้ดหรือเรียบเรียงข้อความ | อ้างอิงทุกแหล่ง และรับผิดชอบความถูกต้องของทุกบรรทัดที่ใช้เครื่องมือช่วย พร้อมเปิดเผยการใช้ตามแนวทางของวารสารหรือสถาบัน |
| ต้นทุนของการคำนวณ | การฝึกแบบจำลองขนาดใหญ่ใช้พลังงานไฟฟ้ามาก | ถ้าวิธีที่ง่ายกว่าให้ผลเท่ากัน ดังที่หลายบทในตำรานี้แสดง ไม่ควรใช้วิธีที่ใหญ่กว่า |

> **ความเข้าใจคลาดเคลื่อนที่พบบ่อย:** "ข้อมูลทางฟิสิกส์เป็นกลาง จึงไม่มีประเด็นจริยธรรม" ข้อมูลทุกชุดผ่านการตัดสินใจของมนุษย์ ทั้งการเลือกว่าจะวัดอะไร วัดที่ไหน วัดด้วยเครื่องมือใด และคัดข้อมูลใดทิ้ง การตัดสินใจเหล่านี้ฝังอคติไว้ในข้อมูล แม้ข้อมูลจะเป็นภาพของกาแล็กซีที่อยู่ห่างออกไปหลายร้อยล้านปีแสง ประเด็นจริยธรรมในงานฟิสิกส์ส่วนใหญ่จึงอยู่ที่การรายงาน การปกปิดอคติที่รู้ การเลือกรายงานเฉพาะผลที่ดี และการไม่ให้เครดิตแก่ผู้สร้างข้อมูล ทำลายความน่าเชื่อถือของวิทยาศาสตร์โดยตรง ไม่ว่าข้อมูลจะมาจากมนุษย์หรือจากดวงดาว

### สัญญาอนุญาตและวิทยาศาสตร์แบบเปิด
ข้อมูลและโค้ดที่ดาวน์โหลดได้จากอินเทอร์เน็ตไม่ได้แปลว่าใช้ได้อย่างอิสระ สิ่งที่กำหนดว่าทำอะไรได้คือสัญญาอนุญาต (license) ซึ่งเจ้าของงานเลือกไว้ เปรียบได้กับหนังสือในห้องสมุด ทุกคนหยิบอ่านได้ แต่การถ่ายสำเนาทั้งเล่มไปขายต่อเป็นอีกเรื่องหนึ่ง

| สัญญาอนุญาต | ใช้กับ | ข้อกำหนดหลัก |
|---|---|---|
| MIT, BSD | ซอฟต์แวร์ | ใช้ ดัดแปลง และเผยแพร่ได้ ต้องคงข้อความลิขสิทธิ์ ไม่มีการรับประกัน |
| Apache 2.0 | ซอฟต์แวร์ | เหมือน MIT และให้สิทธิ์ในสิทธิบัตรที่เกี่ยวข้อง |
| GPL v3 | ซอฟต์แวร์ | งานที่ดัดแปลงและเผยแพร่ต่อต้องใช้ GPL ด้วย |
| CC BY 4.0 | ข้อมูล เอกสาร | ใช้และดัดแปลงได้ ต้องอ้างถึงที่มาและระบุว่าเปลี่ยนแปลงอะไร |
| CC0 | ข้อมูล | สละสิทธิ์ทั้งหมด ไม่มีข้อบังคับ แต่การอ้างอิงยังเป็นมารยาททางวิชาการ |

ชุดข้อมูลที่ใช้ในตำรานี้แสดงความหลากหลายของสัญญาอนุญาต ข้อมูลและเมทาดาทาบนพอร์ทัล CERN Open Data (ข้อมูลเปิดของ CMS ในบทที่ 1) เผยแพร่ภายใต้การสละสิทธิ์แบบ CC0 ขณะที่ซอฟต์แวร์บนพอร์ทัลเดียวกันมักใช้ GPL ส่วน Galaxy10 DECaLS ในบทที่ 11 ใช้ CC BY 4.0 ข้อสังเกตสองข้อตามมา ข้อแรก สัญญาอนุญาตของข้อมูลกับของโค้ดที่ใช้วิเคราะห์ข้อมูลเป็นคนละเรื่อง และต้องตรวจแยกกัน ข้อที่สอง แม้ CC0 จะไม่บังคับให้อ้างอิง การไม่อ้างอิงแหล่งข้อมูลในงานวิชาการยังเป็นการละเมิดจรรยาบรรณ

วิทยาศาสตร์แบบเปิด (open science) ขยายความคิดนี้ไปทั้งกระบวนการ คือเปิดข้อมูล เปิดโค้ด เปิดขั้นตอนการวิเคราะห์ และเผยแพร่ผลในที่ที่ผู้อื่นเข้าถึงได้ ผลที่ผู้อื่นตรวจไม่ได้ก็ไม่ต่างจากคำกล่าวอ้าง การเปิดไม่ได้หมายความว่าทุกอย่างต้องเปิดทันที ข้อมูลที่มีข้อมูลส่วนบุคคลหรือที่อยู่ระหว่างการตีพิมพ์อาจเปิดภายหลังหรือเปิดแบบมีเงื่อนไขได้ แต่ควรมีแผนที่ชัดเจนตั้งแต่ต้น

### ✏️ แบบฝึก 12.B (5 นาที) อ่านสัญญาอนุญาต
เติมคำตอบใน `my_license` ตามตารางข้างบน

2. ข้อมูลที่ใช้ CC0 ต้องอ้างอิงตาม **กฎหมาย** หรือไม่ (`True` / `False`)
3. ตัดภาพบางภาพออกจากชุดข้อมูลที่ใช้ CC BY 4.0 แล้วเผยแพร่ชุดที่ตัดแล้ว ต้องระบุการเปลี่ยนแปลงหรือไม่ (`True` / `False`)
4. สัญญาอนุญาตซอฟต์แวร์ใดในตารางให้สิทธิ์ในสิทธิบัตรที่เกี่ยวข้องด้วย

```python
my_license = {1: None, 2: None, 3: None, 4: None}   # ✏️
```

```python
#@markdown ✅ ตรวจคำตอบ
_say = _gate(['my_license'])
_ok = {1: "GPL v3", 2: False, 3: True, 4: "Apache 2.0"}
_norm = lambda v: v.replace(" ", "").lower() if isinstance(v, str) else v
for k, ok in _ok.items():
    mine = my_license.get(k)
    _say(f"ข้อ {k}:", "(ยังไม่ได้ตอบ)" if mine is None else ("✔" if _norm(mine) == _norm(ok) else "✘"))
```

### ข้อมูล AFM ของผู้เขียน: สายการอ้างอิงที่ครบถ้วน
ข้อมูลเส้นโค้งแรงกับระยะจากกล้องจุลทรรศน์แรงอะตอมที่ใช้ในบทที่ 3 และแนวคิดของตัวอย่างในบทที่ 7 มาจากงานวิจัยของผู้เขียน ชุดข้อมูลกำลังจัดเตรียมเพื่อเผยแพร่บน Zenodo โดยวางแผนใช้สัญญาอนุญาต CC BY 4.0 ระหว่างนี้ ตำราจึงใช้ข้อมูลจำลองที่ออกแบบให้มีลักษณะสำคัญของปัญหาจริง และไม่ใส่ตัวเลขผลที่ยังไม่ได้ตีพิมพ์ การตัดสินใจนี้เป็นตัวอย่างของการจัดการข้อมูลที่อยู่ระหว่างการวิจัยอย่างรับผิดชอบ

เมื่อชุดข้อมูลเผยแพร่แล้ว ผู้ที่นำไปใช้ควรทำสี่อย่าง ซึ่งรวมกันเรียกว่าสายการอ้างอิง (attribution chain)

1. อ้างอิงชุดข้อมูลด้วยตัวระบุวัตถุดิจิทัล (DOI) ของรุ่นที่ใช้ ไม่ใช่เพียงชื่อโครงการ
2. ระบุสัญญาอนุญาตในเอกสารของงาน
3. ระบุว่าเปลี่ยนแปลงข้อมูลอย่างไร เช่น ตัดเส้นโค้งใดทิ้ง ปรับมาตราส่วนอย่างไร และสร้างลักษณะเด่นใดเพิ่ม ซึ่งเป็นข้อกำหนดของ CC BY 4.0 โดยตรง
4. เผยแพร่โค้ด สภาพแวดล้อม และเมล็ดสุ่มที่ใช้ เพื่อให้ผู้อื่นทำซ้ำได้ (คาบที่ 2)

### หลักการ FAIR
การเปิดข้อมูลเพียงอย่างเดียวไม่ได้ทำให้ข้อมูลใช้ได้ วิลคินสันและคณะ (Wilkinson et al., 2016) จึงเสนอหลักการ FAIR

> **นิยาม (หลักการ FAIR)** ข้อมูลที่ดีควร **ค้นพบได้ (Findable)** คือมีตัวระบุถาวร เช่น DOI และมีเมทาดาทาที่ค้นหาได้ ควร **เข้าถึงได้ (Accessible)** คือดาวน์โหลดได้ด้วยโพรโทคอลมาตรฐาน และเมทาดาทายังอยู่แม้ข้อมูลจะถูกถอนออก ควร **ทำงานร่วมกันได้ (Interoperable)** คือใช้รูปแบบไฟล์และคำศัพท์มาตรฐานของสาขา เช่น FITS ในดาราศาสตร์ หรือ HDF5 และควร **นำกลับมาใช้ได้ (Reusable)** คือมีสัญญาอนุญาตที่ชัด มีประวัติที่มาของข้อมูล (provenance) และเป็นไปตามมาตรฐานของชุมชน

เซลล์ถัดไปดาวน์โหลดตัวอย่างย่อยของ Galaxy10 DECaLS ที่ใช้ในบทที่ 1 และ 11 แล้วอ่านเฉพาะเมทาดาทา ได้แก่ ชื่อชุดข้อมูลย่อยในไฟล์ และคุณสมบัติ (attributes) ที่ผู้เขียนบันทึกไว้เมื่อสร้างตัวอย่างย่อย ซึ่งเป็นตัวอย่างขนาดเล็กของการบันทึกสัญญาอนุญาตและประวัติที่มาไว้ในไฟล์ข้อมูลเอง

```python
import shutil, urllib.request, h5py
os.makedirs("outputs", exist_ok=True)
_dst = "outputs/galaxy10_decals_sample.h5"
_local = "../ch01_intro/sample/galaxy10_decals_sample.h5"
_url = ("https://raw.githubusercontent.com/busarapa-stack/dsml-for-physicists/main/"
        "ch01_intro/sample/galaxy10_decals_sample.h5")
if not os.path.exists(_dst):
    if os.path.exists(_local):
        shutil.copy(_local, _dst); print("copied the sample from", _local)
    else:
        urllib.request.urlretrieve(_url, _dst); print("downloaded the sample from GitHub")
with h5py.File(_dst, "r") as f:
    print("datasets in the file:", {k: f[k].shape for k in f.keys()})
    print("\nfile attributes (metadata written with the data):")
    for k, v in f.attrs.items():
        print(f"  {k:12s}: {v}")
```

ไฟล์นี้บันทึกไว้ในตัวเองว่าเป็นชุดย่อยของอะไร ใช้สัญญาอนุญาตใด ภาพและป้ายกำกับมาจากโครงการใด และสร้างด้วยสคริปต์ใด ผู้ที่ได้ไฟล์ไปแม้ไม่มีเอกสารอื่นก็รู้ว่าต้องอ้างถึงใครและใช้ได้ภายใต้เงื่อนไขใด ลองตรวจ Galaxy10 DECaLS ฉบับเต็มตามหลักการ FAIR ชุดข้อมูลมี DOI บน Zenodo จึงค้นพบได้ ดาวน์โหลดได้โดยตรงจึงเข้าถึงได้ ใช้รูปแบบ HDF5 จึงทำงานร่วมกันได้ และมีสัญญาอนุญาต CC BY 4.0 พร้อมการอ้างถึงงานต้นทางของภาพและป้ายกำกับ จึงนำกลับมาใช้ได้ จุดที่ยังปรับปรุงได้คือรายละเอียดของเกณฑ์ที่ใช้คัดภาพจากคะแนนของอาสาสมัคร ซึ่งผู้ใช้ต้องไปค้นในงานต้นทาง

### ✏️ แบบฝึก 12.C (5 นาที) ตรวจตามหลักการ FAIR
ชุดข้อมูลสมมติชุดหนึ่งมีลักษณะดังนี้ "ไฟล์ข้อมูลในรูปแบบเฉพาะของซอฟต์แวร์ที่มากับเครื่องมือวัด อัปโหลดไว้ในเว็บไซต์ส่วนตัวของผู้วิจัย ดาวน์โหลดได้ด้วยลิงก์ HTTPS ธรรมดา ไม่มี DOI ไม่มีคำอธิบายคอลัมน์ และไม่ระบุสัญญาอนุญาต" ตัดสินว่าชุดข้อมูลนี้ผ่านหลักการแต่ละข้อหรือไม่ (`True` / `False`) แล้วเติมใน `my_fair`

```python
my_fair = {"F": None, "A": None, "I": None, "R": None}   # ✏️ True = passes, False = fails
```

```python
#@markdown ✅ ตรวจคำตอบ
_say = _gate(['my_fair'])
_ok = {"F": False, "A": True, "I": False, "R": False}
for k, ok in _ok.items():
    mine = my_fair.get(k)
    _say(f"{k}:", "(ยังไม่ได้ตอบ)" if mine is None else ("✔" if mine == ok else "✘"))
```

```python
#@markdown แนวคำตอบ 12.C
_say = _gate(['my_fair'])
_say('''แนวคำตอบ 12.C
F ไม่ผ่าน: ไม่มีตัวระบุถาวรและเมทาดาทาที่ค้นหาได้ ลิงก์ในเว็บส่วนตัวหายได้เมื่อผู้วิจัยย้ายที่ทำงาน
A ผ่าน (อย่างอ่อน): ดาวน์โหลดได้ด้วยโพรโทคอลมาตรฐาน แต่ไม่มีการรับประกันว่าเมทาดาทาจะยังอยู่
I ไม่ผ่าน: รูปแบบเฉพาะของซอฟต์แวร์เครื่องมือวัดต้องใช้โปรแกรมเฉพาะเปิด และไม่มีคำอธิบายคอลัมน์
R ไม่ผ่าน: ไม่มีสัญญาอนุญาต ผู้อื่นจึงไม่รู้ว่ามีสิทธิ์ใช้หรือไม่ และไม่มีประวัติที่มาของข้อมูล
ข้อมูลที่ "เปิด" แล้วจึงอาจใช้ประโยชน์ไม่ได้เลย การแก้ที่ทำได้ง่ายคือฝากไว้ใน Zenodo (ได้ DOI) แปลงเป็น CSV หรือ HDF5
พร้อมคำอธิบายคอลัมน์ และเลือกสัญญาอนุญาต เช่น CC BY 4.0''')
```

### ธรรมาภิบาลข้อมูล
ธรรมาภิบาลข้อมูล (data governance) คือนโยบาย กระบวนการ และผู้รับผิดชอบที่ดูแลข้อมูลตลอดวงจรชีวิต ตั้งแต่การเก็บ การจัดเก็บ การใช้ การแบ่งปัน จนถึงการลบหรือเก็บถาวร เครื่องมือหลักคือแผนการจัดการข้อมูล (data management plan) ซึ่งแหล่งทุนวิจัยจำนวนมากกำหนดให้ส่งพร้อมข้อเสนอโครงการ แผนนี้ตอบคำถามว่าจะเก็บข้อมูลอะไร ในรูปแบบใด เก็บไว้ที่ไหนและนานเท่าใด ใครเข้าถึงได้ จะสำรองอย่างไร และจะเผยแพร่เมื่อใดภายใต้สัญญาอนุญาตใด สำหรับโครงงานของนิสิต แผนนี้อาจยาวเพียงครึ่งหน้า แต่การเขียนตั้งแต่ต้นช่วยป้องกันปัญหาที่พบบ่อย เช่น ข้อมูลดิบถูกเขียนทับ ไม่รู้ว่าไฟล์ใดเป็นรุ่นล่าสุด หรือเผยแพร่ข้อมูลที่ไม่มีสิทธิ์เผยแพร่

### ✏️ แบบฝึก E12.1 ในชั้นเรียน (10 นาที) ตรวจข้อมูลเปิดของ CMS
แบบฝึกหัด E12.1 ในตำราให้ตรวจชุดข้อมูลที่จะใช้ในโครงงานของตนเอง ในชั้นเรียน ให้ฝึกกับข้อมูลเปิดของ CMS บนพอร์ทัล CERN Open Data ก่อน โดยเติม `my_e121` ตามข้อเท็จจริงในส่วนนี้

- `data_license`: สัญญาอนุญาตของข้อมูลและเมทาดาทา (ชื่อตามตาราง)
- `software_license`: สัญญาอนุญาตที่ซอฟต์แวร์บนพอร์ทัลมักใช้ (ชื่อตามตาราง)
- `must_cite`: ต้องอ้างอิงในงานวิชาการหรือไม่ (`True` / `False`)
- `main_bias`: อคติที่สำคัญที่สุดของข้อมูลการชนของอนุภาค เลือก `'trigger'` (เหตุการณ์ผ่านระบบคัดเลือกแบบทันทีมาแล้ว), `'malmquist'` หรือ `'none'`
- `personal_data`: มีข้อมูลส่วนบุคคลหรือไม่ (`True` / `False`)

การตรวจชุดข้อมูลของโครงงานของตนเองจริงพร้อมหลักฐานเป็นงานนอกชั้น

### วิกฤตการทำซ้ำได้
ในปี 2016 วารสาร Nature สำรวจนักวิจัยกว่า 1,500 คนจากหลายสาขา พบว่ากว่าร้อยละ 70 เคยพยายามทำการทดลองของผู้อื่นซ้ำแล้วไม่สำเร็จ และกว่าครึ่งเคยทำการทดลองของตนเองซ้ำไม่สำเร็จ ผลสำรวจนี้เป็นความเห็นของนักวิจัย ไม่ใช่การวัดอัตราที่แท้จริง แต่สะท้อนปัญหาที่ชุมชนวิทยาศาสตร์ยอมรับกันกว้างขวาง ในงานที่ใช้การเรียนรู้ของเครื่อง ปัญหานี้มีลักษณะเฉพาะ คาปูร์และนารายานัน (Kapoor and Narayanan) รวบรวมงานตีพิมพ์หลายร้อยชิ้นในหลายสาขาที่ผลดีเกินจริงเพราะการรั่วไหลของข้อมูล ซึ่งตำรานี้พบมาแล้วหลายรูปแบบ ทั้งการแบ่งข้อมูลที่สัมพันธ์กันในเวลาแบบสุ่ม การเลือกไฮเพอร์พารามิเตอร์ด้วยชุดทดสอบ (บทที่ 9) และภาพซ้ำในชุดข้อมูล (บทที่ 11)

บทที่ 2 ได้แยกการทำซ้ำได้ (reproducibility) ซึ่งคือการได้ผลเดิมจากข้อมูลและโค้ดเดิม ออกจากการทำซ้ำด้วยข้อมูลใหม่ (replicability) ซึ่งคือการได้ข้อสรุปเดิมจากการเก็บข้อมูลใหม่ และได้วางพื้นฐานของสภาพแวดล้อมเสมือน Git และเมล็ดสุ่มไว้แล้ว คาบนี้ต่อยอดด้วยสามเรื่องที่เฉพาะกับงานการเรียนรู้ของเครื่อง คือเมล็ดสุ่มแต่ละตัวมีผลต่างกันเพียงใด จะบันทึกทุกสิ่งที่จำเป็นต่อการรันซ้ำได้อย่างไร และจะตรวจการรั่วไหลจากข้อมูลซ้ำได้อย่างไร

### เมล็ดสุ่มคนละตัวมีผลคนละขนาด
งานการเรียนรู้ของเครื่องชิ้นหนึ่งมักมีแหล่งความสุ่มหลายแหล่ง ได้แก่ การแบ่งข้อมูล การเริ่มต้นน้ำหนัก การสุ่มชุดย่อย การสุ่มลักษณะเด่นของป่าสุ่ม และดรอปเอาต์ การกำหนดเมล็ดสุ่มทุกแหล่งทำให้ผลทำซ้ำได้ แต่คำถามที่สำคัญกว่าคือผลเปลี่ยนมากเพียงใดเมื่อเปลี่ยนเมล็ดสุ่ม โค้ดที่ 12.1 วัดคำถามนี้กับป่าสุ่มบนข้อมูลฮิกส์จำลอง โดยเปลี่ยนเมล็ดสุ่มทีละแหล่ง (⏱ ราวหนึ่งนาทีบน Colab)

### โค้ดที่ 12.1 — ผลของเมล็ดสุ่มแต่ละแหล่งต่อพื้นที่ใต้เส้นโค้ง ROC

```python
# ⏱ ราวหนึ่งนาทีบน Colab
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

X, y = df[['m_bb', 'pT', 'dR', 'ETmiss']].values, df['label'].values

def run(split_seed, model_seed):
    X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, stratify=y,
                                              random_state=split_seed)
    rf = RandomForestClassifier(n_estimators=100, min_samples_leaf=10,
                                random_state=model_seed, n_jobs=-1).fit(X_tr, y_tr)
    return roc_auc_score(y_te, rf.predict_proba(X_te)[:, 1])

print("same seeds twice:", run(42, 0), run(42, 0))
auc_model = [run(42, s) for s in range(10)]        # only the forest's seed changes
auc_split = [run(s, 0) for s in range(10)]         # only the data split changes
print(f"model seed: mean {np.mean(auc_model):.4f}, range "
      f"{min(auc_model):.4f}-{max(auc_model):.4f}")
print(f"split seed: mean {np.mean(auc_split):.4f}, range "
      f"{min(auc_split):.4f}-{max(auc_split):.4f}")
```

```python
fig, ax = plt.subplots(figsize=(7, 2.6))
for k, (vals, lab) in enumerate([(auc_model, "model seed only (split 42)"), (auc_split, "split seed only (model 0)")]):
    ax.scatter(vals, np.full(len(vals), k) + np.random.default_rng(k).uniform(-0.08, 0.08, len(vals)), s=30)
    ax.hlines(k, min(vals), max(vals), lw=6, alpha=0.25)
ax.set_yticks([0, 1]); ax.set_yticklabels(["model seed only (split 42)", "split seed only (model 0)"])
ax.set_ylim(-0.5, 1.5); ax.set_xlabel("ROC AUC on the test set")
ax.set_title("random forest on toy Higgs data (SYNTHETIC): which seed matters?", fontsize=9)
fig.tight_layout(); plt.show()
```

ในตำรา การรันด้วยเมล็ดสุ่มเดิมสองครั้งให้พื้นที่ใต้เส้นโค้ง ROC เท่ากันทุกหลัก การกำหนดเมล็ดสุ่มจึงทำให้ผลทำซ้ำได้จริง เมื่อเปลี่ยนเฉพาะเมล็ดสุ่มของป่าสุ่มสิบค่า พื้นที่ใต้เส้นโค้งอยู่ในช่วง 0.9759 ถึง 0.9765 กว้างเพียง 0.0006 แต่เมื่อเปลี่ยนเฉพาะเมล็ดสุ่มของการแบ่งข้อมูล พื้นที่ใต้เส้นโค้งอยู่ในช่วง 0.9770 ถึง 0.9820 กว้างเกือบสิบเท่า ความต่างนี้มีเหตุผล การแบ่งข้อมูลเปลี่ยนทั้งข้อมูลที่ใช้ฝึกและข้อมูลที่ใช้วัด ส่วนเมล็ดสุ่มของป่าสุ่มขนาด 100 ต้นเปลี่ยนเพียงรายละเอียดของต้นไม้ ซึ่งเฉลี่ยกันไปมากแล้ว

ผลนี้มีนัยต่อการรายงาน ความกว้าง 0.005 จากการแบ่งข้อมูลใหญ่กว่าผลต่างระหว่างวิธีจำแนกที่บทที่ 9 ทดสอบ (ผลต่างของ SVM กับป่าสุ่ม 0.0028) ผู้วิจัยที่ลองการแบ่งข้อมูลหลายแบบแล้วรายงานแบบที่ได้ผลดีที่สุด จะได้ตัวเลขที่ดีเกินจริงโดยไม่ได้ตั้งใจโกงเลย วิธีป้องกันคือกำหนดการแบ่งข้อมูลไว้ก่อนเริ่มวิเคราะห์ หรือใช้การตรวจสอบไขว้ซ้ำหลายรอบแล้วรายงานการกระจาย

### เมล็ดสุ่มเดิม เครื่องต่างกัน ตัวเลขต่างกัน
การคำนวณแบบขนานบนหน่วยประมวลผลหลายแกนหรือบนการ์ดประมวลผลภาพอาจให้ผลต่างกันในหลักท้าย ๆ แม้ใช้เมล็ดสุ่มเดิม เพราะลำดับการบวกเลขทศนิยมต่างกัน เซลล์ถัดไปแสดงต้นเหตุนี้ในระดับที่เล็กที่สุด คือการบวกเลขทศนิยมชุดเดียวกันในลำดับที่ต่างกัน

```python
rng_ = np.random.default_rng(0)
v = rng_.normal(size=100_000).astype(np.float32)
s_forward = np.float32(0)
for a in v[:2000]:
    s_forward += a                                   # one number at a time, front to back
s_backward = np.float32(0)
for a in v[:2000][::-1]:
    s_backward += a                                  # the same numbers, back to front
print(f"sum, front to back : {s_forward:.8f}")
print(f"sum, back to front : {s_backward:.8f}")
print(f"numpy (pairwise)   : {v[:2000].sum():.8f}")
print("identical?", s_forward == s_backward)
```

โค้ดที่ 12.2 เขียนบันทึกการรัน (run manifest) เป็นไฟล์ JSON และมีฟังก์ชันตรวจว่าชุดทดสอบมีแถวที่ซ้ำกับชุดฝึกหรือไม่ โดยเทียบค่าแฮชของแต่ละแถว โค้ดใช้ `X`, `y` และ `train_test_split` จากโค้ดที่ 12.1 เซลล์ถัดไปย้ายเข้าโฟลเดอร์ `outputs/` ก่อนรันโค้ดที่ 12.2 เพื่อให้ไฟล์ `higgs_toy.csv` และ `run_manifest.json` อยู่ที่นั่น แล้วย้ายกลับ

### โค้ดที่ 12.2 — บันทึกการรันและการตรวจแถวซ้ำข้ามชุดข้อมูล

```python
os.chdir("outputs")                                  # the files of Listing 12.2 go to outputs/
import hashlib, json, platform, sys, datetime
import numpy, scipy, sklearn

def sha256_of(path, chunk=1 << 20):
    h = hashlib.sha256()
    with open(path, 'rb') as f:
        for block in iter(lambda: f.read(chunk), b''):
            h.update(block)
    return h.hexdigest()

def write_manifest(config, data_files, out='run_manifest.json'):
    """Record what is needed to rerun this analysis exactly."""
    manifest = {
        'created': datetime.datetime.now(datetime.timezone.utc).isoformat(),
        'python': sys.version.split()[0],
        'platform': platform.platform(),
        'packages': {m.__name__: m.__version__ for m in (numpy, scipy, sklearn)},
        'config': config,                               # every seed and setting
        'data_sha256': {p: sha256_of(p) for p in data_files},
    }
    with open(out, 'w') as f:
        json.dump(manifest, f, indent=2)
    return manifest

def row_overlap(A, B, decimals=6):
    """Number of rows of B that also appear in A (duplicates across a split)."""
    key = lambda r: hashlib.md5(numpy.round(r, decimals).tobytes()).hexdigest()
    seen = {key(r) for r in A}
    return sum(key(r) in seen for r in B)

df.to_csv('higgs_toy.csv', index=False)                 # the exact data used
config = {'split_seed': 42, 'model_seed': 0, 'test_size': 0.3,
          'n_estimators': 100, 'min_samples_leaf': 10}
m = write_manifest(config, ['higgs_toy.csv'])
print(m['packages'], m['data_sha256']['higgs_toy.csv'][:16])

X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, stratify=y, random_state=42)
print("overlap, clean split:", row_overlap(X_tr, X_te))
X_dup = np.vstack([X, X[:300]])                          # 300 rows recorded twice
y_dup = np.concatenate([y, y[:300]])
X_tr, X_te, _, _ = train_test_split(X_dup, y_dup, test_size=0.3, random_state=42)
print("overlap, data with duplicates:", row_overlap(X_tr, X_te))
os.chdir("..")
```

### ✏️ แบบฝึก 12.E (5 นาที) ลายนิ้วมือของไฟล์
อ่านไฟล์ `outputs/higgs_toy.csv` เป็นไบต์ เปลี่ยนตัวเลขหนึ่งตัวกลางไฟล์ (หนึ่งไบต์) แล้วคำนวณ SHA-256 ของไฟล์ก่อนและหลังการแก้ด้วย `hashlib.sha256(...).hexdigest()` นับว่าเลขฐานสิบหกทั้ง 64 หลักต่างกันกี่หลัก เก็บใน `my_n_diff` (ไม่ต้องเขียนไฟล์ที่แก้แล้วลงดิสก์)

ตัวอย่างการเปลี่ยนหนึ่งไบต์: `raw = bytearray(open(path, "rb").read()); i = len(raw) // 2; raw[i] = ord("7") if raw[i] != ord("7") else ord("8")`

ฝาแฝดเหล่านี้ทำให้ผลดีเกินจริงเพียงใด เซลล์ถัดไปฝึกป่าสุ่มแบบเดียวกับโค้ดที่ 12.1 บนชุดฝึกที่มีแถวซ้ำ แล้วเทียบความแม่นยำกับแถวทดสอบที่มีฝาแฝดในชุดฝึกและแถวอื่น ทั้งกับต้นไม้ที่จำกัดขนาดใบ (`min_samples_leaf=10` ตามโค้ดที่ 12.1) และต้นไม้ที่โตเต็มที่ (`min_samples_leaf=1`)

```python
X_dup = np.vstack([X, X[:300]]); y_dup = np.concatenate([y, y[:300]])
idx = np.arange(len(y_dup))
tr, te = train_test_split(idx, test_size=0.3, random_state=42)          # the same split as Listing 12.2
twin = np.isin(te % len(y), tr % len(y)) & ((te >= len(y)) | (te < 300))
print("test rows with a twin in training:", twin.sum())
for leaf in (10, 1):
    rf_ = RandomForestClassifier(n_estimators=100, min_samples_leaf=leaf, random_state=0, n_jobs=-1).fit(X_dup[tr], y_dup[tr])
    ok_ = rf_.predict(X_dup[te]) == y_dup[te]
    print(f"forest min_samples_leaf={leaf:2d}: accuracy on rows with a twin {ok_[twin].mean():.3f}, on the other rows {ok_[~twin].mean():.3f}")
```

ในตำรา ป่าสุ่มทายแถวที่มีฝาแฝดถูกร้อยละ 97.5 เทียบกับร้อยละ 93.5 ของแถวอื่น และถูกทุกแถวเมื่อให้ต้นไม้โตเต็มที่ ส่วนที่เกินมานี้มาจากความจำ ไม่ใช่จากการเรียนรู้ การตรวจนี้ใช้เวลาไม่ถึงหนึ่งวินาที และควรทำทุกครั้งก่อนประเมินแบบจำลอง สำหรับภาพหรือข้อมูลที่มีสัญญาณรบกวน ข้อมูลซ้ำอาจไม่เหมือนกันทุกบิต จึงต้องใช้การเทียบที่ยอมให้ต่างกันเล็กน้อย หรือแบ่งข้อมูลตามวัตถุตามที่แบบฝึก E11.3 แนะนำ

บทที่ 11 แนะนำการ์ดแบบจำลองไว้แล้ว เอกสารคู่กันคือการ์ดชุดข้อมูล (dataset card) หรือแผ่นข้อมูลของชุดข้อมูล (datasheet for datasets) ซึ่งเกบรูและคณะเสนอโดยได้แรงบันดาลใจจากแผ่นข้อมูลของชิ้นส่วนอิเล็กทรอนิกส์ วิศวกรที่เลือกทรานซิสเตอร์ไม่ต้องทดสอบทุกตัวเอง เพราะแผ่นข้อมูลบอกช่วงแรงดัน อุณหภูมิที่ใช้งานได้ และค่าที่วัดภายใต้เงื่อนไขใด การ์ดชุดข้อมูลตอบคำถามเจ็ดกลุ่ม

1. **แรงจูงใจ:** สร้างขึ้นเพื่ออะไร โดยใคร ด้วยทุนจากใคร
2. **องค์ประกอบ:** มีตัวอย่างกี่ตัว แต่ละตัวมีอะไร มีข้อมูลขาดหายหรือไม่
3. **การเก็บข้อมูล:** เก็บด้วยเครื่องมือใด เมื่อใด และเลือกตัวอย่างอย่างไร
4. **การเตรียมข้อมูล:** ทำความสะอาดอย่างไร และได้ป้ายกำกับมาอย่างไร
5. **การใช้งาน:** ออกแบบมาเพื่องานใด และไม่ควรใช้กับงานใด
6. **การเผยแพร่:** ภายใต้สัญญาอนุญาตใด อ้างอิงอย่างไร
7. **การดูแลรักษา:** ใครเป็นผู้ติดต่อ และจะมีรุ่นใหม่หรือไม่

งานเมล็ดข้าวในบทที่ 11 ให้ตัวอย่างที่ดีของกลุ่มที่สามและสี่ การเก็บภาพด้วยสแกนเนอร์รุ่นเดียว การจัดเรียงเมล็ดด้วยแผ่นอะคริลิก และการตัดเมล็ดที่คะแนนมาตรฐานของพื้นที่เกิน 3 เป็นข้อมูลที่ผู้ใช้ต้องรู้จึงจะตัดสินได้ว่าผลจะใช้กับภาพจากโทรศัพท์มือถือได้หรือไม่

### เริ่มจากคำถาม ไม่ใช่จากวิธี
ข้อผิดพลาดที่พบบ่อยที่สุดในการเริ่มโครงงานคือการเริ่มจากวิธี "อยากลองใช้โครงข่ายคอนโวลูชัน" เป็นแรงจูงใจที่เข้าใจได้ แต่ไม่ใช่คำถามทางวิทยาศาสตร์ โครงงานที่เริ่มแบบนี้มักจบด้วยตัวเลขความแม่นยำที่ไม่มีใครรู้ว่าจะนำไปใช้อย่างไร โครงงานที่ดีเริ่มจากคำถามที่ตอบแล้วมีคนได้ประโยชน์ เช่น "ตัวอย่างวัสดุสองแบบต่างกันที่ความแข็งเชิงกลหรือไม่" หรือ "อุณหภูมิที่สายพอลิเมอร์ยุบตัวขึ้นกับความยาวของสายอย่างไร" แล้วจึงถามว่าเครื่องมือใดตอบคำถามนั้นได้ดีที่สุด

หลักห้าประการของตำราใช้เป็นรายการตรวจของข้อเสนอโครงงานได้โดยตรง

| หลัก | คำถามที่ข้อเสนอต้องตอบ |
|---|---|
| P1 | คำถามมาจากฟิสิกส์หรือไม่ และคำตอบแบบใดจะนับว่าตอบคำถามได้ |
| P2 | รู้สมมติฐานของวิธีที่จะใช้ ทั้งแบบจำลองของสัญญาณรบกวนและสมมติฐานของแบบจำลอง และวางแผนตรวจสมมติฐานเหล่านั้นหรือไม่ |
| P3 | ข้อมูลมาจากการวัดหรือการจำลองที่รู้ที่มา รู้คุณภาพ และมีสิทธิ์ใช้หรือไม่ |
| P4 | ผลที่ได้จะตีความทางฟิสิกส์ได้อย่างไร |
| P5 | จะระบุความไม่แน่นอนด้วยวิธีใด และครอบคลุมแหล่งใด |

ลองเปรียบเทียบข้อเสนอสองแบบ

- **ข้อเสนอ A:** "ใช้โครงข่ายคอนโวลูชันจำแนก Galaxy10 ให้ได้ความแม่นยำสูงกว่าร้อยละ 90"

ข้อเสนอ B ใช้เครื่องมือเดียวกัน แต่มีคำถามทางดาราศาสตร์ มีแผนจัดการความผิดพลาดของตัวจำแนก และมีแผนระบุความไม่แน่นอน ข้อเสนอ B ยังต้องคิดเรื่องอคติแบบมาล์มควิสต์ในคาบที่ 1 ด้วย เพราะกาแล็กซีที่จางกว่าอาจมองเห็นคานได้ยากกว่า ซึ่งเป็นตัวอย่างของการที่ P2 และ P4 เชื่อมกัน

### เส้นฐานก่อน และกำหนดการวิเคราะห์หลักไว้ล่วงหน้า
ตำรานี้แสดงซ้ำหลายครั้งว่าวิธีที่ง่ายกว่าหรือปริมาณทางฟิสิกส์ตัวเดียวอาจให้ผลเท่ากับวิธีที่ซับซ้อน ตั้งแต่การถดถอยโลจิสติกกับลักษณะเด่น $|m_{bb} - 125|$ ในบทที่ 6 รัศมีไจเรชันตัวเดียวในบทที่ 10 ไปจนถึงขนาดของเมล็ดข้าวในบทที่ 11 แผนโครงงานจึงควรเริ่มจากเส้นฐานที่ง่ายที่สุดที่สมเหตุสมผลเสมอ เส้นฐานให้ตัวเลขที่วิธีที่ซับซ้อนต้องชนะ และมักช่วยให้เข้าใจข้อมูลก่อนลงมือกับวิธีที่ตีความยาก

หลักข้อที่สองคือการกำหนดการวิเคราะห์หลักไว้ล่วงหน้า (pre-registration) ก่อนดูผล ได้แก่ ตัวชี้วัดหลัก การแบ่งข้อมูล เกณฑ์เลือกแบบจำลอง และเกณฑ์ที่จะนับว่าผลสนับสนุนหรือไม่สนับสนุนสมมติฐาน งานวิจัย AFM ของผู้เขียนที่บทที่ 7 ยกมาใช้หลักนี้ โดยกำหนดการตั้งค่าไว้ก่อน แล้วรายงานทุกการตั้งค่า ไม่ใช่เฉพาะการตั้งค่าที่ได้ผลดี เมื่อผู้วิจัยลองหลายทางแล้วเลือกทางที่ได้ผลดี ผลที่รายงานจะมีอคติแบบเดียวกับการเลือกเอพอกที่ดีที่สุดในบทที่ 11 และการตั้งสมมติฐานหลังเห็นผล (HARKing) ทำให้การทดสอบไม่มีความหมาย การวิเคราะห์ที่ไม่ได้วางแผนไว้ยังทำได้ แต่ต้องรายงานแยกว่าเป็นการวิเคราะห์เชิงสำรวจ

แผนโครงงานที่ใช้งานได้จริงจึงมีห้าส่วน

1. **คำถามและเกณฑ์ของคำตอบ**
2. **ข้อมูล** พร้อมผลการตรวจสัญญาอนุญาตและหลักการ FAIR (E12.1)
3. **แผนการวิเคราะห์** ตามลำดับขั้นหกขั้น โดยระบุเส้นฐานและการวิเคราะห์หลักที่กำหนดไว้ล่วงหน้า
4. **แผนระบุความไม่แน่นอน**
5. **ความเสี่ยง** ทั้งทางเทคนิค เช่น ข้อมูลน้อยเกินไปหรือคำนวณนานเกินไป และทางจริยธรรม พร้อมแผนสำรองของแต่ละความเสี่ยง

### เลือกวิธีระบุความไม่แน่นอนให้เหมาะกับโครงงาน
ตารางเลือกวิธีระบุความไม่แน่นอนในบทที่ 9 เป็นจุดเริ่มต้น โครงงานที่ปรับเส้นโค้งกับข้อมูลที่ดีใช้เมทริกซ์ความแปรปรวนร่วมได้ โครงงานที่ปริมาณสุดท้ายไม่เชิงเส้นหรือข้อมูลน้อยควรใช้บูตสแตรปหรือการแจกแจงภายหลัง โครงงานที่ใช้ข้อมูลจากการจำลองมอนติคาร์โลหรือพลวัตเชิงโมเลกุลต้องจัดการสหสัมพันธ์ในเวลา และโครงงานที่ใช้โครงข่ายควรใช้การรวมกลุ่ม แต่บทเรียนที่สำคัญที่สุดมาจากบทที่ 10 **ความไม่แน่นอนจากเครื่องมือทางสถิติมักเล็กกว่าความไม่แน่นอนจากการตัดสินใจของผู้วิเคราะห์** ในตัวอย่างพอลิเมอร์ การรวมกลุ่มโครงข่ายให้ความไม่แน่นอนราว 0.03 แต่การเปลี่ยนช่วงของข้อมูลฝึกทำให้ผลเลื่อนราว 0.2 แผนความไม่แน่นอนของโครงงานจึงควรมีส่วนที่เปลี่ยนการตัดสินใจที่สมเหตุสมผลอย่างเป็นระบบ แล้วรายงานว่าผลเปลี่ยนเพียงใด ซึ่งนักฟิสิกส์ทดลองเรียกว่าการประเมินความไม่แน่นอนเชิงระบบ

เซลล์ถัดไปทำการวิเคราะห์นี้กับข้อมูลฮิกส์จำลองเป็นตัวอย่างของรายการที่สามของแบบฝึก E12.4 ฟังก์ชัน `analysis` คือป่าสุ่มของโค้ดที่ 12.1 ที่เปิดให้เปลี่ยนการตัดสินใจสี่อย่าง ได้แก่ เมล็ดสุ่มของการแบ่งข้อมูล ขนาดใบขั้นต่ำของต้นไม้ ชุดของลักษณะเด่น และเกณฑ์ตัดเหตุการณ์ที่มวล $m_{bb}$ สูง ความไม่แน่นอนทางสถิติของพื้นที่ใต้เส้นโค้งวัดด้วยบูตสแตรปของชุดทดสอบ 200 ครั้ง (⏱ ราวหนึ่งนาทีบน Colab)

```python
# ⏱ ราวหนึ่งนาทีบน Colab
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score

def analysis(split_seed=42, leaf=10, features=("m_bb", "pT", "dR", "ETmiss"), m_max=None, return_pred=False):
    # the random forest of Listing 12.1 with four analyst decisions exposed
    d = df if m_max is None else df[df.m_bb < m_max]
    X_, y_ = d[list(features)].values, d["label"].values
    X_tr, X_te, y_tr, y_te = train_test_split(X_, y_, test_size=0.3, stratify=y_, random_state=split_seed)
    rf = RandomForestClassifier(n_estimators=100, min_samples_leaf=leaf, random_state=0, n_jobs=-1).fit(X_tr, y_tr)
    p = rf.predict_proba(X_te)[:, 1]
    return (roc_auc_score(y_te, p), y_te, p) if return_pred else roc_auc_score(y_te, p)

auc_base, y_te_, p_ = analysis(return_pred=True)
rng_b = np.random.default_rng(0)
boot = []
for _ in range(200):                                   # bootstrap of the test set: statistical uncertainty
    i = rng_b.integers(0, len(y_te_), len(y_te_))
    boot.append(roc_auc_score(y_te_[i], p_[i]))
se_boot = np.std(boot, ddof=1)
print(f"baseline AUC {auc_base:.4f}, bootstrap standard error {se_boot:.4f}")

decisions = {                                          # each entry changes ONE analyst decision
    "split seed 7 (not 42)": dict(split_seed=7),
    "min_samples_leaf 1 (not 10)": dict(leaf=1),
    "min_samples_leaf 50 (not 10)": dict(leaf=50),
    "drop ETmiss": dict(features=("m_bb", "pT", "dR")),
    "keep only m_bb < 250 GeV": dict(m_max=250),
}
rows = []
for name, kw in decisions.items():
    a = analysis(**kw)
    rows.append((name, a, a - auc_base, (a - auc_base) / se_boot))
sens = pd.DataFrame(rows, columns=["decision", "AUC", "shift", "shift / SE"]).set_index("decision")
print(sens.round(4))
```

```python
fig, ax = plt.subplots(figsize=(7, 2.8))
ax.barh(sens.index, sens["shift"], color=["C3" if abs(s) > se_boot else "C0" for s in sens["shift"]])
ax.axvspan(-se_boot, se_boot, color="grey", alpha=0.25, label="+/- 1 bootstrap SE (statistical)")
ax.axvline(0, color="k", lw=0.8)
ax.set_xlabel("change in ROC AUC from the baseline analysis"); ax.legend(fontsize=7, loc="lower left")
ax.set_title("sensitivity to analyst decisions vs statistical uncertainty (SYNTHETIC)", fontsize=9)
fig.tight_layout(); plt.show()
```

ผู้เขียนได้ความคลาดเคลื่อนมาตรฐานจากบูตสแตรปราว 0.0017 การเปลี่ยนการตัดสินใจหลายข้อทำให้ผลเลื่อนในขนาดใกล้หรือเกินค่านี้ โดยเฉพาะการเพิ่มขนาดใบขั้นต่ำเป็น 50 ซึ่งลดพื้นที่ใต้เส้นโค้งลงราว 0.004 หรือราวสองเท่าครึ่งของความคลาดเคลื่อนมาตรฐาน ข้อสังเกตสองข้อ ข้อแรก การตัดเหตุการณ์ด้วยมวลเปลี่ยนทั้งข้อมูลฝึกและข้อมูลทดสอบ ตัวเลขจึงไม่ได้เทียบบนชุดทดสอบเดียวกัน การตัดสินใจแบบนี้ต้องมีเหตุผลทางฟิสิกส์ (เช่น ช่วงมวลที่การจำลองเชื่อถือได้) ไม่ใช่เลือกเพราะได้ตัวเลขดี ข้อที่สอง ถ้ารายงานเพียงตัวเลขเดียวพร้อมความคลาดเคลื่อนจากบูตสแตรป ผู้อ่านจะเข้าใจว่าผลแม่นยำกว่าความจริง เพราะการตัดสินใจที่สมเหตุสมผลพอ ๆ กันให้ตัวเลขที่ต่างกันมากกว่าช่วงนั้น

### สรุปบท
ข้อมูลทางฟิสิกส์มีประเด็นเชิงจริยธรรมของตัวเอง ทั้งอคติจากการเลือกตัวอย่าง ความเป็นส่วนตัวในงานที่เกี่ยวกับมนุษย์ การใช้งานสองทาง เครดิตของผู้สร้างข้อมูล และต้นทุนของการคำนวณ ประเด็นเหล่านี้ส่วนใหญ่อยู่ที่การรายงาน และการปกปิดสิ่งที่รู้ทำลายความน่าเชื่อถือของวิทยาศาสตร์โดยตรง สัญญาอนุญาตกำหนดว่าใช้ข้อมูลและโค้ดได้อย่างไร และต้องตรวจแยกกัน หลักการ FAIR เป็นรายการตรวจว่าข้อมูลที่เปิดแล้วใช้ได้จริงหรือไม่ และแผนการจัดการข้อมูลป้องกันปัญหาตั้งแต่ต้นโครงงาน

การทำซ้ำได้ของงานที่ใช้การเรียนรู้ของเครื่องต้องบันทึกสภาพแวดล้อม การตั้งค่า เมล็ดสุ่ม รุ่นของโค้ด และค่าแฮชของข้อมูล การกำหนดเมล็ดสุ่มทำให้ผลทำซ้ำได้ แต่ไม่ได้ทำให้ผลเป็นตัวแทน ตัวอย่างข้อมูลฮิกส์แสดงว่าการแบ่งข้อมูลเปลี่ยนผลได้มากกว่าเมล็ดสุ่มของแบบจำลองเกือบสิบเท่า และข้อมูลซ้ำที่ตรวจได้ง่ายด้วยค่าแฮชทำให้เกิดการรั่วไหลได้โดยไม่รู้ตัว การ์ดชุดข้อมูลและการ์ดแบบจำลองเป็นแผ่นข้อมูลที่ทำให้ผู้อื่นใช้งานของเราได้อย่างถูกต้อง

โครงงานที่ดีเริ่มจากคำถามทางฟิสิกส์ ผ่านการตรวจตามหลักห้าประการ เริ่มจากเส้นฐานที่ง่ายที่สุด และกำหนดการวิเคราะห์หลักไว้ก่อนดูผล การลงมือทำควรให้ท่อทั้งหมดทำงานครบก่อน แล้วปรับทีละขั้น การระบุความไม่แน่นอนต้องรวมผลของการตัดสินใจของผู้วิเคราะห์ ซึ่งมักใหญ่กว่าความไม่แน่นอนทางสถิติ และรายงานที่ดีระบุข้อจำกัดอย่างชัดเจน เพราะข้อจำกัดบอกขอบเขตที่ผลนำไปใช้ได้
