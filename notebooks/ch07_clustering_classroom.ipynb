# บทที่ 7 การเรียนรู้แบบไม่มีผู้สอนและการจัดกลุ่ม: โน้ตบุ๊กสำหรับชั้นเรียน
โน้ตบุ๊กประกอบตำรา *วิทยาศาสตร์ข้อมูลและการเรียนรู้ของเครื่องสำหรับนักฟิสิกส์*

โน้ตบุ๊กนี้ใช้ในชั้นเรียนได้ด้วยไฟล์นี้ไฟล์เดียว ไม่ต้องดึงโค้ดจาก GitHub และไม่ต้องดาวน์โหลดข้อมูล เนื้อหาเรียงตามบทที่ 7 ในตำรา แบ่งเป็นสี่คาบ รวมราว 240 นาที ตามแผนเวลาในคู่มือผู้สอน แต่ละคาบยาวประมาณ 60 นาที กรณีศึกษา AFM (ส่วนที่ 11–13 รวมราว 75 นาที) เริ่มด้วยการอ่านบริบทท้ายคาบที่ 3 แล้วทำต่อทั้งคาบที่ 4 หัวข้อที่มีป้าย **โค้ดที่ 7.N** ตรงกับตัวอย่างโค้ดเลขเดียวกันในตำรา

## วัตถุประสงค์
เมื่อทำโน้ตบุ๊กนี้จนจบ ผู้เรียนควรสามารถ

1. แยกได้ว่าปัญหาใดเหมาะกับการเรียนรู้แบบมีผู้สอนหรือแบบไม่มีผู้สอน (unsupervised learning) และอธิบายได้ว่าทำไมการปรับมาตราส่วนและการเลือกระยะทางจึงสำคัญยิ่งกว่าในการจัดกลุ่ม (clustering)
2. อธิบายหลักการ สมมติฐาน และข้อจำกัดของเคมีนส์ (K-means) DBSCAN HDBSCAN และแบบจำลองผสมเกาส์ (Gaussian mixture model, GMM) และใช้ทั้งหมดใน scikit-learn
3. เลือกจำนวนกลุ่มด้วยหลายวิธีประกอบกัน ได้แก่ วิธีข้อศอก ค่าซิลูเอตต์ เกณฑ์สารสนเทศแบบเบส์ และสถิติช่องว่าง และบอกอคติของแต่ละวิธี
4. ประเมินผลการจัดกลุ่มด้วยตัวชี้วัดภายในและภายนอก และรู้ว่าตัวชี้วัดภายในอาจชี้ไปผิดทาง
5. เชื่อมกลุ่มที่ได้เข้ากับโครงสร้างทางฟิสิกส์ของระบบ และรายงานผลโดยไม่เลือกการตั้งค่าด้วยคำตอบจริง

## ข้อมูลในโน้ตบุ๊กนี้
ข้อมูลทุกชุดในคาบเรียนเป็น **ข้อมูลจำลอง (SYNTHETIC)** ที่สร้างด้วยโค้ดในตำรา และตัวเลขตรงกับตำรา

- **ข้อมูลรูปทรงพื้นฐาน:** กลุ่มก้อนสามกลุ่มจาก `make_blobs` และรูปพระจันทร์เสี้ยวสองดวงจาก `make_moons`
- **พอลิเมอร์:** ค่า $(R_g, R_{ee})$ ของสายพอลิเมอร์จำลองที่มีสถานะกึ่งเสถียรสามสถานะ (โค้ดที่ 7.5)
- **AFM:** แผนที่แรงจาก AFM จำลองขนาด $128 \times 128$ พิกเซล (โค้ดที่ 7.7) ผู้เขียนออกแบบให้มีลักษณะสำคัญของงานวิจัยจริงเรื่องการแยกโครงร่างรองรับเซลล์ออกจากวัสดุรองรับ แต่ **ไม่ใช่ข้อมูลหรือผลของงานวิจัย** ชุดข้อมูลจริงอยู่ระหว่างการจัดเตรียมเพื่อเผยแพร่
- **SDSS (แบบฝึกหัด E7.6 นอกชั้น):** ใช้ตาราง SDSS ของบทที่ 6 ถ้ายังไม่มีไฟล์ โน้ตบุ๊กจะสร้างข้อมูลแทนสังเคราะห์ (synthetic stand-in) แบบเดียวกับบทที่ 6

หัวกราฟที่มาจากข้อมูลจำลองมีคำว่า SYNTHETIC กำกับ ตัวเลขใช้เพื่อการเรียนวิธีการ ไม่ใช่ผลการวิจัย

### การล็อกแนวคำตอบ
เซลล์ ✅ และเซลล์แนวคำตอบในโน้ตบุ๊กนี้แสดงคำตอบอ้างอิงเมื่อผู้เรียนเติมคำตอบในเซลล์ ✏️ ของข้อนั้นครบแล้วเท่านั้น ถ้ายังไม่ได้ตอบ เซลล์จะแสดงเพียงข้อความ 🔒 บอกว่ายังขาดคำตอบใด การคำนวณในเซลล์ยังทำงานตามปกติ ดังนั้น Run all จึงรันได้ครบโดยไม่เปิดเผยคำตอบ แนวคำตอบของงานเขียนและงานนอกชั้นซึ่งตรวจด้วยโค้ดไม่ได้จะเปิดเมื่อผู้สอนเลือกเท่านั้น

**สำหรับผู้สอน:** ติ๊กช่อง `SHOW_ANSWERS` ในเซลล์ถัดไป (หรือแก้เป็น `True`) แล้วรันเซลล์นั้นอีกครั้ง เพื่อแสดงแนวคำตอบทุกข้อ เช่น เมื่อเฉลยในชั้นหรือเตรียมสอน

```python
#@title ⚙️ ตัวเลือกของผู้สอน: แสดงแนวคำตอบ
SHOW_ANSWERS = False  #@param {type:"boolean"}

def _answered(v):
    """True when an exercise variable holds an answer (not None, "?", "", empty or containing blanks)."""
    if v is None or v is Ellipsis:
        return False
    if isinstance(v, str):                       # placeholders such as "?", "?-?", "..." count as blank
        return v.replace("?", "").replace("-", "").replace(".", "").strip() != ""
    if isinstance(v, dict):
        return len(v) > 0 and all(_answered(x) for x in v.values())
    if isinstance(v, (list, tuple, set, frozenset)):
        return len(v) > 0 and all(_answered(x) for x in v)
    if type(v).__name__ == "ndarray":
        return v.size > 0
    return True

def _gate(names=()):
    """Return print if the answers may be shown, otherwise print one lock message and return a no-op."""
    if SHOW_ANSWERS:
        return print
    if names:
        missing = [n for n in names if not _answered(globals().get(n))]
        if not missing:
            return print
        print("🔒 ยังไม่ได้ตอบ:", ", ".join(missing),
              "— เติมคำตอบในเซลล์ ✏️ ด้านบนให้ครบก่อน แล้วรันเซลล์นี้อีกครั้ง")
    else:
        print("🔒 แนวคำตอบข้อนี้ผู้สอนเป็นผู้เปิด (ตั้ง SHOW_ANSWERS = True ในส่วนที่ 0)")
    return lambda *args, **kwargs: None

print("แสดงแนวคำตอบทั้งหมด:", "ใช่ (โหมดผู้สอน)" if SHOW_ANSWERS else "ไม่ (แสดงเมื่อผู้เรียนตอบข้อนั้นครบแล้ว)")
```

# ส่วนที่ 1 แรงจูงใจ นิยาม และการจัดกลุ่มสามตระกูล (หัวข้อ 7.1.1–7.1.3)

### แนวคิดสำคัญ
สองบทก่อนหน้านี้ทุกจุดข้อมูลมาพร้อมคำตอบ $y$ ที่ถูกต้อง แต่งานวิจัยจำนวนมากเริ่มจากจุดที่ไม่มีใครรู้คำตอบ นักฟิสิกส์ที่ถ่ายภาพพื้นผิวของวัสดุชนิดใหม่ด้วยกล้องจุลทรรศน์แรงอะตอม (atomic force microscope, AFM) ไม่รู้ล่วงหน้าว่าพื้นผิวมีบริเวณที่มีสมบัติต่างกันกี่แบบ การเรียนรู้โครงสร้างของข้อมูลโดยไม่มีคำตอบให้เรียกว่า **การเรียนรู้แบบไม่มีผู้สอน (unsupervised learning)** และงานที่พบบ่อยที่สุดคือ **การจัดกลุ่ม (clustering)**

คำถามหลักจึงเปลี่ยนจาก "ป้ายกำกับของจุดใหม่นี้คืออะไร" เป็น "ข้อมูลของเราแบ่งตัวเป็นกลุ่มตามธรรมชาติอย่างไร" ผลของการจัดกลุ่มไม่ใช่ข้อสรุป แต่เป็นสมมติฐานที่ต้องตรวจสอบต่อด้วยฟิสิกส์หรือการทดลองใหม่ ความยากที่การเรียนรู้แบบมีผู้สอนไม่มีคือ **วิธีจัดกลุ่มทุกวิธีคืนกลุ่มออกมาเสมอ แม้ข้อมูลจะไม่มีกลุ่มอยู่จริงเลย**

> **นิยาม (การเรียนรู้แบบไม่มีผู้สอน)** คือการเรียนรู้โครงสร้างของข้อมูล $\{\mathbf x_i\}$ โดยไม่มีค่าเป้าหมาย $y_i$ ผลลัพธ์มีสามรูปแบบหลัก ได้แก่ กลุ่มของจุดข้อมูลที่คล้ายกัน (บทนี้) การแทนข้อมูลด้วยมิติที่น้อยลง (บทที่ 8) และการตรวจหาจุดข้อมูลที่ผิดแผก

### การจัดกลุ่มสามตระกูล
วิธีจัดกลุ่มแต่ละวิธีมีนิยามของคำว่า "กลุ่ม" ต่างกัน และนิยามนั้นกำหนดว่าวิธีนั้นจะเห็นกลุ่มแบบใดได้บ้าง

| ตระกูล | ตัวแทน | นิยามของกลุ่ม | สมมติฐาน | จำนวนกลุ่ม |
|---|---|---|---|---|
| จุดศูนย์กลาง | เคมีนส์ | จุดที่อยู่ใกล้จุดศูนย์กลางเดียวกัน | กลุ่มนูน ขนาดและการกระจายใกล้เคียงกัน | ต้องระบุ |

ตระกูลที่สี่คือ **การจัดกลุ่มแบบลำดับชั้น (hierarchical clustering)** ซึ่งรวมคู่กลุ่มที่ใกล้ที่สุดทีละคู่จนเหลือกลุ่มเดียว **เกณฑ์การรวมแบบวอร์ด (Ward's method)** เลือกรวมคู่ที่ทำให้ผลรวมกำลังสองภายในกลุ่มเพิ่มน้อยที่สุด ให้ผลใกล้เคียงกับเคมีนส์ และจะกลับมาในกรณีศึกษา AFM (ส่วนที่ 12)

แนวทางคร่าว ๆ ในการเลือกวิธี: กลุ่มทรงกลมหรือนูนและรู้จำนวนกลุ่มโดยประมาณ ให้เริ่มจากเคมีนส์ กลุ่มรูปทรงแปลกและมีจุดรบกวน ให้ใช้วิธีที่อาศัยความหนาแน่น ต้องการความน่าจะเป็นของการเป็นสมาชิก ให้ใช้แบบจำลองผสมเกาส์ และถ้าไม่แน่ใจ ให้ใช้ทุกวิธีแล้วเปรียบเทียบ

### ระยะทาง
| ระยะทาง | สูตร | เหมาะกับ |
|---|---|---|
| ยุคลิด ($L^2$) | $\sqrt{\sum_d (x_d - y_d)^2}$ | ปริมาณต่อเนื่องทั่วไป ค่าตั้งต้นของเกือบทุกวิธี |
| แมนฮัตตัน ($L^1$) | $\sum_d \lvert x_d - y_d\rvert$ | ข้อมูลที่มีค่าผิดปกติ เพราะไม่ยกกำลังสองความต่าง |
| โคไซน์ | $1 - \cos\theta$ | เปรียบเทียบรูปร่างของสเปกตรัม เมื่อทิศทางสำคัญกว่าขนาด |
| แฮมมิง | จำนวนตำแหน่งที่ต่างกัน | ข้อมูลแบบสองค่าหรือแบบหมวดหมู่ |

เซลล์ถัดไปเปรียบเทียบสเปกตรัมวัตถุดำของดาวสามดวง ดาว A และ B มีอุณหภูมิเท่ากัน (5,800 K) แต่ B อยู่ไกลกว่าจึงสว่างเพียงหนึ่งในสี่ ดาว C ร้อนกว่า (7,000 K) และมีความสว่างรวมใกล้กับ A

```python
from sklearn.metrics.pairwise import euclidean_distances, cosine_distances

lam = np.linspace(300e-9, 1000e-9, 200)                      # wavelength (m)
def planck(T):
    return 1 / lam**5 / (np.exp(1.4388e-2 / (lam * T)) - 1)   # shape of B_lambda(T)
A = planck(5800); B = 0.25 * planck(5800); C = planck(7000)
C *= A.sum() / C.sum()                                      # similar total brightness to A
S = np.vstack([A, B, C]) / A.max()
print("Euclidean  A-B {:.2f}   A-C {:.2f}".format(*euclidean_distances(S)[0, 1:]))
print("cosine     A-B {:.3f}  A-C {:.3f}".format(*cosine_distances(S)[0, 1:]))
```

ระยะแบบยุคลิดเห็นว่า A ต่างจาก B มากกว่าต่างจาก C ทั้งที่ A กับ B เป็นดาวชนิดเดียวกัน ระยะแบบโคไซน์เห็นว่า A กับ B เหมือนกันทุกประการ (ระยะเป็นศูนย์) ซึ่งตรงกับฟิสิกส์มากกว่า

# ส่วนที่ 3 เคมีนส์และการวนซ้ำของลอยด์ (หัวข้อ 7.2.1–7.2.2)

### แนวคิดสำคัญ
อุปมาของ **เคมีนส์ (K-means)** คือการตั้งร้านสะดวกซื้อ $K$ สาขาในเมืองหนึ่ง ถ้ารู้ว่าลูกค้าแต่ละคนจะไปร้านใด ตำแหน่งที่ดีที่สุดของร้านคือจุดกึ่งกลางของลูกค้ากลุ่มนั้น และถ้ารู้ตำแหน่งร้าน ลูกค้าแต่ละคนก็จะไปร้านที่ใกล้ที่สุด เคมีนส์สลับทำสองขั้นนี้ไปเรื่อย ๆ จนไม่มีอะไรเปลี่ยน

> **นิยาม (เคมีนส์)** หาการแบ่งข้อมูลเป็น $K$ กลุ่ม $C_1, \ldots, C_K$ และ **จุดศูนย์กลาง (centroid)** $\boldsymbol\mu_1, \ldots, \boldsymbol\mu_K$ ที่ทำให้ **ผลรวมกำลังสองภายในกลุ่ม (within-cluster sum of squares, WCSS)**
> $$W = \sum_{k=1}^{K} \sum_{\mathbf x_i \in C_k} \lVert\mathbf x_i - \boldsymbol\mu_k\rVert^2$$
> มีค่าน้อยที่สุด **การวนซ้ำของลอยด์ (Lloyd iteration)** สลับทำ (1) ขั้นกำหนดกลุ่ม: ให้แต่ละจุดอยู่ในกลุ่มของจุดศูนย์กลางที่ใกล้ที่สุด และ (2) ขั้นปรับจุดศูนย์กลาง: ให้จุดศูนย์กลางใหม่เป็นค่าเฉลี่ยของจุดในกลุ่ม ทำซ้ำจนการกำหนดกลุ่มไม่เปลี่ยน

ทั้งสองขั้นลด $W$ หรือคงไว้เท่าเดิมเสมอ การวนซ้ำจึงต้องหยุด แต่จุดที่หยุดเป็นเพียง **ค่าต่ำสุดเฉพาะที่** ขึ้นกับจุดเริ่มต้น ทางแก้คือเริ่มหลายครั้ง (`n_init`) และเลือกจุดเริ่มต้นให้กระจายตัวด้วยวิธี **k-means++** ซึ่งเป็นค่าตั้งต้นของ scikit-learn

### โค้ดที่ 7.1 — เคมีนส์ที่เริ่มด้วย k-means++

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.datasets import make_blobs

X, _ = make_blobs(n_samples=600, centers=3, cluster_std=1.0, random_state=2026)
X_scaled = StandardScaler().fit_transform(X)

pipe = Pipeline([
    ('scaler',  StandardScaler()),
    ('cluster', KMeans(n_clusters=3, init='k-means++', n_init=10,
                       random_state=42)),
])
labels = pipe.fit_predict(X)
centroids = pipe.named_steps['cluster'].cluster_centers_   # in scaled units
```

```python
print("cluster sizes:", np.bincount(labels).tolist())
print("centroids (scaled units):\n", centroids.round(2))
print("centroids (original units):\n", pipe.named_steps['scaler'].inverse_transform(centroids).round(2))
```

จุดศูนย์กลางที่ได้อยู่ในหน่วยที่ปรับเป็นมาตรฐานแล้ว ก่อนตีความทางฟิสิกส์ต้องแปลงกลับเป็นหน่วยเดิมด้วย `inverse_transform` ของตัวปรับมาตราส่วน ดังบรรทัดสุดท้ายของเซลล์บน นอกจากนี้ scikit-learn รุ่นใหม่ตั้ง `n_init='auto'` ซึ่งเริ่มเพียงครั้งเดียวเมื่อใช้ k-means++ การระบุ `n_init=10` ให้ชัดเจนจึงปลอดภัยกว่า

เซลล์ถัดไปเขียนการวนซ้ำของลอยด์เองในไม่กี่บรรทัด โดยเริ่มจากจุดศูนย์กลางที่เลือกไม่ดี (สามจุดแรกของข้อมูลหลังเรียงตามแกนแรก) เพื่อให้เห็นว่า $W$ ลดลงทุกรอบ

```python
mu = X_scaled[np.argsort(X_scaled[:, 0])[:3]].copy()        # a poor start: three left-most points
for it in range(8):
    d2 = ((X_scaled[:, None, :] - mu[None, :, :])**2).sum(axis=2)
    lab_it = d2.argmin(axis=1)                                  # assignment step
    W_it = d2[np.arange(len(X_scaled)), lab_it].sum()
    print(f"iteration {it}: W = {W_it:8.1f}   sizes {np.bincount(lab_it, minlength=3).tolist()}")
    new_mu = np.array([X_scaled[lab_it == k].mean(axis=0) for k in range(3)])   # update step
    if np.allclose(new_mu, mu):
        break
    mu = new_mu
```

### โค้ดที่ 7.2 — การวิเคราะห์ด้วยวิธีข้อศอก
เคมีนส์ต้องการให้ระบุ $K$ แต่ $W$ ลดลงเสมอเมื่อเพิ่ม $K$ **วิธีข้อศอก (elbow method)** จึงดูที่ *อัตรา* การลดลง แล้วหาจุดที่เส้นหักเหมือนข้อศอก

```python
import numpy as np

wcss = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    km.fit(X_scaled)
    wcss.append(km.inertia_)          # inertia_ is W for this K
# plot range(1, 11) against wcss and look for the elbow
```

```python
print("W(K) for K = 1..10:", [round(w, 1) for w in wcss])
fig, axes = plt.subplots(1, 2, figsize=(10, 3.8))
axes[0].scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels, s=8, cmap='viridis')
axes[0].scatter(centroids[:, 0], centroids[:, 1], c='red', marker='x', s=120)
axes[0].set_title('K-Means, K = 3 (scaled units, SYNTHETIC)')
axes[1].plot(range(1, 11), wcss, 'o-'); axes[1].set_xlabel('K'); axes[1].set_ylabel('W (inertia)')
axes[1].set_title('elbow at K = 3')
fig.tight_layout(); plt.show()
```

ข้อมูลนี้มีสามกลุ่มจริง ข้อศอกจึงชัดเจนมาก $W$ ลดจากราว 1,200 ที่ $K = 1$ เหลือราว 28 ที่ $K = 3$ แล้วลดต่อเพียงเล็กน้อย แต่กับข้อมูลจริง ข้อศอกมักไม่ชัด และคนสองคนอาจเห็นข้อศอกคนละตำแหน่ง วิธีที่สองคือค่าซิลูเอตต์ ซึ่งจะใช้ในแบบฝึก E7.2 ต้นคาบหน้า

### ข้อจำกัดของเคมีนส์
ขอบเขตระหว่างสองกลุ่มของเคมีนส์เป็นระนาบที่แบ่งครึ่งระหว่างจุดศูนย์กลางทั้งสองเสมอ ปริภูมิจึงถูกแบ่งเป็นเซลล์รูปหลายเหลี่ยมนูนที่เรียกว่า **แผนภาพโวโรนอย (Voronoi diagram)** เคมีนส์จึงทำงานดีเมื่อกลุ่มจริงมีรูปทรงกลม ขนาดและการกระจายใกล้เคียงกัน แต่จะผิดพลาดกับกลุ่มที่ยืดยาว กลุ่มที่ซ้อนกันเป็นวงแหวน หรือกลุ่มที่มีขนาดต่างกันมาก ค่าผิดปกติเพียงจุดเดียวก็ดึงจุดศูนย์กลางไปได้ และต้องระบุ $K$ ล่วงหน้าเสมอ

เซลล์ถัดไปยืดข้อมูลสามกลุ่มให้เป็นทรงรีเอียงด้วยการแปลงเชิงเส้น แล้วจัดกลุ่มด้วยเคมีนส์ ($K = 3$ ซึ่งเป็นค่าที่ถูกต้อง)

```python
X_aniso, y_aniso = make_blobs(n_samples=600, centers=3, cluster_std=1.0, random_state=170)
X_aniso = StandardScaler().fit_transform(X_aniso @ np.array([[0.6, -0.64], [-0.41, 0.85]]))
lab_aniso = KMeans(3, n_init=10, random_state=42).fit_predict(X_aniso)
print(f"K-Means on stretched clusters: ARI = {adjusted_rand_score(y_aniso, lab_aniso):.3f}")

fig, axes = plt.subplots(1, 2, figsize=(9, 3.6))
axes[0].scatter(*X_aniso.T, c=y_aniso, s=6); axes[0].set_title('true groups (SYNTHETIC)')
axes[1].scatter(*X_aniso.T, c=lab_aniso, s=6); axes[1].set_title('K-Means, K = 3')
fig.tight_layout(); plt.show()
```

แม้จะรู้ $K$ ที่ถูกต้อง เคมีนส์ก็ตัดทรงรีที่ยืดยาวผิด เพราะเคมีนส์วัดความใกล้ด้วยระยะยุคลิดไปยังจุดศูนย์กลาง ซึ่งให้กลุ่มทรงกลมเสมอ แบบจำลองผสมเกาส์ในส่วนที่ 7 แก้ปัญหานี้ได้

**การตีความทางฟิสิกส์ของเคมีนส์** มีสามทาง (1) จุดศูนย์กลางเมื่อแปลงกลับเป็นหน่วยเดิมคือสถานะเฉลี่ยของกลุ่ม ซึ่งเทียบกับค่าในเอกสารได้ (2) $W$ ของแต่ละกลุ่มหารด้วยจำนวนจุดบอกการกระจายภายในกลุ่ม ซึ่งรวมทั้งความแปรปรวนทางฟิสิกส์และสัญญาณรบกวนของการวัด (3) ขนาดของกลุ่มบอกสัดส่วนของสถานะ แต่ต้องระวัง ถ้าข้อมูลเป็นเฟรมจากการจำลองที่อยู่ในสมดุล สัดส่วนนี้เทียบได้กับประชากรตามการแจกแจงโบลต์ซมันน์ ถ้าข้อมูลเป็นพิกเซลของภาพ สัดส่วนนี้คือสัดส่วนพื้นที่ ซึ่งเป็นปริมาณคนละชนิดกัน

# ส่วนที่ 5 DBSCAN และการต่อเป็นโซ่ (หัวข้อ 7.3.1)

### แนวคิดสำคัญ
ลองนึกภาพเกาะในทะเลที่มองจากเครื่องบินตอนกลางคืน บริเวณที่แสงไฟอยู่ชิดกันต่อเนื่องคือเกาะหนึ่งเกาะ ไม่ว่าเกาะจะยาวคดเคี้ยวเพียงใด ส่วนแสงไฟดวงเดี่ยวกลางทะเลไม่ใช่เกาะ **DBSCAN** (density-based spatial clustering of applications with noise) นิยามกลุ่มแบบเดียวกัน ต่างจากเคมีนส์สามข้อ คือไม่ต้องระบุจำนวนกลุ่ม กลุ่มมีรูปร่างใดก็ได้ และจุดไม่จำเป็นต้องอยู่ในกลุ่มใดเลย

> **นิยาม (DBSCAN)** มีพารามิเตอร์ $\varepsilon$ และ `min_samples` จุดที่ย่านใกล้เคียงรัศมี $\varepsilon$ มีจุดอย่างน้อย `min_samples` จุด (รวมตัวเอง) เรียกว่า **จุดแกน (core point)** จุดที่ไม่ใช่จุดแกนแต่อยู่ในย่านใกล้เคียงของจุดแกน เรียกว่า **จุดขอบ (border point)** จุดที่ไม่เข้าทั้งสองกรณีคือ **จุดรบกวน (noise point)** ซึ่งได้ป้ายกำกับ $-1$ กลุ่มหนึ่งกลุ่มคือเซตของจุดแกนที่ต่อถึงกันได้ผ่านจุดแกนที่ห่างกันไม่เกิน $\varepsilon$ รวมกับจุดขอบของจุดแกนเหล่านั้น

### โค้ดที่ 7.3 — DBSCAN กับกลุ่มรูปพระจันทร์เสี้ยว

```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons

X_moon, y_moon = make_moons(n_samples=500, noise=0.1, random_state=2026)
X_moon = StandardScaler().fit_transform(X_moon)

dbscan = DBSCAN(eps=0.3, min_samples=5)
labels = dbscan.fit_predict(X_moon)          # label -1 means noise
n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
n_noise = (labels == -1).sum()
```

```python
print(f"DBSCAN eps=0.3: {n_clusters} clusters, {n_noise} noise point(s), ARI vs y_moon = {adjusted_rand_score(y_moon, labels):.3f}")
lab35 = DBSCAN(eps=0.35, min_samples=5).fit_predict(X_moon)
print(f"DBSCAN eps=0.35: {len(set(lab35)) - (-1 in lab35)} cluster(s)")
km2 = KMeans(n_clusters=2, n_init=10, random_state=42).fit_predict(X_moon)
print(f"K-Means K=2: ARI = {adjusted_rand_score(y_moon, km2):.3f}")

fig, axes = plt.subplots(1, 3, figsize=(12, 3.6))
for ax, lab, title in zip(axes, [km2, labels, lab35], ['K-Means K=2', 'DBSCAN eps=0.3', 'DBSCAN eps=0.35']):
    ax.scatter(X_moon[:, 0], X_moon[:, 1], c=lab, s=6, cmap='viridis')
    ax.scatter(*X_moon[lab == -1].T, c='red', marker='x', s=30)
    ax.set_title(title + ' (SYNTHETIC)')
fig.tight_layout(); plt.show()
```

DBSCAN ที่ $\varepsilon = 0.3$ แยกพระจันทร์ทั้งสองดวงได้เกือบสมบูรณ์ โดยมีจุดรบกวนเพียงหนึ่งจุด (กากบาทสีแดง) ส่วนเคมีนส์ลากเส้นตรงตัดพระจันทร์ทั้งสองดวงตรงกลาง แต่ผลของ DBSCAN ไวต่อ $\varepsilon$ มาก ที่ $\varepsilon = 0.35$ พระจันทร์ทั้งสองดวงเชื่อมกันเป็นกลุ่มเดียว ปรากฏการณ์นี้เรียกว่า **การต่อเป็นโซ่ (chaining)** และเป็นจุดอ่อนหลักของวิธีที่อาศัยความหนาแน่น

### ✏️ แบบฝึก 7.H (7 นาที) สะพานของการต่อเป็นโซ่
สำหรับ $\varepsilon = 0.30$ และ $0.35$ ให้นับว่าจุดแกนกี่จุดมีย่านใกล้เคียงรัศมี $\varepsilon$ ที่ครอบจุดของพระจันทร์ **ทั้งสองดวง** จุดแกนเหล่านี้คือสะพานที่ทำให้สองกลุ่มต่อถึงกัน เซลล์ถัดไปเตรียมรายการดัชนีของเพื่อนบ้าน `nbrs` ของจุดแกนแต่ละจุดไว้ให้แล้ว เติมบรรทัดที่นับ โดยใช้ `y_moon[n]` เพื่อดูว่าเพื่อนบ้านของจุดแกนหนึ่งมาจากดวงใดบ้าง แล้วเก็บใน `my_bridge[eps]`

```python
from sklearn.neighbors import NearestNeighbors

my_bridge = {}
for eps in (0.30, 0.35):
    db = DBSCAN(eps=eps, min_samples=5).fit(X_moon)
    core_idx = db.core_sample_indices_
    nbrs = NearestNeighbors(radius=eps).fit(X_moon).radius_neighbors(X_moon[core_idx], return_distance=False)
    print(f"eps = {eps}: {len(core_idx)} core points")
    # ✏️ my_bridge[eps] = number of core points whose neighbours come from both moons
```

# ส่วนที่ 6 HDBSCAN และความหมายของจุดรบกวน (หัวข้อ 7.3.2–7.3.3)

DBSCAN ใช้ $\varepsilon$ ค่าเดียวกับทั้งข้อมูล ถ้ากลุ่มหนึ่งหนาแน่นมากและอีกกลุ่มเบาบาง ไม่มี $\varepsilon$ ค่าใดที่เหมาะกับทั้งสองกลุ่ม **HDBSCAN** แก้ปัญหานี้ด้วยการพิจารณา $\varepsilon$ ทุกค่าพร้อมกัน ลองนึกถึงภูมิประเทศที่ความสูงคือความหนาแน่นของข้อมูล แล้วค่อย ๆ ลดระดับน้ำทะเลลง ยอดเขาแต่ละยอดโผล่ขึ้นมาเป็นเกาะ บางเกาะเชื่อมกันเมื่อน้ำลดต่อ HDBSCAN บันทึกประวัติการเกิดและการรวมตัวของเกาะเป็นลำดับชั้น แล้วเลือกเกาะที่ "อยู่รอด" ได้นานที่สุดเป็นกลุ่มสุดท้าย ผู้ใช้กำหนดเพียงขนาดเล็กที่สุดของกลุ่ม `min_cluster_size` ซึ่งตั้งได้จากความรู้ของระบบ เช่น ถ้าโดเมนที่เล็กที่สุดบนภาพ AFM ต้องกว้างอย่างน้อยสิบพิกเซล กลุ่มก็ควรมีอย่างน้อยราวหนึ่งร้อยพิกเซล

### โค้ดที่ 7.4 — HDBSCAN ซึ่งไม่ต้องกำหนดรัศมี

```python
from sklearn.cluster import HDBSCAN

clusterer = HDBSCAN(min_cluster_size=15, min_samples=5, copy=True)
labels_h = clusterer.fit_predict(X_moon)
prob_h = clusterer.probabilities_       # strength of membership, 0 for noise
```

```python
nh = len(set(labels_h)) - (-1 in labels_h)
print(f"HDBSCAN: {nh} clusters, {(labels_h == -1).sum()} noise points, ARI = {adjusted_rand_score(y_moon, labels_h):.3f}, "
      f"median membership strength {np.median(prob_h[labels_h >= 0]):.2f}")

fig, ax = plt.subplots(figsize=(4.5, 3.6))
ax.scatter(*X_moon.T, c=labels_h, s=6, cmap='viridis')
ax.scatter(*X_moon[labels_h == -1].T, c='red', marker='x', s=30)
ax.set_title('HDBSCAN (SYNTHETIC)'); plt.show()
```

เซลล์ถัดไปสร้างข้อมูลสองกลุ่มที่ความหนาแน่นต่างกันมาก กลุ่มหนึ่งแน่น อีกกลุ่มกระจายกว้างกว่าเจ็ดเท่า แล้วเปรียบเทียบ DBSCAN ที่ $\varepsilon$ ค่าเดียวกับ HDBSCAN

```python
X_dens, y_dens = make_blobs(n_samples=[300, 300], centers=[[0, 0], [6, 0]], cluster_std=[0.3, 2.0], random_state=1)
for name, lab in [("DBSCAN eps=0.3", DBSCAN(eps=0.3, min_samples=5).fit_predict(X_dens)),
                  ("DBSCAN eps=1.0", DBSCAN(eps=1.0, min_samples=5).fit_predict(X_dens)),
                  ("HDBSCAN", HDBSCAN(min_cluster_size=15, min_samples=5).fit_predict(X_dens))]:
    print(f"{name:15s}: {len(set(lab)) - (-1 in lab)} cluster(s), noise {100 * (lab == -1).mean():5.1f} %, "
          f"noise in the sparse group {100 * (lab[y_dens == 1] == -1).mean():5.1f} %, ARI {adjusted_rand_score(y_dens, lab):.3f}")
```

วิธีที่อาศัยความหนาแน่นเหมาะกับ (1) โดเมนที่มีรูปร่างไม่สม่ำเสมอ เช่น โดเมนบนพื้นผิววัสดุ หรือโครงสร้างเส้นใยของกาแล็กซี (2) ข้อมูลที่มีจุดรบกวนปน และ (3) กลุ่มที่มีความหนาแน่นต่างกันมาก ซึ่งเหมาะกับ HDBSCAN

# ส่วนที่ 7 แบบจำลองผสมเกาส์ ขั้นตอนวิธี EM และสถานะของพอลิเมอร์ (หัวข้อ 7.4.1–7.4.3)

### แนวคิดสำคัญ
เคมีนส์และ DBSCAN ตัดสินเด็ดขาดว่าจุดหนึ่งอยู่กลุ่มใด แต่ในฟิสิกส์บางจุดก้ำกึ่งจริง ๆ เช่น โมเลกุลที่อยู่ระหว่างสถานะพับกับคลาย **แบบจำลองผสมเกาส์ (Gaussian mixture model, GMM)** มองข้อมูลทั้งชุดเป็นการแจกแจงที่เกิดจากการผสมการแจกแจงเกาส์หลายตัว

> **นิยาม (แบบจำลองผสมเกาส์)** แบบจำลองที่มี $K$ ส่วนประกอบสมมติว่า
> $$p(\mathbf x) = \sum_{k=1}^{K} \pi_k \, \mathcal{N}(\mathbf x \mid \boldsymbol\mu_k, \boldsymbol\Sigma_k)$$
> เมื่อ $\pi_k \ge 0$ คือสัดส่วนผสมซึ่งรวมกันได้หนึ่ง **ความรับผิดชอบ (responsibility)** ของส่วนประกอบ $k$ ต่อจุด $\mathbf x_i$ คือความน่าจะเป็นที่จุดนั้นมาจากส่วนประกอบ $k$ ตามทฤษฎีบทของเบส์
> $$\gamma_{ik} = \frac{\pi_k\, \mathcal{N}(\mathbf x_i \mid \boldsymbol\mu_k, \boldsymbol\Sigma_k)}{\sum_{j} \pi_j\, \mathcal{N}(\mathbf x_i \mid \boldsymbol\mu_j, \boldsymbol\Sigma_j)}$$

ความรับผิดชอบคือ **การจัดกลุ่มแบบอ่อน (soft clustering)** จุดกลางกลุ่มได้ $\gamma_{ik}$ ใกล้ 1 จุดระหว่างสองกลุ่มอาจได้ 0.5 กับ 0.5 เมทริกซ์ $\boldsymbol\Sigma_k$ ทำให้แต่ละกลุ่มเป็นทรงรีที่เอียงและยืดได้ต่างกัน เคมีนส์คือกรณีพิเศษที่ทุกกลุ่มมี $\boldsymbol\Sigma_k = \sigma^2\mathbf I$ เท่ากันและ $\sigma \to 0$

### ขั้นตอนวิธีคาดหมายและทำให้มากที่สุด (expectation–maximization, EM)
ลอการิทึมของภาวะน่าจะเป็นมีผลรวมอยู่ในลอการิทึม จึงไม่มีผลเฉลยรูปปิด EM สลับทำสองขั้น **ขั้นคาดหมาย (E-step)** คำนวณ $\gamma_{ik}$ จากพารามิเตอร์ปัจจุบัน **ขั้นทำให้มากที่สุด (M-step)** ปรับพารามิเตอร์โดยใช้ $\gamma_{ik}$ เป็นน้ำหนัก ให้ $N_k = \sum_i \gamma_{ik}$ แล้ว
$$\pi_k = \frac{N_k}{N}, \qquad \boldsymbol\mu_k = \frac{1}{N_k}\sum_i \gamma_{ik}\,\mathbf x_i, \qquad \boldsymbol\Sigma_k = \frac{1}{N_k}\sum_i \gamma_{ik}\,(\mathbf x_i - \boldsymbol\mu_k)(\mathbf x_i - \boldsymbol\mu_k)^\top$$
EM รับประกันว่าลอการิทึมของภาวะน่าจะเป็นไม่ลดลงในทุกรอบ แต่ลู่เข้าได้เพียงค่าสูงสุดเฉพาะที่ จึงควรเริ่มหลายครั้งด้วย `n_init` ถ้าส่วนประกอบหนึ่งไปจับจุดเพียงไม่กี่จุด ความแปรปรวนจะหดเข้าหาศูนย์และภาวะน่าจะเป็นพุ่งไม่มีขอบเขต scikit-learn ป้องกันด้วยการบวก `reg_covar` เข้าในเส้นทแยงของ $\boldsymbol\Sigma_k$

เซลล์ถัดไปใช้ GMM กับข้อมูลทรงรีเอียงของส่วนที่ 4 ที่เคมีนส์ตัดผิด

```python
from sklearn.mixture import GaussianMixture

lab_aniso_gmm = GaussianMixture(3, covariance_type='full', n_init=5, random_state=42).fit(X_aniso).predict(X_aniso)
print(f"stretched clusters: K-Means ARI = {adjusted_rand_score(y_aniso, lab_aniso):.3f}, "
      f"GMM ARI = {adjusted_rand_score(y_aniso, lab_aniso_gmm):.3f}")
```

### ตัวอย่าง: สถานะกึ่งเสถียรของพอลิเมอร์
การจำลองพลวัตเชิงโมเลกุลบันทึกโครงรูปของสายพอลิเมอร์เป็นลำดับเฟรม แต่ละเฟรมสรุปด้วย **รัศมีไจเรชัน (radius of gyration)** $R_g$ ซึ่งบอกขนาดโดยรวมของสาย และ **ระยะปลายถึงปลาย (end-to-end distance)** $R_{ee}$ ถ้าสายมีสามสถานะกึ่งเสถียร คือพับแน่น กึ่งกลาง และคลายตัว ข้อมูลในระนาบ $(R_g, R_{ee})$ จะรวมเป็นสามกลุ่มที่ยืดตามแนว $R_{ee}$ และกลุ่มกึ่งกลางซ้อนกับอีกสองกลุ่มบางส่วน โค้ดที่ 7.5 สร้างข้อมูลจำลอง 3,000 เฟรม โดยตั้งประชากรของสามสถานะเป็น 0.5, 0.2 และ 0.3 (หน่วยความยาวพันธะ)

### โค้ดที่ 7.5 — แบบจำลองผสมเกาส์และความรับผิดชอบของข้อมูลพอลิเมอร์จำลอง

```python
from sklearn.mixture import GaussianMixture

def make_polymer_states(n=3000, seed=2026):
    """Toy (Rg, Ree) samples of a chain in three metastable states."""
    rng = np.random.default_rng(seed)
    means = np.array([[2.2, 2.5], [3.5, 6.0], [5.0, 12.0]])   # folded, mid, open
    covs  = [np.array([[0.04, 0.03], [0.03, 0.30]]),
             np.array([[0.10, 0.20], [0.20, 1.20]]),
             np.array([[0.30, 0.90], [0.90, 4.00]])]
    pops  = np.array([0.5, 0.2, 0.3])                          # equilibrium weights
    state = rng.choice(3, size=n, p=pops)
    X = np.array([rng.multivariate_normal(means[s], covs[s]) for s in state])
    return X, state

X_polymer, state_true = make_polymer_states()
gmm = GaussianMixture(n_components=3, covariance_type='full',
                      n_init=10, random_state=42).fit(X_polymer)
labels = gmm.predict(X_polymer)
gamma = gmm.predict_proba(X_polymer)            # responsibilities
uncertain = gamma.max(axis=1) < 0.7             # candidate transition frames
print(gmm.weights_, gmm.means_, uncertain.sum())
```

```python
names = ['folded', 'mid', 'open']
print(f"uncertain frames: {uncertain.sum()} ({100 * uncertain.mean():.1f} %); "
      f"by true state {dict(zip(names, np.bincount(state_true[uncertain], minlength=3).tolist()))}")
for k in range(3):
    w, v = np.linalg.eigh(gmm.covariances_[k])
    print(f"component {k}: mean {gmm.means_[k].round(2)}, principal eigenvector (Rg, Ree) = {np.abs(v[:, -1]).round(3)}")

fig, ax = plt.subplots(figsize=(5.5, 4))
ax.scatter(X_polymer[:, 0], X_polymer[:, 1], c=labels, s=5, cmap='viridis')
ax.scatter(*X_polymer[uncertain].T, c='red', s=12, label='max gamma < 0.7')
ax.set_xlabel('R_g'); ax.set_ylabel('R_ee'); ax.legend(); ax.set_title('toy polymer states (SYNTHETIC)')
plt.show()
```

GMM ได้สัดส่วนผสม 0.497, 0.299 และ 0.204 และค่าเฉลี่ยตรงกับค่าที่ใช้สร้างข้อมูลเกือบพอดี ลำดับของส่วนประกอบไม่จำเป็นต้องตรงกับลำดับของสถานะจริง เพราะป้ายกำกับของกลุ่มเป็นเพียงชื่อที่อัลกอริทึมตั้งให้ ผลนี้ตีความทางฟิสิกส์ได้สี่ทาง

1. การจัดกลุ่มแบบเด็ดขาดให้สถานะกึ่งเสถียรสามสถานะ
2. เฟรมที่ความรับผิดชอบสูงสุดต่ำกว่า 0.7 มี 46 เฟรม (ราวร้อยละ 1.5) โดย 40 เฟรมมาจากสถานะกึ่งกลางและคลาย ซึ่งซ้อนกันมากที่สุด เฟรมเหล่านี้เป็นผู้สมัครของ **กลุ่มสถานะเปลี่ยนผ่าน (transition state ensemble)**
3. ถ้าการจำลองยาวพอจนอยู่ในสมดุล $\pi_k$ ประมาณประชากรในสมดุล และให้ผลต่างของพลังงานอิสระผ่าน $\Delta G = -k_BT \ln(\pi_i/\pi_j)$ (แบบฝึก 7.C ต้นคาบหน้า)
4. เวกเตอร์เจาะจงหลักของ $\boldsymbol\Sigma_k$ ทุกสถานะชี้ไปทางแกน $R_{ee}$ เกือบทั้งหมด สอดคล้องกับที่ปลายสายแกว่งมากกว่าขนาดโดยรวม

> **ความเข้าใจคลาดเคลื่อนที่พบบ่อย:** "ความรับผิดชอบ $\gamma_{ik}$ คือความน่าจะเป็นจริงที่จุดนั้นอยู่ในสถานะ $k$" ที่จริงเป็นความน่าจะเป็น *ภายใต้แบบจำลอง* ถ้าสถานะจริงไม่ได้มีรูปร่างแบบเกาส์ เช่น มีหางยาวหรือเป็นรูปโค้ง $\gamma_{ik}$ จะผิดไปด้วย และ $\gamma_{ik}$ ไม่ได้รวมความไม่แน่นอนของพารามิเตอร์ของแบบจำลองเอง

**คำถามชวนคิด (คำถามทบทวนข้อ 5):** การจัดกลุ่มแบบเด็ดขาดกับแบบอ่อนต่างกันอย่างไร จงยกตัวอย่างทางฟิสิกส์อีกหนึ่งตัวอย่างที่แบบอ่อนให้ข้อมูลมากกว่า (แนวคำตอบ: พิกเซลบนภาพ AFM ในบริเวณที่เนื้อโครงร่างบางและสมบัติคล้ายวัสดุรองรับ ความรับผิดชอบที่ใกล้ 0.5 บอกว่าพิกเซลใดควรตรวจเพิ่มด้วยวิธีอื่น แทนที่จะรายงานเป็นคำตอบที่แน่นอน)

# ส่วนที่ 8 การเลือกจำนวนส่วนประกอบด้วยเกณฑ์สารสนเทศแบบเบส์ (หัวข้อ 7.4.4)

GMM เป็นแบบจำลองความน่าจะเป็นเต็มรูป จึงมีภาวะน่าจะเป็น และใช้ **เกณฑ์สารสนเทศแบบเบส์ (Bayesian information criterion, BIC)** จากบทที่ 5 เลือกจำนวนส่วนประกอบได้โดยตรง
$$\mathrm{BIC} = -2\ln\hat{\mathcal L} + p\ln N$$
พจน์แรกลดลงเมื่อแบบจำลองเข้ากับข้อมูลดีขึ้น พจน์ที่สองเป็นค่าปรับตามจำนวนพารามิเตอร์ $p$ สำหรับ GMM ที่มี $K$ ส่วนประกอบในข้อมูล $D$ มิติและใช้เมทริกซ์ความแปรปรวนร่วมเต็ม $p = (K - 1) + KD + KD(D+1)/2$ ซึ่งเพิ่มเร็วตาม $D$ เราเลือก $K$ ที่ให้ BIC ต่ำที่สุด ใน scikit-learn คำนวณด้วย `gmm.bic(X)`

พารามิเตอร์ `covariance_type` เลือกได้ด้วย BIC เช่นกัน แบบ `'full'` ให้แต่ละส่วนประกอบมีเมทริกซ์ของตัวเองเต็มรูป แบบ `'tied'` ใช้เมทริกซ์เดียวกันทุกส่วนประกอบ แบบ `'diag'` ใช้เฉพาะเส้นทแยง (ทรงรีที่ไม่เอียง) และแบบ `'spherical'` ใช้ทรงกลม

```python
my_bic = []           # ✏️ BIC for K = 1..6
my_kstar_bic = None   # ✏️ K with the lowest BIC
```

### ตัวชี้วัดสามตระกูล
| ตระกูล | ตัวอย่าง | ต้องการ | ใช้เมื่อ |
|---|---|---|---|
| ภายใน (internal) | ซิลูเอตต์ ดาวีส์–โบลดิน คาลินสกี–ฮาราบาซ | ข้อมูลและผลการจัดกลุ่มเท่านั้น | ทุกกรณี |
| ภายนอก (external) | ดัชนีแรนด์ที่ปรับแล้ว สารสนเทศร่วม | ป้ายกำกับจริง ซึ่งหาได้ยาก | ทดสอบวิธีกับข้อมูลจำลองหรือข้อมูลที่รู้คำตอบ |

### ค่าซิลูเอตต์ (silhouette)
ค่าซิลูเอตต์ของจุด $i$ คือ
$$s(i) = \frac{b(i) - a(i)}{\max\{a(i), b(i)\}}$$
เมื่อ $a(i)$ คือระยะเฉลี่ยจากจุด $i$ ไปยังจุดอื่นในกลุ่มเดียวกัน (ความกระชับ) และ $b(i)$ คือระยะเฉลี่ยไปยังจุดในกลุ่มอื่นที่ใกล้ที่สุด (การแยกตัว) ค่าใกล้ 1 หมายความว่าจุดอยู่ลึกในกลุ่มของตัวเอง ค่าใกล้ 0 หมายความว่าอยู่บนขอบระหว่างสองกลุ่ม และค่าติดลบหมายความว่าอยู่ใกล้กลุ่มอื่นมากกว่า ซึ่งบ่งว่าอาจจัดกลุ่มผิด ค่าเฉลี่ยของ $s(i)$ ทุกจุดใช้เป็นคะแนนของการจัดกลุ่มทั้งชุด

### โค้ดที่ 7.6 — การใช้ค่าซิลูเอตต์เลือกจำนวนกลุ่ม

```python
from sklearn.metrics import silhouette_score, silhouette_samples

sil_scores = []
for k in range(2, 8):
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    labels_k = km.fit_predict(X_scaled)
    sil_scores.append(silhouette_score(X_scaled, labels_k))
best_k = 2 + np.argmax(sil_scores)
```

```python
for k, s in zip(range(2, 8), sil_scores):
    print(f"K = {k}: silhouette {s:.3f}")
print("best K =", best_k)
```

การคำนวณค่าซิลูเอตต์ใช้ระยะระหว่างทุกคู่ของจุด ซึ่งเพิ่มตาม $N^2$ สำหรับข้อมูลหลายหมื่นจุด เช่น พิกเซลของภาพ ควรใช้อาร์กิวเมนต์ `sample_size` หรือคำนวณจากตัวอย่างสุ่มบางส่วน (ส่วนที่ 12 ใช้ตัวอย่างสุ่ม 5,000 พิกเซล) ค่า $s(i)$ รายจุดจาก `silhouette_samples` มีประโยชน์ไม่น้อยกว่าค่าเฉลี่ย ถ้าวาดกลับบนภาพ จุดที่มีค่าต่ำมักอยู่ตามรอยต่อระหว่างกลุ่ม ในข้อมูล AFM จำลองของส่วนที่ 12 พิกเซลที่ค่าซิลูเอตต์ติดลบภายใต้เคมีนส์ที่ $k = 3$ ร้อยละ 93 อยู่ติดรอยต่อระหว่างกลุ่มบนภาพ เทียบกับร้อยละ 42 ของพิกเซลทั้งหมด

เซลล์ถัดไปเปรียบเทียบค่าซิลูเอตต์กับ ARI ของสองการแบ่งบนรูปพระจันทร์เสี้ยวของส่วนที่ 5 คือเคมีนส์ที่ $K = 2$ ซึ่งแบ่งผิด และ DBSCAN ที่ $\varepsilon = 0.3$ ซึ่งแบ่งเกือบถูก (คำนวณค่าซิลูเอตต์เฉพาะจุดที่ไม่ใช่จุดรบกวน)

```python
lab_db = DBSCAN(eps=0.3, min_samples=5).fit_predict(X_moon); ok = lab_db >= 0
print(f"K-Means K=2     : ARI {adjusted_rand_score(y_moon, km2):.3f}   silhouette {silhouette_score(X_moon, km2):.3f}")
print(f"DBSCAN eps=0.3  : ARI {adjusted_rand_score(y_moon, lab_db):.3f}   silhouette {silhouette_score(X_moon[ok], lab_db[ok]):.3f}")
```

# ส่วนที่ 10 ดาวีส์–โบลดิน คาลินสกี–ฮาราบาซ สถิติช่องว่าง และการตีความ (หัวข้อ 7.5.3–7.5.4)

| ตัวชี้วัด | ทิศทาง | อคติที่ควรรู้ |
|---|---|---|
| ซิลูเอตต์ | สูงดี | เอื้อต่อกลุ่มนูนและการแยกกลุ่มเล็กที่อยู่ไกล |
| ดาวีส์–โบลดิน | ต่ำดี | ใช้จุดศูนย์กลาง จึงเอื้อต่อกลุ่มนูน |
| คาลินสกี–ฮาราบาซ | สูงดี | เอื้อต่อกลุ่มทรงกลม และแม้ข้อมูลไม่มีกลุ่มเลย ค่านี้ก็อาจเพิ่มหรือลดตาม $K$ ขึ้นกับจำนวนมิติ |
| สถิติช่องว่าง | ตามกฎการเลือก | ขึ้นกับการเลือกข้อมูลอ้างอิง และคำนวณนาน |

**ดัชนีดาวีส์–โบลดิน (Davies–Bouldin index)**
$$DB = \frac{1}{K}\sum_{k=1}^{K} \max_{j\ne k} \frac{\sigma_k + \sigma_j}{d(\boldsymbol\mu_k, \boldsymbol\mu_j)}$$
เมื่อ $\sigma_k$ คือระยะเฉลี่ยจากจุดในกลุ่ม $k$ ถึงจุดศูนย์กลาง สำหรับแต่ละกลุ่ม ดัชนีหากลุ่มที่ "น่ากังวลที่สุด" คือกลุ่มที่กระจายกว้างเมื่อเทียบกับระยะห่างระหว่างจุดศูนย์กลาง **ดัชนีคาลินสกี–ฮาราบาซ (Calinski–Harabasz index)** เป็นอัตราส่วนระหว่างความแปรปรวนระหว่างกลุ่มกับความแปรปรวนภายในกลุ่ม ปรับด้วยองศาเสรี คล้ายสถิติ $F$ ในการวิเคราะห์ความแปรปรวน

```python
from sklearn.metrics import davies_bouldin_score, calinski_harabasz_score

print(" K   silhouette   Davies-Bouldin   Calinski-Harabasz   (blobs of code 7.1)")
for k in range(2, 8):
    lab_k = KMeans(k, n_init=10, random_state=42).fit_predict(X_scaled)
    print(f"{k:2d}   {silhouette_score(X_scaled, lab_k):9.3f}   {davies_bouldin_score(X_scaled, lab_k):14.3f}   "
          f"{calinski_harabasz_score(X_scaled, lab_k):17.0f}")
```

กับข้อมูลที่มีสามกลุ่มชัดเจน ตัวชี้วัดทั้งสามชี้ไปที่ $K = 3$ ตรงกัน แต่ตัวชี้วัดทั้งสามนิยามไว้เฉพาะเมื่อ $K \ge 2$ จึงบอกไม่ได้ว่าข้อมูล **ไม่มีกลุ่มเลย**

**สถิติช่องว่าง (gap statistic)** เปรียบเทียบ $\ln W_K$ ของข้อมูลจริงกับค่าเฉลี่ยของ $\ln W_K$ จากข้อมูลอ้างอิงที่ไม่มีกลุ่ม ซึ่งสุ่มอย่างสม่ำเสมอในกล่องที่ครอบข้อมูล
$$\mathrm{Gap}(K) = \mathbb{E}_{\text{ref}}[\ln W_K] - \ln W_K$$
กฎการเลือกคือใช้ $K$ ที่เล็กที่สุดซึ่ง $\mathrm{Gap}(K) \ge \mathrm{Gap}(K+1) - s_{K+1}$ เมื่อ $s_{K+1}$ คือส่วนเบี่ยงเบนมาตรฐานของค่าจากข้อมูลอ้างอิง (คูณด้วย $\sqrt{1 + 1/B}$ เมื่อใช้ข้อมูลอ้างอิง $B$ ชุด) scikit-learn ไม่มีสถิติช่องว่างในตัว เซลล์ถัดไปเขียนฟังก์ชันไว้ให้

```python
def gap_k(Xd, kmax=6, B=20, seed=0):
    """Tibshirani et al. (2001): smallest K with Gap(K) >= Gap(K+1) - s(K+1); uniform box reference."""
    r = np.random.default_rng(seed); lo, hi = Xd.min(0), Xd.max(0); G, S = [], []
    for k in range(1, kmax + 1):
        w = np.log(KMeans(k, n_init=3, random_state=42).fit(Xd).inertia_)      # same n_init as the reference sets
        ref = [np.log(KMeans(k, n_init=3, random_state=42).fit(r.uniform(lo, hi, Xd.shape)).inertia_) for _ in range(B)]
        G.append(np.mean(ref) - w); S.append(np.std(ref) * np.sqrt(1 + 1 / B))
    return next((k for k in range(1, kmax) if G[k - 1] >= G[k] - S[k]), kmax)
```

# ส่วนที่ 11 AFM: บริบท ลักษณะเด่น และการเตรียมข้อมูล (หัวข้อ 7.6.1–7.6.2)

### ที่มาและบริบท
โครงร่างรองรับเซลล์เป็นวัสดุพรุนที่ใช้เป็นโครงให้เซลล์ยึดเกาะและเจริญเติบโต สมบัติเชิงกลที่ผิวมีผลต่อการยึดเกาะและการเปลี่ยนแปลงของเซลล์กล้ามเนื้อ ผู้วิจัยฝังโครงร่างในพาราฟิน ตัดเป็นแผ่นบางราว 20 ไมโครเมตร วางบนสไลด์แก้ว แล้วใช้ AFM กดปลายเข็มลงบนผิวเป็นตาราง $128 \times 128$ จุด ครอบคลุม $100 \times 100$ ไมโครเมตร (ราว 0.78 ไมโครเมตรต่อพิกเซล) ที่ทุกจุดได้เส้นโค้งแรงกับระยะหนึ่งเส้น (บทที่ 3) ภาพที่ได้มีทั้งเนื้อโครงร่างและวัสดุรองรับ คือพาราฟินหรือแก้วที่โผล่ตรงรอยฉีกของภาคตัด **งานหลักคือแบ่งภาพเป็นสองประเภท: โครงร่างกับวัสดุรองรับ**

ในงานจริง ผู้วิจัยสร้าง **แผนที่อ้างอิง** ด้วยมือโดยเทียบกับภาพจากกล้องจุลทรรศน์แสงและภาพเอกซเรย์คอมพิวเตอร์ระดับไมโคร แผนที่นี้ใช้ **ประเมิน** ผลเท่านั้น ไม่ได้ใช้ในการจัดกลุ่มหรือการเลือกจำนวนกลุ่ม และพื้นที่ส่วนใหญ่ของภาพเป็นโครงร่าง (ในข้อมูลจำลองราวร้อยละ 71) ความไม่สมดุลนี้เป็นกับดักเดียวกับที่บทที่ 6 เตือนไว้

### ลักษณะเด่น (ชุด HSA)
| ลักษณะเด่น | หน่วย | ที่มาและความหมาย |
|---|---|---|
| ความสูง $H$ | nm | ตำแหน่ง $z$ ที่ปลายเข็มเริ่มสัมผัสผิว เนื้อโครงร่างมักนูนสูงกว่าพาราฟินและแก้ว |
| ความแข็งสัมผัส $S$ | N/m | ความชันของเส้นโค้งแรงในช่วงกดเข้า |
| แรงยึดติด $A$ | nN | แรงดึงสูงสุดขณะถอยปลายเข็ม ขึ้นกับเคมีของผิวและพื้นที่สัมผัส |

> **ความเข้าใจคลาดเคลื่อนที่พบบ่อย:** "ความชันของเส้นโค้งแรงคือมอดุลัสของวัสดุในหน่วยอื่น" ความชันในช่วงสัมผัสขึ้นกับความแข็งของวัสดุ รูปร่างและรัศมีของปลายเข็ม ความลึกที่กด และความแข็งของคานเข็มเอง บนผิวที่แข็งมากอย่างแก้ว ความชันที่วัดได้เข้าใกล้ค่าคงที่สปริงของคานเข็มแทนที่จะสะท้อนความแข็งของแก้ว การแปลงเป็นมอดุลัสต้องใช้แบบจำลองการสัมผัส เช่น แบบจำลองของเฮิรตซ์ และรู้เรขาคณิตของปลายเข็ม $S$ จึงไม่ใช่มอดุลัสของยัง และไม่ควรรายงานในหน่วยปาสคาล

### การเตรียมข้อมูลสามขั้น
1. **พิกเซลที่ไม่มีค่า** (การวิเคราะห์เส้นโค้งล้มเหลว) ต้องทำเครื่องหมายและตัดออก ไม่ใช่เติมค่าเฉลี่ย (บทที่ 3)
2. **การปรับมาตราส่วน** ข้อมูล AFM มีการแจกแจงแบบหางยาว การปรับด้วยค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐานจะถูกค่าสุดโต่งดึง งานนี้จึงใช้ `RobustScaler` ซึ่งลบด้วยค่ามัธยฐานแล้วหารด้วยพิสัยระหว่างควอร์ไทล์ และปรับด้วยค่าเดียวจากพิกเซลของทุกภาพรวมกัน
3. **การเลือกลักษณะเด่น** งานจริงลองทุกชุดย่อยของ H, S และ A ทั้งเจ็ดชุด เพราะไม่รู้ล่วงหน้าว่าช่องใดให้ข้อมูลมากที่สุด

ขั้นที่ดูเล็กน้อยเปลี่ยนผลได้มาก การแปลงแบบไม่เป็นเชิงเส้น เช่น การบีบค่าด้วยฟังก์ชันซิกมอยด์ก่อนเข้าไปป์ไลน์ เปลี่ยนระยะทางระหว่างพิกเซล และตัวปรับมาตราส่วนที่ตามมาแก้คืนไม่ได้ เพราะการปรับมาตราส่วนเป็นการแปลงเชิงเส้น ส่วนที่ 12 จะวัดผลนี้บนข้อมูลจำลอง บทเรียนคือ **เก็บข้อมูลในหน่วยทางฟิสิกส์เดิมไว้เสมอ และบันทึกทุกการแปลงพร้อมพารามิเตอร์** (บทที่ 2)

# ส่วนที่ 12 AFM: เปรียบเทียบวิธีจัดกลุ่มสามวิธี (หัวข้อ 7.6.3)

### การออกแบบการเปรียบเทียบ
1. **วิธีจัดกลุ่มสามวิธี** ที่มีสมมติฐานต่างกัน: เคมีนส์ GMM และวิธีของวอร์ด วิธีที่อาศัยความหนาแน่นไม่อยู่ในการเปรียบเทียบ เพราะเนื้อโครงร่างกับวัสดุรองรับต่อกันผ่านพิกเซลที่ขอบซึ่งถูกปลายเข็มเบลอ ซึ่งเป็นเงื่อนไขของการต่อเป็นโซ่ (ส่วนที่ 5 และท้ายส่วนนี้)
2. **จำนวนกลุ่ม** กวาด $k = 2$ ถึง 5 และเลือกด้วยตัวชี้วัดภายในเท่านั้น
3. **กฎที่แปลงกลุ่มเป็นคำตอบสองประเภท** วิธีจัดกลุ่มคืนกลุ่มที่ไม่มีชื่อ งานนี้ใช้กฎสองแบบและรายงานทั้งคู่ **ผู้ชนะได้ทั้งหมด (winner-take-all, WTA)** ให้เฉพาะกลุ่มที่มีสัดส่วนพิกเซลโครงร่างสูงที่สุดเป็นโครงร่าง และ **การรวมแบบอ่อน (soft)** ให้ทุกกลุ่มที่มีพิกเซลโครงร่างตั้งแต่ครึ่งหนึ่งขึ้นไปเป็นโครงร่าง กฎทั้งสองใช้แผนที่อ้างอิง จึงใช้ได้เฉพาะในขั้นประเมิน กับภาพใหม่ต้องใช้กฎทางฟิสิกส์แทน เช่น กลุ่มที่ความสูงเฉลี่ยสูงที่สุดคือโครงร่าง
4. **การประเมินด้วยแผนที่อ้างอิง** การจัดกลุ่มเป็นสองประเภทเป็นการจำแนกโดยปริยาย จึงใช้ตัวชี้วัดจากบทที่ 6 ได้ งานจริงรายงานคะแนน F1 คู่กับ **ค่าสัมประสิทธิ์สหสัมพันธ์ของแมทธิวส์ (Matthews correlation coefficient, MCC)**
$$\mathrm{MCC} = \frac{\mathrm{TP}\cdot\mathrm{TN} - \mathrm{FP}\cdot\mathrm{FN}}{\sqrt{(\mathrm{TP}+\mathrm{FP})(\mathrm{TP}+\mathrm{FN})(\mathrm{TN}+\mathrm{FP})(\mathrm{TN}+\mathrm{FN})}}$$
ซึ่งใช้ทั้งสี่ช่องของเมทริกซ์ความสับสน มีค่าตั้งแต่ $-1$ ถึง 1 และเป็นศูนย์สำหรับการทายแบบไม่ใช้ข้อมูล และ **ความแม่นยำแบบสมดุล (balanced accuracy)** ซึ่งเป็นค่าเฉลี่ยของประสิทธิภาพในการเก็บทั้งสองประเภท และเป็น 0.5 สำหรับการทายแบบไม่ใช้ข้อมูล

วิธีของวอร์ดต้องเก็บระยะระหว่างทุกคู่ของจุด ซึ่งสำหรับพิกเซลหลายหมื่นพิกเซลต้องใช้หน่วยความจำมาก โค้ดจึงสร้างแผนภาพต้นไม้จากพิกเซลที่สุ่มมา 5,000 พิกเซล แล้วกำหนดกลุ่มให้พิกเซลที่เหลือตามจุดศูนย์กลางที่ใกล้ที่สุด

### โค้ดที่ 7.7 — แผนที่แรงของ AFM จำลองและการเปรียบเทียบวิธีจัดกลุ่มสามวิธี
⏱ เซลล์นี้ใช้เวลาราวหนึ่งนาทีบน Colab เพราะปรับ GMM สิบครั้งต่อค่า $k$ บนพิกเซล 16,384 พิกเซล

```python
from scipy.ndimage import gaussian_filter
from sklearn.preprocessing import RobustScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.neighbors import NearestCentroid
from sklearn.metrics import f1_score, matthews_corrcoef

def make_scaffold_map(n=128, seed=2026):
    """Toy AFM force map: scaffold (label 1) vs substrate (paraffin or glass).
    Channels: height H (nm), contact stiffness S (N/m), adhesion A (nN)."""
    rng = np.random.default_rng(seed)
    phi = gaussian_filter(rng.normal(size=(n, n)), sigma=6); phi /= phi.std()
    scaffold = phi > -0.55                                    # about 70 % scaffold
    g = gaussian_filter(rng.normal(size=(n, n)), sigma=4); g /= g.std()
    glass = (~scaffold) & (g > 0.8)                           # exposed glass
    het = gaussian_filter(rng.normal(size=(n, n)), sigma=3); het /= het.std()
    H = np.where(scaffold, 250 + 72 * het, np.where(glass, 20.0, 130.0))
    S = np.where(scaffold, 0.8 + 0.18 * het, np.where(glass, 1.5, 0.5))
    A = np.where(scaffold, 4.5 - 0.72 * het, np.where(glass, 2.5, 6.5))
    H, S, A = (gaussian_filter(m, sigma=1.0) for m in (H, S, A))   # tip blur
    heavy = lambda s: s * rng.standard_t(df=3, size=(n, n))        # heavy tails
    H += heavy(30.0); S += heavy(0.09); A += heavy(0.65)
    X = np.column_stack([H.ravel(), S.ravel(), A.ravel()])
    return X, scaffold.ravel().astype(int), glass.ravel()

X_afm, y_afm, glass_afm = make_scaffold_map()
Xa = RobustScaler().fit_transform(X_afm)

def ward_labels(k, n_sub=5000, seed=42):
    """Ward on a subsample, then assign every pixel to the nearest centroid."""
    idx = np.random.default_rng(seed).choice(len(Xa), n_sub, replace=False)
    lab = AgglomerativeClustering(n_clusters=k, linkage='ward').fit_predict(Xa[idx])
    return NearestCentroid().fit(Xa[idx], lab).predict(Xa)

algorithms = {
    'K-Means': lambda k: KMeans(k, n_init=20, random_state=42).fit_predict(Xa),
    'GMM':     lambda k: GaussianMixture(k, covariance_type='full', n_init=10,
                                         random_state=42).fit(Xa).predict(Xa),
    'Ward':    ward_labels,
}
sub = np.random.default_rng(0).choice(len(Xa), 5000, replace=False)
for name, fit in algorithms.items():
    for k in range(2, 6):
        lab = fit(k)
        share = {c: y_afm[lab == c].mean() for c in np.unique(lab)}  # evaluation only
        wta  = (lab == max(share, key=share.get)).astype(int)
        soft = np.isin(lab, [c for c, v in share.items() if v >= 0.5]).astype(int)
        print(f"{name:8s} k={k} sil={silhouette_score(Xa[sub], lab[sub]):.3f} "
              f"WTA F1={f1_score(y_afm, wta):.3f} MCC={matthews_corrcoef(y_afm, wta):.3f} "
              f"soft F1={f1_score(y_afm, soft):.3f} MCC={matthews_corrcoef(y_afm, soft):.3f}")
```

### ดูข้อมูลก่อนตีความ
เซลล์ถัดไปวาดแผนที่ของลักษณะเด่นทั้งสาม แผนที่อ้างอิง (1 = โครงร่าง, 0.5 = แก้ว, 0 = พาราฟิน) และผลของเคมีนส์ที่ $k = 3$ แล้ววัดความยาวของหางของการแจกแจง โดยเทียบค่าเบี่ยงเบนจากมัธยฐานที่เปอร์เซ็นไทล์ 99.9 กับมัธยฐานของค่าเบี่ยงเบน การแจกแจงเกาส์ให้อัตราส่วนนี้ราว 4.9

```python
print(f"scaffold fraction {y_afm.mean():.3f}, exposed glass {glass_afm.mean():.3f}, paraffin {1 - y_afm.mean() - glass_afm.mean():.3f}")
for j, ch in enumerate('HSA'):
    dev = np.abs(X_afm[:, j] - np.median(X_afm[:, j]))
    print(f"{ch}: 99.9th percentile of |x - median| / median of |x - median| = {np.percentile(dev, 99.9) / np.median(dev):.1f}  (Gaussian: 4.9)")

lab3 = KMeans(3, n_init=20, random_state=42).fit_predict(Xa)
fig, axes = plt.subplots(1, 5, figsize=(17, 3.4))
for ax, img, title in zip(axes, [X_afm[:, 0], X_afm[:, 1], X_afm[:, 2], y_afm + glass_afm * 0.5, lab3],
                          ['H (nm)', 'S (N/m)', 'A (nN)', 'reference', 'K-Means k=3']):
    img = np.asarray(img, float)
    lo, hi = np.percentile(img, [1, 99])      # heavy tails would wash out the colour scale
    im = ax.imshow(img.reshape(128, 128), cmap='viridis', vmin=lo, vmax=hi); ax.set_title(title); ax.axis('off')
    fig.colorbar(im, ax=ax, fraction=0.046)
fig.suptitle('toy AFM force map (SYNTHETIC)'); fig.tight_layout(); plt.show()
```

ทุกช่องมีหางยาวกว่าการแจกแจงเกาส์ (อัตราส่วนราว 6 ถึง 8 เทียบกับ 4.9) ซึ่งเป็นเหตุผลของการใช้ `RobustScaler` เซลล์ถัดไปวัดผลของการบีบค่าด้วยซิกมอยด์สามรูปแบบก่อน `RobustScaler` โดยเทียบผลของเคมีนส์ที่ $k = 3$ กับผลเดิมด้วย ARI

```python
for name, f in [('(x - median)/std', lambda Z: (Z - np.median(Z, 0)) / Z.std(0)),
                ('(x - mean)/MAD', lambda Z: (Z - Z.mean(0)) / np.median(np.abs(Z - np.median(Z, 0)), 0)),
                ('(x - median)/(IQR/4)', lambda Z: (Z - np.median(Z, 0)) / (0.25 * np.subtract(*np.percentile(Z, [75, 25], 0))))]:
    l = KMeans(3, n_init=20, random_state=42).fit_predict(RobustScaler().fit_transform(1 / (1 + np.exp(-f(X_afm)))))
    print(f"sigmoid of {name:22s}: ARI vs untransformed K-Means k=3 = {adjusted_rand_score(lab3, l):.3f}")
```

การแปลงที่ดูเหมือนเป็นเพียง "การทำให้ค่าอยู่ในช่วงศูนย์ถึงหนึ่ง" เปลี่ยนผลการจัดกลุ่มจน ARI เทียบกับผลเดิมเหลือ 0.83 ถึง 0.95 ขึ้นกับความกว้างของซิกมอยด์

```python
LAB, rows = {}, []
for name, fit in algorithms.items():
    for k in range(2, 6):
        LAB[(name, k)] = lab_k = fit(k)
        rows.append((name, k, silhouette_score(Xa[sub], lab_k[sub]), davies_bouldin_score(Xa, lab_k),
                     calinski_harabasz_score(Xa, lab_k)))
internal = pd.DataFrame(rows, columns=['method', 'k', 'silhouette (high)', 'DB (low)', 'CH (high)']).round(3)
internal['CH (high)'] = internal['CH (high)'].round(0).astype(int)
print(internal.to_string(index=False))
```

3. หาเปอร์เซ็นไทล์ที่ 99 ของระยะเพื่อนบ้านลำดับที่ 5 (`kd` จากส่วนที่ 5) เก็บใน `my_kd99` แล้วอธิบายว่ากราฟระยะเพื่อนบ้านแนะนำ $\varepsilon$ ใด และทำไม $\varepsilon = 0.35$ จึงใหญ่เกินไป

เติมสองบรรทัดในลูปของเซลล์ถัดไป เซลล์จะเก็บค่าเฉลี่ยของ ARI ใน `my_stab[(name, k)]` และนับพิกเซลที่ถูกจัดเข้ากลุ่มเดิมน้อยกว่า 18 ใน 20 ครั้ง (ใช้ `align` จับคู่ชื่อกลุ่มด้วยวิธีฮังการี) ⏱ เมื่อเติมแล้วเซลล์ใช้เวลาราวสามนาทีบน Colab จากนั้นอธิบายว่าพิกเซลที่ไม่เสถียรอยู่ที่ใด และความเสถียรบอกอะไรได้หรือไม่ได้

```python
from scipy.optimize import linear_sum_assignment
from sklearn.metrics import confusion_matrix

def align(ref, lab, k):
    """Relabel lab so that its clusters match ref as well as possible (Hungarian matching)."""
    C = confusion_matrix(ref, lab, labels=range(k))
    r, c = linear_sum_assignment(-C)
    return np.array([dict(zip(c, r))[v] for v in range(k)])[lab]

def model(name, k, seed):
    if name == 'K-Means':
        return KMeans(k, n_init=20, random_state=seed)
    return GaussianMixture(k, covariance_type='full', n_init=10, random_state=seed)

rng_b = np.random.default_rng(2026)
boots = [rng_b.choice(len(Xa), len(Xa), replace=True) for _ in range(20)]
my_stab, unstable_k3 = {}, None
for name, k in [('K-Means', 2), ('K-Means', 3), ('K-Means', 4), ('K-Means', 5), ('GMM', 3)]:
    ref = model(name, k, 42).fit(Xa).predict(Xa)
    ari, same = [], np.zeros(len(Xa))
    for b, idx in enumerate(boots):
        lab_b = None   # ✏️ fit model(name, k, 1000 + b) on Xa[idx], then predict every pixel of Xa
        if lab_b is None:
            break
        ari.append(None)   # ✏️ replace None by the ARI between ref and lab_b
        same += align(ref, lab_b, k) == ref
    if len(ari) == len(boots):
        my_stab[(name, k)] = np.mean(ari)
        print(f"{name:7s} k={k}: ARI mean {np.mean(ari):.3f}  min {np.min(ari):.3f}  unstable pixels {100 * (same < 18).mean():.1f} %")
        if (name, k) == ('K-Means', 3):
            unstable_k3, ref_k3 = same < 18, ref
if not my_stab:
    print("(ยังไม่ได้เติมโค้ด)")
```

การจัดกลุ่มเปลี่ยนคำถามจากการทำนายคำตอบที่รู้อยู่แล้ว ไปเป็นการค้นหาโครงสร้างที่ยังไม่มีใครรู้ วิธีจัดกลุ่มทุกวิธีคืนกลุ่มออกมาเสมอ ภาระในการตัดสินว่ากลุ่มมีความหมายหรือไม่จึงตกอยู่กับผู้วิเคราะห์ การเตรียมลักษณะเด่น การปรับมาตราส่วน และการเลือกระยะทางมีผลต่อคำตอบโดยตรง และเมื่อมิติสูงมาก ระยะทางเองจะเสียความหมาย

วิธีสามตระกูลมีนิยามของกลุ่มต่างกัน เคมีนส์เหมาะกับกลุ่มนูนขนาดใกล้เคียงกันและต้องระบุจำนวนกลุ่ม DBSCAN และ HDBSCAN จับกลุ่มรูปร่างใดก็ได้และแยกจุดรบกวนได้ แต่เสี่ยงต่อการต่อเป็นโซ่ แบบจำลองผสมเกาส์ให้ความรับผิดชอบที่บอกความไม่แน่นอนของแต่ละจุดและเลือกจำนวนกลุ่มด้วย BIC ได้ แต่พยายามอธิบายทุกจุดรวมถึงค่าผิดปกติ ตัวชี้วัดภายในทุกตัวมีนิยามของกลุ่มที่ดีแฝงอยู่ และอาจชี้จำนวนกลุ่มคนละค่า กรณีศึกษา AFM แสดงว่า F1 ของข้อมูลที่ไม่สมดุลต้องเทียบกับการทายประเภทส่วนใหญ่และรายงานคู่กับ MCC การจัดกลุ่มแบ่งตามทิศทางที่ข้อมูลกระจายมากที่สุดซึ่งอาจไม่ใช่คำถามของเรา และการเลือกการตั้งค่าใดด้วยแผนที่อ้างอิงทำให้ผลสูงเกินจริง การตั้งค่าจึงควรกำหนดไว้ล่วงหน้าและรายงานผลทุกการตั้งค่า
